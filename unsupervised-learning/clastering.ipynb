Կլաստերավորումը (Clustering) մեքենայական ուսուցման (Machine Learning) մեթոդ է, որը վերաբերում է «չվերահսկվող ուսուցմանը» (Unsupervised Learning):

Ահա թե ինչ է դա նշանակում պարզ լեզվով.

Խնդիրը՝ Պատկերացնենք, որ ունենք մի մեծ տուփ՝ լցված տարբեր իրերով, և ոչ մի ցուցում չկա, թե որն ինչ է։ Մենք չգիտենք իրերի անունները կամ դասակարգումը (սրանք «չմակնշված տվյալներն» են ):

Կլաստերավորման ալգորիթմը նայում է այդ իրերին և փորձում գտնել դրանց միջև ընդհանրություններ (օրինակ՝ ձևը, գույնը կամ չափսը):

 Ալգորիթմը խմբավորում է նմանատիպ իրերը իրար կողքի: Արդյունքում դուք ստանում եք մի քանի խմբեր (կլաստերներ)։


Այս մեթոդը լայնորեն կիրառվում է իրական կյանքում՝

Մարքեթինգում՝ Հաճախորդներին խմբավորելու ըստ գնումների վարքագծի (օրինակ՝ «ակտիվ գնորդներ», «հազվադեպ գնորդներ»):

Բժշկության մեջ՝ Հիվանդների խմբավորումը՝ ըստ ախտանիշների նմանության:

Պատկերների մշակման մեջ՝ Նույնանման օբյեկտների առանձնացումը լուսանկարներից:

Կարճ ասած կլաստերավորումը օգնում է «հասկանալ» տվյալների կառուցվածքը՝ առանց դրանք նախապես դասակարգելու։



Այս պատկերը հիանալի կերպով ցույց է տալիս, թե ինչպես է կլաստերավորումը (clustering) «քաոսից» ստեղծում կարգուկանոն: Այն տեսողականորեն բացատրում է, թե ինչպես են տվյալների անկանոն կետերը վերածվում կարգավորված խմբերի։

Ահա թե ինչ է կատարվում այս գործընթացում՝

Ձախ կողմը (Original Data). Սա մեր սկզբնական, «չմակնշված» տվյալներն են։ Կետերը ցրված են՝ առանց որևէ ակնհայտ կառուցվածքի։ Մենք տեսնում ենք կետերի հավաքածու, բայց դրանք ոչ մի կերպ դասակարգված չեն:

Աջ կողմը (Clustered Data). Ահա այստեղ է կլաստերավորման իմաստը։ Ալգորիթմը վերլուծել է կետերի միջև եղած հեռավորությունը (նմանությունը) և դրանք բաժանել երեք հստակ գունավոր խմբերի (կլաստերների):

Որպեսզի համակարգիչը կարողանա տարբերակել լավ և վատ կլաստերները, մենք օգտագործում ենք մաթեմատիկական «կանոններ»։ Ահա հիմնական սկզբունքները, որոնցով գնահատվում է կլաստերավորման որակը՝

«Լավ» կլաստերավորման չափանիշները։

Գոյություն ունի երկու հիմնական հասկացություն, որոնցով չափվում է կլաստերի որակը՝

Խտություն (Cohesion): Սա ցույց է տալիս, թե որքան «մոտ» են նույն կլաստերի կետերը իրար։ Լավ կլաստերում բոլոր կետերը հավաքված են իրար շատ մոտ (մինիմալ հեռավորություն կենտրոնից):

Տարանջատում (Separation): Սա ցույց է տալիս, թե որքան «հեռու» են տարբեր կլաստերները միմյանցից։ Լավ կլաստերավորման դեպքում խմբերը հստակ առանձնացված են և չեն «խառնվում» իրար։

Ինչպե՞ս ենք մենք սահմանում այդ «կանոնները»։
Գործնականում մենք օգտագործում ենք որոշակի մետրիկներ (չափումներ)՝ հասկանալու համար, թե որքան լավ է աշխատում ալգորիթմը.

WCSS (Within-Cluster Sum of Squares): Սա հաշվարկում է, թե որքան է յուրաքանչյուր կետի հեռավորությունը իր խմբի կենտրոնից։ Որքան փոքր է այս թիվը, այնքան ավելի «խիտ» են մեր խմբերը։

Արմունկի մեթոդը (Elbow Method): Սա ամենատարածված տեխնիկան է։ Մենք աստիճանաբար մեծացնում ենք կլաստերների քանակը և դիտարկում WCSS-ի արժեքը։ Սկզբում այն կտրուկ ընկնում է, իսկ հետո դանդաղում՝ ստեղծելով «արմունկի» տեսք։ Այդ «արմունկի» կետը համարվում է կլաստերների օպտիմալ քանակը։

Silhouette Score: Սա ցույց է տալիս, թե որքան լավ է յուրաքանչյուր կետ «նստած» իր խմբում՝ համեմատած այլ խմբերի հետ։ Արժեքները տատանվում են -1-ից +1. +1-ին մոտ  արժեքը նշանակում է շատ լավ, հստակ տարանջատված կլաստերներ։





Ինչ է անում կլաստերավորումը։

1. Տվյալների միջև կապերի բացահայտում (Discover relationships)
Կլաստերավորումը թույլ է տալիս տեսնել այնպիսի կապեր, որոնք անզեն աչքով անտեսանելի են։ Օրինակ, եթե դուք ունեք առցանց խանութի գնորդների տվյալներ, կլաստերավորումը կարող է ցույց տալ, որ «ուսանողները» և «երիտասարդ մասնագետները» գնումների տարբեր նախասիրություններ ունեն, նույնիսկ եթե դուք նրանց նախապես պիտակներ չեք տվել։

2. Ինտուիցիայի զարգացում (Intuition building)
Սա հաճախ կոչվում է Exploratory Data Analysis (EDA): Նախքան բարդ մոդելներ կառուցելը, տվյալները կլաստերների բաժանելը օգնում է վերլուծաբանին «զգալ» տվյալների բնույթը. արդյո՞ք դրանք բաշխված են հավասարաչափ, թե՞ կան հստակ առանձնացված խմբեր։

3. Վարկածների առաջադրում (Hypothesis generation)
Երբ դուք տեսնում եք կլաստերներ, դուք սկսում եք հարցեր տալ՝ «Ինչո՞ւ են այս կետերը միասին» կամ «Ինչո՞ւ է այս կլաստերը հեռու մնացածից»։ Այս դիտարկումները դառնում են նոր վարկածներ, որոնք կարող եք ստուգել հետագա վիճակագրական վերլուծությունների միջոցով։

4. Բարձր չափայնության տվյալների կառուցվածքների բացահայտում (Discover structures in high-dimensional data)
Սա կլաստերավորման ամենահզոր կողմերից է։ Մարդու ուղեղը կարող է պատկերացնել միայն 2 կամ 3 չափում (x, y, z), բայց տվյալները կարող են ունենալ հարյուրավոր հատկանիշներ։ Կլաստերավորումը օգնում է «սեղմել» այդ տվյալները՝ ցույց տալով թաքնված կառուցվածքը։

5. Տվյալների ամփոփում և սեղմում (Summarizing / compressing large data)
Միլիոնավոր տվյալներ ունենալու փոխարեն, դուք կարող եք աշխատել միայն կլաստերների «կենտրոններով» (centroids): Սա մեծապես հեշտացնում է հաշվարկները։ Օրինակ, պատկերների սեղմման ժամանակ նմանատիպ գույները խմբավորվում են մի կլաստերի մեջ, ինչը թույլ է տալիս նվազեցնել ֆայլի չափը՝ առանց որակի զգալի կորստի։


Կլաստերավորման սուբյեկտիվությունը

Կլաստերավորման ամենակարևոր փիլիսոփայական կողմը՝ այն սուբյեկտիվ է:

Կլաստերավորման ժամանակ «ճիշտ» պատասխան չկա: Այն, թե ինչպես կխմբավորեք մարդկանց (կամ տվյալները), ամբողջությամբ կախված է ձեր նպատակից:

Եթե ձեր նպատակը աշխատանքային արդյունավետությունն է, կխմբավորեք ըստ «աշխատանքային փորձի»:

Եթե ձեր նպատակը սոցիալական միջոցառում է, կխմբավորեք ըստ «տարիքի» կամ «նախասիրությունների»:

Ամփոփում
Կլաստերավորումը գործիք է, որն օգնում է մեզ կարգուկանոն գտնել քաոսի մեջ: Այն թույլ է տալիս ոչ թե «սովորել» պատրաստի պատասխանները, այլ սեփական ինտուիցիայով և մաթեմատիկական մոտեցումներով հասկանալ՝ ինչպե՞ս են տվյալները կազմակերպված իրական աշխարհում:

Կլաստերավորման ամենակարևոր և գործնական կիրառություններից մեկը՝ անոմալիաների հայտնաբերումը (Anomaly detection)։

Կլաստերավորման ամենակարևոր և գործնական կիրառություններից մեկը՝ անոմալիաների հայտնաբերումը (Anomaly detection)։

Անոմալիաների հայտնաբերում.Կլաստերավորումն օգտագործվում է տվյալների մեջ անոմալիաներ հայտնաբերելու համար, ինչպիսիք են՝

* ֆինանսական գործարքների մեջ խարդախությունները։

* Ցանցային թրաֆիկի (network traffic) անսովոր օրինաչափությունները։

* Բժշկական տվյալների մեջ արտառոց արդյունքները (outliers)։»

Ինչպե՞ս է սա աշխատում
Կլաստերավորման այս մեթոդը հիմնված է «նորմալի» և «արտառոցի» տարբերակման վրա։

Նորմալ վիճակի սահմանում. Ալգորիթմը սկզբում խմբավորում է տվյալները՝ ստեղծելով «նորմալ» վարքագծի կլաստերներ (օրինակ՝ սովորական գնումներ, սովորական ինտերնետային ակտիվություն կամ առողջ մարդու անալիզների ցուցանիշներ)։

Անոմալիայի բացահայտում. Այն տվյալները, որոնք չեն տեղավորվում այս հիմնական խմբերի մեջ կամ գտնվում են դրանցից շատ հեռու, ալգորիթմը նշում է որպես անոմալիա (արտառոց դեպք)։



Օրինակներ՝

1․

Կլաստերային վերլուծությունը (cluster analysis) կիրառվում կենսաբանության և բժշկության մեջ։

Այն ներկայացնում է հետևյալ գործնական օրինակները՝

Գեների բացահայտում. Կլաստերավորումն օգտագործվում է կոնկրետ հիվանդությունների հետ կապված գեները բացահայտելու համար։

Հիվանդների խմբավորում. Այն թույլ է տալիս խմբավորել նմանատիպ կլինիկական բնութագրեր ունեցող հիվանդներին։

Ախտորոշում և բուժում. Այս վերլուծությունն օգնում է բժիշկներին հիվանդությունների ախտորոշման և բուժման գործընթացներում։


2․
«Պատկերների մշակման մեջ կլաստերային վերլուծությունն օգտագործվում է նմանատիպ հատկություններ ունեցող պիքսելները միասին խմբավորելու համար, ինչը թույլ է տալիս բացահայտել օբյեկտներն ու օրինաչափությունները պատկերներում։ Այլընտրանքային տարբերակով՝ կարելի է որոշել պատկերի հիմնական գույները և օգտագործել դրանք վերագունավորման (recoloring) համար»։

Ի՞նչ է սա նշանակում
Պատկերների մշակման մեջ այս մոտեցումը հաճախ կոչվում է պատկերի սեգմենտավորում (segmentation)։ Ահա թե ինչպես է այն աշխատում․

Օբյեկտների բացահայտում. Համակարգիչը վերցնում է պատկերի բոլոր պիքսելները և խմբավորում դրանք՝ ըստ գույնի, պայծառության կամ հյուսվածքի (texture)։ Սա թույլ է տալիս ծրագրին հասկանալ, որ, օրինակ, կանաչ պիքսելների մի խումբը «ծառ» է, իսկ կապույտը՝ «երկինք»։

Գույների կրճատում (Color Quantization). Կլաստերավորման միջոցով կարելի է նկարի մեջ եղած հազարավոր երանգները խմբավորել մի քանի հիմնական «կլաստերային» գույների մեջ։ Սա օգտակար է նկարը վերագունավորելու, ֆայլի չափսը փոքրացնելու (սեղմելու) կամ պարզեցված գեղարվեստական էֆեկտներ ստանալու համար։


3․

«Շուկայի սեգմենտավորում. Կլաստերային վերլուծությունը մարքեթինգում հաճախ օգտագործվում է հաճախորդներին խմբերի բաժանելու համար՝ հիմնվելով նրանց գնումների վարքագծի, ժողովրդագրական տվյալների կամ այլ հատկանիշների վրա։ Այս տեղեկատվությունը կարող է օգտագործվել թիրախային մարքեթինգային արշավներ ստեղծելու կամ նոր ապրանքներ մշակելու համար, որոնք գրավիչ կլինեն հաճախորդների կոնկրետ խմբերի համար»։

Ինչպե՞ս է սա աշխատում գործնականում
Կլաստերավորումն այս դեպքում ընկերություններին թույլ է տալիս ավելի խելացի աշխատել իրենց հաճախորդների հետ.

Հաճախորդների պրոֆիլավորում. Փոխանակ բոլորին առաջարկելու նույն ապրանքը, ընկերությունը տվյալների (օրինակ՝ տարիք, եկամուտ, գնումների պատմություն) հիման վրա բաժանում է հաճախորդներին խմբերի (կլաստերների)։

Թիրախային արշավներ. Յուրաքանչյուր խմբի համար ստեղծվում է առանձին մարքեթինգային հաղորդագրություն։ Օրինակ՝ ուսանողների համար նախատեսված զեղչերը կարող են տարբերվել պրոֆեսիոնալների համար նախատեսված առաջարկներից։

Ապրանքների զարգացում. Հասկանալով, թե որ խումբն է ավելի ակտիվ, ընկերությունը կարող է մշակել ապրանքներ, որոնք կոնկրետ այդ խմբի կարիքներին են համապատասխանում


Կլաստերավորման երկու հիմնական տեսակները՝

Լղոզված (փափուկ) կլաստերավորում (Fuzzy (Soft) clustering): Այս մոտեցման դեպքում յուրաքանչյուր օբյեկտ պատկանում է յուրաքանչյուր կլաստերին՝ որոշակի հավանականությամբ կամ կշռով (այն կարող է լինել նույնիսկ զրո)։

Ոչ-լղոզված (կոշտ) կլաստերավորում (Non-fuzzy (Hard) clustering): Այս դեպքում յուրաքանչյուր օբյեկտ պատկանում է բացառապես մեկ կլաստերի։

Բացատրություն
Այս տարբերությունը շատ կարևոր է տվյալների վերլուծության ժամանակ՝ կախված նրանից, թե որքան հստակ են սահմանները ձեր տվյալների խմբերի միջև։

Կոշտ (Hard) կլաստերավորում. Սա ամենատարածված ձևն է (օրինակ՝ K-Means ալգորիթմում)։ Այստեղ սահմանները հստակ են․ ամեն մի կետ «կամ այս խմբում է, կամ՝ այն»։ Սա հարմար է այն դեպքերում, երբ օբյեկտները հստակ տարբերակելի են։

Փափուկ (Soft/Fuzzy) կլաստերավորում. Սա ավելի ճկուն մոտեցում է։ Այն օգտագործվում է, երբ սահմանները «մշուշոտ» են, և օբյեկտը չի կարող 100%-ով պատկանել միայն մեկ խմբի։ Օրինակ՝ հաճախորդների վարքագիծը վերլուծելիս մի հաճախորդ կարող է 70%-ով պատկանել «ակտիվ գնորդների» խմբին և 30%-ով՝ «հազվադեպ գնորդների» խմբին։

Այսպիսով, եթե տվյալները հստակ են, օգտագործում ենք կոշտ կլաստերավորում, իսկ եթե տվյալները համընկնում են կամ երկիմաստ են՝ փափուկ կլաստերավորումը ավելի ճշգրիտ պատկեր է տալիս։




Կլաստերավորման կառուցվածքային տեսակներն են ՝ Մասնատող (Partitional) և Հիերարխիկ (Hierarchical) կլաստերավորում։

Մասնատող կլաստերավորում (Partitional clustering). Այս մեթոդը գտնում է կլաստերների հստակ, ֆիքսված քանակ։ Պատկերում տեսնում ենք, որ տվյալները բաժանված են առանձին, իրար չհաջորդող խմբերի՝ առանց ներքին ենթախմբերի։ Սա նման է այնպիսի ալգորիթմների, ինչպիսին է K-Means-ը, որտեղ նախապես որոշում ենք կլաստերների քանակը։

Հիերարխիկ կլաստերավորում (Hierarchical clustering). Այս մեթոդը ստեղծում է կլաստերների շարք, որոնք «ներդրված» են իրար մեջ (nested clusters)։ Պատկերում տեսնում ենք, որ ավելի փոքր կլաստերները գտնվում են ավելի մեծ կլաստերների ներսում՝ ստեղծելով ծառանման կառուցվածք։

Այսպիսով, եթե մեզ անհրաժեշտ է միայն հստակ թվով խմբեր ստանալ, ընտրում ենք մասնատող մեթոդը, իսկ եթե ցանկանում ենք հասկանալ տվյալների մեջ առկա ենթախմբերի կապը՝ հիերարխիկ մեթոդը։





Հիերարխիկ կլաստերավորումը (Hierarchical Clustering) տվյալների խմբավորման մեթոդ է, որը ստեղծում է կլաստերների աստիճանակարգված կառուցվածք։ Ահա այս գործընթացի մանրամասն բացատրությունը՝ հիմնված ձեր տրամադրած նյութի վրա.

Հիմնական սկզբունքները
Վիզուալիզացիա (Դենդրոգրամ). Հիերարխիկ կլաստերավորումը սովորաբար պատկերվում է «դենդրոգրամի» (dendrogram) կամ ծառանման գծապատկերի միջոցով։ Այս գծապատկերը օգնում է հասկանալ, թե ինչպես են տվյալները միավորվում կամ բաժանվում։

Կլաստերների կառուցվածք. Ծառի յուրաքանչյուր ենթաճյուղ (subtree) համապատասխանում է մեկ կոնկրետ կլաստերի։

Հեռավորության ցուցիչ. Ճյուղավորման բարձրությունը ցույց է տալիս օբյեկտների միջև եղած հեռավորությունը։ Եթե երկու օբյեկտ միավորվում են ցածր մակարդակում, նրանք շատ նման են իրար, իսկ եթե միավորվում են բարձր մակարդակում՝ ապա նրանք ավելի տարբեր են։

Կլաստերավորման երկու հիմնական ռազմավարությունները

Հիերարխիկ կլաստերավորումն իրականացվում է երկու հիմնական ուղղությամբ՝

1. Ագլոմերատիվ կլաստերավորում (Agglomerative - Bottom-up)
Սա հիերարխիկ կլաստերավորման ամենատարածված ձևն է։ Այն աշխատում է հետևյալ կերպ.

Մեկնարկը (Singletons). Գործընթացը սկսվում է այն կետից, որտեղ յուրաքանչյուր առանձին տվյալ (օրինակ՝ յուրաքանչյուր պիքսել կամ օբյեկտ) համարվում է առանձին կլաստեր։

Միավորման գործընթացը. Ալգորիթմը փնտրում է երկու ամենամոտ կլաստերները և միավորում դրանք մեկ ընդհանուրի մեջ։

Ռեկուրսիվ կրկնություն. Այս քայլը կրկնվում է անընդհատ՝ աստիճանաբար նվազեցնելով կլաստերների ընդհանուր քանակը, մինչև բոլոր տվյալները միավորվեն մեկ մեծ, ընդհանուր կլաստերի մեջ։

Ինչո՞ւ է սա տարածված. Այն հաշվարկային առումով ավելի պարզ է. յուրաքանչյուր քայլում մենք պարզապես ընտրում ենք «ամենամոտ» զույգը։

2. Դիվիզիվ կլաստերավորում (Divisive - Top-down)
Սա հակառակ ընթացքն է և ավելի հազվադեպ է հանդիպում, քանի որ այն պահանջում է շատ ավելի բարդ հաշվարկներ։

Մեկնարկը. Ալգորիթմը սկսում է մեկ խոշոր կլաստերից, որն իր մեջ ներառում է տվյալների բոլոր առկա կետերը։

Տրոհման գործընթացը. Այնուհետև ալգորիթմը փնտրում է այնպիսի բաժանման գիծ (կամ չափանիշ), որը թույլ կտա այդ մեկ մեծ խումբը տրոհել երկու ավելի փոքր և տրամաբանորեն տարբերվող կլաստերների։

Ռեկուրսիվ կրկնություն. Այս գործընթացը շարունակվում է՝ յուրաքանչյուր ստացված նոր կլաստերը հետագայում բաժանելով ավելի փոքր մասերի, մինչև յուրաքանչյուր կլաստեր դառնա մեկ առանձին օբյեկտ։

Ինչո՞ւ է սա բարդ. Մեկ մեծ կլաստերը ճիշտ տրոհելու բազմաթիվ եղանակներ կան, և ալգորիթմը պետք է գնահատի դրանցից յուրաքանչյուրը, ինչը ավելի շատ ժամանակ է պահանջում, քան ագլոմերատիվ մեթոդը։



Ինչպե՞ս հասկանալ «Դենդրոգրամը» (Ծառանման կառուցվածքը)
Երկու մեթոդներն էլ վիզուալիզացվում են դենդրոգրամի միջոցով, որը հասկանալու համար հարկավոր է ուշադրություն դարձնել հետևյալին.

Ճյուղերի բարձրությունը. Ճյուղի բարձրությունը ցույց է տալիս, թե որքան «տարբեր» են կլաստերները միմյանցից։

Եթե երկու կլաստեր միավորվում են ցածր մակարդակում, նրանք շատ նման են իրար։

Եթե միավորումը կատարվում է շատ բարձր մակարդակում, նշանակում է՝ դրանք միմյանցից շատ հեռու են (տարբեր բնույթի են)։

Ենթաճյուղերը. Յուրաքանչյուր առանձին ճյուղ, որը բխում է միավորման կետից, իրենից ներկայացնում է առանձին կլաստեր։



(Այս նկարը հիերարխիկ կլաստերավորման (Hierarchical Clustering) վիզուալ ամփոփումն է, որը ցույց է տալիս, թե ինչպես են օբյեկտները խմբավորվում։ Այն բաղկացած է երկու հիմնական տարրից.

Ձախ կողմ (Ներդրված կլաստերներ). Այստեղ տեսնում ենք շրջաններ, որոնք ընդգրկված են ավելի մեծ շրջանների մեջ: Սա ցույց է տալիս հիերարխիայի էությունը՝ փոքր կլաստերները միավորվում են ավելի մեծ խմբերի մեջ։

Աջ կողմ (Դենդրոգրամ). Սա «ծառանման» գծապատկերն է, որը ցույց է տալիս այդ նույն միավորման գործընթացը գրաֆիկորեն: Ներքևի գույնզգույն կետերը սկզբնական օբյեկտներն են, իսկ վերև բարձրացող ճյուղերը պատկերում են դրանց միավորման պատմությունը՝ մինչև վերջնական, ամենամեծ կլաստերին հասնելը։)






Հիերարխիկ կլաստերավորման «Bottom-up» (կամ Ագլոմերատիվ) մոտեցումը։ 

Այս մեթոդը հիմնված է «ներքևից վեր» սկզբունքի վրա, որտեղ գործընթացը սկսվում է առանձին օբյեկտներից և հանգում մեկ ընդհանուր խմբի։


Այս ալգորիթմի հիմքում ընկած է հետևյալ պարզ տրամաբանությունը։

1․Ալգորիթմի սկզբում ենթադրվում է, որ յուրաքանչյուր առանձին օբյեկտ ինքնին հանդիսանում է առանձին կլաստեր։

2․Ամենամոտ կլաստերների որոնում. Ալգորիթմը համեմատում է կլաստերները՝ օգտագործելով որևէ չափման միավոր (օրինակ՝ Էվկլիդեսյան հեռավորությունը), որպեսզի գտնի երկու ամենանման կլաստերները։

3․Միավորում (Merge). Այդ երկու ամենամոտ կլաստերները միավորվում են մեկ նոր, ավելի մեծ կլաստերի մեջ։

4․ Կրկնություն (Repeat). Այս քայլերը կրկնվում են այնքան ժամանակ, մինչև բոլոր օբյեկտները միավորվեն մեկ մեծ կլաստերի մեջ։

Վիզուալ գործընթացը
Նկարի ներքևի մասում պատկերված գրաֆիկները ցույց են տալիս այս գործընթացը.

Սկզբում  ունենք ցրված, առանձին կետեր, որոնք շրջանակված են որպես մեկական օբյեկտ։
Յուրաքանչյուր հաջորդ քայլում ամենամոտ կանգնած կետերը միավորվում են օղակների մեջ, և այդ օղակները սկսում են միանալ՝ ստեղծելով ավելի մեծ խմբեր։
Վերջին փուլում բոլոր կետերը հայտնվում են մեկ մեծ, ընդհանուր շրջանակի մեջ, ինչը նշանակում է, որ կլաստերավորումն ավարտված է։

Այս մեթոդը հարմար է, քանի որ այն թույլ է տալիս տեսնել, թե ինչպես են օբյեկտները «մտերմանում» իրար հետ՝ ցածր մակարդակի խմբերից մինչև ամենաբարձր մակարդակի խմբավորում։







Հիերարխիկ կլաստերավորման մեջ Linkage (կապակցում) հասկացությունը սահմանում է, թե ինչպես է չափվում հեռավորությունը երկու կլաստերների միջև. Այս ընտրությունը որոշիչ է, քանի որ այն ազդում է կլաստերների վերջնական տեսքի և բովանդակության վրա, և կանխորոշում է, թե որ կլաստերները կմիավորվեն հաջորդ քայլում.

Ահա Linkage-ի երեք հիմնական չափանիշները.

1. Complete (max) Linkage (Լրիվ կամ առավելագույն կապակցում)
Բացատրություն. Հեռավորությունը հաշվարկվում է երկու կլաստերների «երկու ամենաքիչ նման» մասերի միջև, այսինքն՝ նրանց ամենահեռու կետերի միջև.

Թերություն. Այս մեթոդը հաճախ տառապում է «շղթայման» (chaining) խնդրով, ինչը նշանակում է, որ կլաստերները կարող են չափազանց տարածված դառնալ և կորցնել իրենց կոմպակտությունը. Այն խախտում է «կոմպակտության» սկզբունքը, ըստ որի՝ կլաստերի ներսում գտնվող բոլոր օբյեկտները պետք է նման լինեն միմյանց.

2. Single (min) Linkage (Միակի կամ նվազագույն կապակցում)
Բացատրություն. Հեռավորությունը հաշվարկվում է երկու կլաստերների «երկու ամենանման» մասերի միջև, այսինքն՝ նրանց ամենամոտ կետերի միջև.

Թերություն. Թեև այն խուսափում է շղթայումից, սակայն ունի «բազմության» (crowding) խնդիր. Սա նշանակում է, որ կլաստերները լինում են կոմպակտ, բայց իրարից բավարար հեռու չեն. Օբյեկտները կարող են ավելի մոտ լինել այլ կլաստերների անդամներին, քան իրենց սեփական կլաստերի անդամներին.

3. Average Linkage (Միջին կապակցում)
Բացատրություն. Հեռավորությունը հաշվարկվում է կլաստերների կենտրոնների (centroid) միջև.

Առավելություն. Սա համարվում է հավասարակշռված մոտեցում, որի դեպքում կլաստերները լինում են համեմատաբար կոմպակտ և իրարից բավարար հեռու.





Հիերարխիկ կլաստերավորման մաթեմատիկական հիմքը հիմնված է կլաստերների միջև հեռավորության որոշման վրա։ Եթե ունենք երկու կլաստեր՝ $A$ և $B$, և $d(a, b)$-ն ներկայացնում է որևէ երկու կետի՝ $a \in A$ և $b \in B$ հեռավորությունը (օրինակ՝ Էվկլիդեսյան հեռավորությունը), ապա կլաստերների միջև հեռավորությունը ($D(A, B)$) որոշվում է Linkage-ի կողմից։

Linkage-ի երեք հիմնական մեթոդների մաթեմատիկական ձևակերպումները՝
1. Single (min) Linkage

Այս մեթոդը հեռավորությունը սահմանում է որպես երկու կլաստերների ամենամոտ կետերի միջև եղած հեռավորություն։

$$D(A, B) = \min \{d(a, b) : a \in A, b \in B\}$$

Մաթեմատիկական հետևանք.
Պատկերացրեք երկու կլաստեր՝ $A$ և $B$, որոնք հիմնականում իրարից շատ հեռու են։ Սակայն, եթե կա գոնե մեկական կետ այդ կլաստերներում, որոնք պատահաբար մոտ են հայտնվել (կամ կան «աղմուկ»/outliers, որոնք կամրջում են տարածությունը), ալգորիթմը կատարում է հետևյալը.Միավորում. Քանի որ մինիմալ հեռավորությունը ($\min$) փոքր է, ալգորիթմը միավորում է այդ երկու կլաստերները՝ չնայած նրան, որ դրանց մյուս բոլոր կետերը դեռ շատ հեռու են իրարից։«Կամուրջը». Այդ ամենամոտ գտնվող կետը դառնում է «կամուրջ» երկու կլաստերների միջև։


2. Complete (max) Linkage

Այս մեթոդը հեռավորությունը սահմանում է որպես երկու կլաստերների ամենահեռու կետերի միջև եղած հեռավորություն։

$$D(A, B) = \max \{d(a, b) : a \in A, b \in B\}$$

Մաթեմատիկական հետևանք՝

Մաթեմատիկորեն, $A$ կլաստերի տրամագիծը՝ $\text{diam}(A)$-ն, սահմանվում է որպես կլաստերի ներսում գտնվող ցանկացած երկու կետերի միջև եղած առավելագույն հեռավորությունը.
$$\text{diam}(A) = \max_{a_1, a_2 \in A} d(a_1, a_2)$$

Complete Linkage-ի ալգորիթմը որոշում է միավորել երկու կլաստերներ ($A$ և $B$)՝ հիմնվելով նրանց միջև եղած հեռավորության վրա.
$$D(A, B) = \max_{a \in A, b \in B} d(a, b)$$

Երբ մենք միավորում ենք $A$ և $B$ կլաստերները, նոր՝ $(A \cup B)$ կլաստերի տրամագիծը, ըստ էության, սահմանափակվում է այն արժեքով, որը մենք նվազագույնի ենք հասցնում (այսինքն՝ $D(A, B)$-ով)։
Քանի որ ալգորիթմը միշտ փնտրում է այնպիսի միավորումներ, որտեղ $D(A, B)$-ն ամենափոքրն է, այն խուսափում է այնպիսի կետերից, որոնք հեռու են միմյանցից։

Հետևանք. Ալգորիթմը հակված է ստեղծել «գնդաձև» կամ կոմպակտ խմբեր։ Եթե միավորումը կհանգեցնի տրամագծի կտրուկ մեծացմանը, ալգորիթմը հրաժարվում է այդ միավորումից։

Եթե դուք ունեք մի կետ, որը մի փոքր հեռու է կլաստերից, Complete Linkage-ը նրան «չի սիրում», որովհետև այդ մեկ կետը կմեծացնի կլաստերի տրամագիծը (մաքսիմալ հեռավորությունը), և ալգորիթմը կնախընտրի այլ՝ ավելի կոմպակտ կլաստերներ։

3. Խնդիրը՝ «անկանոն» կամ երկարավուն ձևերի դեպքում
Սա այն կետն է, որտեղ այս մեթոդը «սխալվում» է։ Պատկերացրեք տվյալներ, որոնք դասավորված են երկարավուն «բանանի» կամ «երշիկի» տեսքով։

Այդ երկարավուն ձևի մի ծայրը մյուս ծայրից շատ հեռու է։ Complete Linkage-ի համար այդ երկարավուն ձևի «տրամագիծը» շատ մեծ է։ Ալգորիթմը դա դիտարկում է որպես «վատ» կլաստեր, որովհետև այն չի համապատասխանում «կոմպակտ» (փոքր տրամագիծ ունեցող) կլաստերի սահմանմանը։Արդյունքում տրամաբանական կլաստերը (երկարավուն ձևը) ալգորիթմը կարող է «կտրատել» մի քանի մասի և այդ մասերը միավորել հարևանությամբ գտնվող այլ կոմպակտ բլոկների հետ, միայն թե պահպանի տրամագծի նվազագույնությունը






3. Average Linkage

Այս մեթոդը հեռավորությունը սահմանում է որպես կլաստերների բոլոր կետերի զույգերի հեռավորությունների միջին արժեք։

$$D(A, B) = \text{average} \{d(a, b) : a \in A, b \in B\}$$

Եթե նշանակենք $|A|$ և $|B|$-ն որպես կլաստերների տարրերի քանակ, ապա բանաձևը կլինի.

$$D(A, B) = \frac{1}{|A| \cdot |B|} \sum_{a \in A} \sum_{b \in B} d(a, b)$$

Մաթեմատիկական հետևանք. Սա հավասարակշռված մոտեցում է, որը հաշվի է առնում կլաստերի բոլոր անդամների ազդեցությունը, ինչի արդյունքում կլաստերները ստացվում են համեմատաբար կոմպակտ և իրարից բավարար հեռու։

 Hierarchical clustering: Divisive methods (Հիերարխիկ կլաստերավորում. Տրոհող մեթոդներ) 

Դրա հիմնական կետերն են

Վերևից վար մոտեցում. Ի տարբերություն "Bottom-up" (ներքևից վեր) մեթոդի, այս մեթոդը սկսում է «վերևից»՝ մեկ ընդհանուր կլաստերից, և յուրաքանչյուր մակարդակում ռեկուրսիվ կերպով բաժանում է գոյություն ունեցող կլաստերներից մեկը երկու նոր կլաստերի։

Օպտիմալ բաժանում. Բաժանումը կատարվում է այնպես, որ ստացվեն երկու նոր խմբեր՝ միջխմբային առավելագույն տարբերությամբ (largest between-group dissimilarity)։

Ռեկուրսիվ գործընթաց. Ալգորիթմը շարունակաբար տրոհում է առկա կլաստերները «դուստր» կլաստերների՝ top-down (վերևից վար) եղանակով։



Հիերարխիկ կլաստերավորման՝ Divisive (տրոհող) մեթոդներին։ Ի տարբերություն ավելի տարածված ագրեգատիվ (bottom-up) մեթոդների, սա «վերևից-ներքև» (top-down) սկզբունքով աշխատող մոտեցում է։


Գիտական ուսումնասիրվածություն. Այս մոտեցումը կլաստերավորման մասին ընդհանուր գրականության մեջ այդքան լայնորեն չի ուսումնասիրվել, որքան ագրեգատիվ մեթոդները։ Այն հիմնականում ուսումնասիրվել է ինժեներական գրականության մեջ (օրինակ՝ Gersho և Gray, 1992-ի աշխատանքներում), մասնավորապես՝ տվյալների սեղմման (compression) խնդիրների համատեքստում։ Divisive մեթոդները կարող են ավելի նպատակահարմար լինել, երբ հետազոտողի նպատակը տվյալները համեմատաբար փոքր թվով կլաստերների բաժանելն է։Divisive պարադիգմը կարող է իրականացվել այլ կոմբինատորային մեթոդների (ինչպիսիք են K-means-ը կամ K-medoids-ը) ռեկուրսիվ կիրառման միջոցով՝ յուրաքանչյուր քայլում $K=2$ պայմանով։ Այս մոտեցման արդյունքը զգալիորեն կախված է լինում յուրաքանչյուր քայլում սահմանված սկզբնական կոնֆիգուրացիայից։

հիերարխիկ կլաստերավորման առավելություններին (Pros)


Առավելություն (Pro)․ Տվյալների ընկալելիություն. Հիերարխիկ կլաստերավորումը արդյունավետ է հիստոգրամների հետ աշխատելիս, ինչի շնորհիվ տվյալները դառնում են ավելի դյուրըմբռնելի։



Նկարը ներկայացնում է համեմատություն, թե ինչպես է փոխվում տվյալների պատկերացումը (heatmaps) հիերարխիկ կլաստերավորման կիրառումից առաջ և հետո, ցույց տալով, որ կլաստերավորումը օգնում է տեսանելի դարձնել տվյալների թաքնված կառուցվածքը։


Առավելություն (Pro). Կլաստերները կարելի է վիզուալիզացնել (պատկերացնել) և մեկնաբանել դենդրոգրամների միջոցով։



Առավելություն (Pro)։ Ճկունություն կլաստերների քանակի ընտրության հարցում. Ի տարբերություն շատ այլ ալգորիթմների (օրինակ՝ K-means-ի), որտեղ դուք պետք է նախապես սահմանեք կլաստերների քանակը ($K$), հիերարխիկ կլաստերավորման դեպքում դա անհրաժեշտ չէ։Որոշումների կայացում. Դուք կարող եք ստանալ դենդրոգրամը (ծառանման գրաֆիկը), ուսումնասիրել այն և միայն դրանից հետո որոշել, թե քանի կլաստերի բաժանումն է ամենատրամաբանականը ձեր տվյալների համար։



Նկարում պատկերված գծապատկերը ցույց է տալիս, որ դենդրոգրամի վրա կարելի է «հորիզոնական գծեր» անցկացնել տարբեր բարձրությունների վրա՝ ստանալով 2, 3 կամ 4 կլաստերներ՝ կախված նրանից, թե որքան մանրամասն եք ուզում դիտարկել տվյալները։ Սա թույլ է տալիս ավելի լավ հարմարվել տվյալների բնույթին՝ առանց նախնական ենթադրությունների։

Առավելություն (Pro)։ Հիերարխիկ կառուցվածքի նախապատվություն. Որոշ խնդիրներում հիերարխիան ավելի օգտակար է, քան «հարթ» (flat) կլաստերավորումը (ինչպիսին K-means-ն է), քանի որ այն ցույց է տալիս տվյալների միջև առկա խորքային կապերը (point linkages)։

Թերություններ (Cons)

Մեծ տվյալների հետ աշխատանքի սահմանափակում. Ալգորիթմը արդյունավետ չէ մեծ ծավալի տվյալների դեպքում, քանի որ ունի քառակուսային ժամանակային բարդություն՝ $O(n^2)$, ինչը նշանակում է, որ տվյալների աճին զուգընթաց հաշվարկման ժամանակը շատ արագ է աճում։Կլաստերների քանակի որոշում և զգայունություն. Դժվար է հստակ սահմանել կլաստերների ճշգրիտ քանակը, բացի այդ՝ մեթոդը զգայուն է տվյալների «աղմուկի» (noise) և արտանետումների (outliers) նկատմամբ, որոնք կարող են աղավաղել կառուցվող ծառը։Մեկնաբանելիության բացակայություն. Դժվար է մեկնաբանել կլաստերները, քանի որ բացակայում են հստակ «նկարագրիչներ» (cluster descriptors)։Սխալների ուղղման անհնարինություն. Այս մեթոդը «անշրջելի» է. երբ երկու օբյեկտ կամ խումբ միավորվում են (կամ բաժանվում), հետագայում հնարավոր չէ ուղղել այդ որոշումը, նույնիսկ եթե հետագա քայլերում պարզվի, որ այն սխալ էր։

K-means
K-means-ը ուսուցում առանց ուսուցչի (unsupervised learning) ալգորիթմ է, որը նպատակ ունի տվյալների բազմությունը բաժանել $K$ թվով խմբերի (կլաստերների):

K-means-ը ինտերատիվ (կրկնվող) ալգորիթմ է։ 

1. Ստանդարտացում (Standardization)
Նախքան ալգորիթմը սկսելը, մենք պետք է տվյալները բերենք միևնույն մասշտաբի։

Ինչո՞ւ: Եթե մի սյունակը (օրինակ՝ "Տևողություն") ունի 2-5 արժեքներ, իսկ մյուսը ("Սպասման ժամանակ")՝ 50-90, ապա հեռավորության հաշվարկման ժամանակ (Էվկլիդեսյան հեռավորություն) մեծ արժեքները «կգերիշխեն» փոքրերին։
Լուծում: Մենք օգտագործում ենք Z-score normalization ($z = \frac{x - \mu}{\sigma}$), որպեսզի բոլոր հատկանիշները ունենան 0 միջին արժեք և 1 ստանդարտ շեղում։

2. Նախնականացում (Initialization)

Ալգորիթմը պատահականորեն ընտրում է $K$ թվով կենտրոններ (centroid):Ժամանակակից լուծումներում մենք օգտագործում ենք K-means++ մեթոդը։ Այն պատահական չէ ընտրում կենտրոնները, այլ ընտրում է դրանք այնպես, որ սկզբնական կետերը հնարավորինս հեռու լինեն միմյանցից։ Սա զգալիորեն արագացնում է ալգորիթմի աշխատանքը։

3. Հիմնական ցիկլը (The Core Loop)

Սա այն փուլն է, որտեղ տեղի է ունենում «սովորումը»։ Այն բաղկացած է երկու կրկնվող քայլից.

Ա. Վերագրում (Assignment)
Յուրաքանչյուր կետ հաշվարկում է իր հեռավորությունը բոլոր կենտրոններից և «միանում» է ամենամոտ կենտրոնին։ Արդյունքում մենք ստանում ենք կետերի ժամանակավոր խմբեր։
Բ. Թարմացում (Update)
Երբ կետերը արդեն բաշխված են, ալգորիթմը վերահաշվարկում է յուրաքանչյուր կլաստերի կենտրոնը՝ գտնելով այդ կլաստերի բոլոր կետերի միջին արժեքը (մաթեմատիկական կենտրոնը)։

4. Զուգամտություն (Convergence)

Ալգորիթմը կրկնում է «Վերագրում» և «Թարմացում» քայլերը այնքան ժամանակ, մինչև տեղի է ունենում հետևյալներից մեկը.Կայունություն: Կենտրոնները դադարում են տեղաշարժվել (դրանց դիրքը նախորդ քայլի համեմատ չի փոխվել)։

Սահմանափակում: Հասել ենք կրկնությունների (iterations) առավելագույն թվին, որը մենք նախապես սահմանել ենք։
WCSS-ի նվազագույն փոփոխություն: Կլաստերի ներսում հեռավորությունների գումարի փոփոխությունը դարձել է աննշան։




Ինչո՞ւ է ստանդարտացումը պարտադիր

K-means ալգորիթմը աշխատում է՝ հաշվարկելով կետերի միջև Էվկլիդեսյան հեռավորությունը։ Եթե տվյալների մասշտաբները (չափման միավորները) տարբեր են, ապա ալգորիթմը «սխալվում է»։

1.Գրաֆիկի ձախ կողմում տեսնում եք՝X առանցք (Տևողություն/Duration): Արժեքները տատանվում են 2-ից 5-ի սահմաններում։
Y առանցք (Սպասման ժամանակ/Wait Time): Արժեքները տատանվում են 50-ից 90-ի սահմաններում։
Երբ ալգորիթմը հաշվարկում է հեռավորությունը երկու կետերի միջև, այն քառակուսի է բարձրացնում տարբերությունները․

$$d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$

Քանի որ $y$-ի արժեքները շատ ավելի մեծ են, $(y_2 - y_1)^2$-ը կստանա շատ մեծ թիվ, իսկ $(x_2 - x_1)^2$-ը՝ շատ փոքր։ Արդյունքում՝ ալգորիթմը $y$-ի փոփոխությունները կհամարի ավելի կարևոր, քան $x$-ինը, և կստեղծի ձգված, ոչ ճիշտ կլաստերներ։

2. Լուծումը՝ Ստանդարտացում (Z-score)Ստանդարտացումը բոլոր առանցքները բերում է «ընդհանուր հայտարարի»։ Օգտագործելով հետևյալ բանաձևը՝ մենք փոխում ենք տվյալները այնպես, որ դրանց միջինը լինի 0, իսկ ստանդարտ շեղումը՝ 1։
$$z = \frac{x - \mu}{\sigma}$$
Որտեղ $\mu$-ն միջին արժեքն է, իսկ $\sigma$-ն՝ ստանդարտ շեղումը:Արդյունքը՝ աջ կողմի գրաֆիկում երկու առանցքներն էլ ունեն նույն մասշտաբը (գրեթե -2-ից +2)։ Այժմ ալգորիթմը երկու հատկանիշներն էլ (Duration և Wait Time) գնահատում է հավասարաչափ, և կլաստերները ստացվում են ճշգրիտ ու հստակ։




K-means ֆորմալացում (K-means Formalized)

Դիցուք $\mathcal{X}$-ը տարածություն է որևէ $d$ հեռավորության մետրիկայով։
Ամենահաճախ հանդիպող դեպքում՝ $\mathcal{X} = \mathbf{R}^d$ և $d(x, x') = ||x - x'||$ (Էվկլիդեսյան հեռավորություն)-ն է սահմաանված այդ տարածության վրա:
Տվյալների բազմությունը՝ $\mathcal{D} = \{x_1, \dots, x_n\} \subset \mathcal{X}$ այդ տարածությն ենթաբազմությունն է։
Նպատակը՝ բաժանել $\mathcal{D}$ տվյալները $k$ չհատվող (disjoint) բազմությունների՝ $C_1, \dots, C_k$:
Ալգորիթմի նպատակն է ստեղծել $k$ հատ տարբեր խմբեր այնպես, որ յուրաքանչյուր կետ պատկանի միայն մեկ խմբի:
$C_i$ կլաստերի կենտրոնը (ցենտրոիդը) սահմանվում է որպես այն կետը, որը նվազագույնի է հասցնում կլաստերի բոլոր կետերի հեռավորությունների քառակուսիների գումարը մինչև այդ կենտրոնը՝$$\mu_i = \mu(C_i) = \arg \min_{\mu \in \mathcal{X}} \sum_{x \in C_i} d(x, \mu)^2$$
Սա K-means ալգորիթմի հիմնական գաղափարն է․ կլաստերի կենտրոնը պետք է ընտրվի այնպես, որ կլաստերի ներսում գտնվող բոլոր կետերը հնարավորինս մոտ լինեն դրան:
Նշում՝ $\mathbf{R}^d$ տարածությունում Էվկլիդեսյան հեռավորության դեպքում, $\mu(C_i)$-ն $C_i$ բազմության թվաբանական միջինն է:

(K-means ալգորիթմը փորձում է տվյալների բազմությունը բաժանել $k$ խմբի։ Յուրաքանչյուր խմբի համար հաշվարկվում է «կենտրոնը» (centroid): Եթե մենք օգտագործում ենք սովորական հեռավորություն (Եվկլիդեսյան), ապա այդ կենտրոնը պարզապես խմբի կետերի միջին արժեքն է։ Ալգորիթմն անընդհատ թարմացնում է կետերի պատկանելությունը խմբերին և վերահաշվարկում կենտրոնները, մինչև արդյունքը կայունանա։)

K-means-ի ֆունկցիան (The K-means objective)

Մաթեմատիկական բանաձևը՝
$$J_{k-means}(C_1, \dots, C_k) = \sum_{i=1}^{k} \sum_{x \in C_i} d(x, \mu(C_i))^2$$

Ալգորիթմի հիմնական նպատակն է նվազագույնի հասցնել այս ֆունկցիայի արժեքը՝ $J$-ն։
Սա նշանակոմ է որ՝
Մենք ցանկանում ենք տվյալները բաժանել $k$ թվով խմբերի (կլաստերների) այնպես, որ յուրաքանչյուր խմբի ներսում կետերի ($x$) հեռավորությունը իրենց խմբի կենտրոնից ($\mu$) հնարավորինս փոքր լինի։$d(x, \mu(C_i))^2$ մասը ցույց է տալիս կետի և կենտրոնի միջև եղած հեռավորության քառակուսին։Ալգորիթմը ձգտում է «սեղմել» կլաստերները՝ դարձնելով դրանք հնարավորինս կոմպակտ։

Ալգորիթմի քայլերը
Մուտքային տվյալներ (Input)
* $\mathcal{D} = \{x_1, \dots, x_d\} \subset \mathcal{X}$: Սա մեր ամբողջական տվյալների բազմությունն է, որը մենք ուզում ենք խմբավորել։
Ինիցիալիզացիա (Initialize)
* Ընտրել պատահական սկզբնական կենտրոններ ($\mu_1, \dots, \mu_k$): 
Ալգորիթմի սկզբում մենք պատահականորեն ընտրում ենք $k$ հատ կետ, որոնք կծառայեն որպես մեր կլաստերների նախնական կենտրոններ։

Կրկնվող գործընթաց (Repeat until convergence)

Ալգորիթմը շարունակաբար կրկնում է հետևյալ երկու քայլը, մինչև կենտրոնները դադարեն փոխվել (այսինքն՝ հասնենք կայուն վիճակի)։

Վերագրման քայլը (Assignment step):

Բանաձև՝ $C_i = \{x \in \mathcal{D} : i = \text{arg min}_j d(x, \mu_j)\}$

Բացատրություն․ Այս քայլում մենք յուրաքանչյուր տվյալ կետ ($x$) վերագրում ենք այն կլաստերին, որի կենտրոնին ($\mu_j$) այն ամենամոտն է։ Այսինքն՝ «Ո՞ր կենտրոնն է ինձ ամենամոտը, ուրեմն ես գնում եմ այդ խումբ»։

Թարմացման քայլը (Update step):

Բանաձև՝ $\mu_i = \text{arg min}_{\mu \in X} \sum_{x \in C_i} d(x, \mu)^2$

Բացատրություն․ Երբ բոլոր կետերը արդեն բաշխված են խմբերի, մենք վերահաշվարկում ենք խմբի կենտրոնը։ Նոր կենտրոնը դառնում է այդ խմբի մեջ գտնվող բոլոր կետերի միջին թվաբանականը (ինչպես նշված է նկարում՝ 

$\mu_i = \frac{1}{|C_i|} \sum_{x \in C_i} x$):  


 K-means ալգորիթմի գործնական կիրառումներից մեկը՝ Վեկտորային քվանտավորումը (Vector Quantization),  լայնորեն օգտագործվում է տվյալների սեղմման (compression) մեջ։
 
Սովորական պատկերում յուրաքանչյուր պիքսել ունի իր յուրահատուկ գույնը (օրինակ՝ RGB արժեքների միջոցով)։ Դա շատ հիշողություն է պահանջում։
Վեկտորային քվանտավորման դեպքում մենք պատկերը բաժանում ենք $k$ խմբի (կլաստերի)։ Յուրաքանչյուր խումբ ունի մեկ «ներկայացուցիչ»՝ կենտրոնը ($\mu_i$)։ 
Այսուհետ, բոլոր այն պիքսելները, որոնք պատկանում են այդ խմբին, փոխարինվում են այդ կենտրոնի գույնով։

Կորուստներով տվյալների սեղմում (Lossy Data Compression)
 
Սա կոչվում է «կորուստներով սեղմում», քանի որ մենք կորցնում ենք օրիգինալ պատկերի մանրամասնությունը՝ ստանալով ավելի պարզեցված տարբերակ։
 
Վերականգնման սխալ (Reconstruction Error): K-means-ի նպատակային ֆունկցիան, որը նախորդ բացատրության մեջ նշեցինք որպես $J$, այստեղ հենց այն «սխալն» է, որ մենք թույլ ենք տալիս սեղմման ժամանակ։ Որքան մեծ է $J$-ն, այնքան ավելի շատ ենք հեռանում օրիգինալ պատկերից։

Ալգորիթմի հիմնական առավելությունը հիշողության խնայողությունն է։ 
Տրամաբանությունը հետևյալն է.Եթե մենք ունենք $k$ կլաստեր, ապա յուրաքանչյուր պիքսելը նկարագրելու համար մեզ անհրաժեշտ է միայն հիշել, թե որ կլաստերին է այն պատկանում (այսինքն՝ 1-ից $k$ համարը)։Եթե $k = 2^d$, ապա յուրաքանչյուր պիքսելը պահելու համար մեզ պետք է ընդամենը $d$ բիթ։Օրինակ՝ 
* Եթե օգտագործում ենք $k=8$ (քանի որ $8 = 2^3$), ապա յուրաքանչյուր պիքսելի համար պետք է ընդամենը 3 բիթ, մինչդեռ օրիգինալ գունային համակարգում կարող է պահանջվել 24 բիթ (8 բիթ՝ կարմիր, 8՝ կանաչ, 8՝ կապույտ)։ Սա հսկայական խնայողություն է։
Նշում՝ Եթե որոշ կլաստերներ ավելի մեծ են, քան մյուսները (պարունակում են ավելի շատ պիքսելներ), ապա կարելի է օգտագործել փոփոխական երկարությամբ կոդավորում (ինչպես Huffman-ի կոդավորումը) և միջինում ծախսել նույնիսկ ավելի քիչ բիթեր։


Նկարում ցույց է տրված, թե ինչպես է փոխվում պատկերի որակը՝ կախված $k$ արժեքից՝
    

 







Այս պատկերը  ցույց է տալիս K-means (կլաստերավորումը կենտրոնների միջոցով) և Հիերարխիկ (Hierarchical) կլաստերավորման եղանակների տարբերությունները՝ ինչպես իդեալական, այնպես էլ իրատեսական, բարդ տվյալների պարագայում։

Ահա պատկերի և մեթոդների մանրամասն բացատրությունը`

Իդեալական դեպք (Idealized Case). 
Ձախ կողմի գծապատկերներում տվյալները շատ հստակ են և առանձնացված։

Scatter plot (ցրվածության գծապատկեր): Մենք տեսնում ենք երեք հստակ խումբ (դեղին, վարդագույն, կապույտ)՝ առանց իրար խառնվելու։

Dendrogram (դենդրոգրամ): Սա «ծառաձև» գծապատկեր է, որը ցույց է տալիս տվյալների միավորման գործընթացը։ Իդեալական դեպքում այն շատ պարզ է. տեսնում ենք երեք հիմնական ճյուղ, որոնք հստակ բաժանվում են։ Կարմիր գիծը (cut-off) ցույց է տալիս, որ մենք կարող ենք կտրել ծառը 3 կլաստերի մակարդակում և ստանալ կատարյալ բաժանում։

2. Իրատեսական դեպք (Realistic Case)
Աջ կողմի գծապատկերները ցույց են տալիս իրական կյանքի տվյալները, որոնք հաճախ «աղմկոտ» են (noisy) և ունեն սահմանային տարածքներ։

Scatter plot: Խմբերը իրար հետ մասամբ համընկնում են (overlap)։ K-means ալգորիթմը այստեղ ստիպված կլինի «կոշտ» որոշում կայացնել՝ տվյալ կետը վերագրելով այս կամ այն կենտրոնին, նույնիսկ եթե այն գտնվում է սահմանին։

Dendrogram: Սա է հիերարխիկ մոտեցման հիմնական առավելությունը։ Այն ցույց է տալիս, որ տվյալները ունեն բարդ կառուցվածք։

Դուք կարող եք «կտրել» ծառը տարբեր բարձրությունների վրա՝ ստանալով 3, 5 կամ 7 կլաստեր։

Հիերարխիկ մոտեցումը թույլ է տալիս տեսնել տվյալների ներքին «ենթախմբերը»։ Օրինակ՝ 7 կլաստերի դեպքում մենք կարող ենք հասկանալ ավելի մանր դետալներ, քան 3 կլաստերի դեպքում։

(Ինչպե՞ս կարդալ դենդրոգրամը (Dendrogram)
Դենդրոգրամը կարդում ենք ներքևից վերև (կամ հակառակը)։

Հորիզոնական գծերը ցույց են տալիս հեռավորությունը երկու կետերի կամ կլաստերների միջև։

Որքան բարձր է միավորման կետը, այնքան ավելի տարբեր են այդ խմբերը իրարից։

Կարմիր գծերը (cut-offs) այն «դանակներն» են, որոնցով դուք որոշում եք, թե որ մակարդակում կանգ առնել։ Իրատեսական օրինակում կարմիր գիծը ցույց է տալիս, որ կախված նրանից, թե որքան մանրամասն պատկեր ենք ուզում, մենք կարող ենք ընտրել 3, 5 կամ 7 խումբ։)




Ներկայացնենք թե ինչպես է աշխատում K-means ալգորիթմը և ինչու է այն դադարում աշխատել (զուգամիտում)։


J-ն  «ծախսային ֆունկցիան» է (cost function), որը չափում է, թե որքան լավ են կլաստերները նկարագրում տվյալները: $J$-ի արժեքը որքան փոքր է, այնքան լավ. դա նշանակում է, որ տվյալների կետերը շատ մոտ են իրենց կլաստերների կենտրոններին:

Ալգորիթմի երկու քայլերը:
* Վերագրում (Assignment): Յուրաքանչյուր կետ վերագրվում է ամենամոտ գտնվող կենտրոնին: Սա միշտ նվազեցնում է $J$-ն:
* Կենտրոնի թարմացում (Centroid update): Կլաստերի կենտրոնը տեղափոխվում է այդ կլաստերի մեջ գտնվող կետերի միջին դիրքին: Սա նույնպես նվազեցնում է $J$-ն:




Զուգամտություն: Գրաֆիկի վրա տեսնում եք, որ $J$ արժեքը սկզբում կտրուկ ընկնում է, իսկ հետո դառնում է հաստատուն (հորիզոնական գիծ): Երբ կետերը դադարում են «տեղափոխվել» մեկ կլաստերից մյուսը, իսկ կենտրոնները դադարում են շարժվել, ալգորիթմը համարվում է ավարտված: Հետևաբար մենք  ստացել ենք կլաստերավորման արդյունքը:
Կարևոր է հասկանալ, որ ալգորիթմը չի աշխատի անվերջ: Քանի որ հնարավոր վերագրումների քանակը վերջավոր է, իսկ յուրաքանչյուր քայլ բարելավում է արդյունքը, ալգորիթմը վաղ թե ուշ կհասնի կետի, որտեղ այլևս փոփոխություններ չեն լինում:



K-means. $K$-ի ընտրությունը։

K-means ալգորիթմում կարևոր է նախապես որոշել, թե քանի կլաստեր ($K$) պետք է ունենալ: Արմնկային մեթոդը ամենատարածված մոտեցումն է դա անելու համար:

Արմնկային մեթոդ (Elbow method). մեծացնել $K$-ն այնքան ժամանակ, մինչև այն այլևս չի օգնում ավելի լավ նկարագրել տվյալները:
Մենք ձգտում ենք գտնել այնպիսի $K$, որի դեպքում խմբերի ներսում էվկլիդեսյան հեռավորությունների գումարը փոքր լինի:
$$J = \sum_{j=1}^{K} \sum_{i=1}^{n_j} ||\mathbf{x}_i^{(j)} - \mathbf{c}_j||^2$$

որտեղ $\mathbf{c}_j$-ն $j$-րդ կլաստերի կենտրոնն է (միջինը):

Բանաձևը հաշվարկում է կլաստերի մեջ գտնվող յուրաքանչյուր կետի հեռավորությունը իր կենտրոնից ($\mathbf{c}_j$), բարձրացնում քառակուսի և գումարում բոլոր կետերի համար:
* Եթե $K$-ն շատ փոքր է, ապա կետերը շատ հեռու են կենտրոններից, և $J$-ն մեծ է:
* Եթե $K$-ն շատ մեծ է (օրինակ՝ ամեն կետը դառնա առանձին կլաստեր), ապա հեռավորությունները կլինեն զրո, և $J$-ն կդառնա 0: Սակայն սա կլաստերավորում չէ, քանի որ չի տալիս ընդհանրացված տեղեկություն:

Գրաֆիկի վրա, երբ ավելացնում ենք $K$-ն, սկզբում $J$-ի արժեքը կտրուկ ընկնում է (քանի որ ավելի շատ կենտրոններ նշանակում են ավելի ճշգրիտ նկարագրություն): Սակայն մի պահից սկսած՝ $K$-ն ավելացնելն այլևս չի բերում $J$-ի զգալի նվազման:Գրաֆիկի վրա առաջացած այդ «ծունկը» կամ «արմունկը» (elbow point) հենց այն կետն է, որտեղից հետո $K$-ի ավելացումը քիչ օգուտ է տալիս:Օպտիմալ $K$-ն հենց այդ «արմնկային» կետն է:
Այս մեթոդը թույլ է տալիս գտնել հավասարակշռությունը կլաստերների քանակի և ալգորիթմի ճշգրտության միջև:



(WSS-ը և $J$-ը նույն բանն են, պարզապես օգտագործվում են տարբեր համատեքստերում՝ մեկը որպես բովանդակային անվանում, մյուսը՝ որպես մաթեմատիկական նշանակում։)


K-means: Լոկալ մինիմումներ

K-means ալգորիթմը նպատակ ունի տվյալները բաժանել խմբերի (cluster) այնպես, որ յուրաքանչյուր խմբի կենտրոնը (centroid) հնարավորինս մոտ լինի իր խմբի բոլոր կետերին։ Դա հաշվարկվում է $J$ օբյեկտիվ ֆունկցիայի միջոցով (որը հաճախ կոչվում է «ծախսերի ֆունկցիա»՝ cost function)։

Սակայն այստեղ կա խնդիր:
Լոկալ մինիմում (Local Minima): Պատկերացրեք լեռնաշղթա։ Դուք ուզում եք իջնել ամենացածր կետը (գլոբալ մինիմումը)։ Եթե սկսեք քայլել սխալ վայրից, կարող եք հայտնվել մի փոքրիկ հովտում (լոկալ մինիմում) և մտածել, որ հասել եք հատակին, մինչդեռ իրական ամենացածր կետը լեռան մյուս կողմում է։
K-means-ը հենց այսպես է «շփոթվում»։

Ոչ ուռուցիկություն: Մաթեմատիկորեն $J$ ֆունկցիան այնքան բարդ կառուցվածք ունի, որ ալգորիթմը չի կարող միշտ գտնել լավագույն լուծումը։



Նկարի օրինակը: Տեսեք նկարի աջ հատվածը։ Կանաչ և կապույտ կետերը շատ մոտ են միմյանց, և ալգորիթմը դրանց համար երկու առանձին կենտրոն է ստեղծել, մինչդեռ ավելի տրամաբանական կլիներ դրանք մեկ խմբի մեջ միավորել։ Սա վատ լոկալ օպտիմումի օրինակ է։

Լուծումը հետևյալն է։

Քանի որ K-means-ի արդյունքը կախված է նրանից, թե որտե՞ղ է այն սկզբնապես տեղադրում կենտրոնները (random initialization), խորհուրդ է տրվում ալգորիթմը գործարկել մի քանի անգամ՝ տարբեր պատահական սկզբնակետերով։ Այնուհետև ընտրվում է այն տարբերակը, որտեղ $J$ ֆունկցիայի արժեքն ամենափոքրն է։

K-means++ ալգորիթմը

Սովորական $k$-means ալգորիթմը խնդիրներ ունի նախնական կենտրոններն ընտրելիս (սկզբնավորման փուլում): Եթե առաջին կենտրոնները ընտրվեն իրար շատ մոտ կամ անհաջող դիրքերում, ալգորիթմը կարող է «փակվել» վատ լուծման մեջ (տեղային մինիմում) և չգտնել օպտիմալ կլաստերները:

1․ Ընտրեք մեկ կենտրոն հավասարաչափ պատահականությամբ՝ տվյալների կետերի միջից:

Ալգորիթմը պարզապես վերցնում է պատահական մի կետ և հայտարարում այն առաջին «կենտրոն»։

2․ Յուրաքանչյուր դեռ չընտրված $x$ տվյալի կետի համար, հաշվեք $D(x)$-ը՝ $x$-ի և արդեն ընտրված ամենամոտ կենտրոնի միջև եղած հեռավորությունը:

K-means: Առավելություններ և թերություններ

Դրական կողմեր (Pros)
* Պարզ և հեշտ հասկանալի: Ալգորիթմի տրամաբանությունը շատ պարզ է՝ տվյալները բաժանել $K$ թվով խմբերի՝ ըստ իրենց կենտրոններից (centroid) հեռավորության։

* Արագ և արդյունավետ: K-means-ը հաշվարկային առումով շատ արագ է, ինչը թույլ է տալիս այն կիրառել մեծ տվյալների բազաներում։

* Երաշխավորված համախմբում: Ալգորիթմը միշտ հասնում է որոշակի լուծման (տեղային օպտիմումի)՝ իտերացիաների սահմանափակ թվից հետո։
* Scalable: Այն լավ է աշխատում մեծ ծավալի տվյալների հետ։


Բացասական կողմեր (Cons)

* Պահանջում է կլաստերների նախնական քանակ ($K$): Դուք պետք է նախապես որոշեք $K$-ի արժեքը, ինչը միշտ չէ, որ հեշտ է (սովորաբար օգտագործվում է "Elbow method" կամ "Silhouette analysis"՝ լավագույն $K$-ն գտնելու համար)։

* Զգայուն է կենտրոնների նախնական դիրքի նկատմամբ: Եթե սկզբնական կենտրոնները վատ ընտրվեն, ալգորիթմը կարող է հասնել ոչ օպտիմալ արդյունքի։ Սա լուծելու համար հաճախ օգտագործում են K-means++ մոտեցումը։
* Ենթադրում է գնդաձև կլաստերներ: K-means-ը օգտագործում է էվկլիդեսյան հեռավորությունը (Euclidean distance), ուստի այն լավ է աշխատում միայն գնդաձև, խիտ կլաստերների դեպքում։ Այն չի կարող հայտնաբերել բարդ կամ երկարավուն ձևեր (այստեղ ավելի լավ է աշխատում օրինակ՝ DBSCAN-ը)։
* Չի աշխատում կատեգորիալ տվյալների հետ: Քանի որ ալգորիթմը հաշվարկում է "միջինը" (mean), այն աշխատում է միայն թվային տվյալների հետ։ Կատեգորիալ տվյալների համար պետք է նախապես կիրառել encoding (օրինակ՝ One-Hot Encoding):
* Զգայուն է outliers նկատմամբ: Քանի որ կենտրոնները հաշվարկվում են որպես միջին արժեք, եզակի կամ շատ հեռու ընկած տվյալները կարող են մեծապես շեղել կլաստերի կենտրոնը՝ աղավաղելով արդյունքը։


K-means ալգորիթմը կլաստերները գտնելիս փորձում է նվազագույնի հասցնել կետերի և կենտրոնի (centroid) միջև հեռավորությունների քառակուսիների գումարը: Քանի որ մենք քառակուսի ենք բարձրացնում հեռավորությունը, հեռու գտնվող կետերը ստանում են անհամաչափ մեծ «կշիռ» կամ ազդեցություն:

Outlier-ները տվյալների բազմության մեջ այն կետերն են, որոնք շատ հեռու են հիմնական խմբերից (կլաստերներից): Օրինակ՝ եթե ունենք մարդկանց եկամուտների խումբ, միլիարդատերը կլինի «արտանետում»:
Քանի որ K-means-ը հաշվարկում է միջին արժեքը (centroid), այդ մեկ հեռու կետը  կարող է կտրուկ փոխել կլաստերի կենտրոնի դիրքը, ինչի պատճառով արդյունքը դառնում է ոչ ճշգրիտ: Այդ պատճառով ասում են, որ K-means-ը «կայուն չէ» (lack of robustness):


K-medoids ալգորիթմը կենտրոնի համար չի հաշվարկում «միջին» արժեքը (որը հորինված կետ է), այլ ընտրում է իրական տվյալների կետերից մեկը որպես կենտրոն (medoid): Սա շատ ավելի դիմացկուն է  outlier-ների նկատմամբ, քանի որ մեկ հեռու կետը չի կարող «քաշել» կենտրոնը դեպի իրեն:

Սակայն K-means-ը շատ արագ է աշխատում, իսկ K-medoids-ը պահանջում է շատ ավելի շատ հաշվարկներ՝ յուրաքանչյուր քայլին օպտիմալ «մեդոիդ» գտնելու համար: 

K-medoids ալգորիթմի քայլերը
* Ընտրել K-ն՝ կլաստերների քանակը:

* Նախնական մեդոիդների ընտրություն. Տվյալների բազմությունից պատահականորեն ընտրվում են K թվով կետեր, որոնք դառնում են սկզբնական մեդոիդներ:

* Կլաստերավորում. Յուրաքանչյուր կետ (օբյեկտ) վերագրվում է ամենամոտ մեդոիդին՝ օգտագործելով ընտրված անհամապատասխանության չափման միավորը (dissimilarity measure):

* Մեդոիդների թարմացում. Կատարվում է օպտիմալացում: Փորձ է արվում փոխարինել ընթացիկ մեդոիդը կլաստերի մեկ այլ կետով, եթե այդ փոփոխությունը նվազեցնում է կլաստերի ներսում հեռավորությունների ընդհանուր գումարը:

Սա ամենահետաքրքիր մասն է։ Ալգորիթմը յուրաքանչյուր կլաստերի համար հաշվարկում է բոլոր հնարավոր փոխարինումները: Այսինքն՝ յուրաքանչյուր ոչ-մեդոիդ կետի համար հաշվարկվում է, թե արդյոք այն ավելի լավ կենտրոն կլինի՞, քան ներկայիս մեդոիդը։ Սա բավականին ծանր հաշվարկ է (computationally expensive), այդ պատճառով K-medoids-ը դանդաղ է աշխատում մեծ տվյալների բազաների հետ՝ համեմատած K-means-ի հետ:

* Կոնվերգենցիա. 3-րդ և 4-րդ քայլերը կրկնվում են այնքան ժամանակ, քանի դեռ մեդոիդները չեն դադարում փոխվել, կամ կլաստերավորման արդյունքը չի կայունանում։
Դա այն պահն է, երբ ալգորիթմը «հասկանում է», որ այլևս ոչ մի կետի փոխարինում չի բարելավի կլաստերավորման որակը (ծախսը չի նվազի):

Կլաստերավորումը, ըստ էության, կառուցված է «մոտիկություն» հասկացության վրա, և հեռավորության չափումը հենց այն գործիքն է, որը սահմանում է այդ «մոտիկությունը»։

Պատկերացնենք երկու կետեր։ Մենք դրանք «նման» ենք համարում, եթե դրանց հեռավորությունը փոքր է։ Բայց «հեռավորություն» հասկացությունը կարող է տարբեր կերպ սահմանվել՝ կախված ձեր խնդրից։
1. Էվկլիդեսյան հեռավորություն (Euclidean Distance - $L_2$ norm)Սա ամենատարածված «ուղիղ գծով» հեռավորությունն է։ Եթե ձեր տվյալները ներկայացնում են ֆիզիկական տարածություն (օրինակ՝ քարտեզի վրա կոորդինատներ), ապա սա է ճիշտ ընտրությունը։

2. Մանհեթենյան հեռավորություն (Manhattan Distance - $L_1$ norm)Սա հեռավորությունն է, որը չափվում է «քաղաքային թաղամասերով» (օրինակ՝ չեք կարող անցնել շենքերի միջով, միայն փողոցներով)։ Այն շատ ավելի կայուն է արտանցիկ կետերի (outliers) նկատմամբ։ Եթե ունեք շատ չափումներ (high-dimensional data), Մանհեթենյան հեռավորությունը հաճախ ավելի լավ է աշխատում, քան Էվկլիդեսյանը։

3. Կոսինուսային նմանություն (Cosine Similarity)Սա շատ կարևոր է, երբ ձեզ հետաքրքրում է ոչ թե կետի «մեծությունը» (magnitude), այլ դրա «ուղղությունը»։ Սա ստանդարտ է տեքստային վերլուծության մեջ (NLP)։

Եթե դուք ունեք մի խումբ կետեր, որոնք Էվկլիդեսյան հեռավորությամբ կլաստերավորելիս ձևավորում են երկու կլաստեր, ապա Մանհեթենյան հեռավորության անցնելիս դրանց «տեսանելի» հեռավորությունները կփոխվեն։ Արդյունքում, ալգորիթմը կարող է կետերը վերաբաշխել՝ ստեղծելով բոլորովին այլ կլաստերներ։ 
Data Science-ում կլաստերավորում կատարելիս միշտ հարցրեք ինքներդ ձեզ՝ «Ի՞նչ է նշանակում նմանություն այս տվյալների համար»:
* Եթե տվյալները թվային են և ֆիզիկական՝ Euclidean.
* Եթե ունեք բարձր չափայնություն և աղմուկ՝ Manhattan.
* Եթե աշխատում եք տեքստի կամ վեկտորների ուղղվածության հետ՝ Cosine Similarity



Soft K-means
Իրական կյանքում տվյալները հազվադեպ են հստակորեն առանձնացված։ Հաճախ մի կետը կարող է միաժամանակ պատկանել երկու կլաստերների (օրինակ՝ երկու կլաստերների սահմանին գտնվող կետը)։

Սովորական K-means-ում մենք կատարում ենք hard assignment. Օրինակ՝ կետը կամ պատկանում է A կլաստերին (1), կամ B-ին (0)։

Soft K-means-ում մենք կատարում ենք soft assignment. Յուրաքանչյուր կետ ունի «պատասխանատվություն» (responsibility, նշված որպես $r_k^{(n)}$) յուրաքանչյուր կլաստերի նկատմամբ։ Օրինակ՝ կետը կարող է 70% պատկանել կլաստեր 1-ին և 30%՝ կլաստեր 2-ին

Soft K-means-ի նպատակն է հրաժարվել «կոշտ» (hard) որոշումներից, երբ կետը պարտադիր պետք է լինի կամ A կլաստերում, կամ B-ում։ Փոխարենը, այն թույլ է տալիս կետին միաժամանակ «մասնակի» պատկանել մի քանի կլաստերի։

1․ Responsibility (Պատասխանատվություն) — $r_k^{(n)}$:
Այս բանաձևը ցույց է տալիս, թե ինչպես ենք հաշվարկում կետի պատկանելությունը կլաստերին։
Բանաձևը. Օգտագործվում է softmax ֆունկցիան՝

$$r_k^{(n)} = \frac{\exp[-\beta \|m_k - x^{(n)}\|^2]}{\sum_j \exp[-\beta \|m_j - x^{(n)}\|^2]}$$

Softmax ֆունկցիան $z_k$ վեկտորի համար սահմանվում է այսպես.$$\text{softmax}(z_k) = \frac{\exp(z_k)}{\sum_{j} \exp(z_j)}$$

Եթե նայենք ձեր բանաձևին, ապա կարող ենք կատարել հետևյալ փոխարինումը։
Մուտքային արժեք (Input):
$z_k = -\beta \|m_k - x^{(n)}\|^2$
Եթե այս $z_k$-ն տեղադրենք ստանդարտ Softmax-ի մեջ, կստանանք հենց  բանաձևը՝
$$r_k^{(n)} = \frac{\exp(z_k)}{\sum_{j} \exp(z_j)} \implies \frac{\exp[-\beta \|m_k - x^{(n)}\|^2]}{\sum_j \exp[-\beta \|m_j - x^{(n)}\|^2]}$$

Կետը դառնում է $k$ կլաստերի «պատասխանատուն»՝ հիմնվելով նրա և կենտրոնի միջև եղած հեռավորության վրա։ Որքան մոտ է կետը կենտրոնին ($m_k$), այնքան մեծ է $r_k^{(n)}$ արժեքը։

(Հեռավորության հաշվարկը ($\|m_k - x^{(n)}\|^2$)
Սա կետի ($x^{(n)}$) և կլաստերի կենտրոնի ($m_k$) միջև եղած քառակուսային Էվկլիդեսյան հեռավորությունն է։Եթե հեռավորությունը փոքր է՝ կետը շատ մոտ է կենտրոնին։Եթե հեռավորությունը մեծ է՝ կետը հեռու է կենտրոնից։

2. Էքսպոնենտը՝ $\exp[-\beta \times \text{հեռավորություն}^2]$Սա «տրանսֆորմացիան» է, որը հեռավորությունը դարձնում է «գրավչություն» (score)։Մենք օգտագործում ենք բացասական նշանը ($-\beta$), որպեսզի հակադարձ կապ ստեղծենք.
* որքան փոքր է հեռավորությունը, այնքան մեծ է էքսպոնենտի արժեքը։Ըստ էության, կետը «ձգվում» է դեպի այն կլաստերը, որին ավելի մոտ է։
3. Հայտարարը՝ $\sum_j \exp[-\dots]$ (Նորմալացում)Սա ամենակարևոր մասն է։ Մենք հաշվարկում ենք բոլոր հնարավոր կլաստերների ($j=1 \dots K$) «գրավչությունները» և բաժանում դրանց գումարին։Սա ապահովում է, որ բոլոր $r_k^{(n)}$ արժեքների գումարը հավասար լինի 1-ի (100%):Այսպիսով, մենք ստանում ենք հավանականությունների բաշխում, ոչ թե կամայական թվեր։
4. $\beta$ պարամետրի դերը (Կարևոր է!)$\beta$-ն ալգորիթմի «փափկության» կամ «կոշտության» կարգավորիչն է (երբեմն կոչվում է «inverse temperature»):
* Եթե $\beta$-ն շատ մեծ է: Էքսպոնենտը ավելի «շեշտված» է դառնում։ Ամենամոտ կլաստերը ստանում է գրեթե 1 պատասխանատվություն, իսկ մնացածը՝ 0։ Սա Soft K-means-ը դարձնում է նման սովորական Hard K-means-ի։
* Եթե $\beta$-ն փոքր է: Պատասխանատվությունները «ցրվում» են։ Կետը ավելի հավասարաչափ է բաշխվում կլաստերների միջև, քանի որ մոդելը դառնում է «ավելի քիչ վստահ»։


2․ Refitting (Կենտրոնների թարմացում) — $m_k$:
Սա երկրորդ քայլն է, որտեղ թարմացվում են կլաստերների կենտրոնները։Բանաձևը՝
$$m_k = \frac{\sum_n r_k^{(n)}x^{(n)}}{\sum_n r_k^{(n)}}$$

Իմաստը՝ Սա պարզ միջին չէ, այլ կշռված միջին (weighted mean)։ Նոր կենտրոնը հաշվարկվում է այնպես, որ ավելի մեծ կշիռ ունենան այն կետերը, որոնց համար $r_k^{(n)}$-ը (պատասխանատվությունը) բարձր է։

3․Կոնվերգենցիա (Repeat until convergence):Ալգորիթմը անընդհատ պտտվում է այս երկու քայլերի միջև՝ հաշվարկում է պատասխանատվությունը, հետո թարմացնում կենտրոնները, մինչև համակարգը չդառնա կայուն։



Soft K-means ալգորիթմի առավելություններն (Pros) և թերություններն (Cons)

Առավելություններ (Pros)
* Իրար համընկնող կլաստերներ (Overlapping clusters): Սա ամենակարևոր առավելություններից է․ ի տարբերություն Hard K-means-ի, այս ալգորիթմը թույլ է տալիս, որ կետերը մասամբ պատկանեն մի քանի կլաստերի, ինչը շատ ավելի իրատեսական է իրական տվյալների համար։

* Արագ և արդյունավետ (Fast and efficient): Այն հաշվարկային առումով բավականին արագ ալգորիթմ է։

* Երաշխավորված կոնվերգենցիա (Guaranteed convergence): Ալգորիթմը երաշխավորված կերպով հասնում է կայուն վիճակի (convergence)։

Թերություններ (Cons)

* Ինչպե՞ս ընտրել $\beta$ պարամետրը : $\beta$-ն այն պարամետրն է, որը վերահսկում է «փափկությունը» (softness). դժվար է լինում կանխորոշել, թե որ արժեքն է ամենաօպտիմալը կոնկրետ տվյալների համար։

* Անհավասար կշիռ կամ լայնություն ունեցող կլաստերներ (Unequal weight and width): Ալգորիթմը լավ չի աշխատում, երբ կլաստերները չափերով կամ խտությամբ տարբեր են։

Գնահատման երկու եղանակները
Մենք կլաստերավորումը գնահատում ենք երկու հիմնական ձևով.

1․ Ներքին գնահատում (Internal Evaluation):

Կիրառվում է այն դեպքում, երբ մենք չունենք նախնական պիտակներ (labels)։

Մենք գնահատում ենք կլաստերների կառուցվածքային ամբողջականությունը՝ օգտագործելով միայն տվյալների ներքին հատկությունները։

Չափման միավորներն են՝ հեռավորությունները, խտությունը և վարիացիան (variance)։

3․ Արտաքին գնահատում (External Evaluation):

Կիրառվում է այն դեպքում, երբ մենք ունենք նախնական «բենչմարկ» պիտակներ (հայտնի դասեր)։

Մենք ստուգում ենք, թե մեր կատարած կլաստերավորումը որքանով է համապատասխանում իրական, հայտնի դասերին։


Գնահատման ընտրությունը կախված է նրանից՝ ունե՞նք արդյոք նախապես հայտնի ճիշտ դասակարգում, թե՞ ոչ


Կլաստերավորման գնահատման ներքին մեթոդը՝ Inertնալ, որը հայտնի է նաև որպես WCSS (Within-Class Sum of Squares)։

Իներցիան  այն հիմնական օբյեկտիվ ֆունկցիան է, որը K-means ալգորիթմը փորձում է հասցնել նվազագույնի։ Այն չափում է, թե որքան խիտ կամ «մոտ» են դասավորված կլաստերների կետերը։
Բանաձևը՝
$$WCSS = \sum_{j=1}^{k} \sum_{x_i \in C_j} ||x_i - \mu_j||^2$$

Այս բանաձևը հաշվարկում է կլաստերի ներսում գտնվող բոլոր կետերի հեռավորությունների քառակուսիների գումարը  կլաստերի կենտրոնից($\mu_j$)
Սակայն մենք չենք կարող ուղղակիորեն օգտագործել WCSS-ը՝ տարբեր $k$ (կլաստերների քանակ) ունեցող մոդելները համեմատելու համար։ Երբ $k$-ն աճում և մոտենում է $N$-ին (տվյալների ընդհանուր քանակին), WCSS-ը ձգտում է 0-ի։ Թեև 0 արժեքը մաթեմատիկորեն կատարյալ է թվում, գործնականում այն անօգուտ է, քանի որ չի տալիս իրական պատկերացում կլաստերների մասին։
Խնդիրը լուծելու համար կիրառվում է «Արմնկային մեթոդը» (Elbow Method)։ Մենք գծապատկերի վրա կառուցում ենք WCSS-ի կախվածությունը $k$-ից և փնտրում ենք այսպես կոչված «արմնկային կետը» (inflection point), որտեղ գրաֆիկի անկումը դառնում է ավելի մեղմ։



Կլաստերավորման որակը գնահատելու հզոր ներքին մեթոդ է նաև  Silhouette Coefficient-ը (Ուրվագծի գործակից)։
Ի տարբերություն Իներցիայի (որը պարզապես չափում է հեռավորությունը կենտրոնից), Silhouette-ը հաշվի է առնում և՛ կլաստերի «խտությունը» (cohesion), և՛ այլ կլաստերներից «հեռավորությունը» (separation)։

Գործակիցը հաշվարկելու համար նախ պետք է գտնել երկու կարևոր մեծություն յուրաքանչյուր կետի համար՝ 
* $a(o)$ (Cohesion - Կապվածություն): Սա կետի միջին հեռավորությունն է իր սեփական կլաստերի մյուս բոլոր կետերից։ Որքան փոքր է այս արժեքը, այնքան ավելի լավ է կետը «միավորված» իր կլաստերի հետ։

$$a(o) = \frac{\sum_{o' \in C_i, o' \neq o} \text{dist}(o, o')}{|C_i| - 1}$$

Հաշվարկում ենք $o$ կետի հեռավորությունը ($\text{dist}$) նույն կլաստերի ($C_i$) մյուս բոլոր $o'$ կետերից։
Երբ մենք չափում ենք մի կետի ($o$) հեռավորությունը իր կլաստերի մյուս կետերից, մենք փնտրում ենք «կապվածությունը» (cohesion)՝ այսինքն, թե այդ կետը որքանով է մոտ իր «ընկերներին»։Եթե մենք չհանեինք այդ «1»-ը, մենք կստիպեինք կետին չափել հեռավորությունը նաև ինքն իրենից: Ցանկացած կետի հեռավորությունը ինքն իրենից հավասար է 0-ի:


* $b(o)$ (Separation - Տարանջատում): Սա կետի միջին հեռավորությունն է ամենամոտ հարևան կլաստերի կետերից։ Որքան մեծ է այս արժեքը, այնքան ավելի լավ է կետը տարանջատված այլ կլաստերներից։

$$b(o) = \min_{C_j : j \neq i} \left\{ \frac{\sum_{o' \in C_j} \text{dist}(o, o')}{|C_j|} \right\}$$

Նախ՝ հաշվարկում ենք $o$ կետի միջին հեռավորությունը յուրաքանչյուր հարևան կլաստերի ($C_j$) բոլոր կետերից։
Եթե մենք հաշվեինք հեռավորությունը միայն դեպի $C_j$ կլաստերի ամենամոտ կետը, մենք կստանայինք չափազանց սխալ պատկեր։ Հնարավոր է՝ միայն մեկ կետ լինի պատահաբար մոտ մեր $o$ կետին, իսկ մնացած բոլոր կետերը լինեն շատ հեռու։Ինչո՞ւ միջինը. Մենք գումարում ենք բոլոր հեռավորությունները և բաժանում կետերի քանակի վրա ($|C_j|$), որպեսզի ստանանք միջին հեռավորությունը դեպի այդ կլաստեր։ Սա ցույց է տալիս, թե $o$ կետը իրականում որքանով է «մոտ» այդ խմբին։ Սա շատ ավելի հուսալի ցուցանիշ է, քան մեկ կետի հեռավորությունը։

Ենթադրենք՝ մեր կետը ($o$) գտնվում է 1-ին կլաստերում։ Շրջակայքում կան այլ կլաստերներ՝ 2-րդ, 3-րդ, 4-րդ։
Մեր կետը կարող է շատ հեռու լինել 4-րդ կլաստերից, բայց մի փոքր ավելի մոտ լինել 2-րդ կլաստերին։ Silhouette-ի համար «վտանգը», որ կետը սխալ դասակարգված է, գալիս է այն կլաստերից, որին նա ամենամոտն է։Մինիմումը ընտրելով՝ մենք ասում ենք. «Անտեսենք բոլոր հեռու կլաստերները, դրանք մեզ չեն անհանգստացնում։ Մեզ հետաքրքրում է միայն այն ամենամոտ կլաստերը, որի հետ մեր կլաստերը կարող է շփոթվել կամ սահմանակից լինել»։

Երբ ունենք այս երկու մեծությունները, Silhouette գործակիցը ($s(i)$) հաշվարկվում է այսպես՝

$$s(i) = \frac{b(i) - a(i)}{\max(b(i), a(i))}$$

* Եթե $b(i) > a(i)$ (հեռավորությունը դեպի օտար կլաստեր ավելի մեծ է, քան դեպի սեփականը), ապա արժեքը ձգտում է 1-ի (լավ կլաստերավորում)։

* Եթե $b(i) < a(i)$ (կետն ավելի մոտ է օտար կլաստերին, քան՝ իրենը), ապա արժեքը լինում է բացասական (սխալ կլաստերավորում)։

Ստացված $s(i)$ արժեքը ցույց է տալիս, թե որքան «վստահ» կարող ենք լինել, որ կետը ճիշտ կլաստերում է՝
* $s(i) \approx 1$ (Կատարյալ): Կետը շատ մոտ է իր կլաստերի կետերին և շատ հեռու է հարևան կլաստերներից։ Սա ցույց է տալիս հստակ, լավ ձևավորված կլաստերավորում։

* $s(i) \approx 0$ (Սահմանային): Կետը գտնվում է երկու կլաստերների «սահմանագծին»։ Դժվար է ասել, թե որին է այն ավելի շատ պատկանում։

* $s(i) < 0$ (Սխալ): Կետը հավանաբար սխալ կլաստերում է հայտնվել, քանի որ այն ավելի մոտ է հարևան կլաստերի կետերին, քան՝ իր սեփականին։

Ի տարբերություն WCSS-ի (Իներցիայի), որը միշտ նվազում է, երբ $k$-ն մեծանում է, Silhouette գործակիցը թույլ է տալիս գտնել $k$-ի այն արժեքը, որի դեպքում կլաստերները և՛ իրարից լավ են տարանջատված, և՛ ներքուստ լավ են խտացված։ Սա դարձնում է այն ավելի հուսալի գործիք՝ օպտիմալ կլաստերների քանակը ընտրելու համար։




Calinski-Harabasz(որը հայտնի է նաև որպես «Variance Ratio Criterion»)։ Սա կլաստերավորման որակը գնահատելու ևս մեկ ներքին մեթոդ է, որն ունի իր առանձնահատկությունները։

Calinski-Harabasz ինդեքսը համեմատում է կլաստերների միջև եղած ցրվածությունը (between-clusters dispersion) կլաստերների ներսում եղած ցրվածության (inter-cluster/within-cluster dispersion) հետ։

Այն հաշվարկվում է հետևյալ բանաձևով՝

$$CH = \left[ \frac{Tr(B_k)}{Tr(W_k)} \right] \times \left[ \frac{N - k}{k - 1} \right]$$

* $Tr(B_k)$: Կլաստերների միջև եղած ցրվածության չափն է (Trace of the Between-cluster dispersion matrix)։
* $Tr(W_k)$: Կլաստերների ներսում եղած ցրվածության չափն է (Trace of the Within-cluster dispersion matrix)։
* $(N - k) / (k - 1)$: Նորմալացնող գործակից է, որը հաշվի է առնում տվյալների քանակը ($N$) և կլաստերների թիվը ($k$)։


($Tr$-ը նշանակում է Trace (հետք)։ Ցանկացած քառակուսային մատրիցի համար Trace-ը դրա գլխավոր անկյունագծի տարրերի գումարն է։Եթե ունենք $B_k$ մատրից, ապա $Tr(B_k)$ արդյունքը լինելու է ընդամենը մեկ թիվ (սկալյար)։Եթե ունենք $W_k$ մատրից, ապա $Tr(W_k)$ արդյունքը նույնպես լինելու է մեկ թիվ (սկալյար))

Այս մեթոդի հիմնական առավելությունը արագագործությունն է։Ի տարբերություն Silhouette ինդեքսի, որը հաշվարկելիս պահանջվում է չափել հեռավորությունը բոլոր կետերի միջև, Calinski-Harabasz-ը հաշվարկվում է շատ ավելի արագ։Սա այն դարձնում է իդեալական ընտրություն մեծածավալ տվյալների բազաների (massive datasets) համար։ Ավելի բարձր CH գնահատականը նշանակում է ավելի լավ կլաստերավորում։Սա ցույց է տալիս, որ կլաստերները ավելի լավ են սահմանազատված, ավելի խիտ են և ավելի հստակ տարանջատված միմյանցից։



Davies-Bouldin Index-ին (DB ինդեքս), որը կլաստերավորման որակը գնահատելու ևս մեկ կարևոր ներքին չափանիշ է։

Ի տարբերություն Silhouette կամ Calinski-Harabasz ինդեքսների, որոնց դեպքում ավելի մեծ արժեքը նշանակում էր ավելի լավ արդյունք, Davies-Bouldin ինդեքսի դեպքում հակառակն է՝ ավելի ցածր արժեքն է նախընտրելի։
        $$DB = \frac{1}{k} \sum_{i=1}^{k} \max_{j \neq i} \left( \frac{s_i + s_j}{d(c_i, c_j)} \right)$$

Համարիչ ($s_i + s_j$)՝ 
$s_i$ և $s_j$ արժեքները ներկայացնում են տվյալ կլաստերների ներքին ցրվածությունը (internal scatter)։

Պատկերացնեք կլաստերը որպես կետերի մի կույտ։ $s_i$-ն այդ կույտի «շառավիղն» է կամ «հաստությունը»։ Մենք գումարում ենք երկու հարևան կլաստերների չափերը, որպեսզի հասկանանք՝ որքան «տեղ» են դրանք ընդհանուր առմամբ զբաղեցնում տարածության մեջ։ Եթե այս գումարը մեծ է, ուրեմն կլաստերները շատ ցրված են (հաստ են)։ Որքան հաստ են կլաստերները, այնքան ավելի մեծ է հավանականությունը, որ դրանք կսկսեն «հպվել» կամ խառնվել իրար հետ։

Հայտարար ($d(c_i, c_j)$)՝ 
Սա երկու կլաստերների կենտրոնների (centroids) միջև եղած հեռավորությունն է։ Սա այն «բաց տարածությունն» է, որը առանձնացնում է մեկ կլաստերի կենտրոնը մյուսից։ Եթե այս արժեքը մեծ է, ուրեմն կլաստերները իրարից հեռու են։ Որքան հեռու են կենտրոնները, այնքան ավելի քիչ է հավանականությունը, որ կլաստերները կխառնվեն։ 



$$\frac{s_i + s_j}{d(c_i, c_j)}$$

Այս ամբողջ կոտորակը ցույց է տալիս «խառնվելու վտանգը»։ Եթե Ունենք շատ մեծ/հաստ կլաստերներ (մեծ համարիչ), որոնք իրար շատ մոտ են (փոքր հայտարար)։ Սա վատ է։ Եթե  Ունենք կոմպակտ/փոքր կլաստերներ (փոքր համարիչ), որոնք իրարից հեռու են (մեծ հայտարար)։ Սա լավ է։

Բանաձևի այս հատվածը՝ $\max_{j \neq i} \left( \dots \right)$, ասում է հետևյալը.«Վերցրու կոնկրետ $i$ կլաստերը և համեմատիր այն բոլոր մնացած կլաստերների ($j$) հետ: Գտիր այն մեկ $j$ կլաստերը, որի հետ $i$-ն ունի ամենամեծ "խառնվելու վտանգը" (ամենամեծ արժեքը):

Բայց մենք չենք կարող մեր մոդելի որակը դատել միայն մեկ կլաստերի հիման վրա։ Մեզ պետք է իմանալ, թե ինչպես է գործում ամբողջ համակարգը։$\sum_{i=1}^{k}$ նշանը «անցնում» է բոլոր կլաստերների վրայով՝ $i=1$-ից մինչև $k$։Այն վերցնում է 1-ին կլաստերի «վատագույն ցուցանիշը», 2-րդի «վատագույն ցուցանիշը», ..., մինչև $k$-րդ կլաստերը։Այնուհետև գումարում է դրանք բոլորը իրար, որպեսզի ստանա մեկ ընդհանուր թիվ, որն արտացոլում է ամբողջ մոդելի «կուտակված խնդիրները»։  Եթե մենք պարզապես գումարեինք արժեքները, մենք կունենայինք մի խնդիր՝ արժեքը կախված կլիներ կլաստերների քանակից ($k$):Եթե ունենք 10 կլաստեր, գումարը, հավանաբար, շատ ավելի մեծ կլիներ, քան եթե ունենայինք 2 կլաստեր։Դա անարդար կլիներ տարբեր մոդելներ համեմատելիս։$\frac{1}{k}$ գործակիցը կատարում է նորմալացում (normalization)։Այն բաժանում է ընդհանուր գումարը կլաստերների քանակի վրա ($k$):Արդյունքում մենք ստանում ենք միջին «վատագույն ցուցանիշը» մեկ կլաստերի հաշվով։

Adjusted Rand Index (ARI)-ին, որը կլաստերավորման որակը գնահատելու արտաքին (External) մեթոդ է։

Ի տարբերություն մյուս բոլոր մեթոդների, որոնք դիտարկեցինք, այս մեկը կիրառվում է միայն այն դեպքում, երբ ունենք «ճշգրիտ պատասխաններ» (ground truth labels), այսինքն՝ նախապես գիտենք, թե տվյալները իրականում ինչ խմբերի պետք է պատկանեն։

Դասական դասակարգման խնդիրներում մենք սովորաբար օգտագործում ենք ճշգրտություն (Accuracy), սակայն կլաստերավորման դեպքում դա հնարավոր չէ։Կլաստերների ID-ները (օրինակ՝ «Cluster 0», «Cluster 1») պատահական են և կախված չեն իրական դասերի անվանումներից (օրինակ՝ «Class A», «Class B»)։Այսինքն՝ մոդելի «Cluster 0»-ը կարող է իրականում համապատասխանել «Class B»-ին, և սա սխալ չէ, պարզապես պիտակները չեն համընկնում։

ARI-ն աշխատում է զույգերի համեմատության (combinatorics) սկզբունքով։Ալգորիթմը դիտարկում է տվյալների բոլոր հնարավոր զույգերը։Եթե երկու օրինակ (sample) իրականում նույն դասին են պատկանում ԵՎ մեր կլաստերավորման արդյունքում հայտնվել են նույն կլաստերում, ապա դա համարվում է «հաջողություն» (success)։Իսկ «Adjusted» (ճշգրտված) բառը նշանակում է, որ այս ինդեքսը հաշվի է առնում պատահականության գործոնը՝ արդյունքը նորմալացնելով այնպես, որ այն դառնա ավելի ճշգրիտ։

ARI-ի արժեքը սովորաբար տատանվում է -1-ից մինչև 1
* ARI = 1.0: Կատարյալ համընկնում. ձեր կլաստերավորումը լիովին համապատասխանում է իրական դասերին։
* ARI $\approx$ 0.0: Ձեր կլաստերավորումը ոչնչով չի տարբերվում պատահական գուշակությունից։
* ARI < 0.0: Արդյունքը նույնիսկ ավելի վատ է, քան պատահականությունը (կլաստերները սխալ են ձևավորված)։



Normalized Mutual Information (NMI)-ին, որը կլաստերավորման գնահատման արտաքին (External) մեթոդ է՝ փոխառված ուղղակիորեն ինֆորմացիոն տեսությունից (Information Theory)։

$$NMI(Y, C) = \frac{2 \cdot I(Y; C)}{H(Y) + H(C)}$$

NMI-ը կլաստերավորման որակը գնահատելու հանրահայտ մեթոդ է: Այն օգնում է պատասխանել հետևյալ հարցին. «Որքանո՞վ են մեր կողմից ստացված կլաստերները համընկնում տվյալների իրական դասակարգման հետ»:
Հասկանալու համար հիմնական բաղադրիչները՝
$I(Y; C)$ (Փոխադարձ ինֆորմացիա). Սա չափում է, թե որքան տեղեկատվություն է «կիսում» ձեր կլաստերավորումը ($C$) իրական դասերի ($Y$) հետ: Եթե ձեր կլաստերավորումը լավն է, ապա իմանալով, թե որ կլաստերում է օբյեկտը, դուք կիմանաք նաև նրա իրական դասի մասին:
$H(Y)$ և $H(C)$ (Էնտրոպիա). Սա չափում է անորոշությունը:$H(Y)$՝ որքան «անհավասար» կամ խառն են իրական դասերը:$H(C)$՝ որքան «անհավասար» են ձեր կողմից ստեղծված կլաստերները:Նորմալացում (Normalization). Բանաձևի համարիչում առկա $2$ գործակիցը և հայտարարում առկա էնտրոպիաների գումարը ապահովում են, որ արդյունքը միշտ լինի $0$-ից $1$-ի սահմաններում: Սա շատ հարմար է, քանի որ թույլ է տալիս համեմատել տարբեր ալգորիթմների արդյունավետությունը անկախ դասերի կամ կլաստերների քանակից:
Պարզ օրինակ՝
* NMI = 0. Ձեր կլաստերավորումը բացարձակապես որևէ կապ չունի իրական դասերի հետ (ինչպես պատահական գուշակությունը):
* NMI = 1. Ձեր կլաստերավորումը կատարյալ է. յուրաքանչյուր կլաստեր պարունակում է միայն մեկ դասի պատկանող օբյեկտներ, և յուրաքանչյուր դաս ամբողջությամբ առանձնացված է
