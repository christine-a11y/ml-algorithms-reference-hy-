Gaussian Mixture Models


 «Կոշտ նշանակման խնդիրը» (The Hard Assignment Problem)

1. Ի՞նչ է «Կոշտ նշանակումը» (Hard Assignment)


Ստանդարտ K-Means ալգորիթմում յուրաքանչյուր տվյալի կետ ($x_i$) պետք է պատկանի միայն մեկ կլաստերի ($k$): Սա արտահայտվում է $r_{ik}$ պատասխանատվության գործակցով:

Բուլյան արժեք: $r_{ik} \in \{0, 1\}$:

Եթե կետը պատկանում է կլաստերին, արժեքը 1 է, եթե ոչ՝ 0: Սա նշանակում է, որ ալգորիթմը թույլ չի տալիս «մասնակի» պատկանելություն:

2. Երկրաչափական թերությունը (The Geometric Flaw)

Սակայն առաջանում է մի էական խնդիր,  երբ կետը գտնվում է երկու կլաստերների կենտրոններից (centroids) հավասար հեռավորության վրա:

Անորոշության անտեսում: Նույնիսկ եթե կետը գտնվում է հենց մեջտեղում, K-Means-ը ստիպում է այն վերագրել կլաստերներից մեկին (տալով 1 կամ 0 արժեք):

Բացարձակ վստահություն: Ալգորիթմը ենթադրում է բացարձակ վստահություն նույնիսկ որոշումների սահմանագծին (decision boundaries), որտեղ իրականում կա մեծ անորոշություն:

3. Նպատակը. «Փափուկ նշանակում» (Soft Assignment)

Որպեսզի լուծվի այս խնդիրը, առաջարկվում է անցում կատարել ավելի ճկուն մոդելի:

Շարունակական արժեք: $r_{ik}$-ն պետք է լինի 0-ից 1-ի միջև ընկած ցանկացած թիվ (օրինակ՝ 0.7 պատկանելություն մի կլաստերին և 0.3՝ մյուսին):

Պատկանելության աստիճան: Սա թույլ է տալիս արտահայտել, թե որքանով է կետը համապատասխանում տվյալ կլաստերին:

Սահմանափակում: Ցանկացած $i$ կետի համար բոլոր կլաստերներին պատկանելության գումարը պետք է լինի 1.$$\sum_{k} r_{ik} = 1$$


Soft K-Means (կամ Fuzzy K-Means) ալգորիթմի հիմքում ընկած մաթեմատիկական մոտեցումը, որը լուծում է նախորդ սլայդում նշված «կոշտ» նշանակման խնդիրը:

1. Ինչպե՞ս հաշվել շարունակական $r_{ik}$-ն


Որպեսզի կետը ստանա մասնակի պատկանելություն, օգտագործվում է Softmax ֆունկցիան: Ալգորիթմը վերցնում է կետի և կլաստերի կենտրոնի հեռավորության քառակուսին բացասական նշանով (որքան փոքր է հեռավորությունը, այնքան մեծ պետք է լինի պատկանելությունը):

2. Կոշտության պարամետրը ($\beta$)

Ներմուծվում է $\beta$ գործակիցը, որը կոչվում է stiffness parameter (կոշտության պարամետր): Այն որոշում է, թե որքան «զգայուն» է ալգորիթմը հեռավորությունների նկատմամբ:

Soft Responsibility բանաձևը  ունի հետևյալ տեսքը՝

$$r_{ik} = \frac{\exp(-\beta ||x_i - \mu_k||^2)}{\sum_{j} \exp(-\beta ||x_i - \mu_j||^2)}$$

* $r_{ik}$ (Responsibility): Սա այն հավանականությունն է կամ «պատասխանատվության աստիճանը», որ $x_i$ կետը վերագրվում է $k$ կլաստերին: Ի տարբերություն ստանդարտ K-Means-ի, այստեղ $r_{ik}$ արժեքը գտնվում է $[0, 1]$ միջակայքում:

* $||x_i - \mu_k||^2$: Սա կետի և կլաստերի կենտրոնի ($\mu_k$) միջև Էվկլիդեսյան հեռավորության քառակուսին է:

* $\exp(-\beta \dots)$: Սա էքսպոնենցիալ ֆունկցիա է, որի մեջ հեռավորությունը դրվում է բացասական նշանով: Սա նշանակում է՝ որքան փոքր է հեռավորությունը կետի և կենտրոնի միջև, այնքան մեծ կլինի էքսպոնենտի արժեքը և, հետևաբար, պատկանելության հավանականությունը:

* $\beta$ (Stiffness parameter): Սա կառավարում է ալգորիթմի «կոշտությունը»:


2. Ինչպե՞ս է աշխատում Softmax-ը

Բանաձևը կառուցվածքով Softmax ֆունկցիա է:Համարիչը հաշվում է կոնկրետ $k$ կլաստերի համար «կողմ» լինելու կշիռը:

Հայտարարը ($\sum_{j} \dots$) հաշվում է բոլոր հնարավոր կլաստերների համար ստացված կշիռների գումարը:

Այս բաժանման շնորհիվ ստացվում է նորմալիզացիա, ինչը երաշխավորում է, որ բոլոր կլաստերների համար պատկանելության գումարը լինի ուղիղ 1:

$$\sum_{k} r_{ik} = 1$$

3. $\beta$ պարամետրի ազդեցությունը (Inverse Temperature)$\beta$-ն աշխատում է որպես ջերմաստիճանի հակադարձ մեծություն։


![alt text](image.png)


Այս մոտեցումը լուծում է երկրաչափական թերությունը, երբ կետը գտնվում է երկու կենտրոնների միջև: Այն թույլ է տալիս ալգորիթմին «խոստովանել», որ մոդելն ունի անորոշություն, ինչն ավելի ճշգրիտ է տվյալների իրական բաշխումները նկարագրելիս:


Քանի որ մենք արդեն ունենք պատասխանատվության շարունակական արժեքներ ($r_{ik}$), կենտրոնների թարմացման տրամաբանությունը փոխվում է։


1. Կշռված միջին (Weighted Average)

Ստանդարտ K-Means-ում կենտրոնը ($\mu_k$) պարզապես տվյալ կլաստերին պատկանող կետերի թվաբանական միջինն է: Սակայն Soft K-Means-ում կենտրոնը դառնում է ամբողջ տվյալների բազայի բոլոր կետերի կշռված միջինը:

Յուրաքանչյուր կետ ($x_i$) մասնակցում է $k$ կենտրոնի ձևավորմանը՝ կախված իր $r_{ik}$ պատասխանատվության աստիճանից:

Բանաձևը՝

$$\mu_k = \frac{\sum_{i=1}^{N} r_{ik}x_i}{\sum_{i=1}^{N} r_{ik}}$$

* $\mu_k$: $k$-րդ կլաստերի նոր կենտրոնն է (centroid):

* $x_i$: Տվյալների բազայի $i$-րդ կետն է:

* $r_{ik}$: $i$-րդ կետի պատասխանատվության գործակիցն է (responsibility) $k$ կլաստերի համար: Սա այն շարունակական արժեքն է $[0, 1]$ միջակայքում, որը մենք հաշվարկել էինք նախորդ քայլում:

* $N$: Տվյալների բազայում կետերի ընդհանուր քանակը:2. Համարիչի տրամաբանությունը ($\sum_{i=1}^{N} r_{ik}x_i$)Ի տարբերություն ստանդարտ K-Means-ի, որտեղ հաշվի են առնվում միայն տվյալ կլաստերին պատկանող կետերը, այստեղ բոլոր $N$ կետերը մասնակցում են կենտրոնի ձևավորմանը:Յուրաքանչյուր կետ բազմապատկվում է իր $r_{ik}$ կշռով:Եթե կետը բարձր հավանականությամբ պատկանում է $k$ կլաստերին ($r_{ik} \approx 1$), այն ուժեղ «քաշում» է կենտրոնին դեպի իրեն:
Եթե կետը գրեթե չի պատկանում այդ կլաստերին ($r_{ik} \approx 0$), դրա ազդեցությունը կենտրոնի տեղաշարժի վրա աննշան է:

3. Հայտարարի տրամաբանությունը ($\sum_{i=1}^{N} r_{ik}$)

Սա բոլոր կետերի պատասխանատվությունների գումարն է $k$ կլաստերի համար:

* Այն կարելի է դիտարկել որպես տվյալ կլաստերին պատկանող կետերի «էֆեկտիվ քանակ»:

* Հայտարարի վրա բաժանելը նորմալիզացնում է արդյունքը, որպեսզի նոր կենտրոնը ($\mu_k$) լինի տվյալների կշռված միջինը, այլ ոչ թե պարզապես գումարը:

Ինչո՞վ է սա տարբերվում ստանդարտ K-Means-ից

* Կոշտ (Hard) K-Means: Կենտրոնը պարզ միջինն է միայն այն կետերի, որոնց համար $r_{ik} = 1$: Մնացած բոլոր կետերը ($r_{ik} = 0$) ամբողջությամբ անտեսվում են:

* Փափուկ (Soft) K-Means: Կենտրոնը հաշվի է առնում բոլոր կետերը, ինչը թույլ է տալիս կենտրոններին ավելի ճկուն կերպով տեղաշարժվել և չկառչել կոշտ սահմանագծերից:


![alt text](image-1.png)

Պատկերը հիանալի կերպով ցույց է տալիս «կոշտ» (Hard) և «փափուկ» (Soft) կլաստերիզացիայի միջև եղած տեսանելի տարբերությունը, հատկապես այն կետերի համար, որոնք գտնվում են սահմանագծին:

1. Hard Clustering (Ձախ կողմում)
Բացարձակ պատկանելություն: Այս տարբերակում յուրաքանչյուր կետ պատկանում է կամ կապույտ, կամ նարնջագույն կլաստերին:

Test Location (Դեղին խաչը): Տեսնում ենք, որ դեղին խաչով նշված կետը գտնվում է երկու խմբերի արանքում: Սակայն «կոշտ» կլաստերիզացիայի դեպքում այն 100%-ով վերագրվել է կապույտ կլաստերին:

Թերությունը: Մոդելը լիովին անտեսում է այն փաստը, որ կետը շատ մոտ է նաև նարնջագույն խմբին:

2. Soft Clustering (Աջ կողմում)
Հավանականային պատկանելություն: Այստեղ կետերը չունեն հստակ գույն. դրանք ավելի բաց կամ թափանցիկ են՝ կախված իրենց պատկանելության աստիճանից:

Test Location (Դեղին խաչը): Նույն կետը այժմ ունի բաշխված պատասխանատվություն. 45% կապույտ և 55% նարնջագույն: Սա արտացոլում է իրական պատկերը՝ կետը մի փոքր ավելի մոտ է նարնջագույնին, բայց կապույտի ազդեցությունը նույնպես մեծ է:

Առավելությունը: Սա թույլ է տալիս խուսափել սխալներից այնտեղ, որտեղ տվյալները միախառնված են:

$$\mu_k = \frac{\sum_{i=1}^{N} r_{ik}x_i}{\sum_{i=1}^{N} r_{ik}}$$

բացատրում է, թե ինչպես է հաշվվում կենտրոնը (centroid): Աջ կողմի նկարում կենտրոնների դիրքը որոշելիս դեղին կետը իր 45% «կշռով» կմասնակցի կապույտ կենտրոնի ձևավորմանը և 55%-ով՝ նարնջագույնի:


Soft K-Means-ի սահմանափակումները


Իրական հավանականությունների բացակայությունը

Soft K-Means-ը տալիս է մասնակի (fractional) նշանակումներ, բայց դրանք իրական հավանականություններ չեն։

Արժեքները պարզապես հեռավորություններ են, որոնք անցել են կամայական էքսպոնենցիալ հարթեցնող ֆունկցիայի (softmax) միջով։ Սա կոչվում է heuristics (էվրիստիկա), քանի որ այն հիմնված չէ տվյալների բաշխվածության խիստ մաթեմատիկական ենթադրության վրա։

Երկրաչափական հետևանքը (The Geometric Consequence)

Չնայած «փափուկ» մոտեցմանը, Soft K-Means-ը դեռևս հիմնվում է Էվկլիդեսյան հեռավորության ($||x_i - \mu_k||^2$) վրա։Սա նշանակում է, որ ալգորիթմը դեռևս ենթադրում է, որ բոլոր կլաստերները կատարյալ գնդաձև են (spherical) և ունեն նույն շառավիղը։Իրական կյանքում տվյալների խմբերը հաճախ լինում են ձգված (էլիպսաձև) կամ ունենում են տարբեր խտություններ և չափսեր։

Կլաստերիզացիայի գեներատիվ հայեցակարգը

Այս մոտեցումը ներկայացնում է մեթոդաբանական անցում տվյալների երկրաչափական մշակումից դեպի վիճակագրական մոդելավորում:

1. Օբյեկտիվ չափման անհրաժեշտությունը
Մեզ անհրաժեշտ է տվյալները լավ կլաստերիզացնելու իմաստավորված մաթեմատիկական չափանիշ (sensible, mathematical measure): Սա թույլ է տալիս՝

* Օբյեկտիվ գնահատում (Objective judgment): Հնարավորություն է տալիս տարբեր մեթոդները դատել անկողմնակալ կերպով:

* Մոդելի ընտրություն (Model Selection): Օգնում է որոշել կլաստերների քանակը (number of clusters):

2. Մտածելակերպի փոփոխություն (The Paradigm Shift)
Այս մոտեցման հիմքում ընկած է այն գաղափարը, որ տվյալները ստեղծվել են ինչ-որ թաքնված գեներատիվ մոդելի (hidden generative model) կողմից:

Մենք դիտարկում ենք տվյալները ոչ թե որպես պատահական կետեր, այլ որպես մի գործընթացի արդյունք, որն ունի իր ներքին կանոններն ու բաշխումները:

3. Առավելագույն նմանության սկզբունքը
Եթե մենք կարողանանք սահմանել, թե ինչպես է «տիեզերքը» ստեղծել այդ տվյալները, մենք կարող ենք հարմարեցնել մեր մոդելի պարամետրերը (model parameters)՝ օգտագործելով Առավելագույն նմանության մեթոդը (Maximum Likelihood):

Նպատակը: Առավելագույնի հասցնել այն հավանականությունը (maximizing the probability), որ մեր մոդելը կստեղծեր հենց այն տվյալները, որոնք մենք դիտարկել ենք (observed data):

Գաուսյան խառնուրդների մոդելները (Gaussian Mixture Models - GMMs)

Այստեղ  տվյալների առաջացումը բացատրվում է հավանականային բաշխումների միջոցով:

1. Տվյալների առաջացման ենթադրությունըՄոդելը ենթադրում է, որ $D$ տվյալները ստեղծվել են $K$ տարբեր բազմաչափ նորմալ բաշխումների (Multivariate Normal distributions) խառնուրդից: Յուրաքանչյուր տվյալի կետ ($x$) առաջանում է երկփուլ գործընթացով։

1․ Կլաստերի ընտրություն: Ընտրվում է $z$ կլաստերը $\{1, \dots, K\}$ բազմությունից այնպես, որ $p(z=k) = \pi_k$: Այստեղ $\pi_k$-ն կոչվում է խառնուրդի կշիռ (mixing weight):

2․ Կետի ընտրանք (sampling): Երբ $z$-ը հայտնի է, $x$ կետը ստացվում է համապատասխան Գաուսյան բաշխումից՝ $\mathcal{N}(x|\mu_k, \Sigma_k)$:

2. Համատեղ բաշխում և պարամետրերԱյս գործընթացը սահմանում է համատեղ բաշխում (joint distribution)՝ $p(z, x) = p(z)p(x|z)$, որն ունի հետևյալ պարամետրերը ($\theta$)՝

Համատեղ բաշխման իմաստը ($p(z, x)$)Սա հարցնում է՝ «Որքա՞ն է հավանականությունը, որ տիեզերքը կընտրի կոնկրետ $z$ խումբը և դրա ներսում կստեղծի հենց այս $x$ կետը»:

Բանաձևը բաժանված է երկու մասի՝

* $p(z)$ (Prior): Սա կլաստերի ընտրության հավանականությունն է՝ նախքան կետը տեսնելը:

* $p(x|z)$ (Likelihood): Սա հավանականությունն է, որ $x$ կետը կառաջանա հենց ընտրված $z$ կլաստերի կողմից:

Մոդելի պարամետրերը ($\theta$)

Որպեսզի այս ամբողջ մեխանիզմն աշխատի, մեզ պետք են հատուկ «կարգավորումներ» կամ պարամետրեր, որոնք նշանակվում են հունական $\theta$ (թետա) տառով:
Դրանք են՝
* $\pi_k$ (Mixing Weights): Սա ցույց է տալիս յուրաքանչյուր խմբի «չափը» կամ կշիռը: Օրինակ՝ եթե մի խումբը շատ մեծ է, ապա $\pi_k$-ն մեծ կլինի, ինչը նշանակում է, որ նոր կետ ստեղծելիս տիեզերքն ավելի հաճախ կընտրի հենց այդ խումբը:
* $\mu_k$ (Mean): Սա $k$-րդ կլաստերի կենտրոնն է: Այն որոշում է, թե որտեղ է գտնվում կլաստերը տարածության մեջ:
* $\Sigma_k$ (Covariance Matrix): Սա ամենակարևոր մասն է, որը տարբերում է GMM-ը K-Means-ից: Այն որոշում է կլաստերի ձևը և կողմնորոշումը: Դրա շնորհիվ կլաստերը կարող է լինել ձգված էլիպս, այլ ոչ թե պարզապես կլոր շրջան:

3. Մարգինալ հավանականություն (The True Marginal Probability)

Ցանկացած դիտարկված $x$ կետի ճշմարիտ մարգինալ հավանականությունը բոլոր հնարավոր թաքնված կլաստերների հանրագումարն է: 

Հիմնական բանաձևն է՝

$$p(x) = \sum_{z} p(z, x) = \sum_{k=1}^{K} \pi_k \mathcal{N}(x|\mu_k, \Sigma_k)$$

Այս բանաձևն ասում է՝ «Որքա՞ն է հավանականությունը, որ մենք կտեսնենք հենց $x$ կետը»։

$\mathcal{N}(x|\mu_k, \Sigma_k)$: Սա հարցնում է՝ «Եթե մենք վերցնեինք $k$-րդ խումբը, որքա՞ն հավանական կլիներ, որ նա կծներ հենց այս $x$ կետը»։

$\pi_k$: Սա այդ խմբի ընդհանուր «կշիռն» է։

$\sum$ (Գումար): Քանի որ մենք չգիտենք, թե իրականում որ խումբն է ծնել այդ կետը (դա թաքնված է), մենք պետք է գումարենք բոլոր $K$ խմբերի հավանականությունները։


Պարզաբանում 
1. Ի՞նչ է «Գեներատիվ մոդելը»

Սովորաբար մենք ունենք տվյալներ և փորձում ենք դրանք բաժանել խմբերի։ Գեներատիվ մոդելն աշխատում է հակառակ ուղղությամբ. այն փորձում է բացատրել, թե ինչպե՞ս է բնությունը ստեղծել այդ տվյալները:

2. Երկփուլ գործընթացը (Ինչպես է ծնվում կետը)

Պատկերացրու՝ տիեզերքն ունի մի քանի «խմորիչ» (կլաստերներ)։ Յուրաքանչյուր նոր կետ ստեղծելիս նա անում է երկու քայլ.

Քայլ 1. Ընտրում է կլաստերը: Տիեզերքը նետում է «զառը»։ Յուրաքանչյուր խումբ ունի իր հավանականությունը ($\pi_k$)։ Օրինակ՝ եթե $\pi_1 = 0.7$, ապա 70% դեպքերում կետը կստեղծվի առաջին խմբի կանոններով։

Քայլ 2. Ստեղծում է կետը: Երբ խումբն արդեն ընտրված է, տիեզերքը վերցնում է այդ խմբի «կաղապարը» (Գաուսյան բաշխումը) և դրա ներսում մի պատահական կետ է դնում։ Այդ կետի տեղը կախված է խմբի կենտրոնից ($\mu_k$) և դրա ձևից ($\Sigma_k$)։

Ինչո՞վ է սա ավելի լավ, քան K-Means-ը

Ճկունություն: K-Means-ը կարծում է, որ բոլոր խմբերը կլոր են։ GMM-ը, շնորհիվ $\Sigma_k$ պարամետրի, թույլ է տալիս կլաստերներին լինել ձգված էլիպսներ։

Անորոշություն: Սա տալիս է «ճշմարիտ հավանականություն»։ Մենք չենք ասում՝ «կետը 100% այստեղ է», մենք ասում ենք՝ «կա 80% հավանականություն, որ այն ծնվել է այս էլիպսի կողմից»։

Կարճ ասած՝ GMM-ը փորձում է գտնել այնպիսի «ամպեր» (բաշխումներ), որոնց տակ ամենալավը կտեղավորվեն ձեր տվյալները։


Հետին հավանականություն (Posterior Probability)


$p(z=k|x)$ բանաձևը ցույց է տալիս ճշմարիտ հետին հավանականությունը: Այն մեզ ասում է՝ «Տեսնելով $x$ կետը, որքա՞ն է հավանականությունը, որ այն պատկանում է $k$-րդ կլաստերին»: Սա այն «պատասխանատվությունն» է (responsibility), որը յուրաքանչյուր կլաստեր ստանձնում է տվյալ կետի համար:

Լոգարիթմական նմանության ֆունկցիան (Log-Likelihood)

Մեր նպատակն է գտնել այնպիսի պարամետրեր ($\theta$), որոնք առավելագույնի կհասցնեն տվյալների ստացման հավանականությունը: Բանաձևը հետևյալն է՝

$$\ln p(X|\theta) = \sum_{i=1}^{N} \ln \left( \sum_{k=1}^{K} \pi_k \mathcal{N}(x_i|\mu_k, \Sigma_k) \right)$$

* $\ln p(X|\theta)$: Սա կոչվում է Լոգարիթմական նմանության ֆունկցիա (Log-Likelihood): Մեր նպատակն է առավելագույնի հասցնել այս արժեքը՝ գտնելով լավագույն պարամետրերը ($\theta$) մեր ամբողջ տվյալների բազմության ($X$) համար:

* $\sum_{i=1}^{N}$: Այս գումարը նշանակում է, որ մենք դիտարկում ենք բոլոր $N$ տվյալների կետերը հերթով:

* $\ln (...)$: Լոգարիթմն օգտագործվում է հաշվարկները հեշտացնելու համար, քանի որ այն հավանականությունների արտադրյալը վերածում է գումարի:

$\sum_{k=1}^{K} \pi_k \mathcal{N}(x_i|\mu_k, \Sigma_k)$: Սա ներկայացնում է $i$-րդ կետի առաջացման հավանականությունը բոլոր հնարավոր $K$ կլաստերների միջոցով:

«Պատը» (The Wall) – Ինչո՞ւ է սա անլուծելի

Գումարը լոգարիթմի ներսում է: Կլաստերների վրա կատարվող գումարը ($\sum$ ըստ $k$-ի) գտնվում է $\ln$-ի ներսում, ինչը թույլ չի տալիս լոգարիթմին «չեզոքացնել» էքսպոնենցիալ ֆունկցիաները Գաուսյան բաշխման մեջ:

Ածանցյալի բարդությունը: Երբ փորձում ենք վերցնել այս ֆունկցիայի ածանցյալը պարամետրերի նկատմամբ և հավասարեցնել զրոյի, ստացված հավասարումները չունեն փակ տեսքի հանրահաշվական լուծում:

Ուղղակի հաշվարկի անհնարինություն: Մենք չենք կարող պարզապես մեկ բանաձևով հաշվել և ստանալ վերջնական պատասխանը:


Մեր խնդիրն այն է, որ գումարը «փակված» էր լոգարիթմի ներսում։ Որպեսզի լուծենք սա, մենք կառուցում ենք մեր նպատակային ֆունկցիայի ստորին սահմանը (lower bound)։

Յենսենի անհավասարությունը (Jensen's Inequality)Ցանկացած խիստ գոգավոր (concave) ֆունկցիայի համար (ինչպիսին լոգարիթմն է) և ցանկացած պատահական փոփոխականի համար գործում է հետևյալ կանոնը՝

$$f(\mathbb{E}[Y]) \geq \mathbb{E}[f(Y)]$$

Պարզ ասած «Սպասման լոգարիթմը միշտ մեծ է կամ հավասար լոգարիթմի սպասմանը»։

Այս անհավասարությունը մեզ թույլ է տալիս լոգարիթմը «մտցնել» գումարի ներս։ Թեև մենք չենք կարողանում ուղղակիորեն լուծել բարդ ֆունկցիան, մենք փոխարենը լուծում ենք դրա ստորին սահմանը, որն ավելի հեշտ է հաշվարկվում։

![alt text](image-3.png)

Գրաֆիկը ցույց է տալիս, որ կոր ֆունկցիայի արժեքը երկու կետերի միջինում ($f(tx_1 + (1-t)x_2)$) միշտ ավելի բարձր է (կամ հավասար), քան այդ կետերի ֆունկցիաների արժեքների միջինը ($tf(x_1) + (1-t)f(x_2)$)։



Կուլբակ-Լեյբլեր (KL) շեղումը (Kullback-Leibler Divergence)

Այն հայտնի է նաև որպես Հարաբերական էնտրոպիա: Այն առանցքային գործիք է մեքենայական ուսուցման մեջ, հատկապես Gaussian Mixture Models (GMM) օպտիմալացման ժամանակ:

Ի՞նչ է KL շեղումը

KL շեղումը մաթեմատիկական մեթոդ է, որը թույլ է տալիս չափել «հեռավորությունը» երկու հավանականային բաշխումների միջև:

* $P(x)$: Սա ներկայացնում է տվյալների «ճշմարիտ» կամ թիրախային բաշխումը:

* $Q(x)$: Սա մեր մոդելի կողմից առաջարկվող բաշխումն է, որով փորձում ենք մոտարկել (approximate) իրականությունը:

KL շեղումը չափում է կորսված տեղեկատվությունը, երբ մենք օգտագործում ենք $Q$ բաշխումը $P$-ի փոխարեն:


Այն սովորաբար սահմանվում է որպես՝

$$D_{KL}(P \parallel Q) = \sum_{x \in X} P(x) \log \left( \frac{P(x)}{Q(x)} \right)$$

Սա կարելի է հասկանալ որպես $P$ և $Q$ բաշխումների միջև լոգարիթմական տարբերության սպասվող արժեքը:

Հիմնական հատկությունը՝ Ոչ-բացասականություն

KL շեղման կարևորագույն հատկությունը՝ $D_{KL}(P \parallel Q) \geq 0$։
* Արժեքը միշտ մեծ է կամ հավասար զրոյի:

* Այն ճիշտ զրո է միայն այն դեպքում, երբ $P$ և $Q$ բաշխումները լիովին նույնական են:


Մեքենայական ուսուցման մեջ մեր նպատակն է սովորաբար նվազեցնել (մինիմիզացնել) այս շեղումը: Որքան փոքր է KL շեղման արժեքը, այնքան մեր $Q$ մոդելն ավելի լավ է նկարագրում իրական $P$ տվյալները: Սա հատկապես կարևոր է Gaussian Mixture Models-ի դեպքում, որտեղ մենք փորձում ենք գտնել այնպիսի պարամետրեր, որոնք լավագույնս կհամապատասխանեն տվյալների բաշխմանը:

Կարևոր նշում: KL շեղումը «հեռավորություն» է կոչվում չակերտների մեջ, քանի որ այն սիմետրիկ չէ ($D_{KL}(P \parallel Q) \neq D_{KL}(Q \parallel P)$), հետևաբար մաթեմատիկական իմաստով այն մետրիկա չէ:

1. Ինչո՞ւ են օգտագործվում զուգահեռ գծերը (||) ստորակետի փոխարեն

$D_{KL}(P || Q)$ գրելաձևը հատուկ ընտրված է շեշտելու համար, որ սա սիմետրիկ հեռավորության չափման միավոր չէ:

* Ստանդարտ երկրաչափությունում հեռավորությունը A-ից B նույնն է, ինչ B-ից A:

* Հավանականությունների տեսությունում $P$-ի շեղումը $Q$-ից նույնը չէ, ինչ $Q$-ի շեղումը $P$-ից:

* Մաթեմատիկորեն դա արտահայտվում է այսպես. $D_{KL}(P || Q) \neq D_{KL}(Q || P)$:

Սա նշանակում է, որ կարևոր է, թե որ բաշխումն ենք դիտարկում որպես «ճշմարիտ» ($P$) և որը որպես «մոտարկում» ($Q$):

2. Մաթեմատիկական բանաձևերը

Կախված նրանից, թե ինչպիսի տվյալների հետ ենք աշխատում, օգտագործվում են տարբեր բանաձևեր։

Ա) Դիսկրետ բանաձև (The Discrete Formula)

Երբ փոփոխականները ընդունում են առանձին, հաշվելի արժեքներ (օրինակ՝ զառի գլորման արդյունքները), օգտագործվում է գումարման նշանը ($\sum$)՝

$$D_{KL}(P || Q) = \sum_{x} P(x) \log \left( \frac{P(x)}{Q(x)} \right)$$

Բ) Անընդհատ բանաձև (The Continuous Formula)

Երբ փոփոխականները կարող են ընդունել ցանկացած արժեք որոշակի միջակայքում (օրինակ՝ մարդու հասակը կամ ջերմաստիճանը), օգտագործվում է ինտեգրալը ($\int$)՝

$$D_{KL}(P || Q) = \int P(x) \log \left( \frac{P(x)}{Q(x)} \right) dx$$


KL շեղումը կարելի է պատկերացնել որպես «անակնկալի» չափ: Եթե մենք ակնկալում ենք $Q$ բաշխումը, բայց իրականում ստանում ենք $P$-ն, ապա $D_{KL}$-ը ցույց է տալիս, թե որքանով ենք սխալվել մեր ենթադրության մեջ:


Ինչպես է $\log \left( \frac{P(x)}{Q(x)} \right)$ բաղադրիչը «պատժում» մոդելին սխալների համար:

1. Երբ բաշխումները նման են (Case 1)

Եթե տվյալ կետում $P(x)$ և $Q(x)$ արժեքները մոտ են, ապա նրանց հարաբերակցությունը ($\frac{P(x)}{Q(x)}$) մոտ է 1-ի: Քանի որ $\log(1) = 0$, այս դեպքում շեղման արժեքին գրեթե ոչինչ չի գումարվում: Սա նշանակում է, որ մոդելը ճիշտ է կանխատեսել հավանականությունը:

2. Թերագնահատման պատիժը (Case 2)

Սա ամենակարևոր դեպքն է: Երբ իրական հավանականությունը բարձր է ($P(x)$ is high), բայց մեր մոդելը տալիս է ցածր հավանականություն ($Q(x)$ is low):

* Հարաբերակցությունը դառնում է շատ մեծ թիվ:

* Լոգարիթմը դառնում է մեծ դրական թիվ:

* Արդյունքը բազմապատկվում է բարձր $P(x)$-ով, ինչը հանգեցնում է հսկայական տույժի (massive penalty):

* Իմաստը: Մոդելը խիստ պատժվում է այն իրադարձություններն անտեսելու համար, որոնք իրականում հաճախ են տեղի ունենում:

3. Գերագնահատման «մեղմությունը» (Case 3)

Երբ իրական հավանականությունը ցածր է ($P(x)$ is low), բայց մոդելը տալիս է բարձր հավանականություն ($Q(x)$ is high):

* Հարաբերակցությունը մոտ է 0-ի, իսկ լոգարիթմը մեծ բացասական թիվ է:

* Սակայն, քանի որ մենք այն բազմապատկում ենք $P(x)$-ով (որը մոտ է 0-ի), այդ բացասական պատիժը ճնշվում է (suppressed):

* Իմաստը: KL շեղումն ավելի հանդուրժող է այն սխալների հանդեպ, երբ մոդելը հավանականություն է վերագրում այնտեղ, որտեղ իրականում ոչինչ չկա:

Gaussian Mixture Models (GMM) համատեքստում KL շեղման այս հատկությունը հանգեցնում է այն երևույթին, որը հաճախ անվանում են "Mean-Seeking" (միջինին ձգտող) վարքագիծ:

Եթե  իրական տվյալների բաշխումը ($P$) ունի երկու առանձին «կուտակումներ» (գագաթներ), իսկ դուք փորձում եք այն մոտարկել մեկ պարզ Gaussian բաշխմամբ ($Q$), ապա KL շեղումը կստիպի $Q$-ին տարածվել այնպես, որ այն ծածկի երկու գագաթներն էլ:Եթե մոդելը փորձի ծածկել միայն մեկ գագաթը, ապա մյուս գագաթի տիրույթում $Q(x)$-ը կլինի շատ փոքր, մինչդեռ $P(x)$-ը բարձր է:

Արդյունքում, մոդելն ընտրում է ավելի լայն դիսպերսիա (variance), որպեսզի $Q(x)$-ը զրոյին մոտ չլինի այնտեղ, որտեղ $P(x)$-ը նշանակալի է:

«Ծածկել» (covering) հասկացությունը KL շեղման և Gaussian Mixture Models-ի (GMM) համատեքստում վերաբերում է մոդելի ձգտմանը՝ բաշխել իր հավանականության զանգվածը այնպես, որ այն ներառի իրական տվյալների բոլոր տիրույթները, որտեղ կա զգալի խտություն:

GMM-ի դեպքում մենք օգտագործում ենք սովորաբար Expectation-Maximization (EM) ալգորիթմը, որը անուղղակիորեն մինիմիզացնում է KL շեղումը:

* E-քայլում մենք հաշվում ենք «պատասխանատվությունները» (responsibilities)՝ հիմնվելով ընթացիկ բաշխումների վրա:

* M-քայլում մենք թարմացնում ենք Gaussian-ների պարամետրերը ($\mu, \sigma$): Քանի որ մենք ուզում ենք խուսափել մեծ պատժից, ալգորիթմը ձգտում է տեղադրել Gaussian-ները այնպես, որ տվյալների ոչ մի խտացում դուրս չմնա ծածկույթից:

«Ծածկելու» ռազմավարությունը

Որպեսզի այդ պատժից խուսափի, օպտիմալացման ալգորիթմը (օրինակ՝ EM-ը) ստիպում է $Q$ բաշխմանը լինել ավելի «լայնախոհ»:

* Տարածում (Broadening): Մոդելը մեծացնում է իր դիսպերսիան ($\sigma$), որպեսզի «ծածկի» նույնիսկ այն կետերը, որոնցում սկզբում վստահ չէր:

* Միջինացում (Mean-seeking): Եթե տվյալները երկու առանձին խմբերով են (երկու գագաթ), իսկ մոդելը միայն մեկ Gaussian է, այն կկանգնի հենց մեջտեղում և կտարածվի այնքան, որ երկու խմբերն էլ հայտնվեն իր «կամարի» տակ:

Գործնական օրինակ GMM-ում

Պատկերացրեք՝ ունեք տվյալներ, որոնք կազմում են երկու կլաստեր՝ մեկը $x=0$-ի շուրջ, մյուսը՝ $x=10$-ի:Եթե դուք օգտագործում եք $D_{KL}(P || Q)$, մոդելը չի ընտրի միայն $x=0$-ն, քանի որ $x=10$-ի հատվածում տվյալները կմնան «չծածկված», ինչը կբերի անվերջության ձգտող սխալի:Մոդելը կձգտի ծածկել ամբողջ տիրույթը $[0, 10]$՝ ապահովելով, որ ամենուր $Q(x) > 0$:

«Ծածկել» նշանակում է խուսափել զրոյական կամ շատ ցածր հավանականություն տալուց այնտեղ, որտեղ իրականում տվյալներ կան

Ինչպես է KL շեղումը կիրառվում EM (Expectation-Maximization) ալգորիթմի կոնկրետ քայլում՝ կլաստերավորման խնդիրը լուծելու համար:

1. Թաքնված փոփոխականներ ($Z$)

Մեքենայական ուսուցման մեջ մենք հաճախ ունենում ենք տվյալներ ($X$), բայց չգիտենք, թե որ կլաստերին է պատկանում յուրաքանչյուր կետ: Այդ «թաքնված» պատկանելությունը կոչվում է Latent Variable $Z$: Մեր նպատակն է գտնել դրանք:

2. Խնդիրը՝ Intractable Math
Գոյություն ունի իրական հետին հավանականություն ($P(Z|X, \theta)$), որը ցույց է տալիս, թե որքանով է հավանական, որ տվյալը պատկանում է որոշակի կլաստերի: Սակայն սա ուղղակիորեն հաշվարկելը մաթեմատիկորեն չափազանց բարդ է կամ անհնար (intractable):

3. Լուծումը՝ Ենթադրություն ($q(Z)$)

Քանի որ իրական բաշխումը հաշվել չենք կարող, մենք ներմուծում ենք մի կամայական ենթադրյալ բաշխում՝ $q(Z)$: Սա մեր «լավագույն կանխատեսումն» է այն մասին, թե որտեղ են թաքնված կլաստերները:

4. E-քայլի (Expectation Step) էությունը

Սա այն պահն է, երբ KL շեղումը դառնում է գործիք՝

* Մենք փորձում ենք մինիմիզացնել $D_{KL}(q(Z) || P(Z|X, \theta))$ արժեքը:

* Մենք փոփոխում և թյունինգ ենք անում մեր $q(Z)$ ենթադրությունը այնքան ժամանակ, մինչև շեղումը դառնում է 0:

* Երբ շեղումը 0 է, դա նշանակում է, որ մեր $q(Z)$ ենթադրությունը կատարելապես համընկնում է իրական թաքնված կլաստերների հետ:

Այս գործընթացը թույլ է տալիս մեզ լուծել բարդ խնդիրներ՝ դրանք վերածելով օպտիմալացման խնդրի: Մենք չենք լուծում անհնարին հավասարումներ, այլ պարզապես «մոտեցնում» ենք մեր ենթադրությունը իրականությանը՝ օգտագործելով KL շեղումը որպես կողմնացույց:


Դիտարկենք մաթեմատիկական «կախարդանքը», որի շնորհիվ մենք կարողանում ենք օպտիմալացնել բարդ մոդելները (ինչպիսին է GMM-ը)՝ օգտագործելով ELBO-ն (Evidence Lower Bound):


KL Շեղման բանաձևը (Kullback-Leibler Divergence) դիսկրետ դեպքի համար հետևյալն է՝ 

$$D_{KL}(P \parallel Q) = \sum_{x} P(x) \log \left( \frac{P(x)}{Q(x)} \right)$$


Այն կարելի է ներկայացնել նաև որպես տարբերություն՝

$$D_{KL}(P \parallel Q) = \sum_{x} P(x) \log P(x) - \sum_{x} P(x) \log Q(x)$$

Մենք ուզում ենք հաշվել $\ln p(X)$ (լոգարիթմական հավանականությունը կամ Evidence-ը): 

1. Մաթեմատիկական հնարքը (Multiplying by 1)

Մենք սկսում ենք լոգարիթմական հավանականությունից՝ $\ln p(X)$, որը ցույց է տալիս, թե որքան լավ է մոդելը բացատրում տվյալները:

* Քանի որ $p(X) = \sum_Z p(X, Z)$, մենք ունենք լոգարիթմ՝ գումարի տակ ($\ln \sum$): Սա հաշվարկների համար չափազանց դժվար է:

* Մենք ներմուծում ենք կամայական $q(Z)$ բաշխումը և բազմապատկում ու բաժանում ենք դրանով ($\frac{q(Z)}{q(Z)} = 1$): Սա թույլ է տալիս արտահայտությունը գրել որպես Մաթեմատիկական սպասում ($\mathbb{E}_q$):

$$\ln p(X) = \ln \sum_{Z} p(X, Z) = \ln \sum_{Z} q(Z) \frac{p(X, Z)}{q(Z)}$$

2. Յենսենի անհավասարությունը (Jensen's Inequality)

Յենսենի անհավասարությունը թույլ է տալիս լոգարիթմը «մտցնել» գումարի (կամ սպասման) ներսը:

* Քանի որ լոգարիթմը ուռուցիկ (concave) ֆունկցիա է, սպասման լոգարիթմը միշտ մեծ է կամ հավասար լոգարիթմի սպասումից:

* Արդյունքում ստանում ենք $\mathcal{L}(q, \theta)$ արտահայտությունը, որը կոչվում է ELBO:
    $$\ln p(X) \geq \sum_{Z} q(Z) \ln \frac{p(X, Z)}{q(Z)} = \mathcal{L}(q, \theta)$$

Այժմ մենք ունենք լոգարիթմների գումար (sum of logs)՝ նախկին գումարի լոգարիթմի (log of sums) փոխարեն:

Մաթեմատիկորեն սա նշանակում է, որ բարդ արտադրյալները վերածվում են պարզ գումարների:

Սա թույլ է տալիս հեշտությամբ հաշվել ածանցյալներ և գտնել մոդելի լավագույն պարամետրերը:


Evidence Lower Bound (ELBO)

1. Լոգարիթմական հավանականության սահմանումը

Մենք ցանկանում ենք առավելագույնի հասցնել դիտարկվող տվյալների լոգարիթմական հավանականությունը՝ $\ln p(X)$։ Քանի որ մենք ունենք թաքնված փոփոխականներ ($Z$), $p(X)$-ը կարող ենք արտահայտել որպես համատեղ հավանականության գումար ըստ բոլոր հնարավոր $Z$-ների՝

$$\ln p(X) = \ln \sum_{Z} p(X, Z)$$

2. $q(Z)$ բաշխման ներմուծումը

Քանի որ $\ln \sum$ գործողությունը դժվար է հաշվարկել, ներմուծում ենք կամայական հավանականային բաշխում $q(Z)$ թաքնված փոփոխականների վրա։ Մենք բազմապատկում և բաժանում ենք լոգարիթմի ներսի արտահայտությունը $q(Z)$-ով, ինչը մեզ թույլ է տալիս ստանալ Մաթեմատիկական սպասում $q$ բաշխման նկատմամբ՝

$$\ln p(X) = \ln \sum_{Z} q(Z) \frac{p(X, Z)}{q(Z)} = \ln \mathbb{E}_q \left[ \frac{p(X, Z)}{q(Z)} \right]$$

3. Jensen-ի անհավասարության կիրառումը

Լոգարիթմը կոր ֆունկցիա է ($\cap$-աձև), ուստի ըստ Jensen-ի անհավասարության, ֆունկցիայի սպասումը միշտ փոքր է կամ հավասար սպասման ֆունկցիայից ($\ln(\mathbb{E}[x]) \geq \mathbb{E}[\ln(x)]$): Սա թույլ է տալիս լոգարիթմը «մտցնել» գումարի ներսը՝

$$\ln p(X) \geq \sum_{Z} q(Z) \ln \frac{p(X, Z)}{q(Z)}$$

4. ELBO-ի սահմանումը

Անհավասարության աջ կողմը կոչվում է ELBO (Evidence Lower Bound) և նշանակվում է $\mathcal{L}(q, \theta)$։


Մենք լոգարիթմների գումարն (log of sums) արդեն փոխարինել ենք գումարների լոգարիթմով (sum of logs), ինչը մաթեմատիկորեն շատ ավելի հեշտ է օպտիմալացնել։

Նպատակը. Փոխանակ ուղղակիորեն առավելագույնի հասցնենք $\ln p(X)$-ը, մենք առավելագույնի ենք հասցնում նրա ստորին սահմանը՝ $\mathcal{L}$-ը։

Մեքենայական ուսուցման մեջ մենք հաճախ ցանկանում ենք գտնել տվյալների լոգարիթմական հավանականությունը՝ $\ln p(X)$, որպեսզի առավելագույնի հասցնենք մոդելի ճշգրտությունը: Սակայն բարդ մոդելների դեպքում դա հաշվարկային տեսանկյունից անհնար է դառնում:

Այստեղ օգնության է գալիս  նշված հիմնարար հավասարումը՝

$$\ln p(X) = \mathcal{L}(q, \theta) + D_{KL}(q(Z)||p(Z|X, \theta))$$

Այս բանաձևը բաղկացած է երկու հիմնական մասից՝

* $\mathcal{L}(q, \theta)$ (ELBO - Evidence Lower Bound): Սա մեր «ստորին սահմանն» է: Քանի որ մենք չենք կարող ուղղակիորեն հաշվել $\ln p(X)$-ը, մենք աշխատում ենք այս արժեքի հետ:

* $D_{KL}$ (Kullback-Leibler Divergence): Սա չափում է հեռավորությունը (տարբերությունը) մեր կողմից առաջարկված մոտավոր բաշխման ($q(Z)$) և իրական հետին (posterior) բաշխման ($p(Z|X)$) միջև:


$D_{KL} \geq 0$: Սա նշանակում է, որ $\mathcal{L}(q, \theta)$-ն միշտ փոքր է կամ հավասար $\ln p(X)$-ից: Եթե մենք փոքրացնում ենք տարբերությունը ($D_{KL} \to 0$), ապա մեր մոտարկումը դառնում է կատարյալ:

2. Գործնական կիրառությունը (Usage)

Այս տեսությունը գործնականում օգտագործվում է մի քանի կարևոր ուղղություններում՝

* Variational Autoencoders (VAE): Սա այն հիմքն է, որի վրա կառուցվում են պատկերներ գեներացնող նեյրոնային ցանցերը: Մենք սովորեցնում ենք ցանցին առավելագույնի հասցնել $\mathcal{L}$-ը, ինչն անուղղակիորեն ստիպում է մոդելին սովորել տվյալների ճիշտ կառուցվածքը:

* Բարդ մոդելների մոտարկում: Երբ տվյալները շատ են և փոփոխականները բարդ փոխկապակցված են, մենք $q(Z)$-ի միջոցով պարզեցնում ենք խնդիրը:

* Topic Modeling (LDA): Տեքստերի մեջ թեմաներ գտնելու համար օգտագործվող ալգորիթմները հիմնված են հենց այս մաթեմատիկական սկզբունքի վրա:

Մեր նպատակն է մեծացնել $\mathcal{L}$-ը, որպեսզի փոքրացնենք սխալը ($D_{KL}$): Երբ $q(Z) = p(Z|X)$, մենք ստանում ենք ճշգրիտ լուծում:


1. Ինչպե՞ս է աշխատում բանաձևը դատասեթի վրա

Երբ մենք ուսուցանում ենք մոդելը (օրինակ՝ Variational Autoencoder), մենք կատարում ենք հետևյալ քայլերը.Տվյալների մուտքագրում ($X$): Մոդելը վերցնում է դատասեթի օբյեկտները (օրինակ՝ նկարները):

Լատենտ փոփոխականների կանխատեսում ($Z$): Քանի որ մենք չգիտենք տվյալների թաքնված հատկանիշները (օրինակ՝ նկարի մեջի թվի թեքությունը կամ հաստությունը), մենք օգտագործում ենք $q(Z)$ բաշխումը դրանք մոտարկելու համար:

ELBO-ի օպտիմիզացիա: Մենք չենք կարող հաշվել $p(X)$ ամբողջ դատասեթի համար, ուստի ալգորիթմը փորձում է առավելագույնի հասցնել նշված $\mathcal{L}(q, \theta)$ արժեքը:


Գործնական օգտագործումը (Step-by-Step)

Դատասեթի վրա աշխատելիս գործընթացը հետևյալն է՝

* Encoder-ի միջոցով մոտարկում: Մոդելը փորձում է ստեղծել այնպիսի $q(Z)$ բաշխում, որը հնարավորինս մոտ կլինի իրական $p(Z|X)$ հետին բաշխմանը:

* KL Divergence-ի նվազեցում: Մենք օգտագործում ենք  $D_{KL}$ մասը որպես «տուգանք» (penalty): Եթե մեր մոտարկած $q(Z)$-ը շատ է տարբերվում նախնական ենթադրությունից (prior), ապա $D_{KL}$-ը մեծանում է, և մոդելը սկսում է ուղղվել:

* Վերակառուցում (Reconstruction): $\mathcal{L}(q, \theta)$ բաղադրիչը ստիպում է մոդելին այնպես սովորել, որ $Z$ լատենտ կոդից հնարավոր լինի հետ ստանալ սկզբնական դատասեթի տվյալը:

EM (Expectation-Maximization) ալգորիթմը

Սա այն մեխանիզմն է, որը թույլ է տալիս գործնականում կիրառել նախորդ սլայդի մաթեմատիկական կապը դատասեթի վրա։

1. Ի՞նչ է տեղի ունենում E-Step-ի ժամանակ

Եթե նախորդ բանաձևում մենք տեսանք «ճեղքը», ապա E-Step-ը (Expectation step) այն քայլն է, որտեղ մենք փորձում ենք այդ ճեղքը հասցնել զրոյի։

* Պարամետրերի սառեցում ($\theta$): Մենք ժամանակավորապես չենք փոխում մոդելի հիմնական պարամետրերը (օրինակ՝ կլաստերների կենտրոնները):

* KL Divergence-ի զրոյացում: Քանի որ $\ln p(X)$-ը ֆիքսված է, $\mathcal{L}$ ստորին սահմանը մեծացնելու միակ ձևը $D_{KL}$ տարբերությունը փոքրացնելն է:

Լուծումը: Մենք մեր մոտարկված $q(Z)$ բաշխումը դարձնում ենք հավասար իրական հետին բաշխմանը ($p(Z|X, \theta)$): Սա ստիպում է, որ $D_{KL} = 0$:

Կիրառությունը դատասեթի վրա (Ինտուիցիա)

Պատկերացրու՝ դու ունես դատասեթ, որտեղ կետերը պետք է բաժանվեն կլաստերների (խմբերի)։

* Responsibilities (Պատասխանատվություն):  այս քայլում մենք հաշվում ենք, թե յուրաքանչյուր տվյալ (կետ) ինչքանո՞վ է «պատասխանատու» կամ պատկանում այս կամ այն կլաստերին՝ հիմնվելով մեր ընթացիկ գիտելիքի վրա:

Օրինակ: Եթե ունենք կետ $A$ և երկու կլաստեր, E-step-ի ժամանակ մենք ասում ենք. «Հավանականությունը $0.8$ է, որ այս կետը պատկանում է 1-ին խմբին, և $0.2$, որ 2-րդին»:

Ի՞նչ է նշանակում $q(Z) = p(Z|X, \theta)$ դատասեթի համար

Երբ մենք աշխատում ենք տվյալների բազայի հետ, յուրաքանչյուր $X_i$ տվյալի համար մենք պետք է որոշենք նրա «թաքնված» վիճակը $Z_i$:

Հավասարեցում: Մենք մեր ենթադրյալ $q(Z)$ բաշխումը դարձնում ենք ճիշտ այնպիսին, ինչպիսին տվյալ պահին թելադրում է մոդելը՝ հաշվի առնելով մեր ունեցած տվյալները ($X$) և ընթացիկ պարամետրերը ($\theta$)

:$D_{KL} = 0$: Սա նշանակում է, որ մենք ամբողջությամբ վերացնում ենք տեղեկատվական կորուստը մեր մոտարկման և իրականության միջև տվյալ քայլում:


2. Responsibilities (Պատասխանատվությունների բաշխում)

Աա վերաբերում է կլաստերների նշանակմանը: Դատասեթի վրա դա ունի հետևյալ տեսքը.Կլաստերավորում: Եթե ունենք 3 կլաստեր, ապա յուրաքանչյուր կետի համար հաշվում ենք 3 թիվ (օրինակ՝ 0.1, 0.7, 0.2), որոնց գումարը 1 է:


Soft Assignment: Ի տարբերություն K-means-ի, որտեղ կետը պատկանում է միայն մեկ խմբի, այստեղ մենք օգտագործում ենք «փափուկ» նշանակումներ: Սա հենց այն responsibilities-ն է, որի մասին խոսում է սլայդը:

3. Ինչպե՞ս է սա օգնում հաջորդ քայլին

Այս քայլի ավարտին մենք դատասեթի յուրաքանչյուր կետի համար ունենք հստակ հավանականություններ, թե որ խմբին է այն պատկանում:Մենք «լրացրինք» տվյալների բացը:Այժմ մենք պատրաստ ենք անցնել M-քայլին, որտեղ կօգտագործենք այս հաշվարկված հավանականությունները՝ մոդելի պարամետրերը ($\theta$) թարմացնելու համար:Ստացվում է, որ E-քայլը մեր դատասեթը դարձնում է «ամբողջական»՝ հաշվելով թաքնված փոփոխականների սպասվող արժեքները:

M-քայլի նպատակը և տրամաբանությունը

Այս փուլում տեղի է ունենում հետևյալը՝

Կլաստերների նշանակումների սառեցում ($q(Z)$): Մենք այլևս չենք փոխում այն հավասարեցումները (responsibilities), որոնք ստացել էինք E-քայլում։ Այսինքն՝ մենք գիտենք, թե յուրաքանչյուր կետ որքանով է պատկանում այս կամ այն կլաստերին։


Պարամետրերի թարմացում ($\theta$): Մենք փոփոխում ենք մոդելի պարամետրերը (օրինակ՝ միջին արժեքները կամ շեղումները), որպեսզի բարձրացնենք իրական հավանականությունը ($\ln p(X)$)։


Երաշխավորված աճ: Քանի որ KL դիվերգենցիան միշտ $\geq 0$ է, ապա ստորին սահմանի ($\mathcal{L}$) մեծացումը երաշխավորում է, որ մենք մեծացնում ենք նաև տվյալների իրական լոգարիթմական հավանականությունը։


2. Մաթեմատիկական օբյեկտիվը


Բերված բանաձևը ցույց է տալիս, թե ինչպես ենք գտնում նոր պարամետրերը՝

$$\theta^{(new)} = \arg \max_{\theta} \sum_{Z} q(Z) \ln p(X, Z | \theta)$$

Սա նշանակում է, որ մենք փնտրում ենք այնպիսի $\theta$, որը մաքսիմալացնում է համատեղ հավանականության (joint probability) կշռված լոգարիթմը։ Կշիռները հենց մեր $q(Z)$ հավանականություններն են:


Լոգարիթմի շնորհիվ մենք ստանում ենք «գեղեցիկ, փակ տեսքի լուծումներ» (closed-form solutions)։ Դա նշանակում է, որ մենք ստիպված չենք օգտագործել բարդ թվային մեթոդներ՝

Միջին արժեք ($\mu$): Նոր միջինը հաշվվում է որպես տվյալների կշռված միջին։

Կովարիացիոն մատրից ($\Sigma$): Ցույց է տալիս կլաստերի տարածվածությունը։

Խառնուրդի գործակից ($\pi$): Ցույց է տալիս, թե դատասեթի որ մասն է բաժին ընկնում տվյալ կլաստերին։

Գործնական օրինակ դատասեթի վրա.Պատկերացրու՝ E-քայլում պարզեցինք, որ որոշ կետեր մեծ հավանականությամբ պատկանում են «Ա» խմբին: M-քայլում մենք պարզապես վերցնում ենք այդ կետերը և դրանց հիման վրա վերահաշվարկում ենք «Ա» խմբի կենտրոնը, որպեսզի այն ավելի ճշգրիտ համընկնի տվյալների նոր դասավորության հետ:


GMM-ի M-քայլի (Maximization) թարմացման բանաձևերը

1. Էֆեկտիվ քանակի հաշվարկ ($N_k$)

Նախքան պարամետրերը թարմացնելը, մենք հաշվում ենք, թե քանի «վիրտուալ» կետ է բաժին հասել $k$-րդ կլաստերին՝

$$N_k = \sum_{i=1}^{N} r_{ik}$$

Բացատրություն: $r_{ik}$-ն (responsibility) այն հավանականությունն է, որ $i$-րդ կետը պատկանում է $k$ կլաստերին: Սա «soft» (փափուկ) կլաստերավորում է. եթե կետը 0.7 հավանականությամբ 1-ին կլաստերում է, իսկ 0.3-ով՝ 2-րդում, ապա նա իր «լուման» է ներդնում երկու $N_k$-երի մեջ էլ:

2. Խառնուրդի գործակիցների թարմացում ($\hat{\pi}_k$)

Սա ցույց է տալիս, թե որքան մեծ է $k$ կլաստերի կշիռը ամբողջ մոդելում:

$$\hat{\pi}_k = \frac{N_k}{N}$$

Բացատրություն: Մենք պարզապես վերցնում ենք տվյալ կլաստերին վերագրված էֆեկտիվ կետերի քանակը և բաժանում տվյալների ընդհանուր քանակի ($N$) վրա: Սա հնարավորություն է տալիս մոդելին հասկանալ, թե որ կլաստերներն են ավելի «խիտ» կամ հաճախ հանդիպող դատասեթում:

3. Միջին արժեքի թարմացում ($\hat{\mu}_k$)

Սա կլաստերի նոր կենտրոնն է, որը պետք է լինի հնարավորինս մոտ իրեն պատկանող կետերին:

$$\hat{\mu}_k = \frac{1}{N_k} \sum_{i=1}^{N} r_{ik} x_i$$

Բացատրություն: Սա կշռված միջինն է: Եթե $r_{ik}$ արժեքը մեծ է (կետը մոտ է կենտրոնին), ապա այդ $x_i$ կետը ուժեղ ազդեցություն է ունենում նոր $\hat{\mu}_k$-ի դիրքի վրա: Եթե $r_{ik}$-ն մոտ է զրոյին, այդ կետը գրեթե չի փոխում կենտրոնի դիրքը:


4. Կովարիացիոն մատրիցի թարմացում ($\hat{\Sigma}_k$)

Սա ամենակարևոր մասն է, որը որոշում է կլաստերի ձևը (օրինակ՝ երկարավուն էլիպս կամ կլոր շրջան):

$$\hat{\Sigma}_k = \frac{1}{N_k} \sum_{i=1}^{N} r_{ik} (x_i - \hat{\mu}_k)(x_i - \hat{\mu}_k)^T$$

Բացատրություն: Այստեղ մենք հաշվում ենք յուրաքանչյուր կետի շեղումը նոր ստացված կենտրոնից ($\hat{\mu}_k$):$(x_i - \hat{\mu}_k)(x_i - \hat{\mu}_k)^T$ արտադրյալը ստեղծում է մատրից, որը նկարագրում է տարածվածությունը բոլոր ուղղություններով:

Կշռում: Այս շեղումները նույնպես կշռվում են $r_{ik}$-ով, որպեսզի միայն «իրական» անդամները որոշեն կլաստերի ձևը:


Այս բոլոր բանաձևերը ստացվում են Log-Likelihood ֆունկցիայի ածանցյալները զրոյի հավասարեցնելու միջոցով:

E-քայլում մենք ֆիքսում ենք կլաստերների ձևը և հաշվում կետերի պատկանելիությունը (r_ik):

M-քայլում (որը տեսնում ես image_856d00.png-ում) մենք ֆիքսում ենք պատկանելիությունը և «ձգում» կլաստերները այնպես, որ դրանք լավագույնս ծածկեն իրենց վերագրված կետերը:

Այս ցիկլը կրկնվում է այնքան ժամանակ, մինչև բանաձևերով հաշվարկված նոր արժեքները դադարեն էականորեն փոխվել: Սա երաշխավորում է, որ մենք գտել ենք տվյալների բաշխման լավագույն մոտարկումը:


Օրինակ թե ինչպես է EM ալգորիթմը «ինքնուրույն» կողմնորոշվում տվյալների մեջ։

1. Դատասեթի վերլուծությունՄենք ունենք 10 կետից բաղկացած միաչափ (1D) դատասեթ:

$$X = \{1, 2, 2.5, 3, 8, 8.5, 9, 9.5, 10, 10.5\}$$

Եթե ուշադիր նայենք թվերին, կտեսնենք երկու խումբ:

Ձախ խումբ: 1-ից 3 արժեքների շուրջ։

Աջ խումբ: 8-ից 10.5 արժեքների շուրջ։

2. Ինիցիալիզացիա (Iteration 0)

Ալգորիթմը սկսելու համար մենք պետք է տանք կամայական սկզբնական արժեքներ երկու Գաուսյաններին ($K=2$):

Գաուսյան 1 ($G_1$): $\mu_1 = 3.0$, $\sigma_1^2 = 1.0$, $\pi_1 = 0.5$

Գաուսյան 2 ($G_2$): $\mu_2 = 6.0$, $\sigma_2^2 = 1.0$, $\pi_2 = 0.5$

Կարևոր դիտարկում: Երկրորդ Գաուսյանը ($G_2$) տեղադրված է $x=6$ կետում, որը գտնվում է դատասեթի «դատարկ» հատվածում՝ երկու իրական խմբերի արանքում:

E-քայլը (Expectation Step) գործնական թվային օրինակի վրա։
: Սա այն պահն է, երբ մոդելը «չափում է», թե որ կետը որ կլաստերին է պատկանում:

1. Պարամետրերի սառեցում և հաշվարկ

մենք սառեցնում ենք մոդելի պարամետրերը ($\mu, \sigma^2, \pi$) և յուրաքանչյուր կետի համար հաշվարկում ենք պատասխանատվությունները ($r_{i,k}$):Դիտարկենք $x_5 = 8$ կետը, որը գտնվում է աջակողմյան խմբի սկզբում:

2. Հավակնությունների (Claims) հաշվարկ

Յուրաքանչյուր Գաուսյան բաշխում «հավակնում» է տվյալ կետին՝ ելնելով իր հեռավորությունից և կշռի՝

Գաուսյան 1-ի հավակնությունը ($G_1$):

$$Claim_{G_1} = \pi_1 \cdot \mathcal{N}(8|3, 1) \approx 0.5 \cdot (1.48 \times 10^{-6}) \approx 0$$

Քանի որ $x=8$ կետը շատ հեռու է $G_1$-ի կենտրոնից ($\mu_1=3$), հավանականությունը գրեթե զրոյական է:

Գաուսյան 2-ի հավակնությունը ($G_2$):

$$Claim_{G_2} = \pi_2 \cdot \mathcal{N}(8|6, 1) \approx 0.5 \cdot 0.054 = 0.027$$

Չնայած $G_2$-ը նույնպես հեռու է $x=8$-ից, այն տեխնիկապես ավելի մոտ է, քան $G_1$-ը, ուստի նրա հավակնությունն ավելի մեծ է:

3. Նորմալացում և վերջնական պատասխանատվություն

Որպեսզի ստանանք իրական պատասխանատվությունը ($r_{i,k}$), մենք նորմալացնում ենք այս արժեքները այնպես, որ դրանց գումարը լինի 1:

$$r_{5,2} = \frac{0.027}{0 + 0.027} = 1.0$$

4. Կարևոր արդյունք

Ստացվում է մի հետաքրքիր պատկեր՝ չնայած $G_2$-ը սկզբնապես դրված էր դատարկ տարածության մեջ, այն տեխնիկապես ավելի մոտ է աջակողմյան բոլոր կետերին: Արդյունքում, $G_2$-ը վերցնում է $\{8, 8.5, 9, 9.5, 10, 10.5\}$ կետերի գրեթե 100% սեփականատիրությունը:

Սա նշանակում է, որ հաջորդ քայլում (M-step), երբ մենք վերահաշվարկենք $\mu_2$-ը, այն «թռիչք» կկատարի դեպի այս կետերի կենտրոնը, քանի որ այժմ նա է պատասխանատու դրանց համար:


M-քայլը (Maximization Step), որտեղ տեղի է ունենում մոդելի պարամետրերի կտրուկ և էֆեկտիվ թարմացումը՝ հիմնված նախորդ քայլում ստացված պատասխանատվությունների վրա:

1. Պարամետրերի թարմացումը

Այս փուլում մենք սառեցնում ենք պատասխանատվությունները ($r_{i,k}$) և թարմացնում մոդելի պարամետրերը: Կենտրոնանանք երկրորդ Գաուսյանի միջին արժեքի ($\mu_2$) վրա:

2. $\mu_2$-ի հաշվարկը (Updating $\mu_2$)

Քանի որ առաջին Գաուսյանը ($G_1$) իրեն վերցրեց ձախակողմյան բոլոր կետերը, դրանց համար $r_{i,2}$ արժեքները հավասար են 0-ի: Հետևաբար, $\mu_2$-ի նոր արժեքը հաշվելիս մենք գումարում ենք միայն այն կետերը, որոնց համար «պատասխանատու» է $G_2$-ը՝

$$\hat{\mu}_2 = \frac{\sum r_{i,2}x_i}{\sum r_{i,2}} \approx \frac{8 + 8.5 + 9 + 9.5 + 10 + 10.5}{6} = 9.25$$

3. M-քայլի «ֆիզիկան»

Նկարագրվում է այս պրոցեսը որպես «ֆիզիկական» երևույթ.Կտրուկ տեղաշարժ: $\mu_2$-ը «ագրեսիվորեն և ուժգին» քաշվում է դատարկ տարածությունից ($x=6$) ուղղակիորեն դեպի աջակողմյան կլաստերի կենտրոնը՝ $x=9.25$:

Դիսպերսիայի թարմացում: $\sigma_2^2$ դիսպերսիան նույնպես կփոփոխվի (դառնալով $\approx 0.72$), որպեսզի կատարելապես ծածկի այդ 6 կետերի տարածվածությունը:


Ընդամենը մեկ իտերացիայի արդյունքում մեր մոդելը սխալ դիրքից (դատարկ տարածությունից) հայտնվեց ճիշտ տեղում: Սա ցույց է տալիս, որ EM ալգորիթմը չափազանց արագ է կողմնորոշվում, երբ կլաստերները հստակ առանձնացված են:


EM ալգորիթմի աշխատանքի ավարտը և կոնվերգենցիան (զուգամիտությունը)

1. Իտերացիա 2. 

Պատասխանատվությունների հստակեցում

Երկրորդ իտերացիայի E-քայլի ժամանակ պատասխանատվությունները հաշվարկվում են արդեն շատ ճշգրիտ միջին արժեքների հիման վրա՝

$\mu_1 \approx 2.12$ (ձախ խմբի համար)

$\mu_2 \approx 9.25$ (աջ խմբի համար)

Քանի որ Գաուսյաններն այժմ գտնվում են անմիջապես տվյալների կլաստերների վրա, պատասխանատվությունները դառնում են «ամուր» կամ վերջնական (solidify): Սա նշանակում է, որ յուրաքանչյուր կետ արդեն հստակ գիտի իր «տունը»:


2. Ալգորիթմի ավարտը (Convergence)

Ալգորիթմի զուգամիտումը տեղի է ունենում շատ արագ։

Իտերացիա 3. $\mu$ և $\sigma$ պարամետրերի փոփոխությունները դառնում են աննշան կամ «միկրոսկոպիկ» ($< 10^{-4}$):

Կանգառ. Երբ պարամետրերն այլևս չեն փոխվում, ալգորիթմը դադարեցնում է աշխատանքը (Algorithm Terminated):


EM ալգորիթմի զուգամիտման (convergence) տեսողական գործընթացը ավելի բարդ, երկչափ (2D) տվյալների բազայի վրա:


Էքսպեկտացիայի մաքսիմալացման (EM - Expectation-Maximization) ալգորիթմի աշխատանքը, որն օգտագործվում է Գաուսյան խառնուրդների մոդելի (GMM - Gaussian Mixture Model) կլաստերավորման համար:

* 2-րդ իտերացիայի դրությամբ, E-քայլը հաշվարկում է պատասխանատվությունները՝ հիմնվելով նոր, խիստ ճշգրիտ միջին արժեքների վրա:

Բացատրություն՝

* Միջին արժեքները (Means - $\mu$): Սրանք այն կենտրոններն են, թե որտեղ են գտնվում կապույտ ու կարմիր օվալները: Առաջին քայլում ($L=1$) ալգորիթմը մոտավորապես հասկացավ, թե որտեղ են խմբերը և օվալների կենտրոնները տեղափոխեց տվյալների իսկական կենտրոններին շատ մոտ:

* Պատասխանատվություն (Responsibility - $W_{ik}$): Սա մաթեմատիկական հավանականություն է: Երբ ալգորիթմը նայում է որևէ կետի, նա հարցնում է. «Այս կետի համար ո՞վ է պատասխանատու՝ կապո՞ւյտ Գաուսյանը, թե՞ կարմիրը»:Եթե կետը խորը կապույտ է, նշանակում է կապույտ մոդելի պատասխանատվությունը մոտ է 100%-ի:Եթե կետը մանուշակագույն է (մեջտեղում է), ուրեմն երկուսն էլ ունեն 50-ական % պատասխանատվություն:

Քանի որ 1-ին կրկնությունից հետո օվալների կենտրոնները դարձան «խիստ ճշգրիտ», 2-րդ կրկնության E-քայլն արդեն անսխալ հաշվում է, թե ո՛ր կետը ո՛ր խմբին է պատկանում:

Քանի որ Գաուսյան բաշխումներն այժմ գտնվում են անմիջապես տվյալների վրա, պատասխանատվություններն ամրապնդվում (կայունանում) են:

Սկզբնական փուլերում մոդելը երկմտում էր (կետերը մանուշակագույն էին, հավանականությունները՝ 50/50 կամ 60/40)։

Բայց երբ Գաուսյան կորերը (օվալները) նստում են ուղիղ տվյալների «գլխին» և ստանում են դրանց ճիշտ ձևը (երկարությունը, թեքությունը), անորոշությունը վրանում է։ Պատասխանատվությունները դառնում են ծայրահեղ հստակ (օրինակ՝ 99% կապույտ և 1% կարմիր)։ Այսինքն՝ ամեն կետ «ամրապնդում է» իր դիրքը իր կլաստերի մեջ, ու հաջորդ քայլերում օվալներն այլևս տեղից չեն շարժվում։ Մոդելը հասնում է կայունության (Convergence):

Ալգորիթմն ավարտվեց: Մոդելը հաջողությամբ և ամբողջությամբ առանց վերահսկողության նույնականացրեց թաքնված լատենտ փոփոխականները:

Լատենտ (թաքնված) փոփոխականը տվյալների այն հատկանիշն է, որը մենք չունեինք սկզբում: Մեր ունեցածը միայն կետերի կոորդինատներն էին (X և Y): Մենք չգիտեինք՝ սրանք հիվանդ ու առողջ բջիջներ են, թե՞ կեղծ ու իսկական գործարքներ: Խմբի պատկանելությունը (Կլաստեր 1, թե Կլաստեր 2) թաքնված էր:

Առանց վերահսկողության (Unsupervised): Մենք համակարգչին չենք հուշել, թե որ կետն ինչ է: Մենք նրան տվեցինք ուղղակի կանաչ կետերի քաոս:


Գործարկելով EM ալգորիթմը՝ մոդելն ինքնուրույն, մաթեմատիկական հավանականությունների շնորհիվ, հայտնաբերեց այդ թաքնված փոփոխականը (ամեն կետին տվեց իր իրական «պիտակը»՝ կարմիր կամ կապույտ): Ալգորիթմն ավարտեց աշխատանքը, քանի որ այլևս ոչ մի բան չի փոխվում, խնդիրը լուծված է։

![alt text](image-2.png)

Գրաֆիկների մանրամասն բացատրությունը
Գրաֆիկներում պատկերված է երկչափ տվյալների բազմություն (կանաչ/կապույտ/կարմիր կետեր), որոնք կազմված են երկու իրարից անջատ խմբերից (կլաստերներից): Նպատակն է՝ առանց նախնական հուշումների գտնել այդ երկու խմբերը:

Ալգորիթմն անցնում է հետևյալ փուլերով (նկարներ (a)-ից մինչև (f)).

(a) Սկզբնավորում (Initialization)
Տվյալները սկզբում կանաչ են, այսինքն՝ մոդելը դեռ չգիտի, թե որ կետը որ խմբին է պատկանում:

Պատահականորեն ընտրվում են երկու Գաուսյան բաշխումներ (շրջանակներով պատկերված կապույտ և կարմիր օվալները): Ինչպես տեսնում եք, դրանք սկզբում լիովին հեռու են իրական տվյալներից:

(b) E-քայլ (Expectation step) - Առաջին փուլ
Մոդելը հաշվարկում է «պատասխանատվությունը» (թե յուրաքանչյուր կետ որքանով է հավանական, որ պատկանում է կարմիր կամ կապույտ խմբին):

Կետերը ստանում են երանգներ՝ կապույտ կամ կարմիր, իսկ մեջտեղի կետերը դառնում են մանուշակագույն (քանի որ մոդելը դեռ վստահ չէ):

(c) M-քայլ (Maximization step) - $L=1$ (Առաջին կրկնություն)Հիմնվելով կետերի գույների (հավանականությունների) վրա՝ կապույտ ու կարմիր օվալները «ցատկում» են դեպի տվյալների կենտրոնը և ձգվում են տվյալների ուղղությամբ:

(d) և (e) Հաջորդական կրկնություններ ($L=2$ և $L=5$)Ալգորիթմը կրկնում է E և M քայլերը: Ամեն քայլում օվալները (Գաուսյան բաշխումները) ավելի ու ավելի ճշգրիտ են գրկում տվյալների իրական խմբերը: Մանուշակագույն (անորոշ) կետերի քանակը նվազում է. դրանք դառնում են հստակ կապույտ կամ կարմիր:


(f) Ավարտ ($L=20$)20-րդ կրկնության ժամանակ ալգորիթմը հասնում է կոնվերգենցիայի (զուգամիտության): Կապույտ և կարմիր օվալները կատարելապես նստած են տվյալների երկու կլաստերների վրա:

Թաքնված լատենտ փոփոխականները (այն հարցի պատասխանը, թե որ կետը որ խմբին է պատկանում) ամբողջությամբ բացահայտված են՝ առանց մարդու միջամտության (unsupervised):

Գաուսյան խառնուրդների մոդելի (GMM) կենտրոնների (միջին արժեքների) թարմացման մաթեմատիկական տրամաբանությունը M-քայլում (Maximization Step) և  համեմատումը K-Means ալգորիթմի հետ:


Դիտարկենք Գաուսյան միջինի թարմացման հավասարմանը. GMM-ի միջին արժեքի թարմացում:

$$\hat{\mu}_k = \frac{1}{N_k} \sum_{i=1}^{N} r_{ik}x_i$$

Սա մաթեմատիկորեն նույնական է Soft K-Means-ի ցենտրոիդների (կենտրոնների) թարմացմանը:

$\hat{\mu}_k$ — $k$-րդ կլաստերի (Գաուսյան բաշխման) նոր, թարմացված կենտրոնն է (միջինը), որը մենք ուզում ենք հաշվել:

$x_i$ — Տվյալների բազայի $i$-րդ կետն է (կոորդինատը):

$r_{ik}$ — Responsibility (Պատասխանատվություն): Հավանականությունն այն բանի, որ $x_i$ կետը պատկանում է $k$-րդ կլաստերին: Սա $0$-ից $1$ միջակայքի թիվ է (օրինակ՝ $0.85$):

$N_k$ — $k$-րդ կլաստերին պատկանող «վիրտուալ» կետերի քանակն է: Այն հաշվարկվում է որպես տվյալ կլաստերի համար բոլոր կետերի պատասխանատվությունների գումար՝ $N_k = \sum_{i=1}^{N} r_{ik}$:

Սովորական թվաբանական միջին հաշվելիս մենք ուղղակի գումարում ենք բոլոր կետերը և բաժանում դրանց քանակի վրա:Այստեղ, սակայն, յուրաքանչյուր $x_i$ կետ ունի իր «ձայնի իրավունքի կշիռը» ($r_{ik}$)՝

Եթե ինչ-որ կետ $99\%$ հավանականությամբ պատկանում է կապույտ կլաստերին ($r_{i,\text{blue}} = 0.99$), ապա այն շատ ուժեղ դեպի իրեն կձգի կապույտ օվալի կենտրոնը ($\hat{\mu}_{\text{blue}}$):

Եթե կետը հազիվ $1\%$ հավանականությամբ է պատկանում այդ կլաստերին ($r_{i,\text{blue}} = 0.01$), ապա այն գրեթե չի ազդի կենտրոնի տեղաշարժի վրա:

Կոշտ բուլյան վերագրման (1 կամ 0) փոխարեն, Գաուսյան բաշխման կենտրոնը «ձգվում է» տվյալների բազայի յուրաքանչյուր կետի կոտորակային հավանականությամբ (պատասխանատվությամբ՝ $r_{ik}$)

Այսպիսով, GMM-ը կարելի է պատկերացնել որպես Soft K-Means-ի ավելի հզոր ու ընդհանրացված տարբերակ, որը կենտրոններից բացի հաշվի է առնում նաև կլաստերների ձևն ու թեքությունը (դիսպերսիան և կովարիացիան)։



«Մեծ միավորումը» (The Grand Unification) մաթեմատիկորեն ապացուցում է, որ մեզ  ծանոթ սովորական, կոշտ K-Means-ը առանձին  ալգորիթմ չէ, այլ ընդամենը GMM-ի խիստ սահմանափակված, մասնավոր դեպքն է:

Ո՞ր կոնկրետ պայմանների դեպքում է GMM-ը «փլուզվում» (վերածվում) K-Means-ի:

Քայլ 1. Ոլորտային սահմանափակում ($\Sigma_k = \epsilon I$)

GMM-ում յուրաքանչյուր կլաստեր կարող է ունենալ կամայական ձև՝ ձգված, թեքված, մեծ կամ փոքր օվալներ (էլիպսներ):Սա կառավարվում է $\Sigma_k$ կովարիացիոն մատրիցով:

* Երբ մենք սահմանում ենք $\Sigma_k = \epsilon I$ (որտեղ $I$-ն միավոր մատրիցն է, իսկ $\epsilon$-ը՝ դիսպերսիան), մենք արգելում ենք օվալներին թեքվել կամ ձգվել:

* Բոլոր կլաստերները ստիպողաբար դառնում են կատարյալ շրջանագծեր (կամ գնդեր) և ունենում են ճիշտ նույն չափսը:

* Սա հենց K-Means-ի հիմնական թերություններից ու նախապայմաններից է. K-Means-ը ենթադրում է, որ բոլոր կլաստերները սֆերիկ են և ունեն նույն տարածվածությունը:

Քայլ 2. Դիսպերսիայի սահմանը, երբ այն ձգտում է զրոյի ($\epsilon \to 0$)Սա ամենակարևոր մաթեմատիկական հնարքն է: Երբ շրջանակի շառավիղը (դիսպերսիան՝ $\epsilon$) ձգտում է $0$-ի, Գաուսյան բաշխման կորը վերածվում է ծայրահեղ սրածայր «ասեղի» (այն ձգտում է Դիրակի դելտա ֆունկցիային):

Մաթեմատիկորեն, GMM-ում $x_i$ կետի պատասխանատվությունը ($r_{ik}$) հաշվվում է Բայեսի բանաձևով, որտեղ հեռավորությունը հաշվի է առնվում էքսպոնենտի աստիճանում՝ $\exp\left(-\frac{||x_i - \mu_k||^2}{2\epsilon}\right)$:

Երբ $\epsilon \to 0$՝

* Այն կլաստերի կենտրոնը ($\mu_k$), որն ամենամոտն է $x_i$ կետին, կունենա չնչին ավելի փոքր հեռավորություն, քան մյուսները:

* Բայց քանի որ հայտարարում գտնվող $\epsilon$-ը անվերջ փոքրանում է, էքսպոնենտի մեջ այդ փոքր տարբերությունը դառնում է հսկայական:

* Արդյունքում, ամենամոտ կենտրոնի հավանականությունը սլանում է դեպի 1 ($r_{ik} \to 1$), իսկ մնացած բոլոր հեռու կենտրոններինը՝ դեպի 0 ($r_{ij} \to 0$):

Այլևս չկան «փափուկ» հավանականություններ (օրինակ՝ $0.7$ կամ $0.3$): Հենց որ դիսպերսիան հավասարեցնում ենք զրոյի, հավանականությունները վերածվում են կոշտ 0-ների և 1-երի: Կետը պատկանում է բացառապես իրեն ամենամոտ ցենտրոիդին:Սա հենց Hard K-Means-ն է: Այսպիսով, K-Means-ը կարելի է սահմանել որպես GMM՝ զրոյական դիսպերսիայով և սֆերիկ կովարիացիայով:

Սինգուլյարություններ (Եզակիություններ)

Սա մի իրավիճակ է, երբ ալգորիթմը մաթեմատիկորեն «կոտրվում է» (crash է լինում) բաժանարար զրոների կամ անվերջությունների պատճառով:

Պատկերացնենք, որ ունենք տվյալների բազմություն, որտեղ կետերի մեծ մասը խմբավորված են միասին, բայց կա մեկ պատահական, մեկուսացված կետ (աղմուկ կամ outlier)՝ մյուսներից շատ հեռու:

Եթե EM ալգորիթմի աշխատանքի ընթացքում Գաուսյան օվալներից մեկի կենտրոնը ($\mu_k$) պատահականորեն «նստի» ուղիղ այդ մեկուսացված $x_i$ կետի վրա, տեղի կունենա շղթայական աղետ:

1. Դիսպերսիան դառնում է 0

Դիսպերսիան ($\sigma^2$ կամ $\Sigma_k$) չափում է, թե որքան տարածված են կլաստերի կետերը իրենց կենտրոնի շուրջ: Եթե կլաստերում կա ընդամենը մեկ կետ, և կենտրոնն էլ հենց ինքն է ($\mu_k = x_i$), ապա այդ կետի հեռավորությունը կենտրոնից հավասար է $0-ի$՝ $(x_i - \mu_k)^2 = 0$:Մոդելը հասկանում է, որ այս խմբի ներսում տարածվածություն չկա, ուստի սահմանում է $\sigma \to 0$:

2. Հավանականության խտության «պայթյունը»

Գաուսյան բաշխման բանաձևի հանդեպ հաստատուն գործակիցը հավասար է $\frac{1}{\sqrt{2\pi\sigma^2}}$:

Երբ մենք փորձում ենք հաշվել այս արժեքը, տեսնում ենք, որ հայտարարում գտնվող $\sigma$-ն ձգտում է $0-ի$: Ինչպես գիտենք մաթեմատիկական անալիզից`

$$\lim_{\sigma \to 0} \frac{1}{\sqrt{2\pi\sigma^2}} = \frac{1}{0} = +\infty$$

Երկրաչափորեն սա նշանակում է, որ Գաուսյան լայն, գեղեցիկ զանգակաձև կորը վերածվում է անվերջ բարձր և անվերջ նեղ «ասեղի» (սինգուլյարություն):

3. Ինչո՞ւ է սա մղձավանջ ալգորիթմի համար

EM ալգորիթմի նպատակն է մաքսիմալացնել ամբողջ տվյալների համատեղ հավանականության լոգարիթմը (Log-Likelihood): Քանի որ այս մի հատիկ կետի հավանականությունը դարձավ $+\infty$, ամբողջ տվյալների բազայի գումարային ճշմարտանմանությունը նույնպես դառնում է անվերջություն:

Ալգորիթմը «կարծում է», թե գտել է կատարյալ լուծում, քանի որ անվերջությունից մեծ արժեք չկա: Նա դադարում է ուշադրություն դարձնել մնացած հազարավոր կարևոր կետերին, կենտրոնանում է միայն այս մեկ կետի վրա և կոդի մակարդակով տալիս է  DivisionByZero կամ NaN (Not a Number) սխալ ու կանգնում է: 

Որպեսզի GMM-ը կիրառելիս այսպիսի աղետներ չլինեն, Scikit-learn-ում կամ այլ գրադարաններում կովարիացիոն մատրիցի անկյունագծային տարրերին միշտ գումարում են շատ փոքր հաստատուն թիվ (օրինակ՝ $10^{-6}$), որը կոչվում է reg_covar (regularization covariance)։ Սա թույլ չի տալիս, որ դիսպերսիան երբևէ հասնի բացարձակ $0$-ի։

Սինգուլյարությունների խնդրի լուծումը՝ Այն ցույց է տալիս, թե ինչպես կարելի է մաթեմատիկական հնարքի՝ Ռեգուլյարիզացիայի (Regularization) միջոցով փրկել ալգորիթմը պայթյունից:

Կովարիացիոն մատրիցների՝ դեպի զրո դառնալը կանխելու համար մենք պետք է միջամտենք մաթեմատիկորեն:

Շտկումը (Լուծումը): Յուրաքանչյուր M-քայլի ժամանակ մենք կովարիացիոն մատրիցի անկյունագծին գումարում ենք շատ փոքր, դրական հաստատուն ($\epsilon$). Ռեգուլյարիզացված կովարիացիա՝

 $\hat{\Sigma}_k = \hat{\Sigma}_k + \epsilon I$:

Դիտարկենք էե  ինչ է կատարվում մատրիցների մակարդակով, երբ կիրառում ենք 

$\hat{\Sigma}_k = \hat{\Sigma}_k + \epsilon I$ 

բանաձևը:


Մանրամասն մաթեմատիկական բացատրություն

Հասկանանք, թե ինչ է կատարվում մատրիցների մակարդակով, երբ կիրառում ենք   $\hat{\Sigma}_k = \hat{\Sigma}_k + \epsilon I$ բանաձևը:

* ​$\epsilon$-ը (էպսիլոն) չափազանց փոքր դրական թիվ է, օրինակ՝ $10^{-6}$ կամ $10^{-5}$:

* $I$-ն միավոր մատրիցն է (Identity matrix), որի գլխավոր անկյունագծի տարրերը 1-եր են, իսկ մնացած բոլոր տեղերում 0 է:

* Երբ դրանք բազմապատկում ենք, ստանում ենք մի մատրից, որի միայն անկյունագծի վրա գրված է $\epsilon$՝


$$\epsilon I = \begin{pmatrix} \epsilon & 0 & 0 \\ 0 & \epsilon & 0 \\ 0 & 0 & \epsilon \end{pmatrix}$$


2. Ինչպե՞ս է սա փոխում իրավիճակը

Երբ ալգորիթմը փորձում է կլաստերի դիսպերսիան սեղմել և դարձնել $0$ (քանի որ կլաստերում ընդամենը մեկ կետ է մնացել), $\hat{\Sigma}_k$ մատրիցի անկյունագծային տարրերը (որոնք պատասխանատու են յուրաքանչյուր հատկանիշի դիսպերսիայի համար) ձգտում են զրոյի:Բայց մեր գումարումից հետո նոր, ռեգուլյարիզացված մատրիցի անկյունագծային տարրերը կլինեն $0 + \epsilon = \epsilon$:

Արդյունքը: Դիսպերսիան երբեք չի կարող դառնալ բացարձակ $0$: Այն միշտ կլինի առնվազն $\epsilon$: Քանի որ հայտարարը $0$ չի դառնում, հավանականության խտության ֆունկցիան չի պայթում դեպի անվերջություն, և կոդը չի տալիս DivisionByZero սխալ: Երկրաչափորեն՝ Գաուսյան բաշխման կորը չի վերածվում անվերջ բարձր ասեղի, այլ պահպանում է իր փոքրիկ, բայց կայուն ձևը:


Կապը Ridge Regression-ի հետ (The Callback)

Մեքենայական ուսուցման մեջ շատ կարևոր մի փիլիսոփայություն՝ անվերջություններից ու գերհարմարեցումից (overfitting) խուսափելու համար միշտ օգտագործվում է տույժի (penalty) կամ արհեստականորեն փոքր արժեք ավելացնելու մեթոդը:

* Ridge Regression (L2 Regularization): Այստեղ մենք կորստի ֆունկցիային (Loss function) գումարում ենք կշիռների քառակուսիների գումարը՝ $\lambda \sum w_j^2$: Մատրիցային տեսքով սա հանգեցնում է հենց $X^TX + \lambda I$ գործողությանը: Սա արվում է, որպեսզի մատրիցը դառնա հակադարձելի (invertible), և մոդելի կշիռները չպայթեն ու չդառնան անվերջ մեծ:

*  GMM Regularization: Այստեղ նույնպես ավելացվում է $\epsilon I$, որպեսզի կովարիացիոն մատրիցը լինի դրականորեն որոշված (positive-definite) և հակադարձելի: Չէ՞ որ Գաուսյան բաշխման բանաձևում մեզ պետք է գտնել հակադարձ մատրիցը՝ $\Sigma^{-1}$, իսկ զրոյական անկյունագծով մատրիցը հակադարձել հնարավոր չէ (դեթերմինանտը դառնում է 0):

Այսպիսով, $\epsilon I$ ավելացնելը մաթեմատիկական հզոր «անվտանգության բարձիկ» է, որն ապահովում է ալգորիթմների կայուն աշխատանքը իրական, աղմկոտ տվյալների հետ աշխատելիս:


GMM-ի հաջորդ մեծ մարտահրավերը՝ «Գործնական մղձավանջ 2՝ «K»-ի ընտրությունը»: Այն բացատրում է, թե ինչու կլաստերների ճիշտ քանակը ընտրելու հարցում չենք կարող կուրորեն վստահել մոդելի հավանականության ցուցանիշին, և թե ինչ թակարդի մեջ կընկնենք, եթե այդպես անենք:

K-Means-ում մենք օգտագործում էինք «Արմունկի մեթոդը» (Elbow Method) կամ Սիլուետի գործակիցը (Silhouette Score)՝ կլաստերների օպտիմալ քանակը գուշակելու (գտնելու) համար:

Քանի որ GMM-ը գեներատիվ (ծնող) մոդել է, այն արտածում է իրական թվային Լոգարիթմական ճշմարտանմանություն (Log-Likelihood) ամբողջ տվյալների բազմության համար:

1. Ի՞նչ է Log-Likelihood-ը
Ի տարբերություն K-Means-ի, որը հեռավորությունների վրա հիմնված երկրաչափական ալգորիթմ է, GMM-ը հավանականային մոդել է: Այն մեզ տալիս է կոնկրետ մաթեմատիկական մեծություն՝ Log-Likelihood (Լոգարիթմական ճշմարտանմանություն), որը ցույց է տալիս, թե որքանով է հավանական, որ մեր ընտրած Գաուսյան բաշխումները կարող էին ծննել (գեներացնել) այս տվյալները: Որքան մեծ է այս թիվը, այնքան մոդելը «ավելի լավ է գրկում» տվյալները:

2. Ինչո՞ւ առավելագույն Log-Likelihood-ը փնտրելը կտանի աղետի

Բնական հարց է առաջանում. եթե ունենք մոդելի որակը չափող հստակ թիվ, եկեք փորձենք տարբեր $K$-եր (օրինակ՝ 2 կլաստեր, 3 կլաստեր, 5 կլաստեր) և վերցնենք այն մեկը, որտեղ այդ թիվն ամենամեծն է:

Սակայն այստեղ թաքնված է մաթեմատիկական թակարդը՝ $K$-ն 
մեծացնելուն զուգընթաց Log-Likelihood-ը երբեք չի նվազում, այն միշտ աճում է։

Եթե մենք ունենք $N$ հատ կետ, ու ալգորիթմին թույլ տանք վերցնել $K=N$ հատ կլաստեր, ապա ամեն մի հատիկ կետի վրա կնստի մի առանձին Գաուսյան օվալ: Ինչպես արդեն գիտենք սինգուլյարության (singularities) խնդրից, երբ Գաուսյան բաշխումը կենտրոնանում է մեկ կետի վրա, նրա հավանականության խտությունը ձգտում է անվերջության: Արդյունքում ամբողջ մոդելի Log-Likelihood-ը կդառնա անվերջ մեծ ($+\infty$):

Մաթեմատիկորեն սա «կատարյալ» է, բայց գործնականում՝ անպիտան, քանի որ մենք ոչ թե խմբավորել ենք տվյալները, այլ ուղղակի անգիր ենք արել ամեն մի կետի կոորդինատը: Սա գերհարմարեցման (overfitting) դասական օրինակ է:

Ինչպե՞ս է լուծվում այս խնդիրը գործնականումՔանի որ Log-Likelihood-ը միայնակ չի կարող հուշել ճիշտ $K$-ն, մեքենայական ուսուցման մեջ օգտագործում են տեղեկատվական չափանիշներ (Information Criteria), որոնք պատժում (penalty են տալիս) մոդելին բարդության՝ այսինքն կլաստերների քանակի համար:

Ամենահայտնի երկու մոտեցումներն են՝

1․ BIC (Bayesian Information Criterion - Բայեսյան տեղեկատվական չափանիշ):

$$\text{BIC} = \ln(N)d - 2\ln(\hat{L})$$

2․ AIC (Akaike Information Criterion - Ակաիկեի տեղեկատվական չափանիշ):

$$\text{AIC} = 2d - 2\ln(\hat{L})$$

Որտեղ $d$-ն մոդելի պարամետրերի (ազատության աստիճանների) քանակն է (կախված $K$-ից), իսկ $\ln(\hat{L})$ -ը հենց մեր Log-Likelihood-ն է:

Ինչպե՞ս են դրանք աշխատում՝

Երբ $K$-ն մեծանում է, ճշմարտանմանության մասը (աջ կողմը) նվազեցնում է արժեքը (քանի որ բանաձևում $-2\ln(\hat{L})$ է), բայց փոխարենը պարամետրերի քանակի տույժը (ձախ կողմը) կտրուկ մեծանում է: Մեր նպատակն է գտնել այն $K$-ն, որի դեպքում BIC կամ AIC արժեքը հասնում է իր նվազագույնին (minimum): Դա հենց այն կետն է, որտեղ մոդելը և՛ բավարար չափով ճշգրիտ է, և՛ ավելորդ չի բարդացված:


GMM առավելություններն ու Թերությունները (Pros & Cons)


Առավելությունների մանրամասն վերլուծություն
1. Ոչ-սֆերիկ կլաստերների հայտնաբերում
K-Means-ը կուրորեն ենթադրում է, որ բոլոր կլաստերները կատարյալ գնդեր են: Եթե իրական տվյալները ձգված են էլիպսի նման կամ թեքված են ինչ-որ անկյան տակ, K-Means-ը կձախողվի: GMM-ն օգտագործում է կովարիացիոն մատրիցը, ինչը թույլ է տալիս Գաուսյան օվալներին ձգվել և պտտվել ցանկացած ուղղությամբ՝ կատարելապես հարմարվելով տվյալների իրական երկրաչափական ձևին:

2. Փափուկ վերագրում և համընկնում (Overlapping)
Եթե երկու կլաստեր իրար շատ մոտ են կամ նրանց սահմանները խառնված են, K-Means-ը սահմանի վրա գտնվող կետերը կոպիտ կերպով կբաժանի երկու մասի: GMM-ը հասկանում է անորոշությունը. այն սահմանային կետերին տալիս է հավանականություններ (օրինակ՝ 55% Կլաստեր A և 45% Կլաստեր B): Սա շատ ավելի ռեալիստական է գործնական խնդիրներում:

3. Մասշտաբի (Scale) հանդեպ ցածր զգայունություն
Քանի որ K-Means-ը հիմնված է Էվկլիդյան հեռավորության վրա  $d = \sqrt{\sum(x_i - y_i)^2}$),
եթե մի հատկանիշի արժեքները շատ մեծ լինեն մյուսից (օրինակ՝ աշխատավարձը՝ դրամով, իսկ տարիքը՝ տարիներով), ապա մեծ արժեքներով հատկանիշը ամբողջությամբ կճնշի մյուսին: GMM-ը յուրաքանչյուր հատկանիշի դիսպերսիան հաշվարկում է առանձին, ինչի շնորհիվ այն ավելի ճկուն է մասշտաբային տարբերությունների հանդեպ:


Թերությունների մանրամասն վերլուծություն



1. Հաշվողական ծանրություն

K-Means-ը շատ արագ է, քանի որ հաշվում է միայն պարզ հեռավորություններ: GMM-ի EM ալգորիթմը ամեն իտերացիայի ժամանակ բազմաչափ տարածություններում հաշվում է էքսպոնենտներ, ինչպես նաև կատարում է կովարիացիոն մատրիցների հակադարձում ($\Sigma^{-1}$) և դեթերմինանտների հաշվարկ: Մեծ տվյալների (Big Data) դեպքում սա կարող է զգալիորեն դանդաղեցնել աշխատանքը:


2. Լոկալ օպտիմումների թակարդը և Գլոբալ երաշխիքի բացակայությունը


EM ալգորիթմը գրադիենտային վերելքի (gradient ascent) նման է աշխատում. այն քայլ առ քայլ բարձրանում է դեպի Log-Likelihood-ի ամենամոտ «բլուրը»: Եթե մենք պատահական սկզբնավորման (initialization) ժամանակ կենտրոնները անհաջող տեղերում դնենք, ալգորիթմը կբարձրանա ամենամոտ փոքր բլրի վրա (Local Optimum) և կկանգնի՝ այդպես էլ չհասնելով ամենաբարձր լեռանը (Global Optimum):Գործնականում սա բուժելու համար GMM-ը գործարկում են մի քանի անգամ տարբեր սկզբնական կետերով (n_init պարամետրը Scikit-learn-ում), կամ սկզբնական կենտրոնները ընտրում են K-Means++ ալգորիթմի միջոցով:

3. Զգայունություն աղմուկի հանդեպ

Ինչպես արդեն տեսանք նախորդ սլայդներում, եթե տվյալների մեջ կան ծայրահեղ առանձնացված կետեր (outliers), մոդելը կարող է ընկնել սինգուլյարության թակարդը (դիսպերսիան դառնա 0, իսկ հավանականությունը՝ անվերջություն) և ամբողջությամբ խափանվել:


4. Բարձր չափայնության (High Dimensions) խնդիրը

Եթե մեր տվյալներն ունեն շատ հատկանիշներ (features, օրինակ՝ $D=100$), ապա կովարիացիոն մատրիցի չափսը դառնում է $100 \times 100$: Դա նշանակում է, որ մոդելը պետք է գնահատի հսկայական քանակությամբ պարամետրեր (կշիռներ) յուրաքանչյուր կլաստերի համար: Եթե տվյալների քանակը բավարար չէ, մոդելը կտառապի «չափայնության անեծքից» (Curse of Dimensionality), կկորցնի իր երկրաչափական իմաստը և կգերհարմարեցվի (overfit):


Տեղեկատվական չափանիշները (Information Criteria), մասնավորապես՝ AIC (Akaike Information Criterion) և BIC (Bayesian Information Criterion):

Սրանք վիճակագրական և մեքենայական ուսուցման (Machine Learning) մեջ օգտագործվող կարևոր գործիքներ են, որոնք օգնում են ընտրել լավագույն մոդելը մի քանի հնարավոր տարբերակներից (օրինակ՝ կլաստերիզացիայի մեջ օպտիմալ $K$ քանակը կամ ռեգրեսիայի մեջ հատկանիշների լավագույն բազմությունը գտնելու համար)։

1. Հիմնական գաղափարը (The Core Idea)

Երբ մենք բարդացնում ենք մոդելը (ավելացնում ենք նոր պարամետրեր կամ կլաստերներ), այն ավելի լավ է հարմարվում տվյալներին, և մոդելի ճշտությունը (կամ հավանականությունը՝ Likelihood) միշտ աճում է։ Սակայն չափազանց բարդ մոդելը տանում է գերհարմարեցման (overfitting): Օպտիմալ մոդելը գտնելու համար մենք պետք է «պատժենք» (penalize) մոդելին չափից շատ բարդության համար: Այսինքն՝ փնտրում ենք բալանս ճշտության և պարզության միջև (Occam's razor):

Նշանակումներ՝

$\hat{L}$ (Maximum Likelihood)՝ Մոդելի առավելագույն հավանականության ֆունկցիայի արժեքն է: Այն ցույց է տալիս, թե որքան լավ է մոդելը նկարագրում տվյալները (որքան մեծ է $\hat{L}$-ը, այնքան լավ)։

$p$՝ Մոդելի ընդհանուր պարամետրերի (կամ հատկանիշների/կլաստերների) քանակն է: Սա հենց բարդության չափանիշն է:

$N$՝ Տվյալների (նմուշների/տողերի) ընդհանուր քանակն է (Sample size):

2. Akaike Information Criterion (AIC)$$AIC = 2p - 2\ln(\hat{L})$$

$-2\ln(\hat{L})$ (Goodness of fit)՝ Սա մոդելի սխալի կամ անհամապատասխանության չափն է։ Քանի որ մենք ուզում ենք, որ $\hat{L}$-ը մեծ լինի, ապա $-2\ln(\hat{L})$-ը կլինի փոքր։

$2p$ (Penalty term)՝ Սա տուգանքն է (penalty) բարդության համար։ Ինչքան շատ պարամետր ($p$) ունի մոդելը, այնքան մեծանում է այս արժեքը։

Նպատակը՝ AIC-ն փորձում է գտնել այնպիսի մոդել, որը հնարավորինս լավ կկանխատեսի նոր տվյալները (նվազեցնում է կանխատեսման ինֆորմացիոն կորուստը):


3. Bayesian Information Criterion (BIC)

$$ BIC = p\ln(N) - 2\ln(\hat{L}) $$


$ -2\ln(\hat{L}) $      Նույնն է, ինչ AIC-ի դեպքում (ցույց է տալիս, թե որքան լավ է մոդելը նստում տվյալների վրա)։

$p\ln(N)$ (Penalty term)՝ Սա BIC-ի տուգանային տերմն է։ Այստեղ տուգանքը կախված է ոչ միայն պարամետրերի քանակից ($p$), այլ նաև տվյալների քանակից ($N$)՝ $\ln(N)$ գործակցի միջոցով։

4. Կիրառությունը և տարբերությունները (Usage & Key Difference)

Ինչպե՞ս օգտագործել՝ Մենք հաշվում ենք AIC կամ BIC արժեքները տարբեր $K$-երի (օրինակ՝ կլաստերների քանակի) համար և գծում ենք գրաֆիկը:  Քանի որ երկու բանաձևերում էլ առաջին տերմը տուգանքն է, իսկ երկրորդը՝ սխալը, ցածր AIC/BIC արժեքը նշանակում է ավելի օպտիմալ մոդել: Մենք ընտրում ենք այն $K$-ն, որտեղ այս արժեքները հասնում են իրենց բացարձակ մինիմումին:


Գլխավոր տարբերությունը (AIC vs BIC)։

Եթե տվյալների քանակը $N \ge 8$, ապա $\ln(N) > 2$: Սա նշանակում է, որ BIC-ի տուգանքի գործակիցը ($p\ln(N)$) ավելի մեծ է, քան AIC-ինը ($2p$):

BIC-ն ավելի խիստ է պատժում բարդ մոդելներին: Այդ պատճառով BIC-ն հակված է ընտրելու ավելի պարզ մոդելներ (ավելի քիչ պարամետրերով կամ կլաստերներով)։

AIC-ն ավելի մեղմ է, ինչի պատճառով երբեմն կարող է ընտրել մի փոքր ավելի բարդ մոդել (overfitted), բայց այն շատ լավն է, երբ նպատակը զուտ կանխատեսման ճշտությունն է:


«Պարամետրիկ սահմանափակումը» (The Parametric Limitation)


«Պարամետրիկ սահմանափակումը» (The Parametric Limitation) և բացատրում է, թե ինչու տվյալների վերլուծության որոշ մոդելներ (օրինակ՝ GMM-ը) լավ չեն աշխատում բարդ ձև ունեցող տվյալների հետ, և ինչու պարզ լուծումները (օրինակ՝ հիստոգրամը) նույնպես կատարյալ չեն։

GMM-ը (Gaussian Mixture Model) համարվում է պարամետրիկ մոդել։ Դա նշանակում է, որ մենք նախապես ենթադրություն ենք անում տվյալների բաշխվածության տեսքի մասին և պարտավոր ենք նախապես նշել կլաստերների (խմբերի) հստակ քանակը, որը նշանակված է որպես $K$:

Եթե մենք ենթադրենք, թե ունենք 3 կլաստեր ($K=3$), բայց իրականում մեր տվյալները դասավորված են մի մեծ, շարունակական, ոլորվող պարույրի (spiral) տեսքով, ապա GMM մոդելը ձախողվելու է։

Քանի որ GMM-ն աշխատում է էլիպսաձև «ամպիկներով» (Գաուսյան բաշխումներով), այն չի կարողանա հասկանալ պարույրի ամբողջական կորությունը։ Դրա փոխարեն, այն անփութորեն կտեղադրի երեք հատ էլիպսաձև «բլիթներ» պարույրի տարբեր մասերում՝ լիարժեք չարտացոլելով տվյալների իրական պատկերը։

Հիմնական հարցադրումը (The Question)

«Կարո՞ղ ենք արդյոք գնահատել հավանականության խտության ֆունկցիան՝ $p(x)$-ը, առանց տվյալների գլոբալ ձևի կամ կլաստերների քանակի մասին որևէ նախնական ենթադրություն անելու»։

Այսինքն՝ ինչպե՞ս հասկանալ տվյալների բաշխվածությունը, եթե չգիտենք $K$-ի արժեքը և չենք ուզում մեզ սահմանափակել էլիպսներով կամ այլ կոնկրետ երկրաչափական պատկերներով։

Պարզունակ լուծումը և դրա թերությունները (The Naive Solution)

Առաջին միտքը, որ գալիս է գլխում, Հիստոգրամն է (սյունակավոր գրաֆիկը), որը հաճախ օգտագործվում է տվյալների բաշխվածությունը տեսնելու համար։ Սակայն հիստոգրամն ունի լուրջ թերություններ (Flaws)․

«Histograms are discontinuous (blocky)»: Հիստոգրամները անընդհատ չեն. դրանք կտրտված են, բաղկացած են կոշտ սյունակներից («աղյուսներից»)։ Իսկ իրական կյանքում հավանականությունները սովորաբար սահուն (հարթ) տեսք ունեն։

Հիստոգրամի տեսքը կարող է կտրուկ ու արմատապես փոխվել՝ կախված նրանից, թե մենք ինչպես և որտեղ կգծենք սյունակների սահմանները (bin edges)։ Սահմանները մի փոքր աջ կամ ձախ տանելով՝ կարող ենք ստանալ լիովին տարբեր պատկերներ, ինչը հուսալի չէ։

Այսինքն՝

* Պարամետրիկ մոդելները (ինչպես GMM-ը) չափազանց կոշտ են բարդ տվյալների համար։

* Հիստոգրամները չափազանց պարզունակ են և կոպիտ։

* Անհրաժեշտ է մի այնպիսի մեթոդ, որը կլինի և՛ սահուն (անընդհատ), և՛ թույլ կտա նկարագրել ցանկացած բարդ ձևի տվյալներ՝ առանց նախապես կլաստերների քանակ հուշելու։

Այստեղ խոսքը գնում է հիստոգրամը որպես ՄՈԴԵԼ օգտագործելու մասին, որը պետք է գուշակի (գնահատի) $p(x)$ ֆունկցիան ցանկացած կետի համար։

1. Հիստոգրամը որպես «Հավանականության Գուշակ»

Եթե մենք ունենք հիստոգրամ, մենք կարող ենք այն ձևափոխել այնպես, որ սյունակների մակերեսների գումարը լինի 1 (սրան կոչում ենք նորմալացված հիստոգրամ)։ Այդ դեպքում այն դառնում է հավանականության խտության ֆունկցիայի գնահատական ($p(x)$):Եթե մոդելին տանք մի նոր, անհայտ $x$ կետ, մոդելը կնայի, թե այդ $x$-ը հիստոգրամի որ սյունակի (bin) մեջ է ընկնում, և կպատասխանի. «Այս կետի հանդիպելու հավանականության խտությունը $Y$ է»։

Եթե հիստոգրամն օգտագործենք որպես $p(x)$-ը գուշակող մոդել, մենք կբախվենք 2 մեծ խնդրի։

Խնդիր Ա. Կախվածությունը սահմաններից (Bin Edges)
Պատկերացրեք՝ ունեք տվյալների երկու կետ՝ $x_1 = 4.99$ և $x_2 = 5.01$: Իրար շատ մոտ են, չէ՞:Բայց եթե դուք ձեր հիստոգրամի սյունակի սահմանը (edge) պատահականորեն դնեք ուղիղ $5.0$-ի վրա, ապա․$4.99$-ը կընկնի ձախ սյունակի մեջ։$5.01$-ը կընկնի աջ սյունակի մեջ։

Եթե ձախ սյունակում շատ կետեր կան, իսկ աջում՝ քիչ, ապա մեր «հիստոգրամ-մոդելը» կասի, որ $4.99$ ստանալու հավանականությունը շատ մեծ է, իսկ դրանից հազարերորդականով տարբերվող $5.01$-ի հավանականությունը՝ շատ փոքր։ Սա անհեթեթություն է՝ բնության մեջ տվյալները սահուն են փոխվում, այլ ոչ թե թռիչքներով։

Խնդիր Բ. Կտրտվածություն (Blocky / Discontinuous)

Քանի որ սյունակները ունեն հաստատուն բարձրություն, սյունակի ներսում բոլոր կետերի համար մոդելը տալիս է նույն հավանականությունը։ Օրինակ, եթե սյունակը $5$-ից $10$ միջակայքն է, ապա $5.1$-ի և $9.9$-ի համար մոդելը նույն հավանականությունը կգուշակի, ինչը նույնպես ճիշտ չէ, քանի որ իրական $p(x)$ ֆունկցիան պետք է լինի հարթ ու սահուն կոր։

Փոխանակ տվյալները բաժանենք կոշտ սյունակների, մենք կարող ենք ամեն մի տվյալի կետի վրա դնել մի փոքրիկ «զանգակաձև» կոր (Գաուսյան բաշխում) և այդ բոլոր փոքրիկ կորերը գումարել իրար։ Արդյունքում կստանանք շատ սիրուն, սահուն կոր, որը կրկնում է տվյալների ցանկացած բարդ ձև (թեկուզ պարույր), չունի կոշտ սահմաններ և նախապես չի պահանջում իմանալ կլաստերների $K$ քանակը։ Այս մեթոդը կոչվում է KDE (Kernel Density Estimation):

Parzen-Rosenblatt պատուհանը (The Parzen-Rosenblatt Window), որն ավելի հայտնի է որպես KDE (Kernel Density Estimation — Միջուկային խտության գնահատում)։
Սա հենց հիստոգրամի խնդիրները լուծող, անընդհատ ու սահուն մոդելն է

Փոխանակ տվյալների կետերը գցենք ֆիքսված, կոշտ սյունակների (bins) մեջ (ինչպես հիստոգրամում), եկեք ամեն մի առանձին կետի ուղիղ կենտրոնում տեղադրենք մի փոքրիկ, սահուն «մինի-հավանականային բաշխում»։ Այդ փոքրիկ բաշխումն անվանում ենք kernel (միջուկ): Սովորաբար դա փոքրիկ Գաուսյան (զանգակաձև) կոր է։

Երբ մենք մեր բոլոր $N$ հատ կետերի վրա դրված անհատական փոքրիկ կորերը գումարում ենք իրար, դրանք միաձուլվում են։ Արդյունքում ստանում ենք մեկ ընդհանուր, հարթ, սահուն և անընդհատ կոր (խտության լանդշաֆտ), որը իդեալականորեն նկարագրում է տվյալների իրական տեսքը։

KDE-ի հիմնական բանաձևը, որով հաշվվում է հավանականության խտությունը ցանկացած $x$ կետի համար`

$$\hat{f}_h(x) = \frac{1}{N} \sum_{i=1}^{N} K_h(x - x_i) = \frac{1}{Nh} \sum_{i=1}^{N} K\left(\frac{x - x_i}{h}\right)$$


$\hat{f}_h(x)$: Սա մեր մոդելի կողմից գնահատված (դրա համար գլխին ունի «տանիք»՝ $\hat{}$) հավանականության խտությունն է $x$ կետում։

$N$: Տվյալների բոլոր կետերի ընդհանուր քանակն է (Total number of training points): Բանաձևի սկզբում $\frac{1}{N}$-ով բազմապատկելը (կամ $N$-ի վրա բաժանելը) նշանակում է, որ մենք վերցնում ենք բոլոր կորերի միջին արժեքը, որպեսզի վերջնական ֆունկցիայի ընդհանուր մակերեսը հավասար լինի 1-ի (նորմալացում)։

$\sum_{i=1}^{N}$: Գումարման նշանն է։ Այն ասում է՝ անցիր բոլոր $x_i$ կետերով ($i=1$-ից մինչև $N$) և իրար գումարիր նրանց ներդրումները։$K$: Միջուկի ֆունկցիան է (The Kernel function): Այն որոշում է, թե ինչ ձև ունի ամեն մի կետի շուրջը կառուցվող պատուհանը։ Ինչպես նշված է սլայդում, սովորաբար օգտագործվում է ստանդարտ Գաուսյան բաշխումը (զանգակաձև կորը)։

$x - x_i$: Սա ցույց է տալիս հեռավորությունը այն $x$ կետի (որտեղ ուզում ենք հաշվել հավանականությունը) և մեր ընտրանքի $x_i$ կետի միջև։ Ինչքան $x$-ը մոտ է $x_i$-ին, այնքան $K$ ֆունկցիայի արժեքը մեծ կլինի (այսինքն՝ այդ կետը մեծ ազդեցություն ունի $x$ հարևանության վրա)։

Բանաձևում ամենակարևոր դերերից մեկը խաղում է $h$-ը, որը կոչվում է Bandwidth (թողունակության շերտ կամ պատուհանի լայնություն)։

Ի՞նչ է անում $h$-ը: Այն կառավարում է միջուկի (կորի) լայնությունը։

Բանաձևում տեսնում եք, որ հեռավորությունը բաժանվում է $h$-ի վրա՝ $\left(\frac{x - x_i}{h}\right)$:
* Եթե $h$-ը շատ փոքր է, ապա ամեն մի կետի շուրջը ստացվում է շատ նեղ, սուր և բարձր «ասեղ»։ Վերջնական գրաֆիկը կստացվի շատ խորդուբորդ, ամեն մի կետի վրա կունենանք առանձին պիկ (overfitting)։

* Եթե $h$-ը շատ մեծ է, ապա կորերը դառնում են չափազանց լայն և տափակ։ Դրանք գումարելիս ամեն ինչ կհարթվի, ու մենք կկորցնենք տվյալների իրական կառուցվածքը (underfitting)։

Բանաձևի հայտարարում գտնվող $h$-ը ($\frac{1}{Nh}$) նույնպես ապահովում է, որ կորի լայնանալուն կամ նեղանալուն զուգընթաց նրա մակերեսը միշտ մնա ճիշտ նորմալացված։

KDE-ն թույլ է տալիս ստանալ գեղեցիկ, հարթ $p(x)$ ֆունկցիա։ Մեզ պետք չէ նախապես գուշակել կլաստերների $K$ քանակը, քանի որ մոդելն ինքնաբերաբար հարմարվում է տվյալների ցանկացած ձևին՝ պարզապես իրար գումարելով բոլոր կետերի սահուն հարևանությունները։

1. Ի՞նչ են ներքևի փոքրիկ գծիկները (Rug Plot)

Գրաֆիկի ամենաներքևի մասում՝ հորիզոնական առանցքի վրա ($x$-երի վրա), տեսնում եք բազմաթիվ սև, կարճ վերտիկալ գծիկներ։

Սրանք մեր իրական տվյալների կետերն են ($x_i$)։Որտեղ գծիկները խիտ են (օրինակ՝ $-1$-ից մինչև $1$ միջակայքում), նշանակում է տվյալների այդ հատվածում հանդիպելու հավանականությունը մեծ է։

2. Ի՞նչ է «reference»-ը (Մոխրագույն հաստ կորը)

Սա տվյալների իրական, մայր բաշխվածությունն է (Ground Truth / True Density), որից գեներացվել են այս կետերը։ Ինչպես տեսնում եք, այն սիրուն, հարթ Գաուսյան զանգ է։

Մեր նպատակն է՝ ունենալով միայն ներքևի սև կետերը, վերականգնել հենց այս մոդելը։

3. Երեք տարբեր $h$ (Bandwidth) արժեքների համեմատությունըԱյժմ տեսնենք, թե ինչ է լինում, երբ փոխում ենք $h$-ի արժեքը․

Լուծում Ա. $h = 0.05$ (Նարնջագույն կոր) — Չափազանց փոքր $h$ (Overfitting)

Վարքագիծը: Կորը ստացվել է չափազանց անհարթ, սուր պիկերով և խորդուբորդություններով (ներքև-վերև է թռչում)։

Պատճառը: Քանի որ $h$-ը շատ փոքր է, ամեն մի անհատական կետի շուրջ կառուցված Գաուսյան կորը ստացվել է շատ նեղ ու սուր։ Դրանք իրար գումարելիս չեն միաձուլվել սահուն կերպով, այլ ամեն մի կետի վրա առանձին «ատամիկ» են ձևավորել։

Արդյունքը: Մոդելը չափից ավելի է հարմարվել տվյալների աղմուկին (High Variance / Overfitting)։ Այն ցույց է տալիս կառուցվածքային փոփոխություններ այնտեղ, որտեղ իրականում (մոխրագույն կորի վրա) ամեն ինչ հարթ է։


Լուծում Բ. $h = 0.3$ (Մուգ կապույտ կոր) — Չափազանց մեծ $h$ (Underfitting)

Վարքագիծը: Կորը ստացվել է չափազանց հարթեցված, տափակ և լայն։

Պատճառը: $h$-ը մեծ է, ինչի պատճառով անհատական կորերը շատ լայն են ու ցածր։ Գումարվելիս դրանք «լվացել» ու հարթեցրել են ողջ պատկերը։ Ուշադրություն դարձրեք, որ կենտրոնում ($x=0$ կետում) կապույտ կորի գագաթը շատ ավելի ցածր է, քան իրական մոխրագույն կորինը։

Արդյունքը: Մոդելը չափից շատ է ընդհանրացրել և կորցրել է տվյալների կարևոր մանրամասները (High Bias / Underfitting)։


Լուծում Գ. $h = 0.1$ (Կանաչ կոր) — Օպտիմալ $h$ (Good Fit)

Վարքագիծը: Սա ամենահաջողված տարբերակն է։

Պատճառը: $h=0.1$ արժեքը ոսկե միջինն է։ Այն բավականաչափ մեծ է, որպեսզի հարթեցնի առանձին կետերի պատահական աղմուկը, բայց բավականաչափ փոքր է, որպեսզի ճշգրիտ կրկնի իրական (մոխրագույն) բաշխվածության բարձրությունն ու ընդհանուր տեսքը։

Գրաֆիկը տեսողականորեն ապացուցում է, որ KDE մոդելի ճշգրտությունը գրեթե ամբողջությամբ կախված է $h$-ի ճիշտ ընտրությունից։ Մեքենայական ուսուցման մեջ գոյություն ունեն հատուկ մաթեմատիկական մեթոդներ (օրինակ՝ Cross-Validation կամ Silverman's rule of thumb) հենց այս օպտիմալ կանաչ տարբերակի $h$-ը ավտոմատ գտնելու համար։


Մեծ միավորումը՝ KDE-ն հենց GMM է» (The Grand Unification: KDE is a GMM!)

Այն ցույց է տալիս, որ առաջին հայացքից լրիվ տարբեր թվացող երկու մոդելները՝ KDE-ն (որտեղ մենք ամեն կետի վրա փոքրիկ կորեր էինք դնում) և GMM-ը (որտեղ փնտրում էինք $K$ հատ էլիպսաձև կլաստերներ), մաթեմատիկորեն նույն բանն են, եթե GMM-ի վրա դնենք որոշակի սահմանափակումներ։

Ինչպե՞ս է KDE-ն վերածվում GMM-ի

Հիշենք GMM-ի ընդհանուր մաթեմատիկական տեսքը․ այն մի քանի Գաուսյան բաշխումների կշռված գումար է, որտեղ ամեն մի կլաստեր ունի իր քաշը (ապրիոր հավանականությունը՝ $\pi_i$), իր կենտրոնը (միջին արժեքը՝ $\mu_i$) և իր կովարիացիոն մատրիցը ($\Sigma_i$)։

KDE-ն դառնում է GMM-ի հատուկ դեպք, երբ տեղի են ունենում հետևյալ 4 պայմանները։

1․ $K = N$ (Կլաստերների քանակը հավասար է տվյալների քանակին)։ GMM-ում մենք ունեինք քիչ քանակով կլաստերներ (օրինակ՝ $K=3$): KDE-ում մենք ասում ենք, որ յուրաքանչյուր առանձին տվյալի կետ ինքնին հանդիսանում է առանձին կլաստեր։

2․ Ամեն մի կլաստերի կշիռը ($\pi_i$) հավասար է ուղիղ $\frac{1}{N}$։ GMM-ում որոշ կլաստերներ կարող են ավելի մեծ լինել, որոշները՝ փոքր։ KDE-ում բոլոր կետերը հավասարազոր են, ուստի յուրաքանչյուր կլաստեր-կետ ունի նույն կշիռը՝ $\frac{1}{N}$

3․ Յուրաքանչյուր Գաուսյան կորի կենտրոնը ($\mu_i$) ֆիքսված է տվյալների համապատասխան $x_i$ կետի կոորդինատների վրա։ GMM-ում մենք փնտրում ու շարժում էինք կլաստերների կենտրոնները, որպեսզի դրանք լավագույնս նկարագրեին խմբերը։ KDE-ում կենտրոնները գամված են իրենց իրական կետերի վրա ($\mu_i = x_i$) և երբեք չեն շարժվում։


4․ Բոլոր Գաուսյան կորերի կովարիացիոն մատրիցները նույնն են և հավասար են $\Sigma_i = h^2 I$։ Սա նշանակում է, որ բոլոր կլաստերները կատարյալ սիմետրիկ (գնդաձև) են, չունեն թեքություններ, և բոլորի շառավիղը/լայնությունը կառավարվում է միևնույն $h$ (bandwidth) պարամետրով ($I$-ն միավոր մատրիցն է, որն ապահովում է սիմետրիան բոլոր չափումներով)։

Եթե այս 4 պայմանները տեղադրենք GMM-ի բանաձևի մեջ, մաթեմատիկորեն կստանանք ճիշտ և կոռեկտ KDE-ի բանաձևը։


GMM-ն աշխատեցնելու համար մեզ պետք էր բարդ և երկար հաշվարկներ պահանջող EM (Expectation-Maximization) ալգորիթմը, որը քայլ առ քայլ փորձում էր գուշակել կլաստերների կենտրոնները ($\mu_i$), կշիռները ($\pi_i$) և ձևերը ($\Sigma_i$)։ Այսինքն՝ տեղի էր ունենում ուսուցման (learning) պրոցես։KDE-ի դեպքում մենք ամբողջությամբ շրջանցում ենք EM ալգորիթմը, քանի որ այստեղ սովորելու ոչինչ չկա։ Բոլոր պարամետրերը նախապես որոշված ու ֆիքսված են (կենտրոնները հենց կետերն են, կշիռները՝ $\frac{1}{N}$)։

Մեզ միայն պետք է ընտրել $h$-ի արժեքը, և վերջ։ Տվյալներն իրենք արդեն հանդիսանում են մոդելը (Memory-based կամ Instance-based learning, ինչպես, օրինակ, KNN ալգորիթմում)։

Այս սլայդը ցույց է տալիս կապը երկու տարբեր մոտեցումների միջև՝

GMM-ը պարամետրիկ է․ այն փորձում է տվյալները «սեղմել» և ներկայացնել մի քանի պարամետրերի միջոցով ($K << N$)։

KDE-ն ոչ պարամետրիկ է․ այն պահում է բոլոր տվյալները և ամեն մի կետ օգտագործում է հավանականությունը հաշվելու համար ($K = N$)։



«Շեղման և դիսպերսիայի փոխզիջումը KDE-ում» (The Bias-Variance Tradeoff in KDE)

Քանի որ KDE-ն չունի օպտիմալացվող կշիռներ կամ կենտրոններ, մոդելի ողջ հաջողությունը կախված է մեկ միակ հիպերպարամետրից՝ $h$ (bandwidth): Այս պարամետրը կառավարում է մոդելի բարդությունը և ուղիղ ազդում է Bias-Variance բալանսի վրա։

Մեծ պատուհանի լայնություն՝ Large $h$ (High Bandwidth)

Մեխանիզմը: Յուրաքանչյուր կետի շուրջ կառուցվող միջուկները (kernels) դառնում են շատ լայն և տափակ։ Դրանք հսկայական չափերով ծածկում են իրար (overlap massively) և ձուլվում միմյանց մեջ՝ հարթեցնելով ողջ տարածությունը։

Մաթեմատիկական հետևանքը:

* High Bias (Բարձր շեղում): Մոդելը չափից ավելի պարզունակ է։ Այն անտեսում է տվյալների իրական կառուցվածքը և առանձնահատկությունները։

* Low Variance (Ցածր դիսպերսիա): Մոդելը կայուն է։ Եթե տվյալների մի քանի կետ փոխենք կամ ավելացնենք, վերջնական կորի տեսքը գրեթե չի փոխվի։


Արդյունքը (Underfitting):  Ամեն ինչ միախառնվում է ու վերածվում մեկ միասնական, անձև ամպի («everything merges into a single blob»)։


2. Փոքր պատուհանի լայնություն՝ Small $h$ (Low Bandwidth)

Մեխանիզմը: Միջուկները վերածվում են շատ նեղ, սուր ասեղների (narrow spikes)։ Քանի որ դրանք չափազանց նեղ են, դրանք իրար չեն հասնում և չեն ծածկում միմյանց (they don't overlap)։

Մաթեմատիկական հետևանքը:

* Low Bias (Ցածր շեղում): Մոդելը կատարյալ կերպով անցնում է յուրաքանչյուր կետի վրայով՝ ճշգրիտ կրկնելով տվյալների ցանկացած մանրուք։

* High Variance (Բարձր դիսպերսիա): Մոդելը ծայրահեղ զգայուն է։ Եթե ընտրանքից նույնիսկ մեկ կետ հեռացնենք, գրաֆիկի այդ հատվածի պիկը (ատամիկը) լրիվ կանհետանա, այսինքն՝ մոդելը կտրուկ կփոխվի։

Արդյունքը (Overfitting): Հավանականության խտությունը դառնում է պարզապես առանձնացված, մեկուսացված դելտա-ֆունկցիաների (սուր ասեղների) մի շարք՝ ուղիղ տվյալների կետերի վրա («a series of isolated delta spikes exactly at the training points»)։


Ինչո՞ւ է սա կոչվում Tradeoff (Փոխզիջում)


Մեքենայական ուսուցման մեջ մենք չենք կարող միաժամանակ Bias-ը դարձնել 0 և Variance-ը դարձնել 0։

Եթե փորձենք նվազեցնել Bias-ը (փոքրացնելով $h$-ը), ավտոմատ կերպով կաճի Variance-ը (overfitting):

Եթե փորձենք նվազեցնել Variance-ը (մեծացնելով $h$-ը), ավտոմատ կերպով կաճի Bias-ը (underfitting):

Մեր խնդիրն է գտնել այնպիսի օպտիմալ $h$, որն ապահովում է նվազագույն ընդհանուր սխալը (Total Error)՝ պահելով բալանսը այս երկու ուժերի միջև, ինչպես տեսանք նախորդ կանաչ գծի օրինակում։

Summary: The Density Landscape

Այն ցույց է տալիս, թե ինչպես է զարգացել հավանականության խտության գնահատման (density estimation) գաղափարը մեքենայական ուսուցման մեջ՝ պարզագույն մոդելներից մինչև ամենաբարդերը, և հուշում է, թե ուր է գնալու դասախոսությունը հաջորդիվ։

1. Խտության գնահատման էվոլյուցիան (The Evolution)
մոդելների զարգացումը ներկայացնում է որպես փոխզիջում (tradeoff) աշխարհի բարդության ընկալման և հաշվարկային արագության միջև.


Փուլ Ա. Մեկ Գաուսյան բաշխում (Single Gaussian / GDA)
Գաղափարը: «Assuming the whole world is one shape.» (Ենթադրում ենք, որ ողջ աշխարհն ունի ընդամենը մեկ ձև)։ Մենք վերցնում ենք մեր բոլոր տվյալները և դրանց վրա տեղադրում ընդամենը մեկ Գաուսյան «զանգ» (օրինակ՝ Գաուսյան դիսկրիմինանտ վերլուծության՝ GDA-ի ժամանակ)։

Հատկությունները: * Maximum bias (Առավելագույն շեղում/սխալ): Մոդելը չափազանց պարզունակ է՝ եթե տվյալները բարդ ձև ունեն (օրինակ՝ պարույր կամ կիսալուսին), այս մոդելը լիովին կձախողվի։

Infinite efficiency (Անսահմանափակ արդյունավետություն/արագություն): Մաթեմատիկորեն չափազանց արագ է հաշվարկվում, պահանջում է նվազագույն հիշողություն և ռեսուրսներ։


Փուլ Բ. Գաուսյան խառնուրդի մոդել (GMM + EM Algorithm)

Գաղափարը: (Ենթադրում ենք, որ աշխարհը բաղկացած է $K$ հատ ձևերից)։ Սա ավելի ճկուն քայլ է։ Մենք ասում ենք, որ տվյալները ոչ թե մեկ խումբ են, այլ $K$ հատ տարբեր Գաուսյան բաշխումների խառնուրդ։ EM ալգորիթմի միջոցով մենք մաթեմատիկորեն գտնում ենք այդ $K$ հատ էլիպսների օպտիմալ կենտրոններն ու լայնությունները։


Փուլ Գ. Միջուկային խտության գնահատում (KDE)

Գաղափարը: Ենթադրում ենք, որ աշխարհը չափազանց բարդ է պարամետրիկ ձևով նկարագրելու համար։ Մենք հրաժարվում ենք $K$ հատ ֆիքսված կլաստերներ փնտրելուց և յուրաքանչյուր առանձին տվյալի կետ դարձնում ենք ինքնուրույն «մինի-բաշխում» (միջուկ)։

Հատկությունները: 

* Maximum variance (Առավելագույն դիսպերսիա): Մոդելը ծայրահեղ զգայուն է տվյալների ցանկացած փոփոխության նկատմամբ (հատկապես փոքր $h$-ի դեպքում)։

Perfectively captures reality (Կատարելապես որսում է իրականությունը): Այն կարող է կրկնել և պատկերել ցանկացած, նույնիսկ ամենախճճված ու անկանոն ձևի տվյալների բաշխվածությունը։
