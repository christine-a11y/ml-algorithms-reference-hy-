Principal Component Analysis



PCA-ն մեքենայական ուսուցման և վիճակագրության մեջ ամենատարածված մեթոդներից է, որն օգտագործվում է տվյալների չափողականությունը նվազեցնելու (օրինակ՝ 3D-ից 2D կամ 2D-ից 1D դարձնելու) համար՝ միաժամանակ պահպանելով հնարավորինս շատ արժեքավոր 
տեղեկատվություն։

PCA-ն կարելի է սահմանել երկու համարժեք ձևերով, որոնք մաթեմատիկորեն հանգեցնում են նույն արդյունքին՝

Մաքսիմալ վարիացիայի (ցրվածության) պահպանում. Տվյալները պրոյեկտվում են (պատկերվում են) ավելի ցածր չափողականություն ունեցող տարածության վրա այնպիսի ուղղությամբ, որտեղ տվյալների տարածվածությունը (վարիացիան) առավելագույնն է (maximized)։ Պարզ ասած՝ մենք փնտրում ենք այն առանցքը, որի երկայնքով տվյալները ամենաշատն են տարածված, որպեսզի ինֆորմացիա չկորցնենք։

Մինիմալ սխալի (պրոյեկցիայի արժեքի) ապահովում. PCA-ն գտնում է այնպիսի ուղղություն, որի դեպքում օրիգինալ կետերի և նրանց պրոյեկցիաների (նոր ուղղու վրա ընկած կետերի) միջև միջին քառակուսային հեռավորությունը լինի նվազագույնը (minimized)։ Այսինքն՝ կետերը դեպի նոր առանցք տեղափոխելիս մենք ցանկանում ենք, որ դրանք հնարավորինս քիչ տեղաշարժվեն իրենց սկզբնական դիրքից։

Գրաֆիկների բացատրությունը


1. Ձախ գրաֆիկ (Բնօրինակ տվյալներ)

Ինչ է պատկերված. Սա տվյալների սկզբնական վիճակն է երկչափ ($X$ և $Y$) տարածության մեջ:

Առանձնահատկություն. Տեսնում ենք, որ տվյալները կլաստերավորված են և ունեն հստակ անկյունագծային ուղղվածություն։ Առանցքների արժեքները տարբեր են (օրինակ՝ $X$-ը 2-ից 6 է, իսկ $Y$-ը՝ 40-ից 100):


2. Մեջտեղի գրաֆիկ (Կենտրոնացում և Գլխավոր առանցքի ընտրություն)

Ինչ է արվել. Տվյալները նորմալացվել են (standardized) կամ կենտրոնացվել են կոորդինատային համակարգի սկզբնակետում $(0,0)$։ Այժմ երկու առանցքներն էլ ունեն նույն մասշտաբը (մոտավորապես -2-ից մինչև 2)։

Սև հաստ գիծը (Գլխավոր բաղադրիչ - PC1). Սա հենց PCA-ի գտած առաջին գլխավոր առանցքն է։

* Այն անցնում է տվյալների ամենաերկար տարածման երկայնքով (մաքսիմալացնում է վարիացիան)։

* Եթե բոլոր կետերը ուղղահայաց պրոյեկտենք այս գծի վրա, կետերից մինչև գիծ հեռավորությունների քառակուսիների գումարը կլինի նվազագույնը։

3. Աջ գրաֆիկ (Տարածության շրջում / Պտույտ)

Ինչ է պատկերված. Սա վերջնական արդյունքն է, երբ կոորդինատային համակարգը պտտեցվել է այնպես, որ մեջտեղի գրաֆիկի սև անկյունագծային գիծը դարձել է հորիզոնական առանցք ($X$ առանցք)։

Արդյունք. Այժմ հորիզոնական առանցքը ($PC_1$) ցույց է տալիս տվյալների հիմնական փոփոխությունը (ինֆորմացիայի մեծ մասը), իսկ ուղղահայաց առանցքը ($PC_2$) ցույց է տալիս երկրորդական, ավելի քիչ կարևոր փոփոխությունը։

Ամփոփում. Եթե մենք ցանկանայինք այս 2D տվյալները դարձնել 1D (մեկ չափանի), ապա PCA-ից հետո պարզապես կպահեինք միայն աջ գրաֆիկի հորիզոնական առանցքի արժեքները և կանտեսեինք ուղղահայացը՝ կորցնելով նվազագույն տեղեկատվություն։


PCA-ի մաթեմատիկական ձևակերպումը՝ հիմնված առավելագույն վարիացիայի (Maximum Variance Formulation) սկզբունքի վրա։

Տվյալների բազմություն. Ունենք $\{x_n\}$ տվյալների բազմություն, որտեղ $n = 1, \dots, N$ ($N$-ը նմուշների/կետերի ընդհանուր քանակն է)։

Չափողականություն ($D$). Յուրաքանչյուր $x_n$ կետ հանդիսանում է էվկլիդեսյան տարածության վեկտոր՝ $D$ չափողականությամբ ($x_n \in \mathbb{R}^D$)։ Սա նշանակում է, որ յուրաքանչյուր օբյեկտ ունի $D$ հատ հատկանիշ (features):

Մեր նպատակը. Պրոյեկտել (պատկերել) այս տվյալները մի նոր տարածության վրա, որն ունի $M$ չափողականություն, ընդ որում՝ $M < D$ (սովորաբար շատ ավելի փոքր)։

Պայման. Այս տեղափոխությունը պետք է արվի այնպես, որ պրոյեկտված տվյալների վարիացիան (ցրվածությունը) լինի առավելագույնը:

Ենթադրություն. Այս պահին ենթադրվում է, որ նոր տարածության $M$ չափողականությունը մեզ նախապես տրված է կամ հայտնի է։


Պրոյեկցիա դեպի մեկչափ տարածություն ($M = 1$)

Մաթեմատիկական մոդելը կառուցելու և հասկանալու համար նախ դիտարկվում է ամենապարզ դեպքը, երբ տվյալները $D$-չափանի տարածությունից պրոյեկտվում են ընդամենը մեկչափանի տարածության ($M = 1$)՝ այսինքն մի ուղղի վրա։

Ուղղության վեկտորը ($\mathbf{u}_1$)

Որպեսզի սահմանենք այն ուղղությունը (առանցքը), որի վրա պետք է պրոյեկտվեն կետերը, օգտագործում ենք $D$-չափանի $\mathbf{u}_1$ վեկտորը։Քանի որ մեզ հետաքրքիր է միայն ուղղությունը, այլ ոչ թե վեկտորի երկարությունը, հարմարության և ընդհանրությունը չկորցնելու համար ընտրվում է միավոր վեկտոր (unit vector)։Մաթեմատիկորեն դա գրվում է այսպես`$$\mathbf{u}_1^T \mathbf{u}_1 = 1$$(Սա նշանակում է, որ վեկտորի երկարության քառակուսին, այսինքն՝ սկալյար արտադրյալն ինքն իր հետ, հավասար է 1-ի)։


Սկալյար արժեքների ստացումը


Երբ $x_n$ կետը պրոյեկտում ենք $\mathbf{u}_1$ ուղղության վրա, մենք կատարում ենք վեկտորների սկալյար արտադրյալ։ Արդյունքում յուրաքանչյուր $x_n$ վեկտոր վերածվում է սկալյար (թվային) արժեքի`

$$\mathbf{u}_1^T x_n$$

Այս արժեքը ցույց է տալիս, թե սկզբնակետից ի՞նչ հեռավորության վրա է գտնվում պրոյեկտված կետը նոր առանցքի վրա։

Որպեսզի հաշվարկենք վարիացիան (ցրվածությունը), նախևառաջ մեզ անհրաժեշտ է գտնել պրոյեկտված նոր կետերի միջին թվաբանական արժեքը։Սկզբնական տվյալների միջին վեկտորը նշանակվում է $\bar{\mathbf{x}}$-ով և հաշվվում է սովորական ձևով՝ բոլոր վեկտորների գումարը բաժանած դրանց քանակի ($N$-ի) վրա`

$$\bar{\mathbf{x}} = \frac{1}{N} \sum_{n=1}^{N} x_n$$

Քանի որ պրոյեկցիան գծային գործողություն է, պրոյեկտված տվյալների միջին արժեքը կլինի պարզապես $\mathbf{u}_1$ վեկտորի և սկզբնական տվյալների միջին վեկտորի սկալյար արտադրյալը`

$$\mathbf{u}_1^T \bar{\mathbf{x}}$$


Նախորդ քայլում մենք տեսանք, որ յուրաքանչյուր կետի պրոյեկցիան մեկչափ ուղղու վրա հավասար է $\mathbf{u}_1^T x_n$, իսկ դրանց միջին արժեքը՝ $\mathbf{u}_1^T \bar{\mathbf{x}}$։ Ըստ վիճակագրության դասական սահմանման՝ վարիացիան կետերի և միջինի տարբերությունների քառակուսիների միջինն է`

$$\frac{1}{N} \sum_{n=1}^{N} \{\mathbf{u}_1^T x_n - \mathbf{u}_1^T \bar{\mathbf{x}}\}^2$$

Այս արտահայտությունը գծային հանրահաշվի օրենքներով կարելի է ձևափոխել ավելի կոմպակտ մատրիցային տեսքի։ Քանի որ

$\mathbf{u}_1^T x_n - \mathbf{u}_1^T \bar{\mathbf{x}} = \mathbf{u}_1^T (x_n - \bar{\mathbf{x}})$, և ցանկացած սկալյար $a$-ի համար $a^2 = a \cdot a = a \cdot a^T$, կարող ենք գրել`


$$\{\mathbf{u}_1^T (x_n - \bar{\mathbf{x}})\}^2 = \left( \mathbf{u}_1^T (x_n - \bar{\mathbf{x}}) \right) \left( \mathbf{u}_1^T (x_n - \bar{\mathbf{x}}) \right)^T = \mathbf{u}_1^T (x_n - \bar{\mathbf{x}}) (x_n - \bar{\mathbf{x}})^T \mathbf{u}_1$$


Այս արժեքը տեղադրելով գումարման նշանի տակ և հանելով հաստատուն $\mathbf{u}_1$ վեկտորները գումարից դուրս՝ ստանում ենք սլայդի առաջին տողի վերջնական տեսքը`

$$\frac{1}{N} \sum_{n=1}^{N} \{\mathbf{u}_1^T x_n - \mathbf{u}_1^T \bar{\mathbf{x}}\}^2 = \mathbf{u}_1^T \mathbf{S} \mathbf{u}_1$$

տեղում $\mathbf{S}$-ը տվյալների կովարիացիոն մատրիցն է (covariance matrix):

Կովարիացիոն մատրիցի սահմանումը ($\mathbf{S}$)

$\mathbf{S}$-ը $D \times D$ չափանի սիմետրիկ մատրից է, որը ցույց է տալիս, թե ինչպես են սկզբնական հատկանիշները փոփոխվում (կորելացվում) միմյանց նկատմամբ`

$$\mathbf{S} = \frac{1}{N} \sum_{n=1}^{N} (x_n - \bar{\mathbf{x}})(x_n - \bar{\mathbf{x}})^T$$

* $(x_n - \bar{\mathbf{x}})$-ը կենտրոնացված վեկտոր է ($D \times 1$ չափի)։

* $(x_n - \bar{\mathbf{x}})^T$-ը դրա տրանսպոնացված տարբերակն է ($1 \times D$ չափի)։

Դրանց արտադրյալը տալիս է $D \times D$ չափանի մատրից յուրաքանչյուր կետի համար, որոնց միջինացումով էլ ձևավորվում է $\mathbf{S}$-ը։

Մաքսիմալացման խնդիրը և Սահմանափակումը

Մեր նպատակն է գտնել այնպիսի $\mathbf{u}_1$ վեկտոր, որը մաքսիմալացնում է ստացված վարիացիան՝ $\mathbf{u}_1^T \mathbf{S} \mathbf{u}_1 \to \max$:Սակայն այստեղ կա մի խնդիր` եթե մենք ուղղակի մեծացնենք $\mathbf{u}_1$ վեկտորի բաղադրիչների արժեքները (օրինակ՝ դարձնենք դրա երկարությունը $\|\mathbf{u}_1\| \to \infty$), ապա $\mathbf{u}_1^T \mathbf{S} \mathbf{u}_1$ արժեքը կձգտի անվերջության, ինչը անիմաստ է դարձնում օպտիմալացումը։Ուստի մեզ անհրաժեշտ է սահմանափակումով օպտիմալացում (constrained maximization): Այդ սահմանափակումը հենց նախորդ հատվածում նշված նորմավորման պայմանն է՝ վեկտորի երկարությունը պետք է լինի 1՝

$$\mathbf{u}_1^T \mathbf{u}_1 = 1 \quad \implies \quad 1 - \mathbf{u}_1^T \mathbf{u}_1 = 0$$

Լագրանժի բազմապատկիչների մեթոդը (Lagrange Multipliers)

Որպեսզի պայմանով օպտիմալացման խնդիրը վերածենք սովորական (առանց սահմանափակման) խնդրի, ներմուծում ենք Լագրանժի բազմապատկիչը, որը նշանակվում է $\lambda_1$ տառով:

Կազմվում է Լագրանժի ֆունկցիան՝

$$\mathbf{u}_1^T \mathbf{S} \mathbf{u}_1 + \lambda_1 (1 - \mathbf{u}_1^T \mathbf{u}_1)$$

Այժմ մեր խնդիրն է մաքսիմալացնել այս նոր ֆունկցիան ըստ $\mathbf{u}_1$-ի։


Ածանցում. Երբ այս արտահայտության մասնակի ածանցյալը ըստ $\mathbf{u}_1$ վեկտորի հավասարեցնում ենք զրոյի (ստացիոնար կետը գտնելու համար), ստանում ենք հետևյալ հիմնարար կապը՝

$$\mathbf{S}\mathbf{u}_1 = \lambda_1\mathbf{u}_1$$

Մաթեմատիկական իմաստը՝ Այս հավասարումը գծային հանրահաշվի դասական սահմանումն է, որը նշանակում է, որ $\mathbf{u}_1$-ը հանդիսանում է $\mathbf{S}$ կովարիացիոն մատրիցի սեփական վեկտորը (eigenvector), իսկ $\lambda_1$-ը՝ դրան համապատասխանող սեփական արժեքը (eigenvalue):


Պրոյեկտված վարիացիայի կապը սեփական արժեքի հետ


Որպեսզի հասկանանք, թե որքան է կազմում մեր պրոյեկտված տվյալների վարիացիան այս ստացիոնար կետում, կատարում ենք հետևյալ ձևափոխությունը՝

Վերոնշյալ հավասարման ձախ կողմից բազմապատկում ենք $\mathbf{u}_1^T$ տրանսպոնացված վեկտորով՝

$$\mathbf{u}_1^T \mathbf{S} \mathbf{u}_1 = \mathbf{u}_1^T (\lambda_1 \mathbf{u}_1)$$

Քանի որ $\lambda_1$-ը սկալյար (թիվ) է, այն կարող ենք դուրս բերել առաջ, իսկ ըստ նորմավորման պայմանի՝ գիտենք, որ $\mathbf{u}_1^T \mathbf{u}_1 = 1$: 

Հետևաբար՝

$$\mathbf{u}_1^T \mathbf{S} \mathbf{u}_1 = \lambda_1 (\mathbf{u}_1^T \mathbf{u}_1) = \lambda_1 \cdot 1 = \lambda_1$$

Եզրակացություն՝ Պրոյեկտված տվյալների վարիացիան ուղղակիորեն հավասար է սեփական արժեքին ($\lambda_1$)։ Որպեսզի այս վարիացիան լինի առավելագույնը (maximum), մենք պետք է ընտրենք այն սեփական վեկտորը, որը համապատասխանում է ամենամեծ սեփական արժեքին ($\lambda_{\max}$):Այս առաջին ընտրված վեկտորը կոչվում է առաջին գլխավոր բաղադրիչ (first principal component):


Հաջորդական բաղադրիչների ընդհանրացումը $M$-չափանի տարածության համար


Իսկ ի՞նչ անել, եթե մեզ պետք է տվյալները պրոյեկտել ոչ թե մեկ ուղղի, այլ $M$-չափանի տարածության վրա ($M > 1$):

Ինկրեմենտալ (փուլային) մոտեցում. Մենք կարող ենք հերթով ընտրել նոր ուղղություններ։ Յուրաքանչյուր հաջորդ ուղղություն ընտրելիս դրվում է երկու պայման՝
* Այն պետք է լինի օրթոգոնալ (ուղղահայաց) նախորդ բոլոր արդեն ընտրված ուղղություններին (այսինքն՝ չպետք է կրկնի արդեն հաշվի առնված տեղեկատվությունը. $\mathbf{u}_i^T \mathbf{u}_j = 0$, երբ $i \neq j$):

* Այն պետք է մաքսիմալացնի պրոյեկտված վարիացիան մնացած բոլոր հնարավոր ուղղությունների մեջ։

Վերջնական ընդհանրացված կանոնը. Եթե դիտարկենք ընդհանուր դեպքը $M$-չափանի պրոյեկցիոն տարածության համար, ապա օպտիմալ գծային պրոյեկցիան, որը մաքսիմալացնում է ընդհանուր վարիացիան, որոշվում է $\mathbf{S}$ կովարիացիոն մատրիցի այն $M$ հատ սեփական վեկտորներով ($U_1, \dots, U_M$), որոնք համապատասխանում են $M$ ամենամեծ սեփական արժեքներին ($\lambda_1, \dots, \lambda_M$)։

Գործնական իմաստը. Տվյալների չափողականությունը $D$-ից $M$ դարձնելու համար մենք պարզապես հաշվում ենք կովարիացիոն մատրիցի սեփական արժեքները, դասավորում դրանք նվազման կարգով, վերցնում առաջին $M$ հատին համապատասխանող սեփական վեկտորները և տվյալները բազմապատկում այդ վեկտորներից կազմված մատրիցով։



PCA-ի երկրորդ համարժեք սահմանմանը, որը հիմնված է պրոյեկցիայի սխալի մինիմալացման (Minimum-error formulation) սկզբունքի վրա

1. Օրթոնորմալ բազիսի ներմուծումը

Որպեսզի նկարագրենք տվյալների պրոյեկցիան և դրա ընթացքում առաջացած սխալը, նախ անհրաժեշտ է սահմանել նոր կոորդինատային համակարգ՝ օգտագործելով $D$-չափանի բազիսային վեկտորների լրիվ օրթոնորմալ բազմություն՝ $\{\mathbf{u}_i\}$, որտեղ $i = 1, \dots, D$:Օրթոնորմալության պայմանը մաթեմատիկորեն գրվում է Կրոնեկերի սիմվոլի ($\delta_{ij}$) միջոցով՝

$$\mathbf{u}_i^T \mathbf{u}_j = \delta_{ij}$$

Սա նշանակում է՝

Եթե $i = j$, ապա $\mathbf{u}_i^T \mathbf{u}_i = 1$ (վեկտորները նորմավորված են, այսինքն՝ ունեն $1$ երկարություն)։

Եթե $i \neq j$, ապա $\mathbf{u}_i^T \mathbf{u}_j = 0$ (վեկտորները փոխուղղահայաց/օրթոգոնալ են)։

2. Տվյալների կետերի վերլուծությունը ըստ նոր բազիսի

Քանի որ այս $\{\mathbf{u}_i\}$ բազիսը լրիվ է (այն ծածկում է ողջ $D$-չափանի տարածությունը), սկզբնական տվյալների ցանկացած $x_n$ կետ կարող է ճշգրիտ կերպով ներկայացվել որպես այդ բազիսային վեկտորների գծային կոմբինացիա՝

$$x_n = \sum_{i=1}^{D} \alpha_{ni} \mathbf{u}_i$$

* Այստեղ $\alpha_{ni}$ գործակիցները սկալյար մեծություններ են, որոնք ցույց են տալիս $x_n$ կետի կոորդինատները նոր համակարգում։ Դրանք տարբեր կլինեն տարբեր $x_n$ կետերի համար։

* Երկրաչափորեն սա նշանակում է կոորդինատային համակարգի պարզ պտույտ (rotation) դեպի $\{\mathbf{u}_i\}$-ով սահմանված նոր համակարգ։ Սկզբնական բաղադրիչները՝ $\{x_{n1}, \dots, x_{nD}\}$, փոխարինվում են համարժեք $\{ \alpha_{n1}, \dots, \alpha_{nD} \}$ բազմությամբ։

3. Գործակիցների ($\alpha_{ni}$) հաշվարկը

Որպեսզի գտնենք կոնկրետ $\alpha_{nj}$ գործակցի արժեքը, հավասարման երկու կողմերը ձախից սկալյար բազմապատկում ենք $\mathbf{u}_j^T$ վեկտորով՝

$$\mathbf{u}_j^T x_n = \mathbf{u}_j^T \sum_{i=1}^{D} \alpha_{ni} \mathbf{u}_i = \sum_{i=1}^{D} \alpha_{ni} (\mathbf{u}_j^T \mathbf{u}_i)$$

Օգտվելով օրթոնորմալության պայմանից ($\mathbf{u}_j^T \mathbf{u}_i = \delta_{ji}$), գումարի միջից կենդանի կմնա միայն այն անդամը, որտքեղ $i = j$ (քանի որ մնացած բոլոր դեպքերում արտադրյալը $0$ է)։ Հետևաբար ստանում ենք՝

$$\alpha_{nj} = x_n^T \mathbf{u}_j$$

(քանի որ սկալյար արտադրյալը կոմուտատիվ է՝ $\mathbf{u}_j^T x_n = x_n^T \mathbf{u}_j$)։

Վերջնական ճշգրիտ տեսքը

Տեղադրելով ստացված $\alpha_{ni}$ գործակիցը սկզբնական գծային կոմբինացիայի մեջ, ստանում ենք ցանկացած տվյալների կետի վերլուծությունը ըստ նոր օրթոնորմալ բազիսի՝

$$x_n = \sum_{i=1}^{D} (x_n^T \mathbf{u}_i) \mathbf{u}_i$$

Ի՞նչ է սա մեզ տալիս՝
Սա դեռևս PCA-ի կրճատումը չէ, սա տվյալների ամբողջական և անկորուստ ներկայացումն է նոր առանցքներով։


4. Տվյալների կետի մոտարկումը ($\widetilde{x}_n$)

Մեր նպատակն է սկզբնական $D$-չափանի $x_n$ կետը մոտարկել մի նոր վեկտորով, որը կոչվում է $\widetilde{x}_n$ (կարդացվում է՝ «x նշանով» կամ «x տիլդայով»)։ Այս նոր վեկտորը պետք է գտնվի ավելի ցածր՝ $M$-չափանի ենթատարածության մեջ ($M < D$)։

Առանց ընդհանրությունը կորցնելու, ենթադրում ենք, որ այդ $M$-չափանի ենթատարածությունը նկարագրվում է մեր նոր բազիսի առաջին $M$ հատ վեկտորներով ($\mathbf{u}_1, \dots, \mathbf{u}_M$)։ Ուստի, $x_n$ կետի մոտարկումը՝ $\widetilde{x}_n$-ը, գրվում է հետևյալ տեսքով՝

$$\widetilde{x}_n = \sum_{i=1}^{M} z_{ni} \mathbf{u}_i + \sum_{i=1}^{D} b_i \mathbf{u}_i$$


Բանաձևի երկու մասերի իմաստը՝

* Առաջին գումարը ($\sum_{i=1}^{M} z_{ni} \mathbf{u}_i$)՝ Սա ներկայացնում է տվյալների պրոյեկցիան մեր ընտրած $M$-չափանի գլխավոր ենթատարածության վրա։ $z_{ni}$ գործակիցները կախված են կոնկրետ $x_n$ կետից (յուրաքանչյուր կետ ունի իր սեփական կոորդինատները այս նոր $M$ առանցքներում)։

* Երկրորդ գումարը ($\sum_{i=M+1}^{D} b_i \mathbf{u}_i$)՝ Սա ներկայացնում է մնացած $D-M$ չափողականությունները, որոնք մենք ցանկանում ենք անտեսել կամ կրճատել։ Ուշադրություն դարձրեք, որ այստեղ $b_i$ գործակիցները հաստատուններ (constants) են, որոնք նույնն են բոլոր $x_n$ կետերի համար (այսինքն՝ դրանք այլևս չեն կախված $n$ ինդեքսից)։ Սա նշանակում է, որ այդ անտեսված առանցքների երկայնքով անհատական տեղեկատվությունը կորչում է, և բոլոր կետերին տրվում է նույն միջինացված դիրքը։Մեր խնդիրն է ընտրել այնպիսի բազիսային $\{\mathbf{u}_i\}$ վեկտորներ, անհատական $\{z_{ni}\}$ գործակիցներ և հաստատուն $\{b_i\}$ արժեքներ, որոնք մինիմալի կհասցնեն չափողականության նվազեցման հետևանքով առաջացած աղավաղումը (distortion):

5. Աղավաղման ֆունկցիայի սահմանումը ($J$)

Որպես աղավաղման (սխալի) չափման միավոր ընտրվում է սկզբնական $x_n$ կետի և դրա մոտարկված $\widetilde{x}_n$ տարբերակի միջև էվկլիդեսյան հեռավորության քառակուսին, որը միջինացվում է տվյալների ողջ բազմության համար ($N$-ի վրա բաժանելով)։

Մաթեմատիկորեն մեր նպատակային ֆունկցիան (Loss function / Cost function), որը պետք է մինիմալացնել ($J \to \min$), ունի հետևյալ տեսքը՝

$$J = \frac{1}{N} \sum_{n=1}^{N} \|x_n - \widetilde{x}_n\|^2$$

* $\|x_n - \widetilde{x}_n\|$-ը վեկտորների տարբերության նորման է (երկարությունը)։

* Քառակուսին ապահովում է, որ մենք աշխատենք դրական արժեքների հետ և մաթեմատիկորեն ավելի հեշտ ածանցենք ֆունկցիան։

նախորդ քայլում սահմանված աղավաղման $J$ ֆունկցիան ածանցվում է ըստ անհայտ գործակիցների ($z_{nj}$ և $b_j$), որպեսզի գտնվեն դրանց օպտիմալ արժեքները, և ստացվում է վերականգնման սխալի (displacement vector) վերջնական տեսքը։

6. Օպտիմալ $z_{nj}$ գործակիցների գտնելը

Հիշենք նախորդ սլայդի աղավաղման ֆունկցիան՝ $J = \frac{1}{N} \sum_{n=1}^{N} \|x_n - \widetilde{x}_n\|^2$, որտեղ մոտարկումն էր՝ $\widetilde{x}_n = \sum_{i=1}^{M} z_{ni} \mathbf{u}_i + \sum_{i=M+1}^{D} b_i \mathbf{u}_i$:

* Ածանցում ըստ $z_{nj}$-ի՝ Եթե ցանկանում ենք մինիմալացնել $J$-ն ըստ անհատական $z_{nj}$ կոորդինատների (որտեղ $j = 1, \dots, M$), վերցնում ենք մասնակի ածանցյալ ըստ $z_{nj}$-ի և հավասարեցնում զրոյի ($\frac{\partial J}{\partial z_{nj}} = 0$)։

* Տեղադրելով $x_n$-ի բացված տեսքը և օգտվելով բազիսային վեկտորների օրթոնորմալության պայմաններից ($\mathbf{u}_i^T \mathbf{u}_j = \delta_{ij}$), ստանում ենք՝
$$z_{nj} = x_n^T \mathbf{u}_j \quad (j = 1, \dots, M)$$

Իմաստը՝ Սա նշանակում է, որ նույնիսկ չափողականության կրճատման դեպքում, ընտրված $M$-չափանի գլխավոր ենթատարածության մեջ կետի օպտիմալ կոորդինատները պարզապես այդ կետի օրթոգոնալ պրոյեկցիաներն են համապատասխան բազիսային վեկտորների վրա:


7. Օպտիմալ $b_j$ հաստատունների գտնելը

Այժմ անցնենք այն չափողականություններին, որոնք ցանկանում ենք անտեսել, այսինքն՝ $j = M+1, \dots, D$ առանցքներին, որտեղ կոորդինատները բոլոր կետերի համար նույն հաստատուն $b_j$ արժեքներն են:

* Ածանցում ըստ $b_j$-ի՝ Մինիմալացնելու համար $J$ ֆունկցիայի մասնակի ածանցյալը ըստ $b_j$-ի հավասարեցնում ենք զրոյի ($\frac{\partial J}{\partial b_j} = 0$)։

* Օգտվելով օրթոնորմալության հատկություններից՝ ստանում ենք, որ օպտիմալ $b_j$-ն հավասար է տվյալների միջին վեկտորի ($\bar{\mathbf{x}}$) պրոյեկցիային այդ առանցքի վրա՝

$$b_j = \bar{\mathbf{x}}^T \mathbf{u}_j \quad (j = M+1, \dots, D)$$

որտեղ 

$\bar{\mathbf{x}} = \frac{1}{N} \sum_{n=1}^{N} x_n$:

Քանի որ մենք հրաժարվում ենք անհատական տեղեկատվությունից այս $D-M$ հատ առանցքների երկայնքով, աղավաղումը նվազագույնի հասցնելու լավագույն տարբերակը բոլոր կետերին այդ առանցքների վրա տվյալների ընդհանուր միջին արժեքը վերագրելն է։

8. Վերականգնման սխալի (Շեղման վեկտորի) կազմումը

Այժմ տեղադրենք ստացված օպտիմալ $z_{ni}$-ի և $b_i$-ի արժեքները սկզբնական կետի ($x_n$) և դրա մոտարկման ($\widetilde{x}_n$) բանաձևերի մեջ։Գիտենք, որ ամբողջական անկորուստ վերլուծությունն է՝ $x_n = \sum_{i=1}^{D} (x_n^T \mathbf{u}_i)\mathbf{u}_i$:

Իսկ մեր օպտիմալացված մոտարկումն է՝

$$\widetilde{x}_n = \sum_{i=1}^{M} (x_n^T \mathbf{u}_i)\mathbf{u}_i + \sum_{i=M+1}^{D} (\bar{\mathbf{x}}^T \mathbf{u}_i)\mathbf{u}_i$$

Եթե հաշվենք սրանց տարբերությունը ($x_n - \widetilde{x}_n$), ապա առաջին $M$ անդամները ամբողջությամբ կկրճատվեն (քանի որ դրանք նույնական են)։ Կմնան միայն $M+1$-ից մինչև $D$ անդամների տարբերությունները՝

$$x_n - \widetilde{x}_n = \sum_{i=M+1}^{D} (x_n^T \mathbf{u}_i)\mathbf{u}_i - \sum_{i=M+1}^{D} (\bar{\mathbf{x}}^T \mathbf{u}_i)\mathbf{u}_i$$

Գումարները միավորելով և դուրս բերելով $\mathbf{u}_i$-ն՝ ստանում ենք կարևոր հավասարումը՝

$$x_n - \widetilde{x}_n = \sum_{i=M+1}^{D} \{(x_n - \bar{\mathbf{x}})^T \mathbf{u}_i\} \mathbf{u}_i$$

Երկրաչափական եզրակացությունԱյս վերջնական հավասարումից բխում է մի շատ կարևոր երկրաչափական փաստ՝

Եզրակացություն. Շեղման (տեղաշարժի) $x_n - \widetilde{x}_n$ վեկտորը ամբողջությամբ ընկած է գլխավոր ենթատարածությանը օրթոգոնալ (ուղղահայաց) տարածության մեջ։

Ինչո՞ւ. Որովհետև այն հանդիսանում է բացառապես $\{\mathbf{u}_i\}$ բազիսային վեկտորների գծային կոմբինացիա միայն $i = M+1, \dots, D$ ինդեքսների համար։ Քանի որ այս բոլոր վեկտորները ըստ սահմանման ուղղահայաց են առաջին $M$ հատ վեկտորներին (որոնք կազմում են մեր պրոյեկցիոն հարթությունը), ապա ստացված սխալի վեկտորը նույնպես ուղղահայաց է այդ հարթությանը։

Պարզ ասած՝ PCA-ն կետերը պրոյեկտում է հարթության վրա ուղիղ անկյան տակ (ամենակարճ ճանապարհով), ինչն էլ ապահովում է նվազագույն հեռավորություն (սխալ) օրիգինալ կետի և հարթության վրա նրա պատկերի միջև։


 Այս հատվածը նվազագույն սխալի ձևակերպման (Minimum-error formulation) մաթեմատիկական եզրափակիչ փուլը, որտեղ աղավաղման $J$ ֆունկցիան վերջնականապես կապվում է կովարիացիոն մատրիցի հետ, և ցույց է տրվում դրա օպտիմալացումը Լագրանժի բազմապատկիչների մեթոդով։

1. Աղավաղման ֆունկցիայի վերջնական մատրիցային տեսքը ($J$)

Նախորդ քայլում մենք ստացանք շեղման (սխալի) վեկտորի տեսքը՝ $x_n - \widetilde{x}_n = \sum_{i=M+1}^{D} \{(x_n - \bar{\mathbf{x}})^T \mathbf{u}_i\} \mathbf{u}_i$:

Որպեսզի հաշվենք $J = \frac{1}{N} \sum_{n=1}^{N} \|x_n - \widetilde{x}_n\|^2$ աղավաղման չափանիշը, պետք է վերցնենք այս վեկտորի երկարության քառակուսին։ Օգտվելով այն փաստից, որ $\{\mathbf{u}_i\}$ վեկտորները օրթոնորմալ են ($\mathbf{u}_i^T \mathbf{u}_j = \delta_{ij}$), տարբեր ինդեքսներով անդամների արտադրյալները զրոյանում են, և կենդանի են մնում միայն նույնական ինդեքսներով քառակուսիները՝

$$\|x_n - \widetilde{x}_n\|^2 = \sum_{i=M+1}^{D} \{(x_n - \bar{\mathbf{x}})^T \mathbf{u}_i\}^2 = \sum_{i=M+1}^{D} \mathbf{u}_i^T (x_n - \bar{\mathbf{x}})(x_n - \bar{\mathbf{x}})^T \mathbf{u}_i$$

Այս արտահայտությունը տեղադրելով $J$-ի սահմանման մեջ և փոխելով գումարման նշանների հերթականությունը՝ ստանում ենք՝

$$J = \sum_{i=M+1}^{D} \mathbf{u}_i^T \left\{ \frac{1}{N} \sum_{n=1}^{N} (x_n - \bar{\mathbf{x}})(x_n - \bar{\mathbf{x}})^T \right\} \mathbf{u}_i$$

Քանի որ փակագծերի մեջ գտնվող արտահայտությունը հենց տվյալների $\mathbf{S}$ կովարիացիոն մատրիցն է, $J$-ն ընդունում է սլայդի առաջին տողի տեսքը՝

$$J = \sum_{i=M+1}^{D} \mathbf{u}_i^T \mathbf{S} \mathbf{u}_i$$

Չափողականության կրճատման ընթացքում առաջացած ընդհանուր սխալը հավասար է դեն նետված (անտեսված) բոլոր $D-M$ հատ առանցքների երկայնքով ստացված վարիացիաների գումարին։

2. Մինիմալացման խնդիրը և Սահմանափակումները

Մեր նպատակն է մինիմալացնել այս $J$ սխալը ընտրվող $\{\mathbf{u}_i\}$ վեկտորների նկատմամբ ($J \to \min$)։Եթե որևէ սահմանափակում չդնենք, ապա մինիմալացումը մեզ կտա տրիվիալ (անիմաստ) լուծում՝ $\mathbf{u}_i = \mathbf{0}$:Ուստի, ինչպես առավելագույն վարիացիայի դեպքում, այստեղ նույնպես գործում է օրթոնորմալության սահմանափակումը՝ $\mathbf{u}_i^T \mathbf{u}_i = 1$:

Ինտուիցիայի ձևավորում պարզ օրինակով ($D=2, M=1$)

Որպեսզի խնդրի formal (պաշտոնական) լուծումը հեշտ ընկալելի լինի,  դիտարկենք ամենապարզ երկչափ դեպքը, երբ $D=2$ (ունենք երկու հատկանիշ) և ցանկանում ենք անցնել $M=1$ (մեկչափանի ենթատարածության)։Այս դեպքում մենք ունենք ընդամենը երկու բազիսային վեկտոր՝ $\mathbf{u}_1$ և $\mathbf{u}_2$:

Քանի որ $M=1$, ապա առաջին $\mathbf{u}_1$ վեկտորը կազմում է մեր գլխավոր ենթատարածությունը (որը պահելու ենք), իսկ երկրորդ $\mathbf{u}_2$ վեկտորը այն առանցքն է, որը դեն ենք նետելու ($i = M+1 = 2$)։Հետևաբար, ըստ վերևի բանաձևի, սխալի ֆունկցիան բաղկացած է լինելու ընդամենը մեկ անդամից՝

$$J = \mathbf{u}_2^T \mathbf{S} \mathbf{u}_2$$

Որպեսզի մինիմալացնենք այս սխալը $\mathbf{u}_2^T \mathbf{u}_2 = 1$ պայմանի տակ, կազմում ենք Լագրանժի նոր ֆունկցիան ($\widetilde{J}$), որտեղ օգտագործվում է $\lambda_2$ բազմապատկիչը՝

$$\widetilde{J} = \mathbf{u}_2^T \mathbf{S} \mathbf{u}_2 + \lambda_2 (1 - \mathbf{u}_2^T \mathbf{u}_2)$$


Օպտիմալացման լուծումը երկչափ դեպքի համար ($D=2, M=1$)


Լագրանժի ֆունկցիան՝ անտեսվող (դեն նետվող) $\mathbf{u}_2$ առանցքի նկատմամբ`

$$\widetilde{J} = \mathbf{u}_2^T \mathbf{S} \mathbf{u}_2 + \lambda_2 (1 - \mathbf{u}_2^T \mathbf{u}_2)$$

Ստացիոնար կետի գտնումը:

Այս ֆունկցիայի մասնակի ածանցյալը ըստ $\mathbf{u}_2$ վեկտորի հավասարեցնելով զրոյի՝ ստանում ենք`

$$\mathbf{S}\mathbf{u}_2 = \lambda_2\mathbf{u}_2$$

Սա նշանակում է, որ $\mathbf{u}_2$-ը $\mathbf{S}$ կովարիացիոն մատրիցի սեփական վեկտորն է, իսկ $\lambda_2$-ը՝ դրան համապատասխանող սեփական արժեքը։

* Մինիմալ սխալի պայմանը: Տեղադրելով այս լուծումը սխալի ֆունկցիայի մեջ՝ տեսնում ենք, որ $J = \lambda_2$: Որպեսզի աղավաղման $J$ չափանիշը լինի նվազագույնը (minimum), մենք պետք է ընտրենք երկու սեփական արժեքներից ամենափոքրը ($\lambda_2$)։

* Հիմնական ենթատարածության ընտրությունը: Եթե դեն նետվող ուղղությունը ($\mathbf{u}_2$) համապատասխանում է փոքր սեփական արժեքին, ապա պահպանվող (գլխավոր) ենթատարածությունը ($\mathbf{u}_1$) ավտոմատ կերպով պետք է համընկնի ավելի մեծ սեփական արժեք ունեցող սեփական վեկտորի հետ։ Սա լիովին համապատասխանում է առավելագույն վարիացիայի ինտուիցիային։

Ստացված արդյունքի երկրաչափական իմաստը՝

* Պրոյեկցիայի միջին քառակուսային հեռավորությունը մինիմալացնելու համար գլխավոր բաղադրիչների ենթատարածությունը պետք է անցնի տվյալների միջին կետով ($\bar{\mathbf{x}}$):

* Այդ ենթատարածքի ուղղվածությունը պետք է համընկնի տվյալների առավելագույն վարիացիայի (ցրվածության) ուղղությունների հետ։

* Հատուկ դեպք (Հավասար սեփական արժեքներ). Եթե սեփական արժեքները հավասար են ($\lambda_1 = \lambda_2$), դա նշանակում է, որ տվյալները բոլոր ուղղություններով տարածված են հավասարաչափ (ունեն շրջանաձև տեսք)։ Այդ դեպքում գլխավոր առանցքի ցանկացած ընտրություն կտա սխալի միևնույն $J$ արժեքը, և չկա մեկ օպտիմալ ուղղություն։

Ընդհանուր լուծումը կամայական չափողականությունների համար

Սլայդի վերջին մասում բերվում է խնդրի վերջնական ընդհանրացումը կամայական սկզբնական չափողականության ($D$) և կամայական նոր չափողականության ($M < D$) համար։

Աղավաղման $J$ ֆունկցիայի ընդհանուր մինիմալացումը հանգեցնում է նրան, որ բոլոր բազիսային վեկտորները պետք է լինեն կովարիացիոն մատրիցի սեփական վեկտորներ։

$$\mathbf{S}\mathbf{u}_i = \lambda_i\mathbf{u}_i \quad (i = 1, \dots, D)$$

որտեղ $\{\mathbf{u}_i\}$ վեկտորները ընտրվում են օրթոնորմալ։

Որպեսզի սխալը լինի նվազագույնը, ընդհանուր կանոնն է՝

1․ Հաշվել $\mathbf{S}$ մատրիցի բոլոր $D$ հատ սեփական արժեքները և դրանք դասավորել նվազման կարգով՝ $\lambda_1 \ge \lambda_2 \ge \dots \ge \lambda_D$:

2․ Պահել առաջին $M$ հատ սեփական վեկտորները ($\mathbf{u}_1, \dots, \mathbf{u}_M$), որոնք համապատասխանում են ամենամեծ սեփական արժեքներին (սա կլինի մեր գլխավոր ենթատարածությունը)։

3․ Դեն նետել (անտեսել) մնացած $D-M$ հատ վեկտորները, որոնք համապատասխանում են ամենափոքր սեփական արժեքներին։ Այս դեպքում վերջնական մինիմալացված սխալը կլինի պարզապես անտեսված սեփական արժեքների գումարը՝

$$J_{\min} = \sum_{i=M+1}^{D} \lambda_i$$

Այսպիսով, մաթեմատիկորեն ապացուցվեց, որ երկու ձևակերպումներն էլ գործնականում հանգեցնում են ճիշտ նույն ալգորիթմին։


Ինչպես նախորդիվ հիմնավորել էինք, կամայական չափողականության դեպքում աղավաղման չափանիշի մինիմալ արժեքը ստացվում է, երբ անտեսվող ուղղությունները համապատասխանում են կովարիացիոն մատրիցի ամենափոքր սեփական արժեքներին։ Մաթեմատիկորեն այդ նվազագույն սխալը գրվում է հետևյալ տեսքով՝

$$J = \sum_{i=M+1}^{D} \lambda_i$$


Իմաստը. Չափողականությունը $D$-ից $M$-ի բերելու ընթացքում կորսված ինֆորմացիայի (սխալի) քանակը ուղղակիորեն հավասար է դեն նետված $D-M$ հատ սեփական արժեքների գումարին։

Սրանից հետևում է, որ գլխավոր ենթատարածությունը սահմանող վեկտորները հենց այն $M$ հատ սեփական վեկտորներն են, որոնք համապատասխանում են ամենամեծ սեփական արժեքներին։

Սահմանային դեպք. երբ $M = D$

Դիտարկենք այն դեպքը, երբ չափողականության կրճատում չի կատարվում, այսինքն՝ պահվող առանցքների քանակը հավասար է սկզբնական չափողականությանը ($M = D$)։Այս պարագայում սխալի գումարի տակ ոչ մի անդամ չի մնում, հետևաբար $J = 0$ (ինֆորմացիայի կորուստ չկա)։

Գործողության իմաստը. PCA-ն այս դեպքում չի նվազեցնում տվյալների չափողականությունը, այլ պարզապես կատարում է կոորդինատային առանցքների պտույտ (rotation)։ Նոր առանցքները հավասարեցվում են գլխավոր բաղադրիչների (մաքսիմալ վարիացիայի) ուղղությունների հետ, ինչի շնորհիվ նոր տարածության մեջ հատկանիշների միջև կովարիացիան դառնում է $0$ (տվյալները դեկորելացվում են)։




Օրինակ, որտեղ PCA-ն կիրառված է ձեռագիր «3» թվի պատկերների բազմության վրա։

Ի՞նչ են հուշում պատկերները՝

Միջին վեկտոր ($\bar{\mathbf{x}}$) [Mean]. Առաջին նկարում պատկերված է բոլոր տվյալների միջին պատկերը։ Քանի որ այն ստացվել է բազմաթիվ տարբեր ձեռագրերով գրված «3»-ների միջինացումից, այն ունի լվացված, մշուշոտ տեսք, որը ներկայացնում է «3» թվի ընդհանուր կմախքը։


Սեփական վեկտորներ ($\mathbf{u}_1, \dots, \mathbf{u}_4$) [Eigenimages]. Հաջորդ չորս պատկերները համապատասխանում են ամենամեծ սեփական արժեքներին ($\lambda_1, \dots, \lambda_4$)։ Մեքենայական ուսուցման մեջ սրանց հաճախ անվանում են «Eigendigits» կամ «Eigenimages»:


Ուշադրություն դարձնենք սեփական արժեքների չափերին՝ $\lambda_1 = 3.4 \cdot 10^5$, $\lambda_2 = 2.8 \cdot 10^5$, $\lambda_3 = 2.4 \cdot 10^5$, և $\lambda_4 = 1.6 \cdot 10^5$։ Դրանք նվազում են, ինչը նշանակում է, որ յուրաքանչյուր հաջորդ վեկտոր ավելի քիչ վարիացիա (տեղեկատվություն) է պարունակում, քան նախորդը։

Գույների իմաստը (Մանուշակագույն և դեղին). Այս պատկերների մանուշակագույն և դեղնավուն հատվածները ցույց են տալիս դրական և բացասական արժեքները (շեղումները միջինից)։ Սրանք ներկայացնում են «3» գրելու ամենակարևոր փոփոխականությունները (variations) մարդկանց ձեռագրերում․ օրինակ՝ թեքությունը, վերին կամ ստորին կեռիկի հաստությունը, թվի լայնությունը և այլն։

Ամփոփիչ իմաստ. Ցանկացած ձեռագիր «3» պատկեր այժմ կարելի է ստանալ (կամ մոտարկել)՝ վերցնելով ընդհանուր Միջին պատկերը (Mean) և դրան գումարելով կամ հանելով այս հիմնական սեփական պատկերները (Eigenimages)` որոշակի $z_{ni}$ գործակիցներով։ Սա թույլ է տալիս հարյուրավոր պիքսելներից կազմված պատկերը նկարագրել ընդամենը մի քանի թվով (գլխավոր բաղադրիչներով)։


Որպեսզի պատկերացնեք, թե ինչպես է տվյալների գիտության (Data Science) մեջ պատկերը վերածվում մաթեմատիկական օբյեկտի և ինչու է կիրառվում PCA-ն, դիտարկենք հետևյալ հասկացությունները։

1. Ինչպե՞ս է ստացվում $\mathbb{R}^{256}$ տարածությունը

Յուրաքանչյուր պատկեր ունի $16 \times 16$ չափս: Սա նշանակում է, որ պատկերը բաղկացած է երկչափ ցանցից (մատրիցից), որն ունի 16 տող և 16 սյունակ՝ ընդհանուր $16 \times 16 = 256$ պիքսել (pixel):

* Քանի որ պատկերը մոխրագույն երանգների է (grayscale), յուրաքանչյուր պիքսել ունի ընդամենը մեկ արժեք (սովորաբար 0-ից 255 միջակայքում, որտեղ 0-ն սևն է, իսկ 255-ը՝ սպիտակը, կամ նորմավորված՝ 0-ից 1 միջակայքում):

* Մեքենայական ուսուցման մեջ այս երկչափ $16 \times 16$ մատրիցը «փռում» են (flatten) և դարձնում մեկ երկար վեկտոր, որն ունի 256 կոորդինատ:

* Այսպիսով, ամեն մի ձեռագիր «3» դառնում է մի կետ (վեկտոր) 256-չափանի տարածության մեջ ($\mathbb{R}^{256}$): Մենք ունենք այդպիսի 658 կետ:


.Ի՞նչ է նշանակում «զգալի տատանումներ» (considerable variation)
Մարդիկ տարբեր կերպ են գրում «3» թիվը.

Մեկը գրում է թեք, մյուսը՝ ուղիղ (կողմնորոշման փոփոխություն):

Մեկը գրում է հաստ գրիչով, մյուսը՝ բարակ (հաստության փոփոխություն):

Մեկը վերևի կեռիկը մեծ է անում, մյուսը՝ փոքր (ոճի փոփոխություն):

Այս բոլոր տարբերությունները հանգեցնում են նրան, որ 256-չափանի տարածության մեջ այդ 658 կետերը դասավորված են որոշակի ձևով (ամպի նման): PCA-ի նպատակն է գտնել, թե որոնք են այն հիմնական ուղղությունները, որոնց երկայնքով տվյալները ամենաշատն են փոփոխվում:

Ինչո՞ւ է այստեղ պետք PCA-ն


256 չափումը շատ մեծ է մարդու տեսողական ընկալման և շատ ալգորիթմների արագագործության համար (այն կոչվում է «չափողականության անեծք»` curse of dimensionality):

PCA-ն (Գլխավոր բաղադրիչների մեթոդը) թույլ է տալիս.

Չափողականության նվազեցում (Dimensionality Reduction): Այն գտնում է նոր փոխուղղահայաց առանցքներ (գլխավոր բաղադրիչներ), որոնք դասավորված են ըստ տվյալների դիսպերսիայի (variance) նվազման:

Ինֆորմացիայի խտացում: Առաջին 2 կամ 3 գլխավոր բաղադրիչները հաճախ պահպանում են պատկերների միջև եղած տարբերությունների հիմնական մասը (օրինակ՝ թեքությունը և հաստությունը):

Վիզուալիզացիա: 256-չափանի տվյալները հնարավոր չէ գծագրել, բայց եթե PCA-ի միջոցով դրանք պրոյեկտենք առաջին 2 գլխավոր բաղադրիչների վրա, կստանանք սովորական երկչափ (2D) գրաֆիկ, որտեղ հստակ կերևա, թե ինչպես են գրելաոճերով նման «3»-երը խմբավորվում իրար մոտ:





Տրված է հետևյալ մոդելը՝

$$\hat{f}(\lambda) = \bar{x} + \lambda_1 v_1 + \lambda_2 v_2$$

Սա գծային կոմբինացիա է, որտեղ յուրաքանչյուր անդամ ունի իր հստակ իմաստը՝

* $\bar{x}$ (Միջին պատկեր / Mean image): Սա բոլոր 658 հատ «3» թվերի պիքսելային միջինն է: Եթե նայեք բանաձևի տակ գտնվող առաջին նկարին, կտեսնեք մի քիչ լողացող, բայց հստակ «3»: Դա ստացվել է բոլոր «3»-երը իրար վրա դնելով և միջինացնելով: Սա մեր «մեկնարկային կետն» է:

* $v_1$ և $v_2$ (Գլխավոր բաղադրիչներ / Principal Components): Սրանք այն սեփական վեկտորներն են (eigenvectors), որոնք ցույց են տալիս տվյալների փոփոխության հիմնական ուղղությունները: Նկարում դրանք պատկերված են որպես «ուրվականանման» մոխրագույն պատկերներ: Մուգ և բաց տեղամասերը ցույց են տալիս, թե որ պիքսելներն են փոխվում, երբ մենք շարժվում ենք այդ բաղադրիչի երկայնքով:

* $\lambda_1$ և $\lambda_2$ (Կշիռներ կամ կոորդինատներ / Weights): Սրանք սովորական թվեր են (սկալյարներ): Փոխելով այս թվերը՝ մենք որոշում ենք, թե ինչ համամասնությամբ (ինչ ուժգնությամբ) ավելացնել կամ պակասեցնել $v_1$ և $v_2$ հատկանիշները միջին $\bar{x}$ պատկերին:

$v_1$ (Հորիզոնական շարժում): Այս բաղադրիչը հիմնականում պատասխանատու է «3» թվի ներքևի պոչի երկարացման համար: Եթե մենք մեծացնենք $\lambda_1$-ի արժեքը, ապա ստացվող «3»-ի ներքևի մասը կձգվի դեպի ձախ/ներքև:

$v_2$ (Ուղղահայաց շարժում): Այս բաղադրիչը պատասխանատու է տառի հաստության (thickness) համար: $\lambda_2$-ի արժեքը փոխելով՝ մենք կարող ենք միջին «3»-ը դարձնել ավելի հաստ (թավ) կամ ավելի բարակ:


Ընդհանուր գաղափարը (The Big Picture)

Առանց PCA-ի, նոր «3» պատկեր ստեղծելու կամ եղածը նկարագրելու համար մեզ պետք էր փոփոխել բոլոր 256 պիքսելները առանձին-առանձին:

PCA-ի շնորհիվ մենք հասկացանք, որ «3» գրելու հիմնական տարբերությունները հանգում են ընդամենը երկու կարևոր գործոնի՝ պոչի երկարությանը ($v_1$) և տառի հաստությանը ($v_2$): 

Հետևաբար, 256-չափանի տարածության փոխարեն մենք կարող ենք աշխատել ընդամենը երկչափ տարածության մեջ՝ $(\lambda_1, \lambda_2)$ կոորդինատներով:

Ցանկացած ձեռագիր «3» կարելի է մեծ ճշտությամբ նկարագրել ընդամենը այս երկու թվով: Օրինակ՝$\lambda = (2, -1)$ կնշանակի «երկար պոչով, բայց բարակ գրված 3»:$\lambda = (-1, 3)$ կնշանակի «կարճ պոչով, բայց շատ հաստ գրված 3»:



Գրաֆիկի վրա պատկերված են կանաչ կետեր:Ամեն մի կանաչ կետ համապատասխանում է մեր ունեցած 658 ձեռագիր «3» թվերից մեկին:

Նախկինում այս կետերը գտնվում էին 256-չափանի բարդ տարածության մեջ, իսկ հիմա PCA-ի շնորհիվ դրանք պրոյեկտվել են երկչափ հարթության վրա՝

Հորիզոնական առանցք ($X$ առանցք): Առաջին գլխավոր բաղադրիչն է (First Principal Component, որը համապատասխանում է $\lambda_1$-ին):

Ուղղահայաց առանցք ($Y$ առանցք): Երկրորդ գլխավոր բաղադրիչն է (Second Principal Component, որը համապատասխանում է $\lambda_2$-ին):

Գրաֆիկի վրա կանոնավոր քայլերով ընտրվել են որոշակի կետեր և վերցվել են կարմիր շրջանակի մեջ (ընդհանուր $5 \times 5 = 25$ հատ կարմիր օղակ): Այդ օղակների դիրքերին համապատասխանող իրական «3»-ները բերված են աջ կողմի աղյուսակում:


Աջ կողմի աղյուսակը ($5 \times 5$ մատրիցը)

Այս աղյուսակը ցույց է տալիս, թե ինչպես է փոխվում «3» թվի տեսքը, երբ մենք գրաֆիկի վրա շարժվում ենք ձախից ձախ կամ ներքևից վերև: Սա հենց նախորդ կանխատեսումների ապացույցն է՝

Շարժում հորիզոնական առանցքով (Ձախից դեպի Աջ՝ $v_1$-ի փոփոխություն)


Նայենք աջ կողմի աղյուսակի տողերին ձախից ձախ.

* Ձախ ծայրում (բացասական արժեքներ $X$ առանցքի վրա): «3» թվերը գրված են շատ կարճ կամ ներս ընկած ստորին պոչով: Նրանք գրեթե ուղիղ կամ վերև ձգված տեսք ունեն:


* Աջ ծայրում (դրական արժեքներ $X$ առանցքի վրա): Շարժվելով դեպի աջ՝ «3»-ների ստորին պոչը սկսում է աստիճանաբար երկարել, կորանալ և ձգվել դեպի ձախ/ներքև:
