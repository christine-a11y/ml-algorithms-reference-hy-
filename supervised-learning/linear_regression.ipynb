Ունենք որ՝training տվյալների բազմություն, որը բաղկացած է $N$ դիտարկումներից՝ $\{x_n\}$, որտեղ $n = 1, \dots, N$, համապատասխան պիտակների (labels) հետ միասին՝ $\{y_n\}$:


Այսինքն՝ $x$ — Սա մեր մուտքային տվյալն է (օրինակ՝ տան քառակուսի մետրը, գտնվելու վայրը):

$y$ — Սա թիրախն է կամ պատասխանը (օրինակ՝ տան գինը):Մենք ունենք շատ օրինակներ ($N$ հատ), որտեղ արդեն գիտենք ճիշտ պատասխանները:

Մեր նպատակն է՝ Կանխատեսել $y$-ի արժեքը $x$-ի նոր արժեքի համար: 
(Մոդելին պետք չէ պարզապես հիշել հին տվյալները: Նպատակն այն է, որ երբ մենք տանք մի նոր $x$, որը մոդելը նախկինում չի տեսել, նա կարողանա ճիշտ գուշակել դրան համապատասխան $y$-ը:)

 Մենք փորձում ենք կառուցել մի ֆունկցիա՝ 
 
$\hat{y}(x)$, որը կանխատեսում է $y$-ը նոր մուտքային $x$-երի համար, կամ ավելի ընդհանուր՝ մոդելավորել կանխատեսող հավանականային բաշխումը՝ $p(y | x)$:


$\hat{y}(x)$: Սա մեր մոդելն է (օրինակ՝ գիծ կամ նեյրոնային ցանց): Մենք փնտրում ենք այնպիսի մաթեմատիկական կապ, որը մուտքին տալով $x$, ելքում կտա ճշմարտությանը մոտ $\hat{y}$:

$p(y | x)$: Սա ավելի խորը մոտեցում է: Մենք ոչ թե պարզապես ասում ենք «գինը կլինի 1000$», այլ հաշվում ենք հավանականությունը: Օրինակ՝ «Ինչքա՞ն է հավանականությունը, որ գինը կլինի $y$, եթե տունն ունի $x$ հատկանիշները»:



Մոդելի մաթեմատիկական տեսքը

Եթե ունենք մուտքային փոփոխականների վեկտոր $\mathbf{x} = (x_1, \dots, x_D)^T$, ապա գծային մոդելը սահմանվում է հետևյալ կերպ.

$$y(\mathbf{x}, \mathbf{w}) = w_0 + w_1 x_1 + w_2 x_2 + \dots + w_D x_D$$

Կամ ավելի սեղմ վեկտորական տեսքով.

$$y(\mathbf{x}, \mathbf{w}) = w_0 + \sum_{i=1}^{D} w_i x_i$$

$w_1, \dots, w_D$ (Parameters/Weights): Սրանք կշիռներն են, որոնք ցույց են տալիս, թե յուրաքանչյուր մուտքային փոփոխական ($x_i$) որքանով է ազդում վերջնական կանխատեսման վրա։$w_0$ 

(Bias/Intercept): Սա «շեղումն» է կամ ազատ անդամը։ Այն թույլ է տալիս մոդելին կանխատեսել որոշակի արժեք նույնիսկ այն դեպքում, երբ բոլոր մուտքային տվյալները ($x$) զրո են։

Գծայնություն (Linearity): Ինչպես նշեցիք, ամենակարևոր հատկությունն այն է, որ ֆունկցիան գծային է ըստ պարամետրերի ($w$): Սա նշանակում է, որ եթե մենք փոխենք $w$-ն, ելքային արժեքը կփոխվի ուղիղ համեմատական ձևով։

Residual
Մոդելը կառուցելիս մենք հազվադեպ ենք ստանում կատարյալ գիծ, որն անցնում է բոլոր կետերով:Մնացորդը ($y_i - \hat{y}$) իրական դիտարկված արժեքի ($y$) և մոդելի կողմից կանխատեսված արժեքի ($\hat{y}$) տարբերությունն է:

Գրաֆիկի վրա դրանք այն կարմիր ուղղահայաց գծերն են, որոնք միացնում են կապույտ կետերը սև գծին:
Մեր նպատակն է գտնել այնպիսի գիծ, որտեղ այս բոլոր կարմիր գծերի (սխալների) քառակուսիների գումարը կլինի նվազագույնը: 



Մաթեմատիկորեն դա ներկայացվում է այսպես.

$$\min_{\hat{\beta}_0, \hat{\beta}_1} \sum_{i=1}^{N} (y_i - \hat{\beta}_0 - \hat{\beta}_1 x_i)^2$$

 բերված գծային ֆունկցիան՝ $\hat{y} = \hat{\beta}_0 + \hat{\beta}_1 x$,
  որտեղ`

  $\hat{\beta}_0$ (Intercept): $y$-ի արժեքը, երբ $x=0$ է: Այն ցույց է տալիս, թե որտեղ է գիծը հատում ուղղահայաց առանցքը:

  $\hat{\beta}_1$ (Slope): Գծի թեքությունը, որը ցույց է տալիս $y$-ի փոփոխությունը $x$-ի փոփոխության նկատմամբ:

  $(y_i - \hat{\beta}_0 - \hat{\beta}_1 x_i)$: Սա Residual-ն է (Մնացորդը):$y_i$: Սա իրական արժեքն է մեր տվյալների բազմությունից (օրինակ՝ տան իրական գինը):
  
  $\hat{\beta}_0 + \hat{\beta}_1 x_i$: Սա մոդելի կանխատեսած արժեքն է ($\hat{y}_i$):Այսինքն՝ սա (Իրական - Կանխատեսված) տարբերությունն է, կամ այլ կերպ ասած՝ մոդելի սխալը:
  
  $(\dots)^2$: Ինչպես տեսնում եք նկարում, որոշ կետեր գծից վերև են (դրական սխալ), իսկ որոշները՝ գծից ներքև (բացասական սխալ):Եթե մենք ուղղակի գումարենք դրանք, $+0.5$-ը և $-0.5$-ը իրար կոչնչացնեն, ու կստացվի $0$ սխալ:Բայց մենք գիտենք, որ սխալ կա: Քառակուսի բարձրացնելով՝ բոլոր թվերը դառնում են դրական ($(-0.5)^2 = 0.25$ և $0.5^2 = 0.25$), և մենք ստանում ենք իրական պատկերը:
  Քառակուսի ֆունկցիան (պարաբոլը) հարթ է և ունի ընդամենը մեկ նվազագույն կետ (global minimum): Դա թույլ է տալիս ածանցյալի միջոցով շատ հեշտությամբ գտնել այն կետը, որտեղ սխալը նվազագույնն է:

  :$\sum_{i=1}^{N}$: Սա գումարման նշանն է: Այն ասում է, որ պետք է հաշվել սխալի քառակուսին բոլոր $N$ դիտարկումների համար և գումարել իրար:

   $\min_{\hat{\beta}_0, \hat{\beta}_1}$Սա ամբողջ գործողության նպատակն է: Մենք ասում ենք. «Գտի՛ր այնպիսի $\hat{\beta}_0$ (Intercept) և $\hat{\beta}_1$ (Slope) արժեքներ, որոնց դեպքում այս ամբողջ գումարը կլինի նվազագույնը»:

MSE (Mean Squared Error)
Եթե նախորդ բանաձևը ուղղակի սխալների գումարն էր, ապա Միջին քառակուսային սխալը (MSE) այդ սխալների միջինացված տարբերակն է։
Բանաձևը:

 $MSE = \frac{1}{N} \sum_{i=1}^{N} (y_i - \hat{y}_i)^2$։
 
 Այստեղ $\hat{y}_i$-ն մեր մոդելի կանխատեսումն է ($\hat{\beta}_0 + \hat{\beta}_1 x_i$)։
 
 $N$-ը դիտարկումների ընդհանուր քանակն է, որով բաժանելով մենք ստանում ենք միջին արժեքը։

 Առավելությունները:

Դիֆերենցելիություն: MSE-ի գրաֆիկը հարթ է, ինչը թույլ է տալիս հեշտությամբ հաշվել ածանցյալը և օգտագործել այն որպես Loss function (կորստի ֆունկցիա) մոդելն օպտիմալացնելու համար։

Թերությունները:

Զգայունություն արտաքին շեղումների (outliers) նկատմամբ: Եթե տվյալների մեջ կա մի կետ, որը շատ հեռու է ընդհանուր տրամաբանությունից, քառակուսի բարձրացնելու պատճառով MSE-ն այն «պատժում» է շատ խիստ, ինչը կարող է մոդելը շեղել ճիշտ ուղղուց։

Միավորները: MSE-ի արդյունքը ստացվում է քառակուսի միավորներով (օրինակ՝ դոլարի քառակուսի), ինչը միշտ չէ, որ հեշտ է մեկնաբանել։

1D Linear regression-ը ցույց է տալիս այն մաթեմատիկական քայլերը, որոնց միջոցով մենք գտնում ենք լավագույն գիծը:

Քայլ 1. Հաշվել մասնակի ածանցյալները (Partial Derivative

Մենք ունենք սխալի ֆունկցիան՝ 
$\sum_{i=1}^{N} (y_i - \hat{\beta}_0 - \hat{\beta}_1 x_i)^2$:
Քանի որ մենք ունենք երկու անհայտ փոփոխական ($\hat{\beta}_0$ և $\hat{\beta}_1$), մենք պետք է հասկանանք, թե ինչպես է սխալը փոխվում յուրաքանչյուրի փոփոխության հետ:Մասնակի ածանցյալը մեզ ցույց է տալիս այդ փոփոխության արագությունը յուրաքանչյուր պարամետրի համար առանձին:

Քայլ 2. Ածանցյալները հավասարեցնել 0-ի


Մաթեմատիկայից հայտնի է, որ ֆունկցիայի էքստրեումները գտնելու համար անհրաժոշտ է ածանցյալը հավասարեցնել  զրոյի:

Քայլ 3. Լուծել հավասարումները $\hat{\beta}_0$-ի և $\hat{\beta}_1$-ի համար

Այս քայլում մենք ստանում ենք երկու հավասարում երկու անհայտով: Դրանք լուծելով՝ մենք ստանում ենք այն կոնկրետ թվերը, որոնք պետք է դնենք մեր գծի բանաձևի մեջ.$\hat{\beta}_1$ (Slope): Ցույց կտա գծի թեքությունը:$\hat{\beta}_0$ (Intercept): Ցույց կտա, թե որտեղից է սկսվում գիծը:

Մենք ունենք կետեր (տվյալներ) և ուզում ենք գտնել լավագույն ուղիղ գիծը, որն անցնում է դրանց միջով:
Մոդելի բանաձևը:

 $\hat{y} = \hat{\beta}_0 + \hat{\beta}_1 x$:

 
 $\hat{\beta}_0$ (Intercept): Ազատ անդամը, որը ցույց է տալիս, թե որտեղ է գիծը հատում $y$ առանցքը ($x=0$ դեպքում):

 $\hat{\beta}_1$ (Slope): Թեքությունը, որը ցույց է տալիս, թե որքանով է փոխվում $y$-ը, երբ $x$-ը աճում է 1 միավորով:

 Որպեսզի հասկանանք՝ գիծը լավն է, թե վատ, մենք հաշվում ենք մնացորդները (Residuals)՝ $y_i - \hat{y}$ (տարբերությունը իրական կետի և գծի միջև):
 
 MSE (Mean Squared Error): Մենք բարձրացնում ենք այդ սխալները քառակուսի (որպեսզի բացասական թվերը չջնջեն դրականներին) և հաշվում ենք դրանց միջինը:
 
 Նպատակը: Գտնել այնպիսի $\hat{\beta}_0$ և $\hat{\beta}_1$, որոնց դեպքում սխալների քառակուսիների գումարը լինի նվազագույնը (min):


 Գտնենք էքստրեմումները՝
 1․ Ածանցում ըստ $\hat{\beta}_0$-ի:$$\frac{\partial}{\partial \hat{\beta}_0} \sum_{i=1}^{N} (y_i - \hat{\beta}_0 - \hat{\beta}_1 x_i)^2$$

 Առաջին հերթին հիշենք, որ գումարի ($\sum$) ածանցյալը հավասար է ածանցյալների գումարին։ Այսինքն՝ մենք կարող ենք ածանցման նշանը մտցնել գումարման նշանի ներսը և կիրառել այն յուրաքանչյուր $i$-րդ անդամի համար առանձին.
 
 $$\frac{\partial}{\partial \hat{\beta}_0} \sum_{i=1}^{N} (\dots)^2 = \sum_{i=1}^{N} \frac{\partial}{\partial \hat{\beta}_0} (\dots)^2$$
 
 
  Բարդ ֆունկցիայի ածանցում (Chain Rule)
  
  Հիմա դիտարկենք միայն ներսի մասը՝ 
  
  $(y_i - \hat{\beta}_0 - \hat{\beta}_1 x_i)^2$։ 
  
  Սա բարդ ֆունկցիա է, որտեղ արտաքին մասը քառակուսին է ($u^2$), իսկ ներքին մասը՝ փակագծերի մեջ եղած արտահայտությունը ($u$):Կանոնը՝
  
   $[u^2]' = 2 \cdot u \cdot u'$
   
   Արտաքին մասի ածանցում: 
   $2$-ը որպես գործակից գալիս է առաջ, իսկ փակագծի աստիճանը դառնում է $1$.
   
   $$2 \cdot (y_i - \hat{\beta}_0 - \hat{\beta}_1 x_i)^{2-1} = 2(y_i - \hat{\beta}_0 - \hat{\beta}_1 x_i)
   $$
   
   Ներքին մասի ածանցում ըստ $\hat{\beta}_0$-ի:
   
 Հիմա ածանցում ենք փակագծի ներսի պարունակությունը՝ $\frac{\partial}{\partial \hat{\beta}_0} (y_i - \hat{\beta}_0 - \hat{\beta}_1 x_i)$:
 
 $y_i$-ն հաստատուն է $\hat{\beta}_0$-ի նկատմամբ $\to$ ածանցյալը $0$ է։
 
 $-\hat{\beta}_1 x_i$-ն հաստատուն է $\hat{\beta}_0$-ի նկատմամբ $\to$ ածանցյալը $0$ է։
 
 $-\hat{\beta}_0$-ի ածանցյալն ըստ $\hat{\beta}_0$-ի հավասար է $-1$։
 
Այժմ բազմապատկում ենք արտաքին և ներքին մասերի ստացված արդյունքները.
$$2 \cdot (y_i - \hat{\beta}_0 - \hat{\beta}_1 x_i) \cdot (-1) = -2(y_i - \hat{\beta}_0 - \hat{\beta}_1 x_i)$$
Քանի որ մենք ունեինք $\sum$ նշանը, այս արդյունքը տեղադրում ենք դրա ներսում.

$$\sum_{i=1}^{N} -2(y_i - \hat{\beta}_0 - \hat{\beta}_1 x_i)$$

Քանի որ $-2$-ը հաստատուն թիվ է և չի կախված $i$-ից, այն կարող ենք հանել գումարման նշանի դիմաց:

$$-2 \sum_{i=1}^{N} (y_i - \hat{\beta}_0 - \hat{\beta}_1 x_i)$$

Որպեսզի բանաձևն ավելի գեղեցիկ լինի, փակագծերի մեջ կարող ենք խմբավորել մեր մոդելի կանխատեսումը 

($\hat{y}_i$).$$-2 \sum_{i=1}^{N} (y_i - (\hat{\beta}_0 + \hat{\beta}_1 x_i))$$

2․Ածանցում ըստ $\hat{\beta}_1$-ի
Ինչպես նախորդ դեպքում, ածանցյալը մտնում է գումարման նշանի ($\sum$) ներսը և կիրառվում է յուրաքանչյուր անդամի վրա:

$$\sum_{i=1}^{N} \frac{\partial}{\partial \hat{\beta}_1} (y_i - \hat{\beta}_0 - \hat{\beta}_1 x_i)^2$$


Բարդ ֆունկցիայի ածանցում (Chain Rule)Կիրառում ենք նույն ՝
$[u^2]' = 2 \cdot u \cdot u'$ կանոնը

Արտաքին մասի ածանցում: $2$-ը իջնում է առաջ որպես գործակից, իսկ փակագծի աստիճանը դառնում է $1$:

$$2 \cdot (y_i - \hat{\beta}_0 - \hat{\beta}_1 x_i)$$

Ներքին մասի ածանցում ըստ $\hat{\beta}_1$-ի:

 Հիմա ածանցում ենք փակագծի ներսը՝ $\frac{\partial}{\partial \hat{\beta}_1} (y_i - \hat{\beta}_0 - \hat{\beta}_1 x_i)$:
 
 $y_i$ և $\hat{\beta}_0$ անդամները հաստատուն են $\hat{\beta}_1$-ի նկատմամբ, ուստի դրանց ածանցյալը $0$ է:
 
 $-\hat{\beta}_1 x_i$ անդամի ածանցյալն ըստ $\hat{\beta}_1$-ի հավասար է $-x_i$: (Քանի որ $\hat{\beta}_1$-ը փոփոխականն է, իսկ նրան բազմապատկված $-x_i$-ը՝ գործակիցը):
Այժմ իրար ենք բազմապատկում արտաքին և ներքին մասերից ստացված արդյունքները.
$$2 \cdot (y_i - \hat{\beta}_0 - \hat{\beta}_1 x_i) \cdot (-x_i)$$
Սա վերադասավորելով ստանում ենք.
$$-2 x_i (y_i - \hat{\beta}_0 - \hat{\beta}_1 x_i)$$

Տեղադրում ենք սա գումարման նշանի տակ և $-2$-ը հանում ենք գումարման նշանից դուրս:

$$-2 \sum_{i=1}^{N} x_i (y_i - \hat{\beta}_0 - \hat{\beta}_1 x_i)$$

Կամ, ինչպես ձեր նկարում է, խմբավորելով մոդելի մասը՝


($\hat{\beta}_0 + \hat{\beta}_1 x_i$):$$-2 \sum_{i=1}^{N} x_i (y_i - (\hat{\beta}_0 + \hat{\beta}_1 x_i))$$

Այժմ ածանցյալները հավասարեցնենք 0-ի և գտնենք էքստրեմումները։
Լավագույն $\hat{\beta}_0$ և $\hat{\beta}_1$ արժեքները գտնելու համար մենք վերցնում ենք $\hat{\beta}_0$-ի նկատմամբ ստացված մասնակի ածանցյալը և հավասարեցնում $0$-ի:

$$-2 \sum_{i=1}^{N} (y_i - (\hat{\beta}_0 + \hat{\beta}_1 x_i)) = 0$$

Բաժանելով երկու կողմը $-2$-ի վրա, մնում է միայն գումարը:$$\sum_{i=1}^{N} (y_i - \hat{\beta}_0 - \hat{\beta}_1 x_i) = 0$$
Հավասարման ձախ մասը կարող ենք բաժանել երեք առանձին գումարների:

$$\sum_{i=1}^{N} y_i - \sum_{i=1}^{N} \hat{\beta}_0 - \sum_{i=1}^{N} \hat{\beta}_1 x_i = 0$$

քանի որ $\hat{\beta}_0$-ն հաստատուն թիվ է և չի փոխվում ըստ $i$-ի, ապա նրան $N$ անգամ գումարելը նույնն է, ինչ $N \cdot \hat{\beta}_0$:

$$\sum_{i=1}^{N} y_i - N\hat{\beta}_0 - \hat{\beta}_1 \sum_{i=1}^{N} x_i = 0$$

Այժմ տեղափոխենք $N\hat{\beta}_0$-ն հավասարման աջ կողմ:

$$N\hat{\beta}_0 = \sum_{i=1}^{N} y_i - \hat{\beta}_1 \sum_{i=1}^{N} x_i$$

Եվ վերջապես, բաժանելով երկու կողմը $N$-ի վրա, ստանում ենք  վերջնական բանաձևը:

$$\hat{\beta}_0 = \frac{\sum_{i=1}^{N} y_i - \hat{\beta}_1 \sum_{i=1}^{N} x_i}{N}$$



Նույնը կատարում ենք  $\hat{\beta}_1$-ի հմար։

$$-2 \sum_{i=1}^{N} x_i (y_i - (\hat{\beta}_0 + \hat{\beta}_1 x_i)) = 0$$

Բաժանելով երկու կողմը $-2$-ի վրա, ստանում ենք։

$$\sum_{i=1}^{N} x_i (y_i - \hat{\beta}_0 - \hat{\beta}_1 x_i) = 0$$

Բազմապատկում ենք փակագծի ներսի յուրաքանչյուր անդամը $x_i$-ով։

$$\sum_{i=1}^{N} (x_i y_i - \hat{\beta}_0 x_i - \hat{\beta}_1 x_i^2) = 0$$

Այնուհետև գումարման նշանը տարածում ենք յուրաքանչյուր անդամի վրա առանձին։

$$\sum_{i=1}^{N} x_i y_i - \hat{\beta}_0 \sum_{i=1}^{N} x_i - \hat{\beta}_1 \sum_{i=1}^{N} x_i^2 = 0$$

Այս փուլում մենք արդեն գիտենք $\hat{\beta}_0$-ի բանաձևը, որը նախկինում հաշվարկել էինք։

$$\hat{\beta}_0 = \frac{\sum y_i - \hat{\beta}_1 \sum x_i}{N}$$

$\hat{\beta}_1$-ի նկատմամբ մասնակի ածանցյալը զրոյի հավասարեցնելու արդյունքում ստանում ենք ՝
$$\sum_{i=1}^{N} x_iy_i - \hat{\beta}_0 \sum_{i=1}^{N} x_i - \hat{\beta}_1 \sum_{i=1}^{N} x_i^2 = 0$$

Որպեսզի հավասարման մեջ մնա միայն մեկ անհայտ ($\hat{\beta}_1$), մենք տեղադրում ենք $\hat{\beta}_0$-ի արդեն մեզ հայտնի բանաձևը՝ $$\hat{\beta}_0 = \frac{\sum_{i=1}^{N} y_i - \hat{\beta}_1 \sum_{i=1}^{N} x_i}{N}$$

Տեղադրելով սա հիմնական հավասարման մեջ, ստանում ենք.

$$\sum_{i=1}^{N} x_iy_i - \left( \frac{\sum y_i - \hat{\beta}_1 \sum x_i}{N} \right) \cdot \sum x_i - \hat{\beta}_1 \sum x_i^2 = 0$$

Բազմապատկում ենք կոտորակի համարիչը $\sum x_i$-ով և առանձնացնում անդամները:

$$\sum_{i=1}^{N} x_iy_i - \frac{1}{N} \sum y_i \sum x_i + \frac{\hat{\beta}_1}{N} (\sum x_i)^2 - \hat{\beta}_1 \sum x_i^2 = 0$$

Այժմ հավասարման ձախ կողմում թողնում ենք $\hat{\beta}_1$ չունեցող անդամները, իսկ $\hat{\beta}_1$ ունեցողները տեղափոխում ենք աջ կողմ (նշանները փոխելով):

$$\sum_{i=1}^{N} x_iy_i - \frac{1}{N} \sum y_i \sum x_i = \hat{\beta}_1 \sum x_i^2 - \frac{\hat{\beta}_1}{N} (\sum x_i)^2$$

Աջ կողմում $\hat{\beta}_1$-ը հանում ենք փակագծերից դուրս:

$$\sum_{i=1}^{N} x_iy_i - \frac{1}{N} \sum y_i \sum x_i = \hat{\beta}_1 \left( \sum x_i^2 - \frac{1}{N} (\sum x_i)^2 \right)$$

Վերջնական քայլով բաժանում ենք ձախ մասը փակագծի պարունակության վրա բանաձևը:

$$\hat{\beta}_1 = \frac{\sum_{i=1}^{N} x_iy_i - \frac{1}{N} \sum_{i=1}^{N} y_i \sum_{i=1}^{N} x_i}{\sum_{i=1}^{N} x_i^2 - \frac{1}{N} (\sum_{i=1}^{N} x_i)^2}$$





Multivariate Linear Regression

Մենք օգտագործեցինք MSE-ն որպես կորստի ֆունկցիա, քանի որ այն դիֆերենցելի է (հնարավոր է հաշվել ածանցյալը)։ Կատարելով մասնակի ածանցում և դրանք հավասարեցնելով զրոյի՝ մենք ստացանք լավագույն պարամետրերը:

Հատման կետ ($\hat{\beta}_0$): Ստացվեց՝

 $\hat{\beta}_0 = \frac{\sum_{i=1}^{N} y_i - \hat{\beta}_1 \sum_{i=1}^{N} x_i}{N}$։
 
 Թեքություն ($\hat{\beta}_1$): Բարդ ձևափոխություններից հետո ստացվեց 
 
 $\hat{\beta}_1 = \frac{\sum_{i=1}^{N} x_iy_i - \frac{1}{N} \sum_{i=1}^{N} y_i \sum_{i=1}^{N} x_i}{\sum_{i=1}^{N} x_i^2 - \frac{1}{N} (\sum_{i=1}^{N} x_i)^2}$։


 Multivariate Linear Regression
 Իրական կյանքում $y$-ը սովորաբար կախված է ոչ թե մեկ $x$-ից, այլ մի քանի գործոններից (features)։
 
 Տվյալները: Մենք ունենք $n$ տվյալների կետ, բայց յուրաքանչյուր կետ նկարագրվում է $k$ հատկանիշներով ($k$ dimensions)։Օրինակ՝ Տան գինը ($y$) կախված է ոչ միայն մակերեսից ($x_1$), այլև սենյակների քանակից ($x_2$), կենտրոնից հեռավորությունից ($x_3$) և այլն։
 
 Մոդելը: Այս դեպքում մենք այլևս չունենք ուղիղ գիծ 2D հարթության վրա։ Մենք ունենք հիպերհարթություն (hyperplane) բազմաչափ տարածության մեջ։

 Յուրաքանչյուր $i$-րդ դիտարկման համար բանաձևն ունի հետևյալ տեսքը`
 
 $$y_i = \beta_0 + \beta_1 x_{i1} + \beta_2 x_{i2} + \dots + \beta_k x_{ik} + u_i$$
 
 $y_i$: Իրական արժեքը։
 $\beta_0$: Ազատ անդամը (Intercept)։
 $\beta_1, \dots, \beta_k$: Գործակիցները, որոնք ցույց են տալիս յուրաքանչյուր հատկանիշի ազդեցության չափը։
$u_i$: Սխալի անդամը (Error term), որը ցույց է տալիս, թե որքան հեռու է իրական $y$-ը ռեգրեսիայի գծից։



Համակարգչային հաշվարկները հեշտացնելու համար այս ամբողջը ներկայացվում է մատրիցների տեսքով՝ 

$Y = X\beta + u$:


$Y$ ($n \times 1$ մատրից): Բոլոր իրական պատասխանների սյունակը։


$X$ ($n \times (k+1)$ մատրից): Մուտքային տվյալների մատրիցը։ Առաջին սյունակը միշտ $1$-եր են՝ $\beta_0$ գործակցի համար։


$\beta$ ($(k+1) \times 1$ մատրից): Բոլոր այն գործակիցների (կշիռների) սյունակը, որոնք մենք պետք է գտնենք։


$u$ ($n \times 1$ մատրից): Յուրաքանչյուր կետի համար հաշվարկված սխալների սյունակը ($u = Y - X\beta$)։


Բազմաչափ  տվյալների դեպքում մենք նույնպես ցանկանում ենք նվազագույնի հասցնել սխալների քառակուսիների գումարը, բայց օգտագործում ենք մատրիցային վեկտորներ։

 Յուրաքանչյուր կետի համար հաշվարկված սխալների սյունակը՝ $u = Y - X\beta$


 Սխալների քառակուսիների գումարը ($u'u$): Սա համարժեք է՝
  $\sum (y_i - \hat{y}_i)^2$ արտահայտությանը։
  
  $$\min_{\beta} u'u = (Y - X\beta)'(Y - X\beta)$$

  Օգտագործելով մատրիցների տրանսպոնացման (transpose) հատկությունները՝ բացում ենք փակագծերը՝
  
  $(Y - X\beta)' = Y' - \beta'X'$։

  Բազմապատկելով՝ ստանում ենք չորս անդամ՝
  
  $$u'u = Y'Y - \beta'X'Y - Y'X\beta + \beta'X'X\beta$$

Դիտարկենք՝ $Y'X\beta$ արտահայտության չափսերը (dimensions)։
Հիշենք մեր մատրիցների չափսերը ՝$Y$-ը $n \times 1$ է $\implies$ $Y'$ (տրանսպոնացված) կլինի $1 \times n$։

$X$-ը $n \times (k+1)$ է։

$\beta$-ն $(k+1) \times 1$ է։
Այժմ հերթով բազմապատկենք՝ $(Y' \cdot X)$: $(1 \times n) \cdot (n \times (k+1))$ բազմապատկումից ստացվում է $1 \times (k+1)$ չափսի վեկտոր։

$(Y'X \cdot \beta)$: $(1 \times (k+1)) \cdot ((k+1) \times 1)$ բազմապատկումից ստացվում է $1 \times 1$։
$1 \times 1$ մատրիցը հենց սկալյարն է՝ թիվ։

Ըստ ստացվածի շարունակելով դիտարկել մեր  $$u'u = Y'Y - \beta'X'Y - Y'X\beta + \beta'X'X\beta$$ արտահայտությունը կարող ենք պարզեցնել՝  քանի որ $Y'X\beta$ արտահայտությունը թիվ է,  ապա այն հավասար է իր տրանսպոնացված տարբերակին ($Y'X\beta = \beta'X'Y$)։
Սա թույլ է տալիս միավորել երկու մեջտեղի անդամները՝

$$u'u = Y'Y - 2\beta'X'Y + \beta'X'X\beta$$

Ունենալով պարզեցված սխալի ֆունկցիան՝ $u'u = Y'Y - 2\beta'X'Y + \beta'X'X\beta$, մենք հաշվում ենք դրա ածանցյալն ըստ $\beta$ վեկտորի՝

$$\frac{\partial (u'u)}{\partial \beta} = -2X'Y + 2X'X\beta$$

$Y'Y$-ի ածանցյալը $0$ է, քանի որ այն չի պարունակում $\beta$:

$-2\beta'X'Y$-ի ածանցյալը ստացվում է $-2X'Y$:

$\beta'X'X\beta$-ի ածանցյալը ստացվում է $2X'X\beta$ (նման է $ax^2$-ի ածանցյալին՝ $2ax$):

էքստրեմումները գտնելու համար ածանցյալը հավասարեցնում ենք $0$-ի՝

$$-2X'Y + 2X'X\beta = 0$$

Տեղափոխելով անդամները՝ ստանում ենէ $$X'X\beta = X'Y$$
Հավասարման երկու կողմը ձախից բազմապատկում ենք $(X'X)$ մատրիցի հակադարձ մատրիցով՝ $(X'X)^{-1}$՝

$$(X'X)^{-1}(X'X)\beta = (X'X)^{-1}X'Y$$

Քանի որ $(X'X)^{-1}(X'X)$ հացասար է միավոր մատրիցի ($I$), մենք ստանում ենք ձ բանաձևը՝

**$$\beta = (X'X)^{-1}X'Y$$**

**$$\beta = (X'X)^{-1}X'Y$$** սա էքստրեմումի կետ է։ Բայց սա դեռ չի նշանակում, որ դա նվազագույն է, կարող էր լինել նաև մաքսիմումի կետ։ Որպեսզի գտնենք մաքսիմումի թե մինիմումի կետ է անհրաժեշտ է գտնել երկրորդ կարգի ածանցյաը։
Մի քանի փոփոխականի դեպքւմ երկրորդ կարգի ածանցյալը Հեսսյան մատրիցն է՝

Գծային ռեգրեսիայի դեպքում մեր սխալի ֆունկցիայի ($J$) առաջին կարգի ածանցյալը (գրադիենտը) ըստ $\beta$-ի ստացվել էր՝
 $-2X'Y + 2X'X\beta$:
 
 Եթե մենք նորից ածանցենք այս արտահայտությունը ըստ $\beta$-ի, ապա $-2X'Y$ մասը կդառնա $0$ (քանի որ այն հաստատուն է $\beta$-ի նկատմամբ):Մնում է $2X'X\beta$ մասի ածանցյալը, որը պարզապես $2X'X$ է:Եթե հաշվի առնենք նաև տվյալների քանակը ($N$), ստանում ենք 
 
 $\nabla^2 J = \frac{2}{N} X'X$:


Որպեսզի ֆունկցիան ունենա մինիմում ՝ նրա երկրորդ ածանցյալը պետք է լինի մեծ կամ հավասար $0$-ից: Մատրիցների լեզվով սա նշանակում է, որ ցանկացած $z$ վեկտորի համար պետք է տեղի ունենա հետևյալը՝

$$z'(X'X)z \ge 0$$

Ապացույցը՝


Մատրիցների հատկության համաձայն՝ $z'(X'X)z$ կարող ենք գրել որպես $(Xz)'(Xz)$:Սա վեկտորի սկալյար արտադրյալն է ինքն իր հետ, ինչը հավասար է այդ վեկտորի երկարության քառակուսուն՝ $||Xz||^2$:Քանի որ ցանկացած իրական վեկտորի երկարության քառակուսին չի կարող լինել բացասական, ապա $||Xz||^2 \ge 0$ միշտ ճիշտ է:Սա նշանակում է, որ մեր ֆունկցիան երբեք չի կարող ունենալ մաքսիմում:

Եթե $X$ մատրիցի սյուները (մեր հատկանիշները) գծայնորեն անկախ են, դա նշանակում է, որ $Xz$ վեկտորը կարող է լինել $0$ միայն այն դեպքում, եթե $z$ վեկտորն է $0$:Այս դեպքում $||Xz||^2 > 0$ բոլոր $z \neq 0$ համար:Մատրիցը դառնում է որոշյալ դրական, ինչը երաշխավորում է, որ գոյություն ունի միայն մեկ և միակ կետ, որտեղ սխալը նվազագույնն է:




"""------------իսկ  ինչիցա ենթադրվում որ գլոբալ մինիմումնա ոչ թե լոկալ---------"""


Մենք փնտրում ենք այնպիսի կշիռներ ($w$), որոնց դեպքում կանխատեսված արժեքների ($\hat{y}$) և իրական արժեքների ($y$) միջև տարբերությունը (սխալը) լինի նվազագույնը։
Սահմանենք՝

 $\hat{y} = Xw$


 ($\hat{y}$-ը $X$-ի սյունային տարածությունից է։ Այսինքն $y$-ների  կանխատեսված արժեքը՝$\hat{y}$ տրված $X$-երի մատրիցի սյուներից ինչ-որ վեկտոր է՝ $\hat{y} \in \text{Col}(X)$։ Այսինքն խնդիրը երկրաչափական տեսանկյունից գտնել $X$-ի սյուների տարածության մեջ այն կետը որը ամենամոտնա $y$)

Multivariate Linear Regression-ի երկրաչափական մեկնաբանության ավելի մանրամասն բացատրությունը՝


Գծային ռեգրեսիայի հիմքում ընկած է $||y - Xw||^2$ արժեքի նվազեցումը։$y$-ը մեր իրական թիրախային արժեքների վեկտորն է (այն, ինչ մենք ուզում ենք կանխատեսել)։$Xw$-ն մեր մոդելի կանխատեսումն է ($\hat{y}$), որտեղ $X$-ը տվյալների մատրիցն է, իսկ $w$-ն՝ կշիռները։

Երկրաչափորեն սա նշանակում է, որ մենք փնտրում ենք այնպիսի կանխատեսում, որը ամենամոտն է իրական $y$ վեկտորին։



Կարևոր է հասկանալ, թե որտեղ կարող է գտնվել մեր կանխատեսումը ($\hat{y}$)։

Քանի որ $\hat{y} = Xw$, այն հանդիսանում է $X$ մատրիցի սյունակների գծային համակցություն։ Սա նշանակում է, որ $\hat{y}$-ը միշտ գտնվում է $X$-ի սյունային տարածության մեջ ($Col(X)$)։Պատկերացրեք $Col(X)$-ը որպես մի հարթություն բազմաչափ տարածության մեջ։

(X-ի սյունային տարածությունը `  

Մատրիցը ($X$) բաղկացած է սյունակներից (օրինակ՝ $X_1, X_2$), որոնք մեր մոդելի փոփոխականներն են (features)։ Յուրաքանչյուր սյունակ կարելի է պատկերացնել որպես մի վեկտոր (սլաք) բազմաչափ տարածության մեջ։

Երբ մենք գրում ենք $\hat{y} = Xw$, դա նշանակում է, որ մենք վերցնում ենք այդ սյունակ-վեկտորները և բազմապատկում ինչ-որ թվերով (կշիռներով)՝ $\hat{y} = w_1 X_1 + w_2 X_2$։Պատկերացրեք, որ մենք ունեք երկու սլաք հատակի վրա։ Փոխելով $w_1$ և $w_2$ թվերը, դուք կարող եք ստանալ այդ հատակի վրա գտնվող ցանկացած կետ։
$Col(X)$-ը այն ամբողջ «տարածքն» է, որը կարելի է «ծածկել»՝ օգտագործելով $X$-ի սյունակների բոլոր հնարավոր համակցությունները։Եթե ունեք երկու սյունակ, դրանք սովորաբար ձևավորում են մի հարթություն։Այդ հարթությունը հենց $X$-ի սյունային տարածությունն է։ Մեր կանխատեսումը ($\hat{y}$) պարտադիր պետք է լինի այդ հարթության վրա, քանի որ այն ստացվում է $X$ սյունակների միջոցով։

Ռեգրեսիայի համար սա կարըոր էքանի որ՝ Խնդիրն այն է, որ իրական $y$ վեկտորը (մեր թիրախը) հաճախ դուրս է այդ հարթությունից (տես նկարում $y$ սլաքը, որը բարձրանում է վերև)։Քանի որ մենք չենք կարող «դուրս գալ» հարթությունից, մենք փնտրում ենք այդ հարթության վրայի այն կետը ($\hat{y}$), որն ամենամոտն է $y$-ին։ Ամենամոտ կետը գտնելու համար մենք $y$-ից ուղղահայաց գիծ ենք իջեցնում դեպի հարթություն։ Դա կոչվում է օրթոգոնալ պրոյեկցիա։

 օրինակ՝ Պատկերացնենք՝ սենյակի հատակը $X$-ի սյունային տարածությունն է ($Col(X)$)։ Օդում կախված է մի լամպ (սա մեր իրական $y$-ն է)։ Գծային ռեգրեսիան այդ լամպի ստվերն է հատակի վրա։ Ստվերը միշտ հատակի ($Col(X)$) վրա է, և այն լամպին ամենամոտ կետն է հատակի վրա։)

Քանի որ իրական $y$ վեկտորը սովորաբար չի գտնվում այդ հարթության վրա (քանի որ մոդելը կատարյալ չէ), մենք պետք է գտնենք այդ հարթության այն կետը, որն ամենամոտն է $y$-ին։Երկրաչափությունից հայտնի է, որ կետից հարթություն ամենակարճ հեռավորությունը ուղղահայացն է։ Հետևաբար, գծային ռեգրեսիան ոչ այլ ինչ է, քան $y$ վեկտորի օրթոգոնալ պրոյեկցիան $Col(X)$ հարթության վրա։

Մնացորդային վեկտոր (Residual Vector)։


Մնացորդային վեկտորը սահմանվում է որպես սխալանք.$$r = y - Xw$$Եթե մեր կանխատեսումը լավագույնն է, ապա մնացորդային վեկտորը ($r$) պետք է լինի ուղղահայաց (perpendicular) $X$-ի բոլոր սյունակներին։Մաթեմատիկորեն դա գրվում է այսպես՝

$$X^T r = 0$$

Տեղադրելով $r$-ի արժեքը, ստանում ենք Նորմալ հավասարումը՝

$$X^T (y - Xw) = 0$$

$$X^T X w = X^T y$$

Սա այն հիմնական բանաձևն է, որով հաշվարկվում են օպտիմալ $w$ կշիռները



Ի՞նչ անել, եթե $X^T X$ մատրիցը հակադարձելի չէ

Գծային ռեգրեսիայի լավագույն լուծումը գտնելու համար մենք օգտագործում ենք Նորմալ հավասարումը՝

$$(X^T X)w = X^T y$$

Որպեսզի գտնենք $w$ կշիռները, մենք պետք է երկու կողմը բազմապատկենք $(X^T X)^{-1}$ հակադարձ մատրիցով։ Սակայն երբեմն դա հնարավոր չէ։



$X^T X$-ը կոչվում է սինգուլյար (singular), այսինքն՝ չունի հակադարձ մատրից, եթե $X$ մատրիցի ռանգը (rank) լրիվ չէ

 $rank(X) < D$ (Երկրաչափորեն դա նշանակում է, որ այդ երկու սյունակները չեն ավելացնում նոր չափողականություն կամ տարածություն. նրանք իրար վրա են համընկնում։ )
 
  Սա տեղի է ունենում հետևյալ դեպքերում․
  1․Գծային կախվածություն։ Որոշ հատկանիշներ (features) մյուսների գծային համակցություններն են։
  
  Օրինակ՝ եթե ունեք «քաշը կիլոգրամով» և «քաշը ֆունտով» սյունակները, դրանք նույն տեղեկությունն են պարունակում։
  
  2․Նույնական սյունակներ։ Տվյալների մեջ կան կրկնվող փոփոխականներ։$D > N$. Երբ փոփոխականների (հատկանիշների) քանակը ($D$) ավելի մեծ է, քան տվյալների (տողերի) քանակը։ ($N$)
  
  3․Perfect multicollinearity։ Սա այն դեպքն է, երբ փոփոխականների միջև կա ուժեղ գծային կապ։
  
  Երկրաչափական մեկնաբանություն
  
  Պատկերացնենք երկրաչափորեն՝ եթե երկու փոփոխականներ նույնական են, նրանք չեն ձևավորում երկչափ հարթություն, այլ ընդամենը մի գիծ։ Այդ դեպքում $y$ վեկտորը պրոյեկտելու համար մենք չունենք լիարժեք «հարթություն», և լուծումը դառնում է անորոշ կամ անհնարին ստանդարտ մեթոդով։


  Չնայած մատրիցի հակադարձելիության բացակայությանը, նվազագույն արժեքը (minimum) չի անհետանում։ Գծային ռեգրեսիայի կորուստի ֆունկցիան շարունակում է մնալ ուռուցիկ (convex)։Երկրաչափորեն սա նշանակում է, որ ֆունկցիան դեռևս ունի «գավաթի» տեսք և գլոբալ մինիմումը դեռ գոյություն ունի։ Մենք դեռ կարող ենք գտնել $y$ վեկտորի օրթոգոնալ պրոյեկցիան $X$-ի սյունային տարածության վրա, որը կապահովի նվազագույն սխալանքը։

  Այս իրավիճակի հիմնական առանձնահատկությունն այն է, որ լուծումը դադարում է լինել միակը (not unique)։ Սա բացատրվում է $X$ մատրիցի զրոյական տարածությամբ ($Null(X)$).
  Մատրիցի զրոյական տարածությունը բոլոր այն $z$ վեկտորների բազմությունն է, որոնք $X$ մատրիցով բազմապատկվելիս «ոչնչանում» են՝ վերածվելով զրոյական վեկտորի։

  $$Xz = 0$$
  
  Եթե սա պատկերացնենք որպես ֆունկցիա, ապա $Null(X)$-ը այն «սև խոռոչն» է, որտեղ ընկնելով՝ ցանկացած վեկտոր դառնում է անտեսանելի մոդելի համար։

  Սովորաբար, եթե մեր սյունակները ($X_1, X_2, \dots, X_D$) անկախ են, ապա միակ վեկտորը, որը բավարարում է $Xz=0$ պայմանին, զրոյական վեկտորն է ($z=0$): 

  Օրինակ՝  Պատկերացրեք՝ $X_2 = 2X_1$ (օրինակ՝ մետրը և սանտիմետրը)։ Մենք կարող ենք կազմել մի $z$ վեկտոր, որը $z = [2, -1]^T$:Այդ դեպքում՝
  
  $$Xz = X_1(2) + X_2(-1) = 2X_1 - 2X_1 = 0$$
  
  Այս $z$ վեկտորը գտնվում է զրոյական տարածության մեջ։


Եթե գոյություն ունի մի $z$ վեկտոր, որը պատկանում է $Null(X)$-ին ($Xz=0$), ապա մենք կարող ենք այն գումարել մեր օպտիմալ $w$ կշիռներին։Արդյունքում ստացված նոր կանխատեսումը կլինի նույնը՝

$$X(w+z) = Xw + Xz = Xw + 0 = Xw$$

Քանի որ կանխատեսումը չի փոխվում, սխալանքի քառակուսիների գումարը ևս մնում է նույնը՝

$$||y - X(w+z)||^2 = ||y - Xw||^2$$






Սովորաբար մենք կշիռները հաշվում ենք $w = (X^T X)^{-1} X^T y$ բանաձևով։ Սակայն, եթե $X^T X$-ը հակադարձելի չէ (սինգուլյար է), ապա այդ բանաձևը դառնում է անիմաստ, քանի որ մենք չենք կարող բաժանել զրոյի կամ հակադարձել սինգուլյար մատրիցը։Նման դեպքերում օգնության է հասնում $X^+$ կեղծ հակադարձը, որը գոյություն ունի ցանկացած $X$ մատրիցի համար։


Մուր-Պենրոուզի պայմանները

Որպեսզի $X^+$ մատրիցը համարվի Մուր-Պենրոուզի կեղծ հակադարձ, այն պետք է բավարարի հետևյալ չորս մաթեմատիկական պայմաններին՝

1․ $X X^+ X = X$ — Ապահովում է համապատասխանությունը սյունային տարածության հետ:


2․ $X^+ X X^+ = X^+$ — Ապահովում է համապատասխանությունը տողային տարածության հետ:


3․$(X X^+)^T = X X^+$ — Նշանակում է, որ $X X^+$ մատրիցը սիմետրիկ է:


4․$(X^+ X)^T = X^+ X$ — Նշանակում է, որ $X^+ X$ մատրիցը սիմետրիկ է:


Երկրաչափական իմաստ

Այս չորս պայմանները ունեն կարևոր երկրաչափական իմաստ։ $X X^+$-ը հանդիսանում է սիմետրիկ պրոյեկցիա $X$-ի սյունային տարածության վրա ($Col(X)$): Սա նշանակում է, որ այն օգնում է գտնել մեր նախկինում քննարկած «ստվերը» հարթության վրա:

$X^+ X$-ը հանդիսանում է սիմետրիկ պրոյեկցիա $X$-ի տողային տարածության վրա ($Row(X)$):


Երբ մենք ունեինք անվերջ թվով լուծումներ, կեղծ հակադարձ մատրիցի միջոցով ստացված $w^*$ լուծումը հանդիսանում է այն միակ լուծումը, որն ունի նվազագույն նորմ (ամենակարճ վեկտորն է բոլոր հնարավոր լուծումների միջից)։

Այսինքն Մուր-Պենրոուզի կեղծ հակադարձ մատրիցը մաթեմատիկական «փրկօղակ» է, որը թույլ է տալիս հաշվել ռեգրեսիայի կշիռները նույնիսկ այն դեպքում, երբ տվյալները թերի են կամ կրկնվող:



Կեղծ հակադարձ մատրիցը ($X^+$) պարզապես մոտավոր հաշվարկման գործիք չէ: Այն լուծում է $\min_{w} ||y - Xw||^2$ խնդիրը ճշգրիտ կերպով: Այն ապահովում է, որ մեր կանխատեսումը ($\hat{y} = XX^+ y$) լինի $y$ վեկտորի օրթոգոնալ պրոյեկցիան $X$-ի սյունային տարածության վրա:
Երբ $X^T X$ մատրիցը հակադարձելի չէ (սինգուլյար է), օրինակ՝ $D > N$ դեպքում կամ երբ կան կրկնվող սյունակներ, մենք ունենում ենք անվերջ թվով լուծումներ, որոնք բոլորն էլ տալիս են նույն նվազագույն սխալանքը:
Կեղծ հակադարձ մատրիցը բոլոր հնարավոր օպտիմալ լուծումներից ընտրում է այն մեկը, որն ունի ամենափոքր երկարությունը (նորմը՝ $||w||$):
Մաթեմատիկական ձևակերպումը` $$w^* = \text{argmin}_w ||w|| \quad \text{այն պայմանով, որ } w \text{-ն նվազեցնում է } ||y - Xw||^2$$


Երկրաչափական իմաստը՝

Գծային ռեգրեսիայի կորուստի ֆունկցիան (MSE) ունի հետևյալ տեսքը՝

$$L(w) = ||y - Xw||^2$$

Եթե մենք դիտարկենք սրա Հեսսյան մատրիցը (Hessian matrix), որը ցույց է տալիս ֆունկցիայի կորությունը (curvature), ապա կտեսնենք, որ՝

$$\nabla^2 L(w) = 2X^T X$$

Սովորական դեպքում՝ $X^T X$-ը դրական որոշյալ է (positive definite), ինչը նշանակում է, որ բոլոր ուղղություններով կորությունը դրական է (գավաթի տեսք)։



(Ենթադրենք $A \in \mathbb{R}^{n \times n}$ սիմետրիկ մատրից է ($A = A^T$):

1. Դրական որոշյալ (Positive Definite)$A$ մատրիցը կոչվում է դրական որոշյալ, եթե բոլոր ոչ զրոյական $x \in \mathbb{R}^n \setminus \{0\}$ վեկտորների համար տեղի ունի հետևյալ անհավասարությունը՝
$$x^T A x > 0$$

Սեփական արժեքներ: Բոլոր սեփական արժեքները խիստ դրական են՝ $\lambda_i > 0$:

Օպտիմալացում: Եթե կետում Hessian-ը դրական որոշյալ է, ապա այդ կետը խիստ լոկալ մինիմում է:

Երկրաչափություն: Ֆունկցիայի գրաֆիկն ունի «գավաթի» տեսք, որի հատակը $(0,0)$ կետն է:


2. Դրական կիսաորոշյալ (Positive Semidefinite)$A$ մատրիցը կոչվում է դրական կիսաորոշյալ, եթե բոլոր $x \in \mathbb{R}^n$ վեկտորների համար՝

$$x^T A x \ge 0$$

Սեփական արժեքներ: Բոլոր սեփական արժեքները ոչ բացասական են՝ $\lambda_i \ge 0$ (առնվազն մեկը կարող է լինել $0$)
:Երկրաչափություն: Գրաֆիկը կարող է ունենալ «հարթ հատակ» կամ հովտի տեսք որոշակի ուղղություններով:)








Role of the bias term

Մեքենայական ուսուցման մեջ գծային ռեգրեսիան ամենատարածված մեթոդներից է, որի նպատակն է գտնել կապը մուտքային փոփոխականների և թիրախային արժեքի միջև: Մոդելի ճշգրտությունը և կայունությունը կախված են մի քանի առանցքային բաղադրիչներից, որոնցից են Bias (շեղման) անդամը և RSS (մնացորդային քառակուսիների գումար) սխալանքի չափանիշը:
Bias-ը ($w_0$) գծային ռեգրեսիայի այն հաստատունն է, որը ցույց է տալիս, թե որտեղ է ուղիղը հատում y առանցքը: Այն կատարում է հետևյալ կարևոր ֆունկցիաները.
Պատկերացրեք, որ մենք ուզում ենք կանխատեսել տան գինը ($y$): Մենք ունենք բոլոր տների միջին գինը ($\bar{y}$):Եթե մենք փորձենք գինը հաշվել միայն մակերեսով ու սենյակների քանակով ($x_j$ հատկանիշներով), ապա այդ հաշվարկի արդյունքը (կշռված գումարը՝ $\sum w_j \bar{x}_j$) հազվադեպ կհամընկնի տների իրական միջին գնի հետ։ Այստեղ առաջանում է «ճեղք» կամ տարբերություն։Bias-ի ($w_0$) դերը հենց այդ տարբերությունը փակելն է։
$w_0$-ն այն թիվն է, որը մենք գումարում ենք մեր հաշվարկին, որպեսզի մոդելի կանխատեսած միջինը հավասարվի իրական տվյալների միջինին:Կենտրոնացում. Այն «տեղափոխում» է ռեգրեսիայի գիծը այնպես, որ այն անցնի տվյալների հենց մեջտեղով:
 
Պարզ օրինակ՝  Եթե տների միջին գինը 50,000$ է, իսկ մեր հատկանիշների (մակերես, դիրք) հիման վրա ստացված միջինը ստացվում է 40,000$, ապա $w_0$-ն կլինի հենց այդ պակասող 10,000$-ը ($50,000 - 40,000 = 10,000$): Դա այն բազային արժեքն է, որը կա նույնիսկ եթե մյուս գործոնները հաշվի չառնենք։


Առանց այս անդամի, ռեգրեսիայի գիծը միշտ կանցներ կոորդինատների սկզբնակետով $(0,0)$: Սա թույլ չէր տա ճշգրիտ նկարագրել տվյալները, եթե դրանք կենտրոնացված չեն զրոյի շուրջ:
 
Օպտիմալ $w_0$-ն հաշվարկվում է հետևյալ բանաձևով՝
 
  $w_0 = \bar{y} - \sum_{j=1}^{D} w_j \bar{x}_j$:
  
 Որպեսզի հասկանանք, թե որքանով է մոդելը լավ աշխատում տվյալների համար,  օգտագործվում են սխալանքի չափման հատուկ մեթոդներ:
 
 RSS (Residual Sum of Squares). Սա մնացորդային քառակուսիների գումարն է, որը հավասար է փաստացի արժեքների ($y_i$) և կանխատեսված արժեքների ($\hat{y}_i$) տարբերությունների քառակուսիների գումարին՝ 
 
 $RSS = \sum_{i=1}^{N} (y_i - \hat{y}_i)^2$:
 
 
 MSE (Mean Squared Error). Սա նույն RSS-ն է՝ բաժանված տվյալների քանակի ($N$) վրա ($MSE = \frac{1}{N} RSS$), ինչը ցույց է տալիս միջին սխալանքը մեկ արժեքի համար:
 
 
 Տվյալների վերլուծության ժամանակ կարևոր է հաշվի առնել երկու հիմնարար դիտարկում՝
 Ճշգրտություն. Որքան փոքր է RSS-ը, այնքան մոդելը ավելի լավ է համապատասխանում տվյալներին (better fit):
 
 Հատկանիշների ազդեցությունը. Ավելի շատ հատկանիշների (features) ավելացումը միշտ բերում է ավելի փոքր RSS-ի:
 
 Կարևոր զգուշացում. Թեև հատկանիշների ավելացումը նվազեցնում է RSS-ը, դա միշտ չէ, որ լավ է: Չափազանց շատ հատկանիշները կարող են հանգեցնել overfitting-ի (գերուսուցման), երբ մոդելը սկսում է «անգիր անել» տվյալների պատահական աղմուկը՝ կորցնելով կանխատեսման ունակությունը նոր տվյալների վրա:



Ռեգրեսիոն մոդել կառուցելիս առանցքային հարցն այն է, թե որքանով է մոդելը լավ բացատրում առկա տվյալները: Այս հարցին պատասխանում է $R^2$ (դետերմինացիայի գործակիցը)։


$R^2$-ը ցույց է տալիս տվյալների փոփոխության (տատանման) քանի տոկոսն է բացատրում մոդելը: Որքան մեծ է $R^2$-ը, այնքան մոդելը ավելի լավ է բացատրում տվյալները:

Արժեքների մեկնաբանություն.
0 - Մոդելը ոչինչ չի բացատրում:
0.5 - Տվյալների տատանումների 50%-ը բացատրվում է մոդելով:
0.8 - Տվյալների 80%-ը բացատրվում է մոդելով:
1 - Կատարյալ մոդել:

Օրինակ ենթադրենք՝ ուզում ենք կանխատեսել ուսանողի քննության միավորը ($y$)՝ ըստ սովորած ժամերի ($x$)։
Տվյալներ. (1ժ, 40մ), (2ժ, 50մ), (3ժ, 60մ), (4ժ, 70մ)։
Կանխատեսում. Մոդելը տալիս է կանխատեսված արժեքներ՝ $\hat{y}$։Արդյունք. Եթե կանխատեսումները ($\hat{y}$) շատ մոտ են իրական արժեքներին ($y$), ապա մոդելը լավն է, և $R^2$-ը մեծ կլինի։ Եթե սխալները մեծ են, $R^2$-ը փոքր կլինի։

$R^2$ գործակիցը համեմատում է երկու հասկացություն ՝

 Total Variance։ Սա ցույց է տալիս, թե տվյալները որքանով են շեղված իրենց միջինից ($avg(y)$ կամ $\bar{y}$):
 Բանաձևը՝
 
  $\sum (y_i - \bar{y})^2$
  
Սա այն սխալն է, եթե մենք ոչ մի մոդել չունենք և պարզապես միշտ կանխատեսում ենք միջին արժեքը։

Unexplained Variance։ Սրանք մոդելի սխալներն են:

Բանաձևը՝ 
$\sum (y_i - \hat{y}_i)^2$:

Սա նկարագրում է Իրական արժեքի և մոդելի կանխատեսման տարբերությունը։ 

Բանաձևն ունի հետևյալ տեսքը՝

$$R^2 = 1 - \frac{Unexplained\ variance}{Total\ variance}$$

Սա նշանակում է.Եթե սխալները շատ փոքր են, Կոտորակը փոքր է, հետևաբար $1 - (\text{փոքր թիվ})$ արդյունքում $R^2$-ը մոտ է 1-ին։Եթե սխալները շատ մեծ են. Կոտորակը մեծ է, և $R^2$-ը մոտենում է 0-ին։
