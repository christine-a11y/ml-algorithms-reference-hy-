Առավելագույն լուսանցքով դասակարգիչների (Maximum Margin Classifiers)

Հիմնական գաղափարն այն է, որ տվյալների երկու խմբերը բաժանող հարթությունը պետք է լինի այնպիսին, որ առավելագույնի հասցվի հեռավորությունը այդ հարթության և ամենամոտ գտնվող տվյալների կետերի միջև։ Սա ապահովում է մոդելի ավելի բարձր ճշգրտություն և կայունություն նոր տվյալների դեպքում։


Ա. Հեռավորության սահմանումը

Ցանկացած $x$ կետի ուղղահայաց հեռավորությունը $y(x)=0$ բաժանարար հիպերհարթությունից որոշվում է $|y(x)| / \|w\|$ բանաձևով։

Մեզ հետաքրքրում են միայն այն լուծումները, որտեղ բոլոր կետերը ճիշտ են դասակարգված, այսինքն՝ $t_n y(x_n) > 0$ բոլոր $n$ կետերի համար։

Այս պայմանը թույլ է տալիս հեռավորությունը գրել որպես $t_n y(x_n) / \|w\|$, քանի որ ճիշտ դասակարգման դեպքում արտադրյալը միշտ դրական է։

Լուսանցքի (Margin) գաղափարը

Լուսանցքը սահմանվում է որպես տվյալների բազմության ամենամոտ կետի ուղղահայաց հեռավորությունը բաժանարար գծից։

Մոդելի նպատակն է գտնել այնպիսի $w$ (կշիռներ) և $b$ (շեղում) պարամետրեր, որոնք կմեծացնեն (կմաքսիմալացնեն) այդ նվազագույն հեռավորությունը։

Օպտիմալացման խնդիրը

$$\arg \max_{w,b} \left\{ \frac{1}{\|w\|} \min_n [t_n (w^T \phi(x_n) + b)] \right\}$$

Ներքին մասը ($\min_n$): Յուրաքանչյուր հնարավոր հարթության համար մենք գտնում ենք դրան ամենամոտ գտնվող կետը (սա լուսանցքի չափն է)։

Արտաքին մասը ($\arg \max_{w,b}$): Մենք փնտրում ենք այնպիսի $w$ և $b$, որոնց դեպքում այդ «նվազագույն հեռավորությունը» կլինի հնարավորինս մեծ։

Գործակիցը ($1/\|w\|$): Քանի որ $\|w\|$-ն կախված չէ առանձին $n$ կետերից, այն դուրս է բերվել մինիմիզացիայի նշանի տակից՝ հաշվարկը պարզեցնելու համար։


Այս օպտիմալացման խնդրի ուղղակի լուծումը շատ բարդ կլիներ, ուստի մենք այն կփոխակերպենք համարժեք խնդրի, որը շատ ավելի հեշտ է լուծել: Դա անելու համար մենք նկատում ենք, որ եթե կատարենք մասշտաբավորում՝
$w \rightarrow \kappa w$ և $b \rightarrow \kappa b$, ապա ցանկացած $x_n$ կետի հեռավորությունը որոշման մակերևույթից, որը տրվում է $t_n y(x_n) / \|w\|$ բանաձևով, մնում է անփոփոխ: Մենք կարող ենք օգտագործել այս ազատությունը՝ սահմանելու համար

$$t_n (w^T \phi(x_n) + b) = 1$$

այն կետի համար, որն ամենամոտն է մակերևույթին: Այս դեպքում տվյալների բոլոր կետերը կբավարարեն հետևյալ սահմանափակումներին՝

$$t_n (w^T \phi(x_n) + b) \geqslant 1, \quad n = 1, \dots, N.$$

Սա հայտնի է որպես որոշման հիպերհարթության կանոնիկ ներկայացում: Այն տվյալների կետերի դեպքում, որոնց համար գործում է հավասարությունը, սահմանափակումները համարվում են ակտիվ, իսկ մնացածի համար՝ ոչ ակտիվ: Ըստ սահմանման, միշտ կլինի առնվազն մեկ ակտիվ սահմանափակում, քանի որ միշտ կլինի ամենամոտ կետ, և երբ լուսանցքը առավելագույնի հասցվի, կլինեն առնվազն երկու ակտիվ սահմանափակումներ:

Մանրամասն բացատրություն

Պատկերացրեք բաժանարար գիծը։ Եթե մենք նրա բանաձևի բոլոր թվերը (կշիռները և շեղումը) բազմապատկենք նույն թվով (օրինակ՝ 2-ով), գծի դիրքը չի փոխվի, և կետերի հեռավորությունը գծից նույնպես կմնա նույնը։ Քանի որ մենք ունենք այդ ազատությունը, մենք ընտրում ենք այնպիսի մասշտաբ, որ ամենամոտ կետի հեռավորությունը լինի ուղիղ 1:

Ինչո՞ւ հենց 1:
Սա թույլ է տալիս ազատվել բարդ կոտորակներից։ Եթե ամենամոտ կետի արժեքը 1 է, ապա մնացած բոլոր կետերը, որոնք ավելի հեռու են, պետք է լինեն կա՛մ 1, կա՛մ 1-ից մեծ ($\geqslant 1$): Սա ստեղծում է հստակ մաթեմատիկական սահմանափակում (constraint):

Ակտիվ և Ոչ ակտիվ սահմանափակումներ:

Ակտիվ (Active): Այն կետերը, որոնք գտնվում են հենց լուսանցքի (margin) եզրին (այսինքն՝ նրանց արժեքը ուղիղ 1 է)։ Սրանք հենց այն «Հենակետային վեկտորներն» (Support Vectors) են, որոնք պահում են հարթությունը։

Ոչ ակտիվ (Inactive): Այն կետերը, որոնք ավելի հեռու են գծից և չեն ազդում նրա դիրքի վրա։

Տեքստում ասվում է, որ երբ մենք գտնենք լավագույն լուսանցքը, կունենանք առնվազն երկու ակտիվ կետ։ Դա տրամաբանական է. որպեսզի գիծը լինի «մեջտեղում», այն պետք է հավասարապես հեռու լինի երկու կողմերի ամենամոտ կետերից (մեկը՝ մի խմբից, մյուսը՝ մյուս խմբից)։

Օպտիմալացման խնդիրն այնուհետև պարզապես պահանջում է, որ մենք մաքսիմալացնենք (առավելագույնի հասցնենք) $\|w\|^{-1}$-ը, ինչը համարժեք է $\|w\|^2$-ի մինիմալացմանը (նվազագույնի հասցնելուն), և այսպիսով մենք պետք է լուծենք օպտիմալացման խնդիրը՝ հաշվի առնելով վերը նշված սահմանափակումները

$$\arg \min_{w,b} \frac{1}{2} \|w\|^2$$

Թվում է, թե $b$ շեղման պարամետրը անհետացել է օպտիմալացումից: Այնուամենայնիվ, այն որոշվում է անուղղակիորեն՝ սահմանափակումների միջոցով, քանի որ դրանք պահանջում են, որ $w$-ի փոփոխությունները փոխհատուցվեն $b$-ի փոփոխություններով:

Այս սահմանափակված օպտիմալացման խնդիրը լուծելու համար մենք ներմուծում ենք Լագրանժի բազմապատկիչներ $a_n \geqslant 0$, յուրաքանչյուր սահմանափակման համար մեկ բազմապատկիչ $a_n$, ինչը տալիս է Լագրանժի ֆունկցիային հետևյալ տեսքը։

$$L(w, b, a) = \frac{1}{2} \|w\|^2 - \sum_{n=1}^N a_n \{t_n(w^T \phi(x_n) + b) - 1\}$$

որտեղ $a = (a_1, \dots, a_N)^T$։ Նկատեք մինուս նշանը Լագրանժի բազմապատկիչի անդամի առջև, քանի որ մենք մինիմալացնում ենք $w$ և $b$ փոփոխականների նկատմամբ և մաքսիմալացնում ենք $a$-ի նկատմամբ։ $L(w, b, a)$-ի ածանցյալները ըստ $w$-ի և $b$-ի հավասարեցնելով զրոյի՝ մենք ստանում ենք հետևյալ երկու պայմանները՝

$$w = \sum_{n=1}^N a_n t_n \phi(x_n)$$
$$0 = \sum_{n=1}^N a_n t_n$$

Այս պայմանների միջոցով $L(w, b, a)$-ից հեռացնելով $w$-ն և $b$-ն՝ ստանում ենք առավելագույն լուսանցքի խնդրի դուալ ներկայացումը (dual representation), որտեղ մենք մաքսիմալացնում ենք հետևյալ ֆունկցիան։

$$\tilde{L}(a) = \sum_{n=1}^N a_n - \frac{1}{2} \sum_{n=1}^N \sum_{m=1}^N a_n a_m t_n t_m k(x_n, x_m)$$

Պարզաբանում 

Մենք փնտրում ենք Լագրանժի ֆունկցիայի մինիմումը ըստ $w$ և $b$ պարամետրերի։ Ֆունկցիայի էքստրեմումը (այս դեպքում՝ նվազագույն կետը) գտնելու համար վերցնում ենք ածանցյալն ու հավասարեցնում զրոյի։

$w$-ի բանաձևը ցույց է տալիս, որ օպտիմալ բաժանարար գիծը ($w$) իրականում մեր տվյալների կետերի ($\phi(x_n)$) կշռված գումարն է։

$0 = \sum a_n t_n$ պայմանը պարզապես նշանակում է, որ դասերի կշռված հավասարակշռությունը պետք է պահպանվի։

Դուալ ներկայացման առավելությունը

Երբ մենք տեղադրում ենք $w$-ի արժեքը հիմնական ֆունկցիայի մեջ, $w$ և $b$ անհայտները «անհետանում» են։ Մնում է միայն $a$ (Լագրանժի բազմապատկիչները)։

Ամենակարևոր փոփոխությունը $k(x_n, x_m)$-ի հայտնվելն է։ Սա կոչվում է Kernel (կորիզ)։

$k(x_n, x_m) = \phi(x_n)^T \phi(x_m)$։ Սա նշանակում է, որ մեզ պետք չէ իմանալ կետերի կոորդինատները բարձր չափողականության մեջ, մեզ միայն պետք է նրանց սկալյար արտադրյալը (թե որքանով են նրանք նման իրար)։


Ինչո՞ւ է սա կարևոր

Հաշվարկային արագություն: Եթե մենք ունենք հազարավոր հատկանիշներ (features), բայց քիչ տվյալներ, ապա դուալ ձևով հաշվարկն ավելի արագ է։

Ոչ գծային բաժանում: Այս բանաձևը թույլ է տալիս օգտագործել «Kernel trick»-ը․ մենք կարող ենք բարդ, կորագիծ սահմաններ գծել՝ առանց հսկայական մաթեմատիկական ծախսերի։

Սպարսություն (Sparsity): Կետերի մեծ մասի համար $a_n$ կլինի 0: Միայն «Հենակետային վեկտորները» կունենան $a_n > 0$, ինչը նշանակում է, որ մոդելը հիշում է միայն ամենակարևոր կետերը։


Դուալ ներկայացումը ենթակա է հետևյալ սահմանափակումներին։


$$a_n \geqslant 0, \quad n = 1, \dots, N, \quad \sum_{n=1}^N a_n t_n = 0$$

Այստեղ կորիզային ֆունկցիան (kernel function) սահմանվում է որպես $k(x, x') = \phi(x)^T \phi(x')$։ $M$ փոփոխականներով քառակուսային ծրագրավորման խնդրի լուծումը ընդհանուր առմամբ ունի $O(M^3)$ հաշվարկային բարդություն։ Անցնելով դուալ ձևակերպմանը՝ մենք սկզբնական օպտիմալացման խնդիրը վերածեցինք դուալ խնդրի, որն ունի $N$ փոփոխական։ Հիմնական ֆունկցիաների ֆիքսված բազմության համար, որոնց $M$ քանակը ավելի փոքր է, քան տվյալների կետերի $N$ քանակը, անցումը դուալ խնդրին թվում է անշահավետ։ Այնուամենայնիվ, այն թույլ է տալիս մոդելը վերաձևակերպել՝ օգտագործելով կորիզներ (kernels), և այսպիսով առավելագույն լուսանցքով դասակարգիչը կարող է արդյունավետորեն կիրառվել այնպիսի հատկանիշների տարածությունների համար, որոնց չափողականությունը գերազանցում է տվյալների կետերի քանակը, ներառյալ անվերջ չափողականությամբ տարածությունները։ Կորիզային ձևակերպումը նաև հստակեցնում է այն սահմանափակման դերը, որ $k(x, x')$ կորիզային ֆունկցիան պետք է լինի դրական որոշյալ (positive definite), քանի որ սա ապահովում է, որ Լագրանժի ֆունկցիան սահմանափակված լինի ներքևից՝ հանգեցնելով լավ սահմանված օպտիմալացման խնդրի։

Պարզաբանում

Հաշվարկային բարդություն ($O(M^3)$ vs $N$)

Սկզբնական խնդիր (Primal): Կախված է հատկանիշների քանակից ($M$): Եթե դուք ունեք նկար, որն ունի 1 միլիոն պիքսել ($M=1,000,000$), ապա հաշվարկը կլինի չափազանց դանդաղ:

Դուալ խնդիր (Dual): Կախված է տվյալների կետերի քանակից ($N$): Եթե ունեք ընդամենը 500 նկար, ապա հաշվարկը կլինի շատ արագ, անկախ նրանից, թե որքան պիքսել ունի յուրաքանչյուր նկարը:

Անվերջ չափողականություն և Kernel TrickՍա SVM-ի ամենազարմանալի հատկությունն է։ Քանի որ բանաձևում մենք օգտագործում ենք միայն $k(x, x')$ (կետերի նմանությունը), մեզ պարտադիր չէ կետերը իրականում տեղափոխել բարձր չափողականության տարածություն։ Մենք կարող ենք աշխատել այնպիսի տարածություններում, որոնք ունեն անվերջ չափեր (օրինակ՝ RBF կորիզի դեպքում)։

Որպեսզի մաթեմատիկական խնդիրը լուծելի լինի և ունենա հստակ նվազագույն կամ առավելագույն կետ, kernel ֆունկցիան պետք է բավարարի որոշակի պայմանների (Մերսերի պայմաններ)։ Եթե kernel-ը դրական որոշյալ է, ապա մենք երաշխավորված ենք, որ մեր ալգորիթմը կգտնի օպտիմալ լուծումը և «չի մոլորվի» անվերջության մեջ։


Վարժեցված մոդելի միջոցով նոր տվյալների կետերը դասակարգելու համար մենք գնահատում ենք $y(x)$-ի նշանը: Սա կարող է արտահայտվել $\{a_n\}$ պարամետրերի և kernel ֆունկցիայի միջոցով՝ տեղադրելով $w$-ի արժեքը, ինչը տալիս է`

$$y(x) = \sum_{n=1}^N a_n t_n k(x, x_n) + b$$


Այստեղ մենք ունենք 3 սահմանափակում (Կարուշ-Կուն-Տակերի կամ KKT պայմաններ):

$$a_n \geqslant 0$$
$$t_n y(x_n) - 1 \geqslant 0$$
$$a_n \{t_n y(x_n) - 1\} = 0$$

Այսպիսով, յուրաքանչյուր տվյալի կետի համար կամ $a_n = 0$, կամ $t_n y(x_n) = 1$: Ցանկացած կետ, որի համար $a_n = 0$, չի մասնակցի $y(x)$-ի հաշվարկին և, հետևաբար, ոչ մի դեր չի խաղա նոր տվյալների համար կանխատեսումներ անելիս: Մնացած տվյալների կետերը կոչվում են հենակետային վեկտորներ (support vectors), քանի որ դրանք բավարարում են $t_n y(x_n) = 1$ պայմանին:

Պարզաբանում

Երբ մոդելը ստանում է նոր $x$ կետ, նա չի փորձում հիշել ամբողջ հարթությունը։ Նա պարզապես հաշվում է այդ նոր կետի «նմանությունը» (կորիզը՝ $k(x, x_n)$) միայն այն կետերի հետ, որոնք մոդելը համարել է կարևոր ($a_n > 0$): Եթե վերջնական գումարը դրական է՝ կետը պատկանում է մի դասին, եթե բացասական է՝ մյուսին։

Երրորդ պայմանը՝ $a_n \{t_n y(x_n) - 1\} = 0$, կոչվում է լրացնող ոչ-խստություն (complementary slackness): Այն ասում է հետևյալը.՝ Եթե կետը գտնվում է լուսանցքից (margin) հեռու և ապահով տեղում, ապա նրա $a_n$ գործակիցը պետք է լինի 0:Եթե $a_n = 0$, ապա այդ կետը «անհետանում» է բանաձևից։ Այն այլևս պետք չէ պահել հիշողության մեջ։

Հենակետային վեկտորները այն սակավաթիվ կետերն են, որոնք գտնվում են հենց լուսանցքի գծի վրա ($t_n y(x_n) = 1$): Դրանք ամենադժվար դասակարգվող կետերն են, որոնք էլ հենց «պահում են» բաժանարար պատնեշը։ Եթե դուք ջնջեք ձեր տվյալների 90%-ը, բայց թողնեք միայն հենակետային վեկտորները, SVM-ի արդյունքը բացարձակապես չի փոխվի։


1. Հենակետային վեկտորների դերը (Sparsity)
SVM-ի ամենամեծ առավելություններից մեկն այն է, որ վարժեցումից հետո մեզ պետք չէ հիշել մեր ողջ տվյալների բազան:

Տվյալների մեծ մասը հեռու է բաժանարար գծից և չի ազդում դրա դիրքի վրա ($a_n = 0$):

Հենակետային վեկտորներ ($S$ բազմություն): Սրանք այն սակավաթիվ կետերն են, որոնք գտնվում են լուսանցքի (margin) հենց եզրին: Մոդելը հիմնվում է միայն սրանց վրա: Սա թույլ է տալիս խնայել հիշողությունը և արագացնել կանխատեսումները:

2. Ինչպե՞ս գտնել $b$-ն

$b$-ն այն թիվն է, որը որոշում է բաժանարար հարթության ճիշտ տեղադրությունը (շեղումը կոորդինատների սկզբնակետից)։Մենք գիտենք, որ հենակետային վեկտորների համար արժեքը պետք է լինի ուղիղ 1 (կամ -1) $t_n y(x_n) = 1$:

Եթե փոխարինենք $y(x_n)$-ը իր կորիզային ($kernel$) տեսքով, ստանում ենք`

$$t_n \left( \sum_{m \in S} a_m t_m k(x_n, x_m) + b \right) = 1$$

Տեսականորեն կարելի է վերցնել ընդամենը մեկ support vector և գտնել $b$-ն։ Սակայն իրական տվյալներում կարող են լինել չնչին սխալներ կամ աղմուկ։
Դրանից խուսափելու համար մենք`
* Բազմապատկում ենք $t_n$-ով (քանի որ $t_n^2=1$, այն վերանում է փակագծի առջևից):
* Հաշվում ենք $b$-ի արժեքը բոլոր $N_S$ քանակությամբ հենակետային վեկտորների համար:
* Վերցնում ենք դրանց միջին թվաբանականը:

Վերջնական բանաձևը`

$$b = \frac{1}{N_S} \sum_{n \in S} \left( t_n - \sum_{m \in S} a_m t_m k(x_n, x_m) \right)$$


Որպեսզի $b$-ի արժեքը լինի հնարավորինս ճշգրիտ, ալգորիթմը հաշվում է $b$-ն բոլոր հասանելի հենակետային վեկտորների համար և վերցնում դրանց միջին թվաբանականը ($\frac{1}{N_S} \sum$): Սա նվազեցնում է հաշվարկային սխալի հավանականությունը և դարձնում մոդելն ավելի կայուն։


SVM-ի նպատակն է գտնել այնպիսի բաժանարար հիպերհարթություն ($y(x)=0$), որն առավելագույնի կհասցնի հեռավորությունը երկու դասերի ամենամոտ կետերից:


Այլ մոդելների հետ հետագա համեմատության համար մենք կարող ենք արտահայտել Maximum margin classifair-ը որպես սխալի ֆունկցիայի մինիմալացում՝ պարզ քառակուսային ռեգուլյարիզատորով, հետևյալ տեսքով՝
$$\sum_{n=1}^{N} E_{\infty}(y(\mathbf{x}_n)t_n - 1) + \lambda\|\mathbf{w}\|^2$$

$E_{\infty}(z)$ (Անվերջ սխալի ֆունկցիա): Սա «տուգանային» ֆունկցիա է։ Եթե տվյալը ճիշտ է դասակարգված և գտնվում է լուսանցքից դուրս կամ դրա վրա ($y(\mathbf{x}_n)t_n \geq 1$), ապա սխալը $0$ է։ Բայց եթե թեկուզ մեկ կետ սխալ է դասակարգված կամ ներխուժել է լուսանցքի մեջ, սխալը դառնում է անվերջ ($\infty$): Սա ստիպում է ալգորիթմին գտնել այնպիսի լուծում, որտեղ բոլոր կետերը կատարյալ բաժանված են։

$\lambda\|\mathbf{w}\|^2$ (Ռեգուլյարիզացիա): Այս մասը պատասխանատու է «լուսանցքը» (margin) մաքսիմալացնելու համար։ Մեքենայական ուսուցման մեջ $\|\mathbf{w}\|^2$-ի մինիմալացումը համարժեք է դասակարգող գծերի միջև հեռավորությունը մեծացնելուն։


Նկատի ունեցեք, որ քանի դեռ ռեգուլյարիզացիայի պարամետրը բավարարում է $\lambda > 0$ պայմանին, դրա հստակ արժեքը որևէ դեր չի խաղում։
Ինչու՞ $\lambda$-ի արժեքը կապ չունի: Քանի որ առաջին մասը (գումարը) կարող է լինել կամ $0$, կամ $\infty$, ապա ցանկացած դրական $\lambda$-ի դեպքում մինիմալացման միակ ձևը նախ $E_{\infty}$-ն $0$ դարձնելն է (այսինքն՝ բոլոր սահմանափակումները պահելը), իսկ հետո $\|\mathbf{w}\|^2$-ը փոքրացնելը։


Այս բանաձևը ասում է. «Գտիր այնպիսի գիծ, որը ոչ մի սխալ թույլ չի տա (անվերջ սխալից խուսափելու համար) և միաժամանակ կունենա հնարավոր ամենալայն դատարկ տարածությունը (լուսանցքը) տարբեր դասերի կետերի միջև»։



1. Ոչ գծային բաժանում (Gaussian Kernel)

Պատկերում տեսնում եք, որ կապույտ և կարմիր կետերը հնարավոր չէ բաժանել ուղիղ գծով։ SVM-ն օգտագործում է «կորիզային հնարքը»՝ տվյալները տեղափոխելով ավելի բարձր չափողականություն ունեցող տիրույթ, որտեղ դրանք դառնում են գծայնորեն բաժանելի։ Երբ այդ բաժանումը պրոյեկտում ենք հետ՝ մեր սկզբնական 2D տիրույթի վրա, այն հայտնվում է կորերի տեսքով (ինչպես տեսնում եք նկարում)։

2. Հենակետային վեկտորներ (Support Vectors)

Նկարում կանաչ շրջանակների մեջ առնված կետերը հենց հենակետային վեկտորներն են։ Դրանք այն կետերն են, որոնք գտնվում են լուսանցքի (margin) եզրին։

Ինչու՞ է սա կարևոր. SVM-ի ամբողջ մոդելը կառուցվում է միայն այս մի քանի կետերի հիման վրա։

Sparsity (Նոսրություն). Եթե դուք ջնջեք կամ տեղաշարժեք մնացած բոլոր կետերը (որոնք շրջանակի մեջ չեն), դասակարգող գիծը չի փոխվի։ Սա նշանակում է, որ մոդելը շատ արդյունավետ է և «հիշում» է միայն ամենակարևոր օրինակները, որոնք սահմանում են դասերի միջև սահմանը։

3. Որոշման սահման (Decision Boundary)
Հաստ սև գիծը հենց որոշման սահմանն է, իսկ դրանից երկու կողմ գտնվող բարակ գծերը սահմանում են լուսանցքը։ SVM-ի նպատակն է այդ լուսանցքը դարձնել հնարավորինս լայն։


1. Կապույտ և կարմիր խաչերը ($x$)Սրանք քո տվյալներն են (տարբեր դասեր)։ Ինչպես տեսնում ես, դրանք դասավորված են խմբերով, և հնարավոր չէ մեկ ուղիղ գծով բաժանել կարմիրները կապույտներից։2. Հաստ սև կորը (Decision Boundary)Սա հենց որոշման սահմանն է։Այս գիծը կառուցվել է այնպես, որ դրա մի կողմում հայտնվեն միայն կապույտ կետերը, իսկ մյուս կողմում՝ կարմիրները։Քանի որ օգտագործվել է Gaussian kernel, սահմանը ոչ թե ուղիղ է, այլ «ճկվում է» տվյալների խտության համաձայն՝ ստեղծելով կղզյակներ հիշեցնող տեսք։3. Բարակ սև գծերը և «Լուսանցքը» (Margin)Հաստ գծի երկու կողմերում տեսնում ես ավելի բարակ գծեր։ Դրանց միջև ընկած տարածությունը լուսանցքն է։Ալգորիթմը փորձում է այս տարածությունը դարձնել հնարավորինս լայն։Որքան լայն է այս «ճանապարհը», այնքան ավելի վստահ է մոդելը իր որոշման մեջ։4. Կանաչ շրջանակները (Support Vectors)Սա նկարի ամենակարևոր մասն է։ Նկատիր, որ շրջանակված են միայն այն կետերը, որոնք դիպչում են լուսանցքի եզրերին (բարակ գծերին)։Սրանք հենակետային վեկտորներն են։ Սահմանը կառուցվել է հենց սրանց հիման վրա։Եթե դու փոխես մնացած կետերի դիրքը (նրանց, որոնք շրջանակված չեն), սև հաստ գիծը չի շարժվի։ Բայց եթե թեկուզ մի փոքր շարժես կանաչ շրջանակով կետը, ամբողջ սահմանը կփոխվի։5. Գրադիենտային գծերը (Contour lines)Նկարի վրայի մյուս թույլ գծերը ցույց են տալիս «դաշտի ուժգնությունը»։ Որքան հեռանում ես սահմանից դեպի դասի խորքը, այնքան մոդելը ավելի մեծ հավանականությամբ է կանխատեսում տվյալ դասը։Ամփոփելով նկարը. Սահմանը կառուցված է այնպես, որ առավելագույնի հասցնի հեռավորությունը ամենամոտ կանգնած կապույտ և կարմիր կետերի միջև (կանաչ շրջանակների միջև), ինչը թույլ է տալիս ալգորիթմին լինել կայուն և ճշգրիտ։

(Overlapping Class Distributions):

Մինչ այժմ մենք ենթադրում էինք, որ ուսուցման (training) տվյալների կետերը գծայնորեն բաժանելի են հատկանիշների տարածությունում՝ ϕ(x)։ Ստացված աջակցող վեկտորների մեքենան (Support Vector Machine, SVM) կապահովի ուսուցման տվյալների ճշգրիտ բաժանում սկզբնական մուտքային տարածությունում x, թեև համապատասխան որոշման սահմանը կլինի ոչ գծային։

Սակայն գործնականում դասերին համապատասխան բաշխումները (class-conditional distributions) կարող են հատվել, և այդ դեպքում ուսուցման տվյալների ճշգրիտ բաժանումը կարող է բերել վատ ընդհանրացման (poor generalization)։ Այդ պատճառով անհրաժեշտ է ձևափոխել SVM-ը այնպես, որ որոշ ուսուցման կետեր թույլատրվեն սխալ դասակարգվել։

Բաժանելի դասերի դեպքում մենք անուղղակիորեն օգտագործում էինք սխալի ֆունկցիա, որը տալիս էր անվերջ մեծ սխալ, եթե տվյալ կետը սխալ էր դասակարգվում, և զրոյական սխալ, եթե այն ճիշտ էր դասակարգվում։ Այնուհետև օպտիմալացնում էինք մոդելի պարամետրերը՝ margin-ը առավելագույն դարձնելու համար։

Այժմ մենք փոփոխում ենք այս մոտեցումը այնպես, որ տվյալների կետերին թույլատրվի գտնվել margin-ի սահմանի «սխալ կողմում», բայց դրա դիմաց կիրառվի տուգանք (penalty), որը մեծանում է այդ սահմանից ունեցած հեռավորության աճին զուգահեռ։ Հետագա օպտիմալացման խնդրի համար հարմար է այդ տուգանքը սահմանել որպես այդ հեռավորության գծային ֆունկցիա։



Դրա համար մենք ներմուծում ենք լրացուցիչ (սլաք) փոփոխականներ՝ $\xi_n \ge 0$, որտեղ $n = 1, ..., N$: Յուրաքանչյուր տվյալի կետի համար գոյություն ունի մեկ լրացուցիչ փոփոխական: Դրանք սահմանվում են որպես $\xi_n = 0$ այն կետերի համար, որոնք գտնվում են ճիշտ մարժայի (սահմանի) վրա կամ դրանից ներս, և $\xi_n = |t_n - y(x_n)|$ մյուս կետերի համար: Այսպիսով, այն կետը, որը գտնվում է որոշման սահմանի վրա ($y(x_n)=0$), կունենա $\xi_n = 1$, իսկ $\xi_n > 1$ ունեցող կետերը կլինեն սխալ դասակարգված: Դասակարգման ճշգրիտ սահմանափակումները փոխարինվում են հետևյալով՝

$$t_n y(\mathbf{x}_n) \ge 1 - \xi_n, \quad n = 1, \dots, N$$

որտեղ լրացուցիչ փոփոխականները պետք է բավարարեն $\xi_n \ge 0$ պայմանին:

Տվյալների այն կետերը, որոնց համար՝$\xi_n = 0$, ճիշտ են դասակարգված և գտնվում են կա՛մ մարժայի վրա, կա՛մ մարժայի ճիշտ կողմում:

$0 < \xi_n \le 1$, գտնվում են մարժայի ներսում, բայց որոշման սահմանի (decision boundary) ճիշտ կողմում:

$\xi_n > 1$, գտնվում են որոշման սահմանի սխալ կողմում և սխալ են դասակարգված:



Գծապատկերը հիանալի կերպով ցույց է տալիս կապը $\xi$-ի և կետի դիրքի միջև՝

$\xi = 0$: Կետերը գտնվում են կապույտ գծերի վրա կամ դրանցից դուրս (ճիշտ տիրույթում):

$0 < \xi < 1$: Կետը հատել է կապույտ գիծը, բայց դեռ չի հասել մեջտեղի կարմիր գծին ($y=0$): Սա նշանակում է, որ այն դեռ ճիշտ է դասակարգված, բայց խախտել է «անվտանգության գոտին»:

$\xi > 1$: Կետը հատել է կարմիր գիծը և հայտնվել է հակառակ կողմում: Սա արդեն սխալ դասակարգում է:

Սակայն այս համակարգը դեռևս զգայուն է աութլայերների (outliers - շեղվող արժեքներ) նկատմամբ, քանի որ սխալ դասակարգման համար տուգանքը աճում է գծային ձևով $\xi$-ին զուգահեռ:

Քանի որ տուգանքը հաշվարկվում է որպես $C \sum \xi_n$, եթե ունենք մեկ կետ, որը շատ հեռու է ընկած իր խմբից (աութլայեր), նրա $\xi$ արժեքը կլինի շատ մեծ:

Քանի որ տուգանքն աճում է գծային (ուղիղ համեմատական $\xi$-ին), ալգորիթմը կարող է «զոհաբերել» ընդհանուր մարժայի լայնությունը և թեքել բաժանարար գիծը միայն այդ մեկ սխալ կետի տուգանքը փոքրացնելու համար: Արդյունքում ստացվում է ոչ թե օպտիմալ բաժանում, այլ մի գիծ, որը չափազանցված կերպով հարմարվել է պատահական աղմուկին (outlier):

Հիմնական ալգորիթմ:

Մեր նպատակն այժմ մարժայի առավելագույնի հասցումն է՝ միաժամանակ «մեղմորեն» տուգանելով այն կետերը, որոնք գտնվում են մարժայի սահմանի սխալ կողմում: Հետևաբար, մենք մինիմալացնում ենք՝

$$C \sum_{n=1}^{N} \xi_n + \frac{1}{2} \|\mathbf{w}\|^2$$

որտեղ $C > 0$ պարամետրը վերահսկում է հավասարակշռությունը (trade-off) սխալների տուգանքի և բաժանարար գծի լայնության միջև: Քանի որ ցանկացած սխալ դասակարգված կետի համար $\xi_n > 1$, հետևում է, որ $\sum \xi_n$-ը սխալ դասակարգված կետերի քանակի վերին սահմանն է: $C$ պարամետրը, հետևաբար, համանման է ռեգուլյարիզացիայի գործակցին (ավելի ճիշտ՝ դրա հակադարձին), քանի որ այն կարգավորում է կապը վարժեցման սխալների նվազեցման և մոդելի բարդության միջև: $C \to \infty$ սահմանում մենք կստանանք նախկինում դիտարկված SVM-ը՝ գծայնորեն բաժանելի տվյալների համար:


Պարզաբանում

1. Ի՞նչ է խնդիրը

Սովորաբար SVM-ը փորձում է գտնել այնպիսի գիծ, որը հստակ կբաժանի երկու խումբ տվյալները: Բայց իրական կյանքում տվյալները հաճախ խառնված են: Եթե մենք փորձենք ստեղծել «կատարյալ» բաժանարար գիծ, այն կարող է շատ ծուռումուռ ստացվել (overfitting), ինչը վատ է նոր տվյալների դեպքում:

2. Բանաձևի երկու մասերը

Մոդելը փորձում է միաժամանակ լուծել երկու հակադիր խնդիր՝

Մաս 1: $\frac{1}{2} \|\mathbf{w}\|^2$ (Լայն ճանապարհ)Սա պատասխանատու է բաժանարար գծի շուրջ գտնվող «անվտանգության գոտին» (margin) հնարավորինս լայն պահելու համար: Ինչքան լայն է այս գոտին, այնքան մոդելն ավելի վստահ է աշխատում:

Մաս 2: $C \sum \xi_n$ (Սխալների տուգանք)$\xi$ (քսի) կոչվող փոփոխականները ցույց են տալիս, թե որքանով է կետը «խախտել» սահմանը: Եթե կետը սխալ կողմում է, $\xi$-ն մեծանում է: Մենք ուզում ենք, որ այս գումարը փոքր լինի:

3. $C$ պարամետրի դերը (Կարևոր է!)$C$-ն այն «կարգավորիչն» է, որով դուք ասում եք ալգորիթմին, թե ինչն է ավելի կարևոր ձեզ համար՝

* Եթե $C$-ն շատ մեծ է. Մոդելը շատ խիստ է դառնում: Նա ասում է. «Ես չեմ հանդուրժի ոչ մի սխալ կետ!»: Արդյունքում կստանաք նեղ մարժա, բայց քիչ սխալներ վարժեցման տվյալների վրա:
* Եթե $C$-ն փոքր է. Մոդելը դառնում է «հանդուրժող»: Նա ասում է. «Ոչինչ, թող մի քանի կետ սխալ լինեն, կարևորը ես ունենամ լայն և հարթ բաժանարար գիծ»: Սա օգնում է, որ մոդելը հետագայում ավելի լավ աշխատի նոր, անծանոթ տվյալների հետ:

4. Ի՞նչ է $\xi > 1$-ը

Եթե կետը պարզապես մի փոքր մտել է «անվտանգության գոտի», $\xi$-ն 0-ից 1-ի արանքում է: Բայց եթե կետը հատել է բուն բաժանարար գիծը և հայտնվել հակառակորդի դաշտում (սխալ դասակարգում), ապա $\xi > 1$: Դրա համար էլ ասվում է, որ դրանց գումարը սխալների քանակի «վերին սահմանն» է:

Երբ մենք ունենք մինիմալացման խնդիր (նվազեցնել սխալները և մեծացնել մարժան) և որոշակի սահմանափակումներ (որ կետերը պետք է լինեն ճիշտ կողմում), մենք ստեղծում ենք մեկ միասնական ֆունկցիա՝ $L(\mathbf{w}, b, \mathbf{a})$:


$$L(\mathbf{w}, b, \mathbf{a}) = \frac{1}{2}\|\mathbf{w}\|^2 + C \sum_{n=1}^{N} \xi_n - \sum_{n=1}^{N} a_n \{t_n y(\mathbf{x}_n) - 1 + \xi_n\} - \sum_{n=1}^{N} \mu_n \xi_n$$

Առաջին երկու անդամները այն են, ինչ մենք ուզում էինք մինիմալացնել։

Երրորդ անդամը (հանվող) ներկայացնում է հիմնական սահմանափակումը։ 

$a_n \ge 0$ գործակիցները կոչվում են Լագրանժի բազմապատկիչներ։ Եթե կետը խախտում է սահմանը, այս մասը «տուգանում» է ֆունկցիային։

Չորրորդ անդամը ապահովում է, որ $\xi_n$ (սխալի չափը) միշտ լինի դրական կամ զրո ($\xi_n \ge 0$)։

KKT Պայմանները (KKT Conditions)
Սրանք այն անհրաժեշտ պայմաններն են, որոնք պետք է բավարարվեն օպտիմալ կետում (այսինքն՝ երբ գտել ենք լավագույն բաժանարար գիծը)։

* $a_n \ge 0$ և $\mu_n \ge 0$: Սրանք պարզապես ասում են, որ բազմապատկիչները չեն կարող բացասական լինել։

* $t_n y(\mathbf{x}_n) - 1 + \xi_n \ge 0$։ Soft Margin -
Երբ ավելացնում ենք $\xi_n \ge 0$, պայմանը դառնում է.$$t_n y(\mathbf{x}_n) \ge 1 - \xi_n$$Այստեղ $\xi_n$-ը գործում է որպես «տուգանային միավոր», որը կետին թույլ է տալիս խախտել կանոնները։ Տեսնենք, թե ինչ է լինում տարբեր արժեքների դեպքում.
Եթե $\xi_n = 0$:Կետը պետք է բավարարի $t_n y(\mathbf{x}_n) \ge 1$ պայմանին։ Այսինքն՝ այն ճիշտ տեղում է (սահմանի վրա կամ դրանից դուրս)։Եթե $0 < \xi_n \le 1$:Աջ կողմի արժեքը դառնում է $1$-ից փոքր (օրինակ՝ $0.3$): Սա նշանակում է, որ կետը «մտել է» բաժանարար գոտու (margin) ներսը, բայց դեռ ճիշտ կողմում է։ Այն դեռ սխալ չէ, բայց արդեն «կասկածելի» է։Եթե $\xi_n > 1$:Աջ կողմի արժեքը դառնում է բացասական (օրինակ՝ $1 - 1.5 = -0.5$): Սա նշանակում է, որ կետը հատել է բուն բաժանարար գիծը և հայտնվել է հակառակորդի դաշտում։ Սա արդեն դասակարգման սխալ է։

* $a_n \{t_n y(\mathbf{x}_n) - 1 + \xi_n\} = 0$ (Complementary Slackness):Սա ամենակարևոր պայմանն է։ Այն ասում է՝ կա՛մ $a_n = 0$, կա՛մ կցված արտահայտությունը զրո է։Եթե $a_n = 0$, այդ կետը չի ազդում բաժանարար գծի վրա (այն հեռու է սահմանից)։Եթե $a_n > 0$, ապա այդ կետը հենակետային վեկտոր է (Support Vector), որը կա՛մ ընկած է մարժայի վրա, կա՛մ սխալ կողմում է։

*$\mu_n \xi_n = 0$: Նմանապես, եթե $\xi_n > 0$ (կետը սխալ կողմում է), ապա $\mu_n$ պետք է լինի 0:

Այս մաթեմատիկական ապարատը թույլ է տալիս SVM-ին «որոշել», թե որ տվյալներն են կարևոր բաժանարար գիծը կառուցելու համար և որոնք կարելի է անտեսել։ Գործնականում սա հնարավորություն է տալիս լուծել խնդիրը Dual Form-ով, որն ավելի արդյունավետ է հաշվարկների համար, հատկապես երբ օգտագործում ենք Kernel-ի մեթոդները։

րպեսզի գտնենք Լագրանժի ֆունկցիայի մինիմումը, մենք վերցնում ենք մասնակի ածանցյալներն ըստ մեր փոփոխականների ($\mathbf{w}, b, \xi_n$) և հավասարեցնում զրոյի։

* $\frac{\partial L}{\partial \mathbf{w}} = 0$: Սա մեզ տալիս է $\mathbf{w} = \sum a_n t_n \phi(\mathbf{x}_n)$։

Իմաստը: Այն ասում է, որ օպտիմալ բաժանարար գիծը ($\mathbf{w}$) տվյալների կետերի կշռված գծային համակցությունն է։ Ընդ որում, միայն այն կետերն են մասնակցում գծի կառուցմանը, որոնց համար $a_n > 0$ (հենակետային վեկտորներ)։

* $\frac{\partial L}{\partial b} = 0$: Սա տալիս է $\sum a_n t_n = 0$։

Իմաստը: Սա հավասարակշռության պայման է, որը կապում է դասակարգման պիտակները և Լագրանժի բազմապատկիչները։

* $\frac{\partial L}{\partial \xi_n} = 0$: Սա տալիս է $a_n = C - \mu_n$։

Իմաստը: Սա չափազանց կարևոր կապ է։ Քանի որ $\mu_n \ge 0$, սա նշանակում է, որ $a_n \le C$:Հիմա տեսեք՝ նախորդ քայլում մենք ասացինք, որ $a_n \ge 0$, իսկ այստեղից ստացանք $a_n \le C$: Այսինքն՝ $a_n$ գործակիցը միշտ գտնվում է $[0, C]$ միջակայքում։

Երբ այս ստացված արդյունքները տեղադրում ենք սկզբնական ֆունկցիայի մեջ, $\mathbf{w}$-ն, $b$-ն և $\xi$-ն «անհետանում են», և մենք ստանում ենք մի ֆունկցիա, որը կախված է միայն $a_n$ գործակիցներից։

Երկակի Լագրանժյան (Dual)
$$\tilde{L}(\mathbf{a}) = \sum_{n=1}^{N} a_n - \frac{1}{2} \sum_{n=1}^{N} \sum_{m=1}^{N} a_n a_m t_n t_m k(\mathbf{x}_n, \mathbf{x}_m)$$

Այստեղ հայտնվում է $k(\mathbf{x}_n, \mathbf{x}_m)$ տերմինը, որը կոչվում է Kernel ։

Օպտիմալացումն այժմ կախված է միայն տվյալների կետերի քանակից ($N$), այլ ոչ թե այն բանից, թե քանի հատկանիշ (feature) ունի յուրաքանչյուր կետը։ Սա թույլ է տալիս աշխատել հսկայական չափողականություն ունեցող տվյալների հետ։

ենք կարող ենք փոխարինել պարզ սկալյար արտադրյալը բարդ կորիզային ֆունկցիաներով (օրինակ՝ RBF կամ Gaussian): Սա թույլ է տալիս SVM-ին գտնել ոչ գծային բաժանարար սահմաններ՝ տվյալները մտովի տեղափոխելով ավելի բարձր չափողականության տարածություն, բայց առանց իրականում այդ բարդ հաշվարկները կատարելու։

Մենք այժմ ընդամենը պետք է մաքսիմալացնենք այս ֆունկցիան $0 \le a_n \le C$ և $\sum a_n t_n = 0$ պայմանների դեպքում։ Սա քառակուսային ծրագրավորման (Quadratic Programming) խնդիր է, որը համակարգիչները լուծում են շատ արդյունավետ։

Երկակի ձևը նույնական է գծայնորեն բաժանելի դեպքին, բացառությամբ այն բանի, որ սահմանափակումները փոքր-ինչ տարբեր են: Տեսնելու համար, թե որոնք են այդ սահմանափակումները, մենք նշում ենք, որ $a_n \ge 0$ պահանջվում է, քանի որ դրանք Լագրանժի բազմապատկիչներ են: Ավելին, $\mu_n \ge 0$ պայմանի հետ միասին դա նշանակում է, որ $a_n \le C$: Հետևաբար, մենք պետք է մինիմալացնենք (նվազեցնենք) Լագրանժյանը երկակի փոփոխականների ($\{a_n\}$) նկատմամբ՝ հետևյալ սահմանափակումներով:

$$0 \le a_n \le C, \quad \sum_{n=1}^{N} a_n t_n = 0$$

$n = 1, \dots, N$ համար, որտեղ առաջին հավասարումը հայտնի է որպես box constraints (արկղային սահմանափակումներ):Այժմ մենք կարող ենք մեկնաբանել ստացված լուծումը: Ինչպես նախկինում, տվյալների կետերի մի մասը կարող է ունենալ $a_n = 0$, որի դեպքում դրանք չեն նպաստում կանխատեսող մոդելի կառուցմանը: Մնացած տվյալների կետերը հանդիսանում են հենակետային վեկտորներ (support vectors): Սրանք ունեն $a_n > 0$ և, հետևաբար:

$$t_n y(\mathbf{x}_n) = 1 - \xi_n$$


Ի՞նչ է «Box Constraints»-ը

Սա SVM-ի ամենակարևոր սահմանափակումն է։ Այն ասում է, որ յուրաքանչյուր $a_n$ գործակից պետք է լինի «արկղի» մեջ՝ $0$-ից մինչև $C$։

Ստորին սահմանը (0): Եթե կետը շատ հեռու է բաժանարար գոտուց և ճիշտ կողմում է, նրա $a_n$-ը կլինի 0: Այն մոդելի համար «անտեսանելի» է։

Վերին սահմանը ($C$): $C$-ն դնում է «առավելագույն ուժի» չափը, որով կետը կարող է ազդել գծի վրա։ Եթե կետը սխալ կողմում է, նրա ազդեցությունը չի կարող անվերջ մեծ լինել. այն կանգ կառնի $C$ արժեքի վրա։ Սա թույլ չի տալիս, որ մեկ «աղմկոտ» կետը (outlier) փչացնի ողջ մոդելը։


Երբ մենք ասում ենք $t_n y(\mathbf{x}_n) = 1 - \xi_n$, մենք բաժանում ենք հենակետային վեկտորները երկու տիպի.

* Եթե $0 < a_n < C$: Այստեղից հետևում է, որ $\xi_n = 0$: Սա նշանակում է, որ կետը գտնվում է ճշգրիտ մարժայի (margin) գծի վրա: Սրանք այն իդեալական կետերն են, որոնք «պահում են» ճանապարհի եզրերը։
* Եթե $a_n = C$: Այստեղից հետևում է, որ $\xi_n > 0$: Այս կետերը գտնվում են մարժայի ներսում կամ նույնիսկ սխալ կողմում։ Չնայած դրանք «սխալներ» են, դրանք դեռ համարվում են հենակետային վեկտորներ, քանի որ մասնակցում են բաժանարար գծի դիրքի որոշմանը։

* Եթե $a_n < C$, դա նշանակում է, որ $\mu_n > 0$ (քանի որ $a_n + \mu_n = C$): Ըստ KKT պայմանների, սա պահանջում է, որ $\xi_n = 0$: Հետևաբար, նման կետերը գտնվում են անմիջապես մարժայի (սահմանի) վրա:

Այն կետերը, որտեղ $a_n = C$, կարող են լինել մարժայի ներսում: Դրանք կլինեն ճիշտ դասակարգված, եթե $\xi_n \le 1$, կամ սխալ դասակարգված, եթե $\xi_n > 1$:

$b$-ն գտնելու համար մենք օգտագործում ենք այն հենակետային վեկտորները, որոնք ընկած են հենց մարժայի վրա ($0 < a_n < C$): Այդ կետերի համար $\xi_n = 0$, հետևաբար նրանք բավարարում են հետևյալ հավասարմանը`

$$t_n \left( \sum_{m \in \mathcal{S}} a_m t_m k(\mathbf{x}_n, \mathbf{x}_m) + b \right) = 1$$

որտեղ $\mathcal{S}$-ը բոլոր support vctor-ների բազմությունն է:

Գործնականում, հաշվարկային սխալներից խուսափելու և ավելի ճշգրիտ արդյունք ստանալու համար, $b$-ն հաշվում են ոչ թե մեկ կետի համար, այլ միջինացնում են բոլոր այն կետերի միջև, որոնք գտնվում են մարժայի վրա ($n \in \mathcal{M}$): Վերջնական բանաձևն է՝

$$b = \frac{1}{N_{\mathcal{M}}} \sum_{n \in \mathcal{M}} \left( t_n - \sum_{m \in \mathcal{S}} a_m t_m k(\mathbf{x}_n, \mathbf{x}_m) \right)$$

Multiclass SVMs

Հենակետային վեկտորների մեթոդը (SVM) ի սկզբանե նախատեսված է երկդասային (binary) դասակարգման համար (օրինակ՝ «այո» կամ «ոչ»): Սակայն գործնականում հաճախ հանդիպում են խնդիրներ, որտեղ դասերի քանակը $K > 2$ է: Որպեսզի SVM-ը կարողանա աշխատել բազմաթիվ դասերի հետ, օգտագործվում են մի քանի երկդասային մոդելների համակցություններ:

One-versus-the-rest (Մեկը բոլորի դեմ) մոտեցումըՍա ամենատարածված մեթոդներից մեկն է: Այն աշխատում է հետևյալ կերպ.Կառուցվում է $K$ հատ առանձին SVM մոդել:Յուրաքանչյուր $k$-րդ մոդելը ($y_k(x)$) մարզվում է այնպես, որ $C_k$ դասի տվյալները համարվում են դրական, իսկ մնացած բոլոր $K-1$ դասերի տվյալները միասին՝ բացասական:

Այս մոտեցումն ունի թերություն. հնարավոր է իրավիճակ, երբ միևնույն մուտքային տվյալը ($x$) միաժամանակ վերագրվի մի քանի դասերի կամ ոչ մի դասի:Այս խնդիրը լուծելու համար կիրառվում է հետևյալ բանաձևը՝
$$y(\mathbf{x}) = \max_{k} y_k(\mathbf{x})$$

Սա նշանակում է, որ մենք ընտրում ենք այն դասը, որի համար համապատասխան SVM-ը տվել է ամենաբարձր արժեքը:

Սակայն քանի որ յուրաքանչյուր SVM մարզվել է տարբեր տվյալների բազայի վրա (տարբեր «առաջադրանքներով»), նրանց տված իրական արժեքները ($y_k(x)$) կարող են ունենալ տարբեր սանդղակներ (scales):Օրինակ՝ մի դասակարգիչի համար «0.5»-ը կարող է նշանակել շատ բարձր վստահություն, իսկ մյուսի համար՝ շատ ցածր: Հետևաբար, պարզապես առավելագույն արժեքը ($max$) վերցնելը միշտ չէ, որ արդարացի է կամ ճշգրիտ:

Այսինքն՝բազմադասային դասակարգում ստանալու համար մենք ստեղծում ենք շատ փոքր SVM-ներ, բայց պետք է զգույշ լինենք, քանի որ դրանց արդյունքները կարող են իրար հետ համեմատելի չլինել տարբեր չափողականության պատճառով:

«Մեկը բոլորի դեմ» (one-versus-the-rest) մոտեցման հիմնական թերություններից մեկն այն է, որ մարզման հավաքածուները դառնում են անհամաչափ:Օրինակ՝ Եթե ունենք 10 դաս, և յուրաքանչյուրը պարունակում է հավասար քանակությամբ տվյալներ, ապա յուրաքանչյուր SVM-ի համար դրական օրինակները կլինեն ընդամենը 10%, իսկ բացասականները (մնացած 9 դասերը միասին)՝ 90%:Սա խաթարում է խնդրի սիմետրիան: Լուծումներից մեկն այն է, որ փոխում են թիրախային արժեքները. դրական դասի համար սահմանվում է $+1$, իսկ բացասականի համար՝ $-1/(K-1)$:

One-versus-one (Մեկը մեկի դեմ) մոտեցումը

Սա այլընտրանքային մեթոդ է, որտեղ SVM-ներ են կառուցվում դասերի բոլոր հնարավոր զույգերի համար:

Քանակը: Եթե ունենք $K$ դաս, ապա կառուցվում է $\frac{K(K-1)}{2}$ հատ դասակարգիչ:

Ինչպես է աշխատում: Յուրաքանչյուր մոդել համեմատում է միայն երկու դաս (օրինակ՝ Դաս 1-ը ընդդեմ Դաս 2-ի):

Որոշման կայացում: Նոր տվյալի դասակարգման համար օգտագործվում է «քվեարկության» սկզբունքը. այն դասը, որը կհավաքի ամենաշատ «ձայները» բոլոր զույգային համեմատությունների արդյունքում, համարվում է հաղթող:

Միադասանի SVM-ներ (Single-class SVMs)

Կիրառումը: Հիմնականում օգտագործվում է նորմալ տվյալների շրջանը գտնելու համար: Եթե նոր տվյալը դուրս է մնում այդ սահմանից, այն համարվում է անոմալիա (outlier detection): Այն փորձում է գտնել տվյալների բարձր խտություն ունեցող տիրույթը սահմանափակող հարթ կորը:



SVM for Regression
Այժմ մենք տարածում ենք SVM-ը ռեգրեսիոն խնդիրների վրա՝ միևնույն ժամանակ պահպանելով նոսրության (sparseness) հատկությունը: Պարզ գծային ռեգրեսիայի դեպքում մենք մինիմիզացնում ենք ռեգուլյարացված սխալի ֆունկցիան, որը տրվում է հետևյալ տեսքով.

$$\frac{1}{2} \sum_{n=1}^{N} \{y_n - t_n\}^2 + \frac{\lambda}{2} \| \mathbf{w} \|^2$$

Առաջին մասը քառակուսային սխալն է (Mean Squared Error)։ Այստեղ ցանկացած շեղում կանխատեսված $y_n$ արժեքի և իրական $t_n$ թիրախի միջև բերում է սխալի աճի։

Երկրորդ մասը ($L_2$ regularization) օգտագործվում է մոդելը չափից ավելի բարդացնելուց (overfitting) խուսափելու համար։

Խնդիրը: Սովորական ռեգրեսիան «նոսր» (sparse) չէ։ Դա նշանակում է, որ բոլոր տվյալների կետերը հավասարապես մասնակցում են սխալի հաշվարկին, ինչը կարող է մոդելը դարձնել զգայուն աղմուկի նկատմամբ։

Սովորական ռեգրեսիայի ժամանակ բոլոր տվյալները (ասենք՝ 1000 կետ) ազդում են գծի դիրքի վրա։ SVR-ի դեպքում մենք ուզում ենք այնպիսի լուծում, որտեղ միայն առանցքային կետերը (support vectors) կորոշեն արդյունքը։ Սա թույլ է տալիս խնայել հիշողությունը և հաշվողական ռեսուրսները, քանի որ մոդելը «հիշում է» միայն ամենակարևոր օրինակները։

Փակ ձևի (closed-form) լուծումներ ստանալու նպատակով քառակուսային սխալի ֆունկցիան փոխարինվում է $\epsilon$-անզգայուն ($\epsilon$-insensitive) սխալի ֆունկցիայով։ Այս ֆունկցիան զրոյական սխալ է վերագրում այն դեպքերում, երբ կանխատեսված արժեքի՝ $y(\mathbf{x})$-ի և իրական թիրախային արժեքի՝ $t$-ի միջև բացարձակ տարբերությունը փոքր է $\epsilon$-ից, որտեղ $\epsilon > 0$։ Այլ կերպ ասած՝ մինչև $\epsilon$ չափի շեղումները չեն տուգանվում։

$\epsilon$-անզգայուն սխալի ֆունկցիայի ամենապարզ տարբերակներից մեկը սահմանում է գծային տուգանք (linear cost) այն սխալների համար, որոնք գտնվում են անզգայուն տիրույթից դուրս, և տրվում է հետևյալ կերպ․

$$E_{\epsilon}(y(\mathbf{x}) - t) =
\begin{cases}
0, & \text{եթե } |y(\mathbf{x}) - t| < \epsilon \
|y(\mathbf{x}) - t| - \epsilon, & \text{հակառակ դեպքում}
\end{cases}$$

1. Ի՞նչ է $y(\mathbf{x}) - t$-նՍա կանխատեսված արժեքի ($y(\mathbf{x})$) և իրական թիրախային արժեքի ($t$) տարբերությունն է, այսինքն՝ սխալը։ Բացարձակ արժեքը՝ $|y(\mathbf{x}) - t|$, ցույց է տալիս, թե կետը որքանով է հեռու մեր կառուցած ռեգրեսիայի գծից։

2. Առաջին դեպքը. $E = 0$, եթե $|y(\mathbf{x}) - t| < \epsilon$

Սա ամենակարևոր մասն է։Իմաստը: Եթե սխալը (հեռավորությունը գծից) փոքր է $\epsilon$ արժեքից, ապա մոդելը դա համարում է «ներելի» և սխալի ֆունկցիան հավասարեցնում է 0-ի։Վիզուալիզացիա: Պատկերացրեք մի «խողովակ» ռեգրեսիայի գծի շուրջը, որն ունի $2\epsilon$ լայնություն։ Բոլոր այն կետերը, որոնք ընկնում են այս խողովակի ներսում, չեն պատժվում։ Մոդելը դրանց վրա չի կենտրոնանում։Արդյունքը: Սա ապահովում է մոդելի կայունությունը փոքր աղմուկների նկատմամբ։

3. Երկրորդ դեպքը. $E = |y(\mathbf{x}) - t| - \epsilon$, հակառակ դեպքումՍա այն դեպքն է, երբ կետը դուրս է գալիս $\epsilon$-խողովակի սահմաններից։Իմաստը: Եթե կետը դուրս է, մենք սկսում ենք «պատժել» մոդելին։ Սակայն պատիժը ոչ թե ամբողջ հեռավորությունն է, այլ միայն այն մասը, որը դուրս է $\epsilon$ շերտից։Օրինակ: Եթե $\epsilon = 1$, իսկ կետի իրական հեռավորությունը գծից $1.5$ է, ապա սխալը կլինի $1.5 - 1 = 0.5$։

Գծային աճ: Ի տարբերություն սովորական ռեգրեսիայի (որտեղ սխալը քառակուսի է բարձրացվում), այստեղ սխալն աճում է գծային տարբերակով։ Սա նվազեցնում է խիստ շեղվող կետերի (outliers) ազդեցությունը մոդելի վրա։

Ինչո՞ւ է սա կոչվում «նոսր» (sparse) լուծում

Քանի որ խողովակի ներսում գտնվող բոլոր կետերի սխալը $0$ է, դրանք չեն մասնակցում մաթեմատիկական օպտիմիզացիային (դրանց համապատասխանող գործակիցները դառնում են $0$):

Մոդելը «պահում է» միայն այն կետերը, որոնք՝

Գտնվում են հենց խողովակի սահմանի վրա։

Գտնվում են խողովակից դուրս։

Այս սահմանային կետերն էլ հենց Support Vectors-ներն են։ Եթե դուք ունեք 10,000 տվյալ, բայց միայն 50-ն են դուրս կամ սահմանին, ապա ձեր մոդելը կախված կլինի միայն այդ 50 կետից։ Սա մոդելը դարձնում է չափազանց արդյունավետ

Սովորաբար մենք օգտագործում ենք $(y_n - t_n)^2$։ Եթե կանխատեսումը իրականությունից շեղվում է նույնիսկ 0.01-ով, քառակուսային սխալը դա «պատժում է»։
SVR-ն ասում է. «Եթե շեղումը փոքր է $\epsilon$-ից, եկեք համարենք, որ սխալ չկա»։

Էպսիլոն-խողովակի» (Epsilon-tube) գաղափարը

Պատկերացրեք ձեր ռեգրեսիայի գիծը, որի շուրջը կա $\pm\epsilon$ լայնությամբ մի «միջանցք» կամ «խողովակ»։

Խողովակի ներսում: Կետերը համարվում են ճիշտ կանխատեսված։ Դրանց սխալը $0$ է։

Խողովակի սահմանին կամ դրանից դուրս: Այս կետերը կոչվում են Support Vectors։ Միայն սրանք են մասնակցում մոդելի կառուցմանը։ Եթե կետը դուրս է, սխալը հաշվարկվում է որպես հեռավորություն խողովակի սահմանից ($|y-t| - \epsilon$)։


Ինչո՞ւ է սա ավելի լավ, քան սովորական ռեգրեսիան


Կայունություն (Robustness): Քանի որ մենք անտեսում ենք փոքր շեղումները ($\epsilon$-ի շնորհիվ), մոդելը չի փորձում «կպնել» ամեն մի փոքր աղմուկի։Ճկունություն: Օգտագործելով «Kernel trick» (որը SVM-ի հզոր կողմն է), այս նույն տրամաբանությունը կարելի է կիրառել ոչ գծային տվյալների համար՝ ուղիղ գիծը փոխարինելով բարդ կորերով։



Այս գրաֆիկը վիզուալ կերպով ցույց է տալիս այն տարբերությունը, որի մասին խոսում էինք  բանաձևի քննարկման ժամանակ։ 


Այստեղ համեմատվում են սովորական ռեգրեսիան և SVM ռեգրեսիան (SVR)։

1. Առանցքների նշանակությունը

Հորիզոնական առանցք ($z$): Սա ներկայացնում է սխալը կամ շեղումը ($z = y(\mathbf{x}) - t$)։ $0$ կետը նշանակում է կատարյալ կանխատեսում (կետը գտնվում է հենց գծի վրա)։

Ուղղահայաց առանցք ($E(z)$): Սա «պատժի» չափն է կամ սխալի ֆունկցիայի արժեքը։ Որքան բարձր է գիծը, այնքան ավելի շատ է մոդելը «տուժում» այդ շեղման համար։

2. Կանաչ կորը՝ Քառակուսային սխալ (Quadratic Error)

Սա սովորական գծային ռեգրեսիայի (OLS) պատկերն է։

Անընդհատ պատիժ: Ինչպես տեսնում ես, կորը միայն $0$ կետում է դիպչում առանցքին։ Նույնիսկ ամենափոքր շեղման դեպքում (0-ից մի փոքր ձախ կամ աջ) կանաչ գիծը բարձրանում է։

Արագ աճ: Քանի որ սխալը քառակուսի է բարձրացվում, մեծ շեղումների դեպքում պատիժը դառնում է հսկայական (կորը կտրուկ գնում է վերև)։ Սա նշանակում է, որ մոդելը ամեն գնով փորձելու է մոտենալ շատ հեռու գտնվող կետերին (outliers), ինչը երբեմն վնասում է ընդհանուր ճշտությանը։

3. Կարմիր գիծը՝ $\epsilon$-անզգայուն սխալ ($\epsilon$-insensitive error)

Սա հենց մեր քննարկած SVR-ի ֆունկցիան է։

Անզգայունության գոտի ($-\epsilon$ -ից $\epsilon$): Ուշադրություն դարձնենք, որ $-\epsilon$ և $\epsilon$ միջակայքում կարմիր գիծը համընկնում է հորիզոնական առանցքի հետ ($E=0$): Սա այն «խողովակն» է, որի մասին խոսում էինք։ Այս տիրույթում գտնվող կետերը մոդելի համար «անտեսանելի» են։

Գծային աճ: $\epsilon$-ից դուրս կարմիր գիծը բարձրանում է ուղիղ գծով (գծային), այլ ոչ թե կորով։ Սա նշանակում է, որ SVR-ն ավելի հանգիստ է վերաբերվում մեծ շեղումներին, քան քառակուսային ռեգրեսիան։

Ինչու՞ է սա կարևոր պրակտիկայում
Robustness (Կայունություն): Կարմիր ֆունկցիան ավելի քիչ է զգայուն այն կետերի նկատմամբ, որոնք պատահական սխալներ են կամ «աղմուկ» (outliers), քանի որ դրանց քառակուսիով չի պատժում։

Sparseness (Նոսրություն): Քանի որ կա մի ամբողջ հատված, որտեղ սխալը զրո է, մաթեմատիկական հաշվարկների ժամանակ այդ տիրույթի կետերը դուրս են մնում, և մոդելը ստացվում է ավելի թեթև ու արագ։




(SVR) մաթեմատիկական ձևակերպումը

SVM-ի կիրառումը ռեգրեսիոն խնդիրների լուծման համար հիմնված է նոսրության (sparseness) հատկության պահպանման վրա: Ի տարբերություն ավանդական գծային ռեգրեսիայի, որտեղ օգտագործվում է քառակուսային սխալի ֆունկցիան, SVR-ում կիրառվում է $\epsilon$-անզգայուն սխալի ֆունկցիա:

Օպտիմիզացիայի նպատակային ֆունկցիան
SVR մոդելի կառուցման նպատակը հետևյալ ռեգուլյարացված սխալի ֆունկցիայի մինիմիզացումն է՝

$$C \sum_{n=1}^{N} E_{\epsilon}(y(\mathbf{x}_n) - t_n) + \frac{1}{2} \| \mathbf{w} \|^2$$

Այստեղ $C$ պարամետրը կարգավորում է հավասարակշռությունը մոդելի հարթության ($\| \mathbf{w} \|^2$) և թույլատրելի սխալների միջև:


$\epsilon$-անզգայուն սխալի ֆունկցիա
Այս ֆունկցիան թույլ է տալիս անտեսել այն շեղումները, որոնք փոքր են նախապես սահմանված $\epsilon$ արժեքից՝ ստեղծելով մի «անզգայունության գոտի» ռեգրեսիայի գծի շուրջ՝

$$E_{\epsilon}(y(\mathbf{x}) - t) =
\begin{cases}
0, & \text{եթե } |y(\mathbf{x}) - t| < \epsilon \
|y(\mathbf{x}) - t| - \epsilon, & \text{հակառակ դեպքում}
\end{cases}$$

Թուլացնող փոփոխականներ (Slack Variables)

Որպեսզի հնարավոր լինի մոդելավորել կետերի շեղումները $\epsilon$-անզգայուն գոտուց (խողովակից) դուրս, ներմուծվում են երկու օժանդակ դրական փոփոխականներ՝ $\xi_n$ և $\hat{\xi}_n$՝

$\xi_n \ge 0$՝ արտահայտում է շեղումը խողովակից վերև ($t_n > y(\mathbf{x}_n) + \epsilon$):

$\hat{\xi}_n \ge 0$՝ արտահայտում է շեղումը խողովակից ներքև ($t_n < y(\mathbf{x}_n) - \epsilon$):

Սահմանափակումներ (Constraints)Յուրաքանչյուր տվյալի կետի համար պետք է տեղի ունենան հետևյալ անհավասարությունները՝

$$t_n \le y(\mathbf{x}_n) + \epsilon + \xi_n$$

$$t_n \ge y(\mathbf{x}_n) - \epsilon - \hat{\xi}_n$$

Այս մոտեցումը հնարավորություն է տալիս ձևակերպել ռեգրեսիայի խնդիրը որպես քառակուսային ծրագրավորման խնդիր: Արդյունքում ստացված մոդելը կախված է լինում միայն այն կետերից, որոնք գտնվում են $\epsilon$-խողովակի սահմանին կամ դրանից դուրս: Այդ կետերն էլ հենց հանդիսանում են Suport վեկտորները, ինչն ապահովում է մոդելի արդյունավետությունը և կայունությունը աղմուկի նկատմամբ:



Հենակետային վեկտորների ռեգրեսիայի էությունը պատկերված է վերոհիշյալ գրաֆիկում, որտեղ կարմիր կորը ներկայացնում է մեր կանխատեսող ֆունկցիան՝ $y(x)$: Կորի շուրջը գտնվող վարդագույն շերտը հենց $\epsilon$-անզգայուն խողովակն է, որի լայնությունը սահմանվում է որպես $y \pm \epsilon$:

Կետերի դասակարգումը ըստ դիրքի`

1. Խողովակի ներսում գտնվող կետեր. Այս կետերի համար սխալը համարվում է զրոյական: Մաթեմատիկորեն դա նշանակում է, որ երկու slack փոփոխականներն էլ հավասար են զրոյի՝ $\xi = \hat{\xi} = 0$: Այս տվյալները չեն մասնակցում մոդելի կառուցմանը և չեն հանդիսանում suport վեկտորներ:

2. Խողովակից վերև գտնվող կետեր. Եթե տվյալի կետը գտնվում է խողովակի վերին սահմանից ($y + \epsilon$) բարձր, ապա ներմուծվում է $\xi > 0$ փոփոխականը: Այն ցույց է տալիս կետի ուղղահայաց հեռավորությունը խողովակի վերին սահմանից: Այս դեպքում $\hat{\xi} = 0$:

3. Խողովակից ներքև գտնվող կետեր. Եթե կետը գտնվում է ստորին սահմանից ($y - \epsilon$) ցածր (ինչպես երևում է կապույտ կետի օրինակով), ապա $\hat{\xi} > 0$, իսկ $\xi = 0$: $\hat{\xi}$-ն չափում է հեռավորությունը խողովակի ստորին եզրից:

Մոդելի վերջնական տեսքը որոշվում է բացառապես այն կետերով, որոնց համար $\xi > 0$ կամ $\hat{\xi} > 0$ (խողովակից դուրս գտնվողներ), ինչպես նաև այն կետերով, որոնք գտնվում են հենց խողովակի սահմանների վրա: Սա ապահովում է մոդելի կայունությունը տվյալների բազմության հիմնական մասում առկա փոքր տատանումների նկատմամբ և թույլ է տալիս ստանալ առավել ընդհանրացված լուծում:

Դիտարկենք այն ֆունկցիան, որը պետք է մինիմալացնել՝

$$C \sum_{n=1}^{N} (\xi_n + \widehat{\xi}_n) + \frac{1}{2} \|\mathbf{w}\|^2$$

$\frac{1}{2} \|\mathbf{w}\|^2$: Սա կանոնավորման (regularization) անդամն է, որը ձգտում է մոդելը դարձնել ավելի «հարթ»՝ խուսափելու համար գերուսուցումից (overfitting):

$\xi_n$ և $\widehat{\xi}_n$: Սրանք կոչվում են «թույլ տվող» փոփոխականներ (slack variables): SVR-ում մենք սահմանում ենք մի «խողովակ» (tube) $\epsilon$ շառավղով: Եթե կետը գտնվում է այդ խողովակի մեջ, սխալը զրո է: Եթե կետը դուրս է գալիս խողովակից վերև ($\xi_n$) կամ ներքև ($\widehat{\xi}_n$), ապա այդ շեղումը հաշվի է առնվում որպես տույժ:

$C$: Սա հիպերպարամետր է, որը հավասարակշռում է մոդելի հարթության և թույլատրելի սխալների միջև:

Լագրանժի ֆունկցիան (Lagrangian)

Որպեսզի լուծվի այս մինիմալացման խնդիրը սահմանափակումների պայմաններում ($\xi_n \geq 0$ և $\widehat{\xi}_n \geq 0$), ներմուծվում են Լագրանժի բազմապատկիչներ ($a_n, a_n^*, \mu_n, \mu_n^*$):

$$L = C \sum_{n=1}^{N} (\xi_n + \hat{\xi}_n) + \frac{1}{2} \|\mathbf{w}\|^2 - \sum_{n=1}^{N} (\mu_n \xi_n + \hat{\mu}_n \hat{\xi}_n) - \sum_{n=1}^{N} a_n (\epsilon + \xi_n + y_n - t_n) - \sum_{n=1}^{N} \hat{a}_n (\epsilon + \hat{\xi}_n - y_n + t_n)$$

Որտեղ $y_n = \mathbf{w}^T \phi(\mathbf{x}_n) + b$: Եթե տեղադրենք $y_n$-ի արժեքը, բանաձևը կստանա իր վերջնական մանրամասն տեսքը՝

$$L = C \sum_{n=1}^{N} (\xi_n + \hat{\xi}_n) + \frac{1}{2} \|\mathbf{w}\|^2 - \sum_{n=1}^{N} (\mu_n \xi_n + \hat{\mu}_n \hat{\xi}_n) - \sum_{n=1}^{N} a_n (\epsilon + \xi_n + \mathbf{w}^T \phi(\mathbf{x}_n) + b - t_n) - \sum_{n=1}^{N} \hat{a}_n (\epsilon + \hat{\xi}_n - \mathbf{w}^T \phi(\mathbf{x}_n) - b + t_n)$$



Բաղադրիչների նշանակությունը.
* $C \sum (\xi_n + \hat{\xi}_n) + \frac{1}{2} \|\mathbf{w}\|^2$: Սա հիմնական նպատակային ֆունկցիան է (մինիմալացվող մասը):

* $a_n, \hat{a}_n, \mu_n, \hat{\mu}_n$: Լագրանժի բազմապատկիչներն են (բոլորը $\ge 0$):

* $\epsilon$: Էպսիլոն-անզգայուն գոտու (epsilon-insensitive tube) շառավիղն է: Այս գոտու ներսում գտնվող կետերը տույժի չեն ենթարկվում:

* $\xi_n, \hat{\xi}_n$: Շեղումներն են «խողովակից» (slack variables) դեպի վերև և դեպի ներքև:

* $t_n$: Թիրախային (իրական) արժեքն է:$y_n$: Մոդելի կողմից կանխատեսված արժեքն է

:Այս ֆունկցիան օպտիմալացնելու համար անհրաժեշտ է հաշվել մասնակի ածանցյալներն ըստ $\mathbf{w}, b, \xi_n$ և $\hat{\xi}_n$ փոփոխականների և դրանք հավասարեցնել զրոյի, ինչը մեզ թույլ կտա անցնել դուալ (dual) խնդրին:

Ածանցյալն ըստ $\mathbf{w}$-ի

$$\frac{\partial L}{\partial \mathbf{w}} = 0 \Rightarrow \mathbf{w} = \sum_{n=1}^{N} (a_n - \hat{a}_n) \phi(\mathbf{x}_n)$$

Սա ամենակարևոր արդյունքն է։ Այն ցույց է տալիս, որ օպտիմալ կշիռների վեկտորը ($\mathbf{w}$) հանդիսանում է մուտքային տվյալների ($\phi(\mathbf{x}_n)$) գծային համակցություն։ Սա թույլ է տալիս հետագայում օգտագործել Kernel trick-ը, քանի որ մոդելը կախված է միայն տվյալների սկալյար արտադրյալից։

2. Ածանցյալն ըստ $b$-ի

$$\frac{\partial L}{\partial b} = 0 \Rightarrow \sum_{n=1}^{N} (a_n - \hat{a}_n) = 0$$

Սա լրացուցիչ սահմանափակում է Լագրանժի բազմապատկիչների համար։ Այն ասում է, որ դրական և բացասական շեղումների գործակիցների տարբերությունների գումարը պետք է լինի զրո։


3. Ածանցյալներն ըստ $\xi_n$ և $\hat{\xi}_n$

$$\frac{\partial L}{\partial \xi_n} = 0 \Rightarrow a_n + \mu_n = C$$

$$\frac{\partial L}{\partial \hat{\xi}_n} = 0 \Rightarrow \hat{a}_n + \hat{\mu}_n = C$$

Քանի որ $\mu_n \ge 0$ և $\hat{\mu}_n \ge 0$, այս հավասարումներից հետևում է, որ $a_n$ և $\hat{a}_n$ գործակիցները սահմանափակված են $[0, C]$ միջակայքով։ Սա կոչվում է box constraint։

Այս արդյունքները տեղադրելով սկզբնական Լագրանժիանի մեջ, մենք ազատվում ենք $\mathbf{w}, b, \xi, \hat{\xi}$ փոփոխականներից և ստանում ենք Դուալ խնդիրը (Dual Problem)։ Դուալ խնդիրը կախված է լինելու միայն $a_n$ և $\hat{a}_n$ փոփոխականներից, ինչն անհամեմատ ավելի հեշտ է լուծել քառակուսային ծրագրավորման (Quadratic Programming) մեթոդներով։Բացի այդ, շատ կարևոր է նկատել, որ կետերի մեծ մասի համար $a_n$ և $\hat{a}_n$ լինելու են զրո։ Միայն այն կետերը, որոնց համար այս գործակիցները զրոյից տարբեր են, կոչվում են Հենակետային վեկտորներ (Support Vectors), և հենց դրանք են որոշում ռեգրեսիայի գիծը։

Նախորդ քայլերում ստացված ածանցյալները տեղադրելով Լագրանժիանի մեջ, մենք ստանում ենք նոր ֆունկցիա ($\tilde{L}$), որն այժմ պետք է մաքսիմալացնել միայն $a_n$ և $\hat{a}_n$ փոփոխականների 

նկատմամբ՝


$$\tilde{L}(\mathbf{a}, \hat{\mathbf{a}}) = -\frac{1}{2} \sum_{n=1}^{N} \sum_{m=1}^{N} (a_n - \hat{a}_n)(a_m - \hat{a}_m) k(\mathbf{x}_n, \mathbf{x}_m) - \epsilon \sum_{n=1}^{N} (a_n + \hat{a}_n) + \sum_{n=1}^{N} (a_n - \hat{a}_n) t_n$$


Այս բանաձևի գեղեցկությունը Kernel trick-ի ($k(\mathbf{x}_n, \mathbf{x}_m)$) կիրառման մեջ է։ Մենք կարիք չունենք ուղղակիորեն հաշվելու $\phi(\mathbf{x})$ վեկտորները (որոնք կարող են լինել անվերջ չափանի):Մեզ անհրաժեշտ է միայն տվյալների զույգերի սկալյար արտադրյալը կեռնելի միջոցով: Սա թույլ է տալիս լուծել խիստ ոչ գծային ռեգրեսիայի խնդիրներ:

Սահմանափակումները (Box Constraints)

Մաքսիմալացումը կատարվում է հետևյալ խիստ պայմաններում՝

$0 \le a_n \le C$
$0 \le \hat{a}_n \le C$

Այս $C$ պարամետրը կառավարում է հավասարակշռությունը՝ մոդելի պարզության (կանոնավորման) և մարզման տվյալներին առավելագույնս մոտ լինելու միջև:

$$y(\mathbf{x}) = \sum_{n=1}^{N} (a_n - \hat{a}_n) k(\mathbf{x}, \mathbf{x}_n) + b$$

Սա մոդելի վերջնական տեսքն է։ Նոր $x$ մուտքագրելիս մոդելը հաշվում է դրա նմանությունը (կեռնելի միջոցով) բոլոր մարզման կետերի հետ։ Սակայն, քանի որ կետերի մեծ մասի համար $(a_n - \hat{a}_n) = 0$ է, կանխատեսումը կախված է լինում միայն հենակետային վեկտորներից։

Կարուշ-Կուն-Տակերի (KKT) պայմաններ

KKT պայմանները մաթեմատիկորեն սահմանում են, թե որ դեպքում Լագրանժի բազմապատկիչները ($a_n, \hat{a}_n$) պետք է լինեն զրո։ 

Սրանք ամենակարևոր պայմաններն են, որոնք պետք է տեղի ունենան օպտիմալ կետում։ Դրանք ցույց են տալիս կապը Լագրանժի բազմապատկիչների և սահմանափակումների միջև՝

$$a_n (\epsilon + \xi_n + y_n - t_n) = 0$$
$$\hat{a}_n (\epsilon + \hat{\xi}_n - y_n + t_n) = 0$$
$$(C - a_n) \xi_n = 0$$
$$(C - \hat{a}_n) \hat{\xi}_n = 0$$

Այս պայմաններից հետևում են շատ կարևոր եզրակացություններ՝

$a_n (\epsilon + \xi_n + y_n - t_n) = 0$Եթե կետը գտնվում է $\epsilon$-խողովակի ներսում, ապա $(\epsilon + \dots) \neq 0$, հետևաբար $a_n$-ը պարտադիր պետք է լինի 0։ Նույնը վերաբերում է $\hat{a}_n$-ին։ Սա նշանակում է, որ խողովակի ներսի կետերը մոդելի վրա ոչ մի ազդեցություն չունեն։

$(C - a_n) \xi_n = 0$ Եթե կետը դուրս է եկել խողովակից ($\xi_n > 0$), ապա $a_n$-ը պետք է հավասար լինի $C$-ի։ Սրանք այն կետերն են, որոնք մոդելը դժվարությամբ է մոտարկում (outliers կամ եզրային կետեր)։


Support Vector Regression (SVR) մոդելի կառուցման վերջնական փուլում առանցքային նշանակություն ունեն Լագրանժի բազմապատկիչների փոխազդեցությունը և մոդելի երկրաչափական սահմանափակումները:

Մաթեմատիկական օպտիմալացման արդյունքում տվյալների յուրաքանչյուր $(\mathbf{x}_n, t_n)$ զույգի համար ստացվում են երկու գործակիցներ՝ $a_n$ և $\hat{a}_n$: Դրանց միջև գործում է փոխբացառման կանոնը՝
Ցանկացած կետի համար կա՛մ $a_n = 0$, կա՛մ $\hat{a}_n = 0$, կա՛մ երկուսն էլ հավասար են զրոյի:Սա ունի պարզ ֆիզիկական բացատրություն․ տվյալը չի կարող միաժամանակ գտնվել $\epsilon$-անզգայուն խողովակի և՛ վերին, և՛ ստորին սահմաններից դուրս:

SVR-ի հիմնական առավելությունը նրա նոսրությունն է (sparsity): Մոդելը «հիշում» է միայն այն կետերը, որոնք անմիջականորեն ազդում են ռեգրեսիայի գծի ձևավորման վրա։
* Support Vectors: Այն կետերն են, որոնք ընկած են $\epsilon$-խողովակի սահմանի վրա կամ դրանից դուրս ($a_n \neq 0$ կամ $\hat{a}_n \neq 0$):
* Ներքին կետեր: Խողովակի ներսում գտնվող բոլոր կետերի համար $a_n = \hat{a}_n = 0$: Սրանք որևէ կերպ չեն մասնակցում կանխատեսմանը, ինչը թույլ է տալիս ալգորիթմին աշխատել մեծածավալ տվյալների հետ՝ պահպանելով միայն կարևորագույն վեկտորները:

$b$ պարամետրի որոշման համար օգտագործվում են այն հենակետային վեկտորները, որոնք գտնվում են հենց խողովակի սահմանի վրա (երբ $0 < a_n < C$): Այս դեպքում շեղման փոփոխականը $\xi_n = 0$ է, և տեղի ունի հետևյալ պայմանը՝
    $$\epsilon + y_n - t_n = 0 \implies \epsilon + (\mathbf{w}^T \phi(\mathbf{x}_n) + b) - t_n = 0$$

Այստեղից ստացվում է $b$-ի հաշվարկման վերջնական բանաձևը՝ կեռնելի միջոցով։
    $$b = t_n - \epsilon - \sum_{m=1}^{N} (a_m - \hat{a}_m) k(\mathbf{x}_n, \mathbf{x}_m)$$

Նշում: Գործնականում, հաշվողական ճշգրտությունն ապահովելու համար, $b$-ն հաշվարկվում է բոլոր հնարավոր սահմանային կետերի համար, և վերջնական արժեքը վերցվում է որպես դրանց միջին թվաբանական:



«$\nu$-SVM ռեգրեսիայի պատկերում՝ կիրառված սինուսոիդալ սինթետիկ տվյալների բազայի վրա՝ օգտագործելով Գաուսյան միջուկներ (Gaussian kernels): Կանխատեսված ռեգրեսիոն կորը պատկերված է կարմիր գծով, իսկ $\epsilon$-անզգայուն խողովակը (tube) համապատասխանում է ստվերագծված տիրույթին: Նաև տվյալների կետերը պատկերված են կանաչով, իսկ հենակետային վեկտորներ (support vectors) հանդիսացող կետերը նշված են կապույտ շրջանակներով:»

Հիմնական հասկացությունների բացատրությունը

1. $\nu$-SVM (Nu-SVM) ռեգրեսիաՍա սովորական SVR-ի կատարելագործված տեսակն է: Եթե սովորական ռեգրեսիայի ժամանակ մենք ինքներս ենք որոշում «խողովակի» լայնությունը ($\epsilon$), ապա $\nu$-SVM-ի դեպքում $\nu$ պարամետրի միջոցով մոդելը ինքնաբերաբար հաշվում է օպտիմալ լայնությունը: $\nu$-ն սահմանափակում է սխալների քանակը և հենակետային վեկտորների մասնաբաժինը:

2. $\epsilon$-անզգայուն խողովակ ($\epsilon$-insensitive tube)Սա նկարի ստվերագծված հատվածն է: Մոդելի տրամաբանությունն այն է, որ եթե տվյալների կետը գտնվում է այս «խողովակի» ներսում, ապա դրա սխալը համարվում է զրո: Մոդելը ուշադրություն է դարձնում միայն այն կետերին, որոնք դուրս են այս տիրույթից կամ գտնվում են դրա սահմանին:

3. Գաուսյան միջուկներ (Gaussian Kernels/RBF)Քանի որ տվյալները սինուսոիդալ են (ալիքաձև), ուղիղ գիծը չէր կարող դրանք ճշգրիտ նկարագրել: Գաուսյան միջուկը թույլ է տալիս մոդելին լինել ճկուն և կառուցել հարթ, կոր գծեր (ինչպես նկարի կարմիր գիծը):

4. Հենակետային վեկտորներ (Support Vectors)Սրանք այն ամենակարևոր կետերն են (կապույտ շրջանակները), որոնք «պահում» են կորը: Եթե մյուս կանաչ կետերը ջնջենք, բայց այս կապույտները թողնենք, կարմիր կորի տեսքը չի փոխվի: Դրանք հիմնականում գտնվում են «խողովակի» սահմաններին կամ դրանից դուրս:


Այս մեթոդը շատ արդյունավետ է Machine Learning-ում, երբ ցանկանում ենք ստանալ այնպիսի մոդել, որը չի գերուսուցվի (overfitting) և հաշվի չի առնի չնչին աղմուկը տվյալների մեջ՝ շնորհիվ այդ $\epsilon$ «խողովակի»:

Այս մոդելը լավագույնս դրսևորվում է այնպիսի խնդիրներում, որտեղ տվյալները պարունակում են որոշակի «աղմուկ», և մեզ պետք է ստանալ ընդհանրացված կոր, որը հետագայում նոր (չտեսնված) տվյալների վրա կտա ճշգրիտ արդյունքներ (Generalization):

Ոչ գծային SVM դասակարգումը (Non-linear SVM classification)

Եթե նախորդ օրինակը ռեգրեսիայի (կորի կառուցման) մասին էր, ապա այստեղ խոսքը երկու տարբեր խմբերի (կարմիր x-եր և կապույտ x-եր) իրարից բաժանելու մասին է:



Նկարում երևում է, որ կարմիր և կապույտ կետերը հնարավոր չէ բաժանել մեկ ուղիղ գծով: Սակայն, օգտագործելով Գաուսյան միջուկը (Gaussian kernel), մոդելը տվյալները «տեղափոխում է» ավելի բարձր չափողականություն ունեցող տարածություն, որտեղ դրանք դառնում են գծայինորեն բաժանելի: Արդյունքում, սկզբնական 2D տարածության մեջ մենք ստանում ենք բարդ, կորացած սահմանագիծ (Decision Boundary):

SVM-ի ամենակարևոր առանձնահատկություններից է՝

Հենակետային վեկտորներ (Support Vectors): Ուշադրություն դարձրեք կանաչ շրջանակների մեջ առնված կետերին: Միայն այս կետերն են որոշում բաժանարար գծի դիրքը:

Անկախություն մյուս կետերից: Եթե դուք տեղաշարժեք կամ ջնջեք այն կետերը, որոնք շրջանակի մեջ չեն (քանի դեռ դրանք չեն հատում սահմանը), բաժանարար գիծը չի փոխվի: Սա կոչվում է «sparsity» (նոսրություն), քանի որ լուծումը կախված է տվյալների միայն մի փոքր ենթաբազմությունից:

Այս օրինակը ցույց է տալիս, թե ինչպես է SVM-ը կարողանում.

Ճշգրիտ բաժանել բարդ դասավորվածություն ունեցող խմբերը:

Լինել էֆեկտիվ՝ կենտրոնանալով միայն սահմանին մոտ գտնվող «դժվար» կետերի (Support Vectors) վրա:

Ապահովել առավելագույն լայնության միջնապատ (Maximum Margin) խմբերի միջև, ինչը նվազեցնում է սխալվելու հավանականությունը նոր տվյալների դեպքում:

Լրացուցիչ

Գաուսյան կեռնելը (Gaussian Kernel)

Հիմնական գաղափարն այն է, որ կեռնելը չափում է նմանությունը երկու կետերի միջև։ Գաուսյան կեռնելն ասում է հետևյալը. եթե երկու կետեր իրար մոտ են տարածության մեջ, դրանց նմանության աստիճանը բարձր է (մոտ է 1-ին), իսկ եթե հեռու են՝ նմանությունը կտրուկ նվազում է և ձգտում 0-ի։

Այն թույլ է տալիս SVM-ին լուծել ոչ գծային խնդիրներ։ Պատկերացրեք՝ տվյալները դասավորված են շրջանաձև (մի խումբը կենտրոնում է, մյուսը՝ օղակաձև դրսում)։ Ուղիղ գծով դրանք չենք բաժանի։ Գաուսյան կեռնելը «բարձրացնում» է այդ կետերը երրորդ չափողականություն՝ ստեղծելով մի տեսակ «բլուրներ», որտեղ խմբերն արդեն հեշտությամբ բաժանվում են հարթությամբ։

Գաուսյան կեռնելը հաշվարկվում է հետևյալ բանաձևով՝

$$K(x, x') = \exp\left( -\frac{\|x - x'\|^2}{2\sigma^2} \right)$$

Կամ ավելի հաճախ օգտագործվող տեսքը $\gamma$ (գամմա) պարամետրով՝

$$K(x, x') = \exp(-\gamma \|x - x'\|^2)$$

Որտեղ՝$x$ և $x'$: Երկու տվյալների կետերն են (վեկտորները)։

$\|x - x'\|^2$: Էվկլիդեսյան հեռավորության քառակուսին է այդ երկու կետերի միջև։

$\gamma$ (Gamma): Սա ամենակարևոր պարամետրն է, որը որոշում է կեռնելի «ազդեցության շառավիղը»։

3. $\gamma$ (Gamma) պարամետրի դերը

Սա այն «կոճակն» է, որով մենք կարգավորում ենք մոդելի վարքը՝
Մեծ $\gamma$ (Low variance): Կեռնելը դառնում է շատ «սուր»։ Այսինքն՝ մի կետը ազդեցություն ունի միայն իրեն շատ մոտ գտնվող հարևանների վրա։ Սա կարող է բերել Overfitting-ի (գերուսուցման), երբ մոդելը չափից շատ է հարմարվում յուրաքանչյուր կետին։

Փոքր $\gamma$ (High variance): Կեռնելը դառնում է «հարթ» և լայն։ Մի կետի ազդեցությունը տարածվում է հեռու գտնվող կետերի վրա։ Կորը ստացվում է ավելի հարթ և ընդհանրացված։

4. Ինչո՞ւ է այն կոչվում «Անվերջ չափողականություն»

Սա հետաքրքիր տեսական փաստ է։ Մաթեմատիկորեն ապացուցվում է (Թեյլորի շարքի միջոցով), որ Գաուսյան կեռնելն օգտագործելիս մենք տվյալները պրոյեկտում ենք անվերջ չափողականություն ունեցող տարածություն։ Դա է պատճառը, որ այն կարողանում է բաժանել գրեթե ցանկացած բարդության տվյալներ։

Kernel Trick

Մեր «Երկակի ձևակերպումից» (Dual Formulation) մենք գիտենք, որ SVM-ին հետաքրքրում է միայն տվյալների կետերի սկալյար արտադրյալը՝

$$\sum \alpha_i \alpha_j y_i y_j (x_i^T x_j)$$

SVM-ի մաթեմատիկական ձևակերպման մեջ (Dual Formulation) պարզվում է, որ մեզ պետք չեն հենց իրենք՝ տվյալների կետերը, այլ միայն դրանց սկալյար արտադրյալը ($x_i^T x_j$): Սա նշանակում է, որ մոդելին հետաքրքիր է միայն կետերի միջև կապը կամ նմանությունը։

Նաիվ մոտեցումը 

Եթե մեր տվյալները գծային չեն բաժանվում (ինչպես նախորդ նկարներում), մենք ստիպված ենք դրանք պրոյեկտել ավելի բարձր չափողականության մեջ ($x \rightarrow \phi(x)$): Այնուհետև մենք փոխարինում ենք սկալյար արտադրյալը $\phi(x_i)^T \phi(x_j)$-ով:

«Մղձավանջը» (The Nightmare)
Խնդիրը: Եթե ունենք ընդամենը 100 պիքսելանոց պատկեր և ուզում ենք հաշվի առնել բոլոր հնարավոր փոխազդեցությունները (օրինակ՝ 5-րդ աստիճանի պոլինոմ), ապա մեր վեկտորը կունենա 75 միլիոն չափողականություն։

Հաշվարկային աղետ: Նման հսկայական վեկտորներ հաշվելը և դրանց սկալյար արտադրյալը գտնելը կպահանջի ահռելի ժամանակ և համակարգչային հզորություն։


Ի՞նչ է Kernel Trick-ը

Սահմանումը (The Definition): «Կեռնելը» (Kernel) պարզապես $K(x, z)$ ֆունկցիա է, որը հաշվում է $\phi(x)$-ի և $\phi(z)$-ի սկալյար արտադրյալը՝ առանց երբևէ իրականում հաշվարկելու $\phi(x)$-ը կամ $\phi(z)$-ը:

Սա «մաթեմատիկական կարճ ճանապարհն է»։ Կեռնելը ($K(x, z)$) ֆունկցիա է, որը տալիս է մեզ նույն արդյունքը, ինչ կտար բարձր չափողականության մեջ սկալյար արտադրյալ հաշվելը, բայց առանց այդ բարձր չափողականության մեջ իրականում մտնելու։Այսինքն՝ մենք ստանում ենք բարդ, ոչ գծային բաժանարար սահմաններ՝ կատարելով պարզ հաշվարկներ սկզբնական (ցածր) չափողականության մեջ։

Ընդհանուր եզրակացություն

Առանց Kernel Trick-ի: Մենք պետք է կատարեինք $x \rightarrow \phi(x)$ (շատ թանկ է) և հետո հաշվեինք $\phi(x)^T \phi(z)$ (ավելի թանկ է)։

Kernel Trick-ով: Մենք ուղղակի հաշվում ենք $K(x, z)$ (շատ արագ է)։


Դիտարկենք Kernel Trick-ի մաթեմատիկական ապացույցը մի պարզ օրինակի վրա: Այն ցույց է տալիս, թե ինչպես կարելի է հաշվել 3D տարածության սկալյար արտադրյալը՝ մնալով 2D տարածության մեջ:

Մենք ուզում ենք գտնել մի այնպիսի մոգական $K(x, z)$ ֆունկցիա, որը հավասար լինի սովորական սկալյար արտադրյալի քառակուսուն՝

$$K(x, z) = (x^T z)^2$$

Սա մեր «հնարքն» է։ Հիմա արի տեսնենք, թե ինչ տեսք կունենար տվյալների արտապատկերումը ($\phi$), եթե մենք փորձեինք սա բացել սովորական ձևով։

Եթե $x = [x_1, x_2]^T$ և $z = [z_1, z_2]^T$, ապա նրանց սկալյար արտադրյալը կլինի $x_1z_1 + x_2z_2$:
Հիմա այդ ամբողջը բարձրացնենք քառակուսի (օգտվելով $(a+b)^2 = a^2 + 2ab + b^2$ բանաձևից)`

$$(x_1z_1 + x_2z_2)^2 = (x_1z_1)^2 + 2(x_1z_1)(x_2z_2) + (x_2z_2)^2$$

$$= x_1^2z_1^2 + 2x_1z_1x_2z_2 + x_2^2z_2^2$$

Որպեսզի սա ներկայացնենք որպես երկու նոր վեկտորների սկալյար արտադրյալ ($\phi(x)^T \phi(z)$), մենք պետք է այն գրենք այսպես՝

$$(x_1^2)(z_1^2) + (x_2^2)(z_2^2) + (\sqrt{2}x_1x_2)(\sqrt{2}z_1z_2)$$


Նկատենք, որ $2x_1z_1x_2z_2$ անդամը մենք բաժանեցինք երկու մասի՝ $\sqrt{2}x_1x_2$ և $\sqrt{2}z_1z_2$, որպեսզի երբ դրանք բազմապատկվեն, նորից ստանանք $2$:

Այստեղից էլ հենց երևում է, թե ինչ տեսք պետք է ունենա մեր նոր վեկտորը 3D տարածությունում, որպեսզի նրա բաղադրիչների արտադրյալը տա վերը նշված արդյունքը՝

$$\phi(x) = \begin{bmatrix} x_1^2 \\ x_2^2 \\ \sqrt{2}x_1x_2 \end{bmatrix}$$

Վեկտորները (The Vectors): Դիցուք $x = [x_1, x_2]^T$ և $z = [z_1, z_2]^T$:

Հատկանիշների արտապատկերումը (The Feature Map): Սահմանենք կոնկրետ $\phi$ արտապատկերում, որը մեր 2D վեկտորները տեղափոխում է 3D տարածություն՝
$$\phi(x) = [x_1^2, x_2^2, \sqrt{2}x_1x_2]^T$$

Երկար ճանապարհը (սկալյար արտադրյալ հատկանիշների տարածությունում):

$$\phi(x)^T\phi(z) = (x_1^2)(z_1^2) + (x_2^2)(z_2^2) + (\sqrt{2}x_1x_2)(\sqrt{2}z_1z_2)$$

$$\phi(x)^T\phi(z) = x_1^2z_1^2 + x_2^2z_2^2 + 2x_1x_2z_1z_2$$


Վերլուծումը արտադրիչների (The Factorization): Նկատեք, որ վերևի արդյունքը լրիվ քառակուսի է.$$x_1^2z_1^2 + x_2^2z_2^2 + 2x_1x_2z_1z_2 = (x_1z_1 + x_2z_2)^2$$


նարքը (The Trick): Փակագծերի ներսի արտահայտությունը հենց սկզբնական 2D սկալյար արտադրյալն է.$$\phi(x)^T\phi(z) = (x^Tz)^2 = K(x, z)$$

3D-ում սկալյար արտադրյալը գտնելու համար մենք պարզապես հաշվեցինք սկալյար արտադրյալը 2D-ում և այն բարձրացրեցինք քառակուսի: $O(D^2)$ բարդությունը կրճատվեց մինչև $O(D)$: Երբ չափողականությունները մեծանում են, այս հնարքը խնայում է միլիարդավոր հաշվարկներ:



Պատկերացրու՝ ունես երկու տարբեր սորտի խնձորներ, որոնք սեղանի վրա խառնված են։ Դու չես կարող մի ուղիղ քանոն դնել սեղանին այնպես, որ մի կողմում մնան միայն կարմիրները, մյուս կողմում՝ կանաչները, որովհետև նրանք շրջանաձև խառնված են։1. Ի՞նչն է «սարսափելի գաղափարը»Որպեսզի դրանք բաժանես, դու պետք է խնձորները «թռցնես» օդ (բարձրացնես 3-րդ չափողականություն)։ Օդում դու կարող ես մի թուղթ մտցնել խնձորների արանքով ու բաժանել դրանք։Բայց եթե խնձորները միլիոնավոր են, ամեն մեկին օդ բարձրացնելը և այնտեղ դրանց իրար հետ համեմատելը սարսափելի շատ էներգիա ու ժամանակ կխլի։ Համակարգիչը պարզապես «կկախի»։2. Ի՞նչ է անում Kernel Trick-ըKernel Trick-ը նման է մի մոգական բանաձևի։ Դու դեռ սեղանի մոտ նստած ես (2D տարածությունում), խնձորները սեղանին են, բայց դու ունես մի հաշվարկ, որը քեզ ասում է.«Եթե ես այս խնձորները օդ բարձրացնեի, նրանց հեռավորությունն իրարից կլիներ այսքան»։Այսինքն՝ դու ստանում ես օդում եղած արդյունքը՝ առանց խնձորները իրականում օդ բարձրացնելու։3. Սլայդի օրինակըՍլայդի վրա գրված է.Մենք ունենք $x$ և $z$ (մեր խնձորները 2D-ում)։Մենք ուզում ենք իմանալ դրանց կապը 3D-ում ($\phi(x)^T \phi(z)$)։Փոխանակ ամեն մեկին հատ-հատ սարքենք 3D ու հետո բազմապատկենք (երկար ճանապարհ), մենք ուղղակի բազմապատկում ենք դրանք 2D-ում ու արդյունքը քառակուսի բարձրացնում ($(x^T z)^2$)։Այդ քառակուսի բարձրացնելը հենց այդ «մոգական բանաձևն» է (Kernel-ը)։ Այն քեզ տալիս է նույն պատասխանը, ինչ կտար 3D-ի բարդ հաշվարկը։

Մերսերի թեորեմը (Mercer’s Theorem)

Մերսերի թեորեմը (Mercer’s Theorem)

Հարցը (The Question): Եթե մեզ պետք չէ հաշվել $\phi(x)$-ը, կարո՞ղ եմ ես պարզապես հնարել ցանկացած ֆունկցիա, օրինակ՝ $K(x, z) = \sin(x^T z)$, և օգտագործել այն որպես SVM կեռնել։


Պատասխանը (The Answer): Ոչ։ $K$ ֆունկցիան վավեր կեռնել է միայն այն դեպքում, եթե այն համապատասխանում է սկալյար արտադրյալին ինչ-որ օրինական հատկանիշների տարածությունում։

Մերսերի թեորեմը (Mercer's Theorem): $K(x, z)$ ֆունկցիան վավեր կեռնել է այն և միայն այն դեպքում, եթե ցանկացած վերջավոր տվյալների բազայի $\{x_1, \dots, x_n\}$ համար ստացված $N \times N$ չափանի Կեռնել մատրիցը (կամ Գրամի մատրիցը) Դրական կիսաորոշյալ է (Positive Semi-Definite - PSD):

Հիշեցում Գծային հանրահաշվից: Մատրիցը PSD է, եթե նրա սեփական արժեքները (eigenvalues) ոչ բացասական են, ինչը նշանակում է $v^T K v \geq 0$ ցանկացած ոչ զրոյական $v$ վեկտորի համար։

Ինտուիցիան (The Intuition): Սկալյար արտադրյալը չափում է վեկտորների միջև անկյան կոսինուսը (հեռավորության կամ նմանության չափանիշ)։ Եթե մատրիցը PSD չէ, դա նշանակում է, որ ենթադրյալ հատկանիշների տարածությունը պարունակում է «բացասական երկարություններ» կամ երևակայական երկրաչափություններ։ Մերսերի թեորեմը երաշխավորում է, որ մեր մաթեմատիկան մնում է հիմնավորված իրական Հիլբերտյան տարածությունում:

Պատկերացնենք՝ դու ցանկանում ես ստեղծել նմանության մի չափանիշ (կեռնել)։ Մերսերի թեորեմն ասում է, որ քո «նմանության կանոնը» պետք է լինի տրամաբանական:

Եթե մատրիցը PSD չլինի, դա նման կլինի մի իրավիճակի, որտեղ՝

Ա կետը նման է Բ-ին:

Բ կետը նման է Գ-ին:

Բայց Ա-ն ու Գ-ն այնքան աննման են, որ նրանց հեռավորությունը ստացվում է «բացասական» կամ անհնարին:

Մաթեմատիկորեն, SVM-ը փորձում է լուծել օպտիմիզացիայի խնդիր (գտնել մաքսիմալ միջնապատ): Եթե քո կեռնելը չբավարարի Մերսերի պայմանին, ապա այդ խնդիրը կարող է չունենալ լուծում, կամ ալգորիթմը կարող է երբեք կանգ չառնել (չկոնվերգացվել):

Մենք ինքներս ենք  որոշում, թե որ ֆունկցիան օգտագործենք՝ հիմնվելով  տվյալների բնույթի վրա՝

* Linear Kernel (Գծային): Եթե տվյալներդ արդեն պարզ են և գծով բաժանվող։

* Polynomial Kernel (Պոլինոմիալ): Եթե գիտես, որ տվյալների միջև կապը քառակուսային է կամ խորանարդային (ինչպես սլայդի 3D-ի օրինակում)։

* RBF / Gaussian Kernel: Սա «ամենագնացն» է։ Եթե չգիտես ինչ ընտրել, սովորաբար սա լավագույնն է աշխատում, քանի որ այն կարող է հարմարվել գրեթե ցանկացած բարդ կորի։

ամենատարախվածներն են։


Ֆունկցիան ընտրելուց հետո դու պետք է որոշես նաև դրա «ուժգնությունը»։ Օրինակ՝ Գաուսյան կեռնելի դեպքում դու պետք է որոշես $\gamma$ (gamma)-ն.Եթե $\gamma$-ն շատ մեծ դնես, մոդելը կդառնա շատ կպչուն ու կփորձի ամեն մի կետին առանձին հարմարվել (Overfitting):Եթե $\gamma$-ն շատ փոքր դնես, մոդելը կդառնա չափից շատ պարզ ու կկորցնի կարևոր դետալները (Underfitting):


RBF Kernel (Radial Basis Function)

Եկեք դիտարկենք ամենատարածված կեռնելը՝ Գաուսյան շառավղային բազիսային ֆունկցիան (RBF):

$$K(x, z) = e^{-\gamma \|x - z\|^2}$$

Պնդումը (The Claim): RBF կեռնելի կողմից ենթադրվող $\phi(x)$ հատկանիշների տարածությունն ունի անվերջ թվով չափողականություններ։ 

Ինչպե՞ս է սա հնարավոր։Ապացույցը (The Proof): Եկեք պարզեցնենք և դիտարկենք 1D սկալյար կեռնել $K(x, z) = e^{xz}$:

Վերլուծենք $e^{xz}$-ը՝ օգտագործելով նրա անվերջ Թեյլորի շարքը՝

$$e^{xz} = 1 + xz + \frac{(xz)^2}{2!} + \frac{(xz)^3}{3!} + \dots + \frac{(xz)^\infty}{\infty!}$$

Վերլուծությունը (The Breakdown): Այս անվերջ գումարի յուրաքանչյուր անդամ պոլինոմիալ կեռնել է։

$$e^{xz} = (1)(1) + (x)(z) + \left(\frac{x^2}{\sqrt{2!}}\right)\left(\frac{z^2}{\sqrt{2!}}\right) + \left(\frac{x^3}{\sqrt{3!}}\right)\left(\frac{z^3}{\sqrt{3!}}\right) + \dots$$

Իրականությունը (The Reality): Օգտագործելով $e^x$ ֆունկցիան՝ RBF կեռնելը հաշվում է սկալյար արտադրյալը մի $\phi(x)$ վեկտորի համար, որը պարունակում է $x$-ի բոլոր հնարավոր պոլինոմիալ աստիճանները՝ ձգվելով մինչև անվերջություն։

Kernel Trick-ի իսկական մոգությունը: Մենք գտնում ենք օպտիմալ հիպերհարթություն անվերջ չափողական տարածությունում, և դա մեզ վրա նստում է ընդամենը մեկ $O(1)$ էքսպոնենցիալ գործողություն՝ յուրաքանչյուր զույգ կետի համար։

Սա մաթեմատիկական հանճարեղ լուծում է։  Նախորդ խնդրում, որտեղ ասվում էր, որ 5-րդ աստիճանի պոլինոմի համար պետք է 75 միլիոն չափողականություն, ապա պատկերացրու, թե քանի միլիարդ չափողականություն պետք կլիներ 100-րդ կամ 1000-րդ աստիճանի համար։RBF կեռնելն ասում է. «Ես ձեզ կտամ անվերջ աստիճանի պոլինոմիալ ճկունություն»։ Բայց փոխանակ իրականում հաշվելու այդ անվերջ թվերը, նա ուղղակի օգտագործում է $e^x$ ֆունկցիան, որը շատ արագ է հաշվարկվում։Արդյունքում մենք ստանում ենք մի մոդել, որը կարող է հարմարվել ցանկացած բարդության տվյալների, որովհետև նրա մաթեմատիկական «զինանոցը» անվերջ է։ Սա է պատճառը, որ RBF-ը համարվում է «ամենագնաց» կեռնել։

Ի՞նչ է Պոլինոմիալ կեռնելը

Սա մի ֆունկցիա է, որը թույլ է տալիս SVM-ին տեսնել տվյալների միջև եղած աստիճանային կապերը։ Եթե գծային կեռնելը միայն ուղիղ գծեր է քաշում, ապա պոլինոմիալը կարող է քաշել կորեր (պարաբոլներ, հիպերբոլներ և այլն)։


Բանաձևը՝
$$K(x, z) = (x^T z + r)^d$$

Այստեղ՝$

x^T z$-ն սովորական սկալյար արտադրյալն է։

$r$-ը հաստատուն է (coefficient), որը որոշում է, թե որքանով են բարձր աստիճանի անդամները գերիշխում ցածրերին։

$d$-ն հենց պոլինոմի աստիճանն է (degree)։


2. Ի՞նչ է պոլինոմիալ աստիճանը ($d$)

Աստիճանը ցույց է տալիս, թե որքան «բարդ» կամ «ճկուն» է մեր կորը։ Այն ուղղակիորեն կապված է այն չափողականության հետ, որտեղ մենք «տեղափոխում» ենք տվյալները։

$d = 1$: Սա սովորական գծային (Linear) կեռնելն է։ Ստացվում է ուղիղ գիծ։

$d = 2$ (Քառակուսային): Տվյալները պրոյեկտվում են այնպիսի տարածություն, որտեղ հաշվի են առնվում քառակուսիները ($x_1^2, x_2^2$) և զույգ առ զույգ արտադրյալները ($x_1x_2$)։ Սա թույլ է տալիս ստանալ էլիպսաձև կամ պարաբոլիկ սահմաններ։

$d = 3$ (Խորանարդային): Ավելի բարդ կորեր։ Հաշվի են առնվում $x^3$ անդամները։

$d = 10, 20 \dots$: Կորը դառնում է չափազանց բարդ և սկսում է «ոլորվել» ամեն մի կետի շուրջ։

3. Ինչպե՞ս է աստիճանը ազդում չափողականության վրա

Սա հենց այն «մղձավանջն» է։ Երբ աստիճանը ($d$) մեծանում է, հնարավոր փոխազդեցությունների քանակը էքսպոնենցիալ աճում է։

Օրինակ, եթե ունես ընդամենը 2 հատկանիշ ($x_1, x_2$).$d=2$-ի դեպքում նոր տարածությունը կունենա 3 չափողականություն ($x_1^2, x_2^2, \sqrt{2}x_1x_2$):

$d=10$-ի դեպքում չափողականությունների քանակը կդառնա տասնյակներ։

$d=100$ և ավելի հատկանիշների դեպքում (ինչպես պատկերների պիքսելները)՝ չափողականությունը դառնում է միլիոնավոր։


Չնայած թվում է, թե ինչքան մեծ լինի $d$-ն, այնքան լավ, իրականում ML-ում հազվադեպ են օգտագործում $d=3$-ից մեծ աստիճաններ։

Overfitting: Մեծ $d$-ի դեպքում մոդելը սկսում է անգիր անել մարզվող տվյալները և նոր տվյալների վրա սխալվում է։Հ

աշվարկային արագություն: Չնայած Kernel Trick-ը օգնում է, բայց շատ մեծ աստիճանները դեռևս կարող են դանդաղեցնել հաշվարկները։

Կեռնելների հանրահաշիվը (The Algebra of Kernels)

Կեռնելների հանրահաշիվը (The Algebra of Kernels) և այն, թե ինչպես կարելի է ստեղծել նոր կեռնելներ՝ համակցելով արդեն գոյություն ունեցողները:

Ինչու է Մերսերի թեորեմը կարևոր գործնականում. Քանի որ Մերսերը երաշխավորում է PSD (դրական կիսաորոշյալ) մատրիցների առկայությունը, մենք կարող ենք օգտագործել PSD մատրիցների հատկությունները՝ պարզ կեռնելներից նոր, բարդ կեռնելներ կառուցելու համար՝ առանց երբևէ կարիք ունենալու իմանալ, թե ինչ տեսք ունի նրանց $\phi(x)$-ը:

* Մասշտաբավորումը (Scaling): $cK_1$-ը վավեր է (երբ $c > 0$):
* Գումարումը (Addition): $K_1 + K_2$-ը վավեր է:
