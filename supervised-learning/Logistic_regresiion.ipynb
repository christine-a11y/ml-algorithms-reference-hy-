LOGISTIC Regression 


  EVALUATION METRIC

Ենթադրենք՝ մենք մարզել ենք մի քանի մոդել, որոնք լավ են աշխատում ուսումնական տվյալների (train data) վրա: Ինչպե՞ս կարող ենք համեմատել այդ մոդելները: Ի՞նչ չափանիշներ կարող ենք օգտագործել դրանցից լավագույնը ընտրելու կամ տեսնելու համար, թե արդյոք մեր մոդելը կենսունակ է «իրական աշխարհում»: Դասակարգման (classification) մոդելների համար առավել հաճախ օգտագործվող չափանիշներն են.

1. Ճշգրտություն (Accuracy)

2. Պրեցիզիա (Precision)

3. Հիշողություն կամ Ընդգրկում (Recall)

4. F1 գնահատական (F1 Score)

5. Կորի տակ գտնվող տարածք (AUC) և ROC կոր


Հակիրճ նկարագրենք յուրաքնչյուրը և հերթով մանրմասն ներկայացնենք։

1․ Ճշգրտություն (Accuracy). Ցույց է տալիս, թե ընդհանուր կանխատեսումներից քանիսն են եղել ճիշտ: Այն օգտակար է միայն այն դեպքում, երբ տվյալների հավաքածուի դասերը հավասարակշռված են:

2․ Պրեցիզիա (Precision). Պատասխանում է այն հարցին, թե մոդելի դրական կանխատեսումներից քանիսն են իրականում եղել դրական: Կարևոր է, երբ սխալ դրական արդյունքը (False Positive) մեծ վնաս է հասցնում (օրինակ՝ սպամի ֆիլտրում):

4․ Հիշողություն (Recall). Ցույց է տալիս, թե իրական դրական դեպքերից քանիսն է մոդելը կարողացել հայտնաբերել: Շատ կարևոր է, երբ սխալ բացասական արդյունքը (False Negative) վտանգավոր է (օրինակ՝ հիվանդությունների ախտորոշում):

5․ F1 գնահատական (F1 Score). Սա պրեցիզիայի և հիշողության միջին թվաբանականն է (հարմոնիկ միջին): Այն լավագույնն է այն դեպքերում, երբ ունենք անհավասարակշռված տվյալներ և պետք է պահպանել հավասարակշռությունը պրեցիզիայի և հիշողության միջև:

6․ ROC կոր և AUC. ROC կորը ցույց է տալիս մոդելի աշխատանքը տարբեր շեմերի (thresholds) դեպքում՝ համեմատելով ճիշտ դրականների և սխալ դրականների հարաբերակցությունը: AUC-ն մեկ թվով ներկայացնում է մոդելի որակը. որքան մոտ է 1-ին, այնքան լավ է մոդելը տարբերակում դասերը:


confusion matrix-ը  մեքենայական ուսուցման մեջ մոդելի ճշգրտությունը գնահատելու հիմքն է: Հետևյալ օրինակը  ցույց է տալիս, թե ինչպես է մոդելը տարբերակում «սպամ» և «ոչ սպամ» էլեկտրոնային նամակները:

1․ True Positive (TP - Ճշմարիտ դրական). Մոդելը ճիշտ է կռահել, որ նամակը սպամ է, և այն ուղարկել է սպամի թղթապանակ:

2․ False Positive (FP - Սխալ դրական). Մոդելը սխալմամբ նորմալ նամակը նշել է որպես սպամ: Սա կոչվում է նաև «Type I error» (Առաջին տեսակի սխալ):

3․ False Negative (FN - Սխալ բացասական). Մոդելը սպամ նամակը սխալմամբ ընդունել է որպես նորմալ և թողել է այն մուտքային նամակների (inbox) մեջ: Սա կոչվում է նաև «Type II error» (Երկրորդ տեսակի սխալ):

4․ True Negative (TN - Ճշմարիտ բացասական). Մոդելը ճիշտ է կռահել, որ նամակը սպամ չէ, և այն ուղարկել է մուտքային նամակների թղթապանակ:

Այս աղյուսակը օգնում է հասկանալ, թե որտեղ է մոդելը սխալվում: Օրինակ՝ եթե քո նպատակն է, որ ոչ մի կարևոր նամակ չհայտնվի սպամում, ապա պետք է նվազեցնել FP-ի քանակը: Եթե նպատակն է, որ ընդհանրապես սպամ նամակ չհայտնվի inbox-ում, պետք է նվազեցնել FN-ի քանակը:


Ճշգրտություն (Accuracy)

Ճշգրտությունը հիմնարար չափանիշ է, որն օգտագործվում է դասակարգման (classification) մոդելի արդյունավետությունը գնահատելու համար: Այն ցույց է տալիս մոդելի կողմից արված ճիշտ կանխատեսումների որքան են կազմում կանխատեսումների ընդհանուր թվի մեջ:

Ճշգրտություն = Ճիշտ կանխատեսումների քանակը / Կանխատեսումների ընդհանուր քանակը։

Թերություններ՝ 

Կարող է ապակողմնորոշիչ լինել անհավասարակշռված տվյալների (imbalanced datasets) դեպքում:

Ստեղծում է բարձր ճշգրտության կեղծ տպավորություն:

Մեքենայական ուսուցման մեջ «Accuracy»-ն ամենահեշտ հասկանալի ցուցանիշն է, բայց այն կարող է լուրջ սխալներ առաջացնել:

Պատկերացնենք, որ ունենք 100 նամակ, որոնցից 99-ը սովորական են, իսկ 1-ը՝ սպամ: Եթե  մոդելը ոչինչ չսովորի և պարզապես միշտ կանխատեսի «սովորական», ապա այն կունենա 99% ճշգրտություն (քանի որ 99 դեպքում կռահել է): Սակայն այս մոդելը գործնականում անօգուտ է, քանի որ այն չի կարողանում հայտնաբերել սպամը:

Պրեցիզիա (Precision)

Պրեցիզիան չափում է, թե մոդելի կողմից արված դրական կանխատեսումներից քանիսն են իրականում ճիշտ: Այն օգտակար է այն դեպքերում, երբ «սխալ դրական» (False Positive) արդյունքները շատ են, օրինակ՝ բժշկական ախտորոշումների ժամանակ, որտեղ առանց հիվանդության առկայության այն ախտորոշելը կարող է լուրջ հետևանքներ ունենալ:
Precision = TP / (TP + FP)

Առավելություններ՝

Արդյունավետ է անհավասարակշռված տվյալների դեպքում:

Ցույց է տալիս դրական կանխատեսումների ճշգրտությունը:

Թերություններ՝

Անտեսում է «սխալ բացասական» (False Negative) դեպքերը, ինչը կարող է հանգեցնել չափազանց պահպանողական մոդելների, որոնք բաց են թողնում բազմաթիվ իրական դրական դեպքեր:

Կենտրոնանում է միայն դրական դասի վրա, քանի որ չափանիշը հաշվի չի առնում մոդելի կարողությունը՝ գտնելու բոլոր համապատասխան դեպքերը:

Պրեցիզիան մեզ ասում է. «Երբ մոդելը ասում է՝ այո, որքա՞ն հաճախ է այն իրականում ճիշտ»:


Պատկերացնենք, որ ունենք էլեկտրոնային փոստի ֆիլտր: Եթե Պրեցիզիան ցածր է, դա նշանակում է, որ մոդելը շատ նորմալ նամակներ է սխալմամբ տեղափոխում «սպամ» թղթապանակ: Սա նյարդայնացնող է օգտատիրոջ համար, քանի որ կարող է բաց թողնել կարևոր նամակ:

Սակայն այն ունի թերություններ՝
Քանի որ բանաձևի մեջ (TP + FP) հաշվի չի առնվում FN-ը (այն դեպքերը, երբ մոդելը չի նկատել սպամը), մոդելը կարող է «շատ զգույշ» լինել: Այն կարող է ընտրել միայն ամենաակնհայտ սպամ նամակները՝ ապահովելով բարձր Պրեցիզիա, բայց միևնույն ժամանակ բաց թողնելով շատ այլ սպամ նամակներ, որոնք հայտնվում են մուտքային նամակների (inbox) մեջ:

Recall
Հիշողությունը կամ զգայունությունը (Sensitivity) չափում է, թե իրական դրական դեպքերից քանիսն են ճիշտ նույնականացվել մոդելի կողմից: Այն կարևոր է այն դեպքերում, երբ դրական դեպքը բաց թողնելը (սխալ բացասական - false negative) ավելի թանկ է արժենում, քան սխալ դրականները:




Հիշողություն = TP / (TP + FN)

Առավելություններ`

Այն սցենարներում, որտեղ կարևոր է հայտնաբերել բոլոր դրական դեպքերը (օրինակ՝ հիվանդությունների ախտորոշումը), Հիշողությունը առանցքային չափանիշ է:

Թերություններ`

Անտեսում է սխալ դրականները (False Positives), ինչը կարող է հանգեցնել չափազանց ազատ (liberal) մոդելների, որոնք ամեն ինչ կանխատեսում են որպես դրական:

Օրինակ՝
Պատկերացնենք քաղցկեղի ախտորոշման մոդել: Այստեղ հիմնական նպատակն է, որ ոչ մի հիվանդ մարդ չմնա չախտորոշված: Եթե մոդելը սխալմամբ առողջ մարդուն ասի, որ նա հիվանդ է (False Positive), դա տհաճ է, բայց լուծելի (հավելյալ հետազոտությամբ): Իսկ եթե մոդելը հիվանդ մարդուն ասի, որ նա առողջ է (False Negative), դա կարող է ճակատագրական լինել: Հիշողության բարձր ցուցանիշը ապահովում է, որ մենք բաց չենք թողնում դրական դեպքերը։

Եթե մոդելը պարզապես բոլոր դեպքերը նշի որպես «դրական», ապա այն կունենա 100% Հիշողություն, քանի որ կգտնի բոլոր իրական դրական դեպքերը: Բայց այն նաև կունենա շատ սխալ դրականներ: Հենց այստեղ է, որ հարկավոր է հաշվի առնել նաև Precision-ը:



F1 Score

F1 Score-ը պրեցիզիայի և հիշողության (recall) հարմոնիկ միջինն է: Այն օգտակար է, երբ մեզ անհրաժեշտ է հավասարակշռություն պրեցիզիայի և հիշողության միջև, քանի որ այն երկուսն էլ միավորում է մեկ թվի մեջ:

1. Ցածր հիշողությունը և բարձր պրեցիզիան տալիս են մեծ ճշգրտություն (accuracy), բայց այդ դեպքում մոդելը բաց է թողնում դեպքերի մեծ մասը:

2. Բարձր F1 score-ը նշանակում է, որ մոդելը լավ է աշխատում երկու չափանիշներով էլ: Դրա տիրույթը [0, 1] է:

3. Որքան բարձր է F1 score-ը, այնքան ավելի լավ է մոդելի աշխատանքը: Այն մաթեմատիկորեն կարելի է արտահայտել այսպես.

F1 Score = 2 * (Precision * Recall) / (Precision + Recall)

Պարզ միջինը (թվաբանական) կարող է շփոթեցնող լինել: Եթե մոդելը ունի 1.0 (100%) պրեցիզիա և 0.0 (0%) հիշողություն, ապա թվաբանական միջինը կլինի 0.5, ինչը կարող է թվալ ընդունելի: Իսկ հարմոնիկ միջինը (F1 score) այս դեպքում կլինի 0, ինչը ճիշտ է ցույց տալիս, որ մոդելը անօգուտ է (քանի որ հիշողությունը 0 է):
Այն լավագույնն է անհավասարակշռված տվյալների (imbalanced datasets) համար։

(Հարմոնիկ միջին $= \frac{2}{\frac{1}{a} + \frac{1}{b}} = \frac{2ab}{a + b}$)

AUC և ROC կոր

Նախ սահմանենք մի քանի լրացուցիչ տերմիններ՝ Specificity (TNR - Ճշմարիտ բացասականների տոկոս), որը չափում է, թե իրական բացասական դեպքերից քանիսն են ճիշտ նույնականացվել մոդելի կողմից: Այն պատասխանում է հետևյալ հարցին՝ «Իրական բացասական դեպքերից քանիսի՞ն է մոդելը ճիշտ գտել»:

TNR = TN / (TN + FP)



Եվ FPR (Սխալ դրականների տոկոս) - այն չափում է, թե իրական բացասական դեպքերից քանիսն են սխալմամբ դասակարգվել որպես դրական:

FPR = FP / (FP + TN)



Բացատրություն
ROC կորը կառուցելու համար մենք օգտագործում ենք երկու հիմնական ցուցանիշ՝ Recall (կամ TPR - True Positive Rate) և FPR (False Positive Rate):

TNR (Specificity)։ Սա Recall-ի նույն իմաստն  է,  դրականի փոխարեն բացասական դասի համար: Այն մեզ ասում է՝ «Որքանո՞վ է մոդելը լավ ճանաչում նորմալ, ոչ սպամ նամակները»:

FPR (False Positive Rate)։  Սա մեզ ասում է՝ «Որքա՞ն հաճախ է մոդելը սխալվում և նորմալ նամակը համարում սպամ»:


ROC կորը հենց այս երկու ցուցանիշների՝ TPR (y-առանցք) և FPR (x-առանցք) փոխկապակցվածությունն է տարբեր «շեմերի» (thresholds) դեպքում:

Եթե մենք շատ խիստ ենք լինում, մենք կունենանք ցածր FPR, բայց նաև՝ ցածր Recall:

Եթե մենք շատ զիջող ենք լինում, մենք կունենանք բարձր Recall, բայց նաև՝ բարձր FPR:

ROC կորը թույլ է տալիս գտնել այն իդեալական կետը, որտեղ մենք ունենք բարձր Recall և ցածր FPR: 



ROC կորը (Receiver Operating Characteristic Curve) ճշմարիտ դրական արդյունքների հարաբերակցության (TPR) և կեղծ դրական արդյունքների հարաբերակցության (FPR) գրաֆիկական պատկերումն է՝ կախված դասակարգման տարբեր շեմերից: Այս կորը օգնում է մեզ պատկերացնել զգայունության (TPR) և սպեցիֆիկության (1 - FPR) միջև առկա կապերը տարբեր շեմերի դեպքում:

Շեմեր (Thresholds). Դասակարգման մոդելները սովորաբար արդյունքում տալիս են հավանականություն (օրինակ՝ 0-ից 1): Շեմը այն սահմանագիծն է, որից բարձր դեպքը համարում ենք «դրական», իսկ ցածր՝ «բացասական»: ROC կորը ցույց է տալիս, թե ինչպես է փոխվում մոդելի ճշգրտությունը, եթե մենք փոխենք այդ սահմանագիծը:

AUC-ն (Area Under Curve) քանակապես գնահատում է մոդելի ընդհանուր կարողությունը՝ դրական և բացասական դասերը միմյանցից տարբերելու համար:

* AUC = 1. Կատարյալ մոդել (միշտ ճիշտ է դասակարգում դրական և բացասական դասերը):

* AUC = 0.5. Մոդելը ոչնչով չի գերազանցում պատահական գուշակությանը:

* AUC < 0.5. Մոդելն աշխատում է ավելի վատ, քան պատահական գուշակությունը (դա ցույց է տալիս, որ մոդելը «շրջված» է, այսինքն՝ դրականը համարում է բացասական և հակառակը):

ROC կորերը հատկապես օգտակար են, երբ.

* Տվյալների բազաները անհավասարակշիռ են (imbalanced):

* Անհրաժեշտ է կարգավորել դասակարգման շեմերը:



Նարնջագույն գիծը (ROC կորը): Սա ձեր մոդելի «փաստացի աշխատանքն» է։ Ինչպես տեսնում եք, այն կորանում է դեպի վերին աջ անկյուն։ Որքան կորը ավելի մոտ է վերին աջ անկյանը, այնքան մոդելն ավելի լավ է տարբերակում «դրական» և «բացասական» դասերը։

Կապույտ կտրատված գիծը: Սա «պատահական գուշակության» գիծն է (AUC = 0.5): Եթե ձեր մոդելի գիծը լիներ այս կապույտ գծի վրա, դա կնշանակեր, որ ձեր մոդելը ոչինչ չի հասկանում և պարզապես պատահական է ընտրում։

Area = 0.79: Սա հենց AUC-ի արժեքն է։ Քանի որ 0.79-ը զգալիորեն մեծ է 0.5-ից, սա նշանակում է, որ ձեր մոդելը բավականին լավ է տարբերակում դասերը։ 0.79-ը նշանակում է, որ 79% հավանականությամբ մոդելը ավելի բարձր գնահատական կտա պատահական ընտրված «դրական» օրինակին, քան պատահական ընտրված «բացասական» օրինակին։

Դիտարկենք հետևյալ պատկերը՝ 


Մոդելի բարդություն (x-առանցք). Սա մոդելի «կարողությունն» է սովորելու տվյալների օրինաչափությունները։ Շատ պարզ մոդելները (ձախ կողմում) կարող են բաց թողնել կարևոր դետալները, իսկ չափազանց բարդ մոդելները (աջ կողմում) կարող են սովորել նույնիսկ պատահական աղմուկը։

Training Error (ուսուցման սխալ). Ինչպես տեսնում եք, որքան բարդանում է մոդելը, այնքան այս գիծը իջնում է։ Սա տրամաբանական է. եթե մոդելին տանք անսահմանափակ «հիշողություն», այն կկարողանա անգիր անել ամբողջ ուսուցողական տվյալների բազան՝ հասցնելով սխալը գրեթե զրոյի։

Test Error (թեստային սխալ). Սա մեր «իրական» ցուցիչն է։ Սկզբում այն նույնպես իջնում է, քանի որ մոդելը սովորում է օգտակար օրինաչափություններ։ Սակայն, որոշակի պահից հետո այն սկսում է աճել։ Սա հենց Overfitting-ն է (գերհարմարեցում)։

«Sweet spot»-ը կամ «քաղցր կետը» այն վայրն է, որտեղ մոդելը հավասարակշռված է։ Այն բավականաչափ բարդ է՝ կարևոր օրինաչափությունները հասկանալու համար, բայց ոչ այնքան բարդ, որպեսզի սկսի անգիր անել աղմուկը։


«Regularization-ը օգնում է մեզ հասնել այս կետին՝
Regularization-ը մոդելին «պատժում» է չափազանց մեծ բարդության համար։ Դա մոդելի համար դժվարացնում է «հիշել» յուրաքանչյուր մանրուք և ստիպում է նրան կենտրոնանալ ավելի ընդհանուր, ճիշտ օրինաչափությունների վրա։ Պատկերացրեք, որ մենք մոդելին ասում ենք. «Դու կարող ես բարդանալ, բայց միայն այն դեպքում, եթե դա իսկապես մեծացնում է ճշգրտությունը»։

Այսինքն՝

* Եթե մոդելը գտնվում է «Sweet spot»-ից ձախ, մենք ունենք Underfitting (մոդելը շատ պարզ է)։

* Եթե գտնվում է աջ կողմում, մենք ունենք Overfitting (մոդելը շատ բարդ է և անգիր է արել տվյալները)։



Classification

Ի՞նչ է դասակարգումը

Պատկերացնենք, որ  ունենք մի խումբ իրեր, որոնք պետք է բաժանել խմբերի։ Դասակարգումը այն գործընթացն է, որի միջոցով մենք մոդելին սովորեցնում ենք՝ որտե՞ղ է ավարտվում մի խումբը և սկսվում մյուսը։


(Training data). Մենք ունենք տվյալներ (օրինակ՝ $N$ հատ նկար կամ չափում), որոնցից յուրաքանչյուրի մասին արդեն գիտենք, թե որ խմբին է այն պատկանում ։ Գրաֆիկում սա կապույտ շրջանակներն ու կանաչ եռանկյուններն են։

Մեր նպատակն է  ստեղծել մի կանոն (ֆունկցիա), որը թույլ կտա նոր օրինակը որը երբեք չի տեսել,  նայելիս ասել՝ «Սա կապույտ է» թե «Սա կանաչ»։

Դրա համար մենք կառուցում ենք $y(x)$ ֆունկցիա։ Գրաֆիկում դա այն ուղիղ գիծն է, որը անցնում է կապույտ և կանաչ պատկերների միջով։ Այդ գիծը սահմանն է (decision boundary)։


Մոդելը հաճախ չի ասում պարզապես «սա կապույտ է»։ Այն ասում է՝ «Որքանո՞վ եմ ես համոզված, որ սա կապույտ է»։Եթե մենք նոր կետ դնենք գծի կապույտ կողմում՝ մոդելը կասի, օրինակ՝ «90% հավանականությամբ սա կապույտ է» դա հենց  $p(t | x)$-ն է։Եթե կետը շատ մոտ է գծին, մոդելը կարող է ասել՝ «50/50 հավանականություն է, ես վստահ չեմ»։Այսպիսով, դասակարգումը ընդամենը տարածության բաժանումն է այնպիսի մասերի, որտեղ ամեն մի մասին համապատասխանում է իր «անունը» (դասը)։



Մաթեմատիկորեն ներկայացումը հետևյալ կերպ՝

Մենք ունենք տվյալների բազմություն, որը կազմված է $N$ դիտարկումներից՝ $\{(x_n, t_n)\}_{n=1}^N$, որտեղ՝
* $x_n \in \mathbb{R}^D$ (մուտքային վեկտոր է $D$ չափանի տարածությունում),
* $t_n \in \{C_1, \dots, C_K\}$ (class-ի label-ն է)։Մեր նպատակն է ստեղծել $y(x)$ ֆունկցիա, որը $x$-ը համապատասխանցնում է  դասի $t$ պիտակին։

Փոխանակ ուղղակի կանխատեսելու պիտակը, մենք մոդելավորում ենք պայմանական հավանականությունը՝ $p(t|x)$։ Սա նշանակում է՝ ի՞նչ հավանականությամբ է $x$ մուտքային տվյալը պատկանում $k$-րդ դասին՝ $p(t=C_k|x)$։Այսպիսով, դասակարգումը դառնում է հետևյալ խնդիրը՝

$$\hat{t} = \arg\max_{k} p(t=C_k|x)$$

Մենք ընտրում ենք այն դասը, որի հավանականությունը տվյալ $x$-ի համար առավելագույնն է։




(Input space): Սա այն բազմաչափ տարածությունն է, որտեղ գտնվում են մեր տվյալները: Յուրաքանչյուր կետ (օրինակ՝ լուսանկարի պիքսելների արժեքները) ունի իր կոորդինատները: 
Որոշման մակերեսներ (Decision Surfaces): Երբ մենք ուզում ենք դասակարգել տվյալները, մենք պետք է դրանք բաժանենք խմբերի: Մոդելը ստեղծում է սահման, որը տարանջատում է դասերը: Գծային մոդելների դեպքում այս սահմանը միշտ «ուղիղ» է (հիպերհարթություն):
Գծային բաժանելիություն (Linearly Separable): Սա ամենահեշտ դեպքն է: Եթե կարող ես մեկ ուղիղ գծով (2D-ում) կամ հարթությամբ (3D-ում) տարանջատել կարմիր կետերը կապույտներից՝ առանց սխալվելու, ուրեմն տվյալները գծայնորեն բաժանելի են:


Մեքենայական ուսուցման մեջ հաճախ մենք չենք ասում «սա ճիշտ է կամ սխալ», այլ տալիս ենք հավանականություն, թե որքանով է օբյեկտը պատկանում տվյալ դասին:Երկու դասի դեպքում (Binary classification): Սա ամենապարզն է: Մենք ունենք մեկ թիրախ՝ $t$:

*$t = 1$: Օբյեկտը պատկանում է $C_1$ դասին:
*$t = 0$: Օբյեկտը պատկանում է $C_2$ դասին:

Սա նման է անջատիչի՝ «միացված է» կամ «անջատված է»:Շատ դասերի դեպքում (Multi-class classification - 1-of-K): Երբ դասերը 2-ից շատ են, $0$ և $1$ թվերը բավարար չեն: Այստեղ կիրառվում է «1-ից K» կոդավորումը:Մեք ստեղծում ենք վեկտոր, որի երկարությունը հավասար է դասերի քանակին: Եթե ունենք 5 տեսակի տվյալ ($K=5$), ապա 2-րդ տեսակի տվյալը կնշվի $(0, 1, 0, 0, 0)$ վեկտորով: Սա կարևոր է Որովհետև համակարգչային մաթեմատիկական մոդելները (օրինակ՝ Նեյրոնային ցանցերը) հիմնված են վեկտորային հաշվարկների վրա: Վեկտորի մեջ «1»-ի դիրքը ցույց է տալիս, թե որ դասն է ընտրված:


Մեքենայական ուսուցման մեջ դասակարգման մոդելների դասակարգման հիմնական մոտեցումներն են՝ դիսկրիմինատիվ և գեներատիվ։
 * Դիսկրիմինատիվ (Direct approach): Այս մոդելները կենտրոնանում են միայն դասերի միջև որոշումների սահմանի (decision boundary) վրա: Նրանք չեն փորձում հասկանալ, թե ինչպես են տվյալները բաշխված իրականում, այլ պարզապես փնտրում են այնպիսի գիծ կամ մակերես, որը լավագույնս կբաժանի դասերը: Դա նման է ճանապարհի վրա գծանշում անելուն, որը ցույց է տալիս, թե որ կողմն է աջը և որը՝ ձախը:
 * Գեներատիվ (Generative approach): Այս մոդելները փորձում են «հասկանալ» յուրաքանչյուր դասի ներքին կառուցվածքը: Նրանք ուսումնասիրում են $p(x|C_k)$ բաշխումը, այսինքն՝ թե ինչպիսի տեսք ունեն $C_k$ դասի տվյալները: Դա նման է ոչ միայն ճանապարհի գծանշումը իմանալուն, այլ նաև ամբողջ տեղանքի քարտեզը ունենալուն, ինչը հնարավորություն է տալիս նաև նոր տվյալներ «ստեղծել»: Օգտագործում է Բայեսի թեորեմը։
 Բայեսի թեորեմը գրվում է այսպես՝
 $$p(C_k|x) = \frac{p(x|C_k) \cdot p(C_k)}{p(x)}$$
 
 * Որտեղ՝$p(C_k|x)$ (Posteriory/Պոստերիոր)՝ այն, ինչ ուզում ենք իմանալ։ հավանականությունը, որ տվյալ $x$-ը պատկանում է $C_k$ դասին։
 
 * $p(x|C_k)$ (Likelihood/Լայքլիհուդ)՝ սա այն է, ինչ մոդելավորում են գեներատիվ մոդելները։ Սա ցույց է տալիս, թե որքանով է հավանական $x$ տվյալի առկայությունը, եթե մենք արդեն գիտենք, որ այն պատկանում է $C_k$ դասին։
 
 * $p(C_k)$ ՝ դասի նախնական հավանականությունը (ինչքա՞ն հաճախ է այս դասը հանդիպում ընդհանուր տվյալների մեջ)
 ։
 
 * $p(x)$ ՝ նորմալիզացնող գործակից, որը ապահովում է, որ բոլոր հավանականությունների գումարը հավասար լինի 1-ի։



 Դիտակենք հետևյալ խնդիրը՝

Գծային ռեգրեսիան կանխատեսում է անընդհատ արժեքներ (օրինակ՝ գին, ջերմաստիճան), որոնք կարող են լինել ցանկացած թվի։ Սակայն դասակարգման դեպքում մեր նպատակը դիսկրետ (առանձնացված) դասերի կամ հավանականությունների որոշումն է։ Օրինակ՝ «Այո» կամ «Ոչ», որը մաթեմատիկորեն կարելի է ներկայացնել որպես 0 կամ 1։

 (Եթե մենք օգտագործենք սովորական գծային ֆունկցիա՝ $y = w^Tx + w_0$, ապա արդյունքը կարող է լինել ցանկացած թիվ՝ $-\infty$-ից $+\infty$։
Բայց մենք դասակարգման համար ուզում ենք հավանականություն, որը պետք է լինի 0-ից 1-ի միջև (օրինակ՝ 0.85 հավանականություն, որ սա «շուն» է)։ Մենք կիրառում ենք ակտիվացման ֆունկցիա ($f$), օրինակ՝ Սիգմոիդը, որը ցանկացած թիվ «սեղմում» է (0, 1) միջակայքի մեջ։Գծային մասը ($w^Tx + w_0$): Որոշում է, թե «որքան ուժեղ» է պատկանելությունը դասին։Ոչ գծային մասը ($f$): Այդ ուժը վերածում է հավանականության։)

Մենք վերցնում ենք գծային կոմբինացիան՝ $w^T x + w_0$, և անցկացնում այն ոչ գծային $f(\cdot)$ ֆունկցիայի միջով`

$$y(x) = f(w^T x + w_0)$$


Այս ֆունկցիան «սեղմում» է մեր հաշվարկած արժեքը որոշակի սահմաններում (օրինակ՝ 0-ից 1-ի միջակայքում, ինչպես սիգմոիդ ֆունկցիայի դեպքում)։

(ինչու մոնոտոն։
Մոնոտոն ֆունկցիան (օրինակ՝ սիգմոիդը) կամ միշտ աճող է, կամ միշտ նվազող։ Դա նշանակում է, որ յուրաքանչյուր $y$ արժեքի համար գոյություն ունի միայն մեկ $z$ արժեք (որտեղ $z = w^T x + w_0$), որի դեպքում $f(z) = y$։)


Գծային դասակարգման երկրաչափական մեկնաբանությունը

Մեքենայական ուսուցման մեջ դասակարգումը (Classification) հանդիսանում է վերահսկվող ուսուցման առանցքային խնդիրներից մեկը: Գծային դասակարգիչների աշխատանքի հիմքում ընկած է $D$-չափանի հատկանիշների տարածությունը երկու կամ ավելի տիրույթների բաժանելու գաղափարը՝ օգտագործելով հիպերհարթություններ:


Դասակարգման պարզագույն ձևը երկու դասերի բաժանումն է ($C_1$ և $C_2$): Մոդելը ներկայացվում է որպես գծային ֆունկցիա.$$y(x) = w^T x + w_0$$որտեղ $w$ վեկտորը կոչվում է կշիռների վեկտոր, իսկ $w_0$-ն՝ շեղում (bias): Որոշման կանոնը հետևյալն է. եթե $y(x) \geq 0$, ապա օբյեկտը պատկանում է $C_1$ դասին, հակառակ դեպքում՝ $C_2$ դասին:

Այն բազմությունը, որի համար $y(x) = 0$, կոչվում է որոշման մակերևույթ (decision surface): Երկչափ տարածության մեջ սա ուղիղ գիծ է, եռաչափում՝ հարթություն, իսկ ավելի բարձր չափայնության դեպքում՝ հիպերհարթություն:

$w$ վեկտորի երկրաչափական դերը

Դիտարկենք երկու կամայական կետ՝ $x_A$ և $x_B$, որոնք գտնվում են որոշման հարթության վրա: Համաձայն սահմանման՝ $y(x_A) = 0$ և $y(x_B) = 0$: Սրանից հետևում է`

$$(w^T x_A + w_0) - (w^T x_B + w_0) = 0 \implies w^T(x_A - x_B) = 0$$

Այս հավասարումը ցույց է տալիս, որ $w$ վեկտորը ուղղահայաց է հարթության վրա ընկած ցանկացած վեկտորի ($x_A - x_B$): Հետևաբար, $w$ վեկտորը հանդիսանում է հարթության նորմալը և որոշում է դրա կողմնորոշումը տարածության մեջ:


Շեղման գործակիցը պատասխանատու է հարթության դիրքի համար կոորդինատային համակարգի սկզբնակետի նկատմամբ: Սկզբնակետից մինչև հիպերհարթություն ընկած ամենակարճ (ուղղահայաց) հեռավորությունը տրվում է հետևյալ բանաձևով՝

$$\text{dist}_{\text{origin}} = \frac{w_0}{\|w\|}$$

Եթե $w_0 > 0$, հարթությունը շեղված է սկզբնակետից հակառակ ուղղությամբ, իսկ եթե $w_0 = 0$, հարթությունն անցնում է հենց սկզբնակետով:

Եթե $x$-ը որոշման հարթության վրա գտնվող կետ է, ապա կոորդինատների սկզբնակետից մինչև այդ հարթություն եղած նորմալ (ուղղահայաց) հեռավորությունը հաշվվում է հետևյալ բանաձևով:$$\frac{w^T x}{\|w\|} = -\frac{w_0}{\|w\|}$$

Այստեղ $w_0$-ն (bias) որոշում է, թե որքանով է հարթությունը շեղված սկզբնակետից:


Ցանկացած $x$ կետի համար մենք կարող ենք հաշվել դրա «նշանով հեռավորությունը» ($r$) հարթությունից: Եթե $x$-ը պրոյեկտենք հարթության վրա (ստանալով $x_{\perp}$ կետը), ապա կապը կլինի.$$x = x_{\perp} + r \frac{w}{\|w\|}$$

Տեղադրելով սա $y(x) = w^T x + w_0$ բանաձևի մեջ և հաշվի առնելով, որ $y(x_{\perp}) = 0$, ստանում ենք՝

$$r = \frac{y(x)}{\|w\|}$$

Սա առանցքային նշանակություն ունի այնպիսի ալգորիթմների համար, ինչպիսին է Support Vector Machines (SVM)-ը, քանի որ այն թույլ է տալիս օպտիմալացնել «մարժան» (margin)՝ առավելագույնի հասցնելով հեռավորությունը դասերի միջև:

Կարևոր եզրահանգումներ՝
* $w$ կշիռները որոշում են բաժանարար գծի անկյունը:
* $w_0$ շեղումը որոշում է գծի հեռավորությունը սկզբնակետից:
* $y(x)$ արժեքի բացարձակ մեծությունը ցույց է տալիս, թե որքան հեռու է կետը սահմանից, ինչը կարելի է մեկնաբանել որպես դասակարգման վստահության աստիճան:




Նկարում պատկերված է երկչափ հատկանիշների տարածություն ($x_1, x_2$), որը բաժանված է երկու տիրույթի ($R_1$ և $R_2$):

Կարմիր գիծը ($y=0$): Սա մեր որոշման սահմանն է (Decision Boundary): Այն բաժանում է տարածությունը $y > 0$ (դրական կանխատեսում) և $y < 0$ (բացասական կանխատեսում) տիրույթների:

$w$ վեկտորը (կանաչ սլաք): Ինչպես տեսնում ենք, $w$ վեկտորը ուղղահայաց է կարմիր գծին: Սա հաստատում է այն թեզը, որ կշիռների վեկտորը հանդիսանում է հիպերհարթության նորմալը և ցույց է տալիս դեպի դրական դասի ($R_1$) կողմը:

Հեռավորությունը սկզբնակետից$-w_0 / \|w\|$ (սև սլաք): Գծապատկերը ցույց է տալիս սկզբնակետից ($0,0$) մինչև որոշման սահմանն ընկած ամենակարճ հեռավորությունը: Ուշադրություն դարձրեք, որ եթե $w_0$-ն փոխվի, գիծը կշարժվի $w$ վեկտորի ուղղությամբ կամ հակառակ, բայց չի փոխի իր թեքության անկյունը:

Կամայական $x$ կետի վերլուծությունըՆկարում կապույտով պատկերված է կամայական $x$ վեկտորը, որը դասակարգվում է:$x_{\perp}$ (պրոյեկցիա): Սա $x$ կետի պրոյեկցիան է որոշման սահմանի վրա: Այն ներկայացնում է հարթության վրա այն ամենամոտ կետը, որի համար $y(x_{\perp}) = 0$:$y(x) / \|w\|$ (կապույտ կետագիծ): Սա հենց այն $r$ հեռավորությունն է, որի մասին խոսեցինք: Այն ցույց է տալիս, թե որքան է $x$ կետի ուղղահայաց շեղումը սահմանից: Քանի որ այս դեպքում $x$-ը գտնվում է այն կողմում, ուր ուղղված է $w$-ն, ապա $y(x) > 0$:


Բազմադաս դասակարգման մեթոդները

Երբ դասերի քանակը գերազանցում է երկուսը, գծային դասակարգիչների կիրառումը դառնում է ավելի բարդ: Գոյություն ունեն մի քանի հիմնական ռազմավարություններ։

1. «Մեկը բոլորի դեմ» (One-vs-Rest)

Այս մոտեցման ժամանակ կիրառվում են $K-1$ երկու դասանի դասակարգիչներ:Յուրաքանչյուր դասակարգիչ առանձնացնում է կոնկրետ $C_k$ դասի կետերը մնացած բոլոր կետերից, որոնք չեն պատկանում այդ դասին:Սա կարող է առաջացնել տարածության այնպիսի հատվածներ, որոնք միաժամանակ վերագրվում են մի քանի դասերի կամ ոչ մի դասի:


2. «Մեկը մեկի դեմ» (One-vs-One)

Այս դեպքում ներմուծվում են $K(K-1)/2$ բինար դիսկրիմինանտ ֆունկցիաներ՝ յուրաքանչյուր հնարավոր դասերի զույգի համար:Յուրաքանչյուր կետ դասակարգվում է ըստ բոլոր ֆունկցիաների միջև տեղի ունեցած «ձայների մեծամասնության»:Այս մեթոդը նույնպես զերծ չէ երկիմաստ տիրույթների առաջացումից, որտեղ դժվար է որոշել վերջնական դասը:

$K$-դասանի դասակարգում (K-class Classification)

Սա առավել կատարելագործված մոտեցում է, որն օգտագործում է մեկ միասնական դիսկրիմինանտ՝ բաղկացած $K$ գծային ֆունկցիաներից:

Որոշման կանոն: $x$ կետը վերագրվում է $C_k$ դասին, եթե $y_k(x) > y_j(x)$ բոլոր $j \neq k$ դեպքերում:

Որոշման սահման: $C_k$ և $C_j$ դասերի միջև բաժանարար սահմանը տրվում է $y_k(x) = y_j(x)$ պայմանով:

Երկրաչափական առավելությունը: Այս մեթոդով ստացված որոշման տիրույթները միշտ կապակցված են և ուռուցիկ (convex), ինչը բացառում է դասակարգման հակասությունները հատկանիշների տարածության մեջ:



1. Ձախ պատկեր. «Մեկը բոլորի դեմ» (One-vs-Rest) խնդիրը

Այստեղ փորձ է արվում առանձնացնել երեք դաս ($C_1, C_2, C_3$) երկու բինար դասակարգիչների միջոցով։Կանաչ տիրույթը (հարցականով): Այս հատվածը ներկայացնում է երկիմաստության գոտի։ Քանի որ մենք ունենք դասակարգիչ, որն ասում է «սա $C_1$ է» և մեկ այլ դասակարգիչ, որն ասում է «սա $C_2$ է», ապա կանաչ գոտում հայտնված կետերը միաժամանակ «հավակնում են» պատկանել և՛ $C_1$-ին, և՛ $C_2$-ին։

Հետևանք: Մոդելը չի կարողանում միանշանակ որոշում կայացնել այս տիրույթի համար, ինչը հանգեցնում է սխալների։

2. Աջ պատկեր. «Մեկը մեկի դեմ» (One-vs-One) խնդիրը

Այստեղ օգտագործվում են զույգ առ զույգ դասակարգիչներ ($C_1$ vs $C_2$, $C_1$ vs $C_3$, և $C_2$ vs $C_3$)։Կանաչ եռանկյունին (հարցականով): Սա կենտրոնական անորոշության գոտին է։ Այս տիրույթում յուրաքանչյուր զույգային դասակարգիչ տալիս է հակասական պատասխան։ Օրինակ՝ $C_1$ vs $C_2$ դասակարգիչը կարող է ընտրել $C_1$-ը, $C_2$ vs $C_3$-ը՝ $C_2$-ը, իսկ $C_3$ vs $C_1$-ը՝ $C_3$-ը։

Հետևանք: Առաջանում է «փակ շղթա», և ձայների մեծամասնության սկզբունքը այստեղ չի աշխատում։


Որոշման տիրույթների ուռուցիկությունը (Convexity)

«Նման դիսկրիմինանտի (տարբերիչ ֆունկցիայի) որոշման տիրույթները միշտ միակապ են և ուռուցիկ: Դա տեսնելու համար դիտարկենք երկու կետ՝ $x_A$ և $x_B$, որոնք երկուսն էլ գտնվում են $R_k$ որոշման տիրույթի ներսում: Ցանկացած $\hat{x}$ կետ, որը գտնվում է $x_A$ և $x_B$ միացնող գծի վրա, կարող է արտահայտվել հետևյալ տեսքով. $\hat{x} = \lambda x_A + (1 - \lambda)x_B$: Գծային դիսկրիմինանտ ֆունկցիաների (LDF) համար հետևում է, որ.$$y_k(\hat{x}) = \lambda y_k(x_A) + (1 - \lambda)y_k(x_B)$$Քանի որ և՛ $x_A$, և՛ $x_B$ գտնվում են $R_k$-ի ներսում, հետևում է, որ $y_k(x_A) > y_j(x_A)$ և $y_k(x_B) > y_j(x_B)$ բոլոր $j \neq k$ համար, և հետևաբար $y_k(\hat{x}) > y_j(\hat{x})$։ Այսպիսով, $\hat{x}$-ը նույնպես գտնվում է $R_k$-ի ներսում: Հետևաբար, $R_k$-ն միակապ է և ուռուցիկ:»

բացատրություն

Ի՞նչ է որոշման տիրույթը (Decision Region):

Դա տարածության այն հատվածն է, որտեղ դասակարգիչը որոշում է, որ բոլոր կետերը պատկանում են որոշակի «k» դասին: Օրինակ՝ նկարում $R_k$ տիրույթում գտնվող ցանկացած կետ կհամարվի $k$ դասի անդամ:


Ի՞նչ է նշանակում «ուռուցիկ» (Convex):

Երկրաչափության մեջ պատկերը համարվում է ուռուցիկ, եթե դրա ներսում վերցված ցանկացած երկու կետ միացնող հատվածը ամբողջությամբ ընկած է այդ նույն պատկերի ներսում: Տեքստը մաթեմատիկորեն ապացուցում է, որ գծային դասակարգիչների դեպքում սա միշտ այդպես է:

Ապացույցի տրամաբանությունը:

Մենք ունենք երկու կետ՝ $x_A$ և $x_B$, որոնք պատկանում են $R_k$ դասին (այսինքն՝ այդ կետերում $y_k$ ֆունկցիայի արժեքն ամենամեծն է):Մենք վերցնում ենք մի նոր կետ՝ $\hat{x}$, որը գտնվում է նրանց միացնող ուղիղ գծի վրա:Քանի որ դասակարգիչը գծային է, ապա այդ նոր կետի համար ֆունկցիայի արժեքը լինելու է $x_A$ և $x_B$ կետերի արժեքների կշռված միջինը:Քանի որ երկու ծայրակետերում էլ $k$ դասը «հաղթող» էր, ապա նրանց միջև ընկած ցանկացած կետում $k$ դասը նույնպես կլինի «հաղթող» ($y_k(\hat{x}) > y_j(\hat{x})$):



Պատկերում տեսնում ենք երեք տարբեր տիրույթներ՝ $R_i$, $R_j$, և $R_k$: Յուրաքանչյուր տիրույթ համապատասխանում է մի որոշակի «դասի»: Երբ մոդելին տալիս ենք նոր տվյալ (կետ), այն ընկնում է այս տիրույթներից մեկի մեջ, և մոդելը հայտարարում է, թե որ դասին է այն պատկանում:2
Հիմնական գաղափարն այն է, որ գծային դասակարգիչների ստեղծած տիրույթները միշտ ուռուցիկ (convex) են:

Փորձարկումը: Վերցնենք նույն տիրույթում ($R_k$) գտնվող երկու կետ՝ $x_A$ և $x_B$:

Գիծը: Եթե մենք գիծ քաշենք այդ երկու կետերի միջև, ապա այդ գծի վրա գտնվող ցանկացած $\hat{x}$ կետ նույնպես պետք է գտնվի $R_k$ տիրույթի ներսում:

Արդյունքը: Քանի որ գիծը երբեք դուրս չի գալիս տիրույթի սահմաններից, դա նշանակում է, որ տիրույթը «ուռուցիկ» է:3. Ինչու՞ է սա կարևորՍա նշանակում է, որ գծային դասակարգիչները (ինչպիսիք են Լոգիստիկ ռեգրեսիան կամ SVM-ը) չեն կարող ունենալ «բարդ» կամ կտրտված տիրույթներ: Դրանք սահմանափակված են պարզ երկրաչափական ձևերով


Լոգիստիկ Ռեգրեսիա

Գծային ռեգրեսիան (Linear Regression) նախատեսված է թվեր կանխատեսելու համար (օրինակ՝ տան գինը կամ ջերմաստիճանը): Դրա պատասխանը կարող է լինել ցանկացած թիվ՝ շատ փոքր բացասականից մինչև շատ մեծ դրական:

Սակայն, եթե մենք ուզում ենք պարզել՝ արդյոք էլեկտրոնային նամակը «սպամ» է, թե ոչ, մեզ պետք է հավանականություն: Մաթեմատիկորեն հավանականությունը չի կարող լինել 0-ից փոքր կամ 1-ից մեծ:


Գծային ռեգրեսիայում մենք կանխատեսում ենք իրական արժեք ունեցող ելք:
$$y = w^Tx + b$$

Սակայն դասակարգման ժամանակ մենք ցանկանում ենք ստանալ հավանականություն:$$p(y=1|x)$$

Խնդիրը՝

Գծային ռեգրեսիայի ելքային արժեքները գտնվում են $(-\infty, \infty)$ տիրույթում:
Հավանականությունները պետք է լինեն $[0, 1]$ տիրույթում:

Որպեսզի իրական թիվը «սեղմենք» և դարձնենք հավանականություն, օգտագործում ենք լոգիստիկ սիգմոիդ ֆունկցիան.$$\sigma(z) = \frac{1}{1 + e^{-z}}$$

Այս ֆունկցիայի շնորհիվ ստանում ենք $y=1$ դասին պատկանելու հավանականությունը՝ $$P(y=1|x) = \sigma(w^T x + b) = \frac{1}{1 + e^{-(w^T x + b)}}$$

Երեք կարևոր հատկանիշներ, որոնք սիգմոիդը դարձնում են իդեալական դասակարգման համար՝

Ելքը միշտ 0-ի և 1-ի միջև է (Output is always between 0 and 1): Սա թույլ է տալիս արդյունքը մեկնաբանել որպես հավանականություն:

Հարթ է և դիֆերենցելի (Smooth and differentiable): Սա չափազանց կարևոր է մոդելի մարզման ժամանակ (օրինակ՝ Gradient Descent օգտագործելիս), քանի որ մենք կարող ենք հաշվել ածանցյալը ցանկացած կետում:

Մոնոտոն աճող (Monotonically increasing): Որքան մեծ է մուտքային արժեքը, այնքան մեծ է հավանականությունը:


լգորիթմի հիմքում ընկած է հետևյալ մաթեմատիկական բանաձևը.$$p(y=1|x) = \sigma(w^Tx + b)$$

Արդյունքը մեկնաբանվում է որպես հավանականություն: Մոդելը ոչ թե ուղղակի ասում է «այո» կամ «ոչ», այլ հաշվարկում է հավանականությունը ($p$), թե որքանով է տվյալ մուտքային փոփոխականը ($x$) պատկանում որոշակի դասի:
* Եթե $p > 0.5 \rightarrow$ կանխատեսել 1-ին դասը: Եթե հավանականությունը 50%-ից մեծ է, մոդելը համարում է, որ օբյեկտը պատկանում է թիրախային խմբին:

* Եթե $p < 0.5 \rightarrow$ կանխատեսել 0-ին դասը: Եթե հավանականությունը 50%-ից փոքր է, ապա այն դասակարգվում է որպես բացասական կամ այլ դաս:

Չնայած մենք օգտագործում ենք ոչ գծային ֆունկցիա ($\sigma$ — սիգմոիդ), որոշման սահմանը դեռևս գծային է:

$$w^Tx + b = 0$$

Սա այն գիծն է (կամ հարթությունը), որը տարանջատում է երկու դասերը միմյանցից:

Հավանականությունների մոդելավորում (Modeling Probabilities)

Լոգիստիկ ռեգրեսիան ենթադրում է հետևյալը՝ $y \in \{0, 1\}$: Սա նշանակում է, որ մեր թիրախային փոփոխականը (ելքը) երկուական է (binary)՝ այն կարող է լինել կամ $0$, կամ $1$:Յուրաքանչյուր ելք հետևում է Բեռնուլիի բաշխմանը (Bernoulli distribution): Սա վիճակագրական բաշխում է այնպիսի փորձերի համար, որոնք ունեն միայն երկու հնարավոր արդյունք (օրինակ՝ մետաղադրամի նետումը):

$$p(y|x) = p^y(1 - p)^{1-y}$$

Սա մեկ միասնական բանաձև է, որը նկարագրում է $y=1$ և $y=0$ լինելու հավանականությունները՝ 

Եթե $y = 1$, ապա $p^1(1 - p)^0 = p$:Եթե $y = 0$, ապա $p^0(1 - p)^1 = 1 - p$:Այստեղ $p$ հավանականությունը հաշվարկվում է որպես՝ $$p = \sigma(w^Tx + b)$$

Սա նույն սիգմոիդ ֆունկցիան է, որը գծային արդյունքը ($w^Tx + b$) վերածում է $0$-ից $1$ միջակայքի թվի:

Ենթադրություն (Assumption):
Տվյալների կետերը անկախ են (Data points are independent):
Սա շատ կարևոր վիճակագրական ենթադրություն է: Այն նշանակում է, որ մի տվյալի (օրինակ՝ մեկ հիվանդի ախտորոշման) արդյունքը չի ազդում և կապված չէ մեկ այլ տվյալի արդյունքի հետ: Մաթեմատիկորեն սա թույլ է տալիս մեզ բազմապատկել առանձին հավանականությունները՝ մոդելի ընդհանուր լավագույն պարամետրերը գտնելու համար:


Մենք ոչ թե պարզապես գիծ ենք քաշում, այլ փորձում ենք գտնել այնպիսի $w$ (կշիռներ) և $b$ (շեղում), որոնք առավելագույնի կհասցնեն ճիշտ դասակարգման հավանականությունը մեր ունեցած տվյալների համար:

Log Loss (Լոգարիթմական կորուստ / Cross-Entropy)

Մեքենայական ուսուցման մեջ մենք ուզում ենք առավելագույնի հասցնել հավանականությունը (likelihood), ինչը մաթեմատիկորեն համարժեք է բացասական լոգարիթմական հավանականությունը նվազեցնելուն:

Կորստի բանաձևը:

$$\mathcal{L} = -\sum_{i} [y_i \log p_i + (1 - y_i) \log(1 - p_i)]$$

Այս բանաձևը չափում է տարբերությունը կանխատեսված հավանականության ($p_i$) և իրական արժեքի ($y_i$) միջև:

Հիմնական ինտուիցիան (Key Intuition):Խստորեն պատժում է վստահ, բայց սխալ կանխատեսումները:Եթե մոդելը 99% վստահությամբ ասում է, որ արդյունքը 1 է ($p=0.99$), բայց իրականում այն 0 է ($y=0$), ապա Log Loss-ը տալիս է շատ մեծ տուգանք (կորուստը կտրուկ աճում է): Սա ստիպում է մոդելին լինել ավելի զգույշ:Խրախուսում է ճիշտ հավանականություններ, ոչ միայն ճիշտ պիտակներ:Ի տարբերություն պարզ ճշգրտության (accuracy), որը միայն նայում է՝ «ճի՞շտ էր, թե՞ սխալ», Log Loss-ը մոդելին մղում է դեպի կատարելություն: Ավելի լավ է կանխատեսել 0.9 հավանականություն ճիշտ դասի համար, քան 0.6, նույնիսկ եթե երկու դեպքում էլ դասակարգումը համարվում է ճիշտ:

Բանաձևն ունի երկու «ռեժիմ», որոնք աշխատում են ավտոմատ՝ կախված իրական $y$ արժեքից՝

Եթե իրական պատասխանը $y=1$ է:Բանաձևի երկրորդ մասը դառնում է զրո ($(1-1)\log...$), և մնում է միայն $-\log(p_i)$: Եթե $p_i$ մոտ է 1-ին, կորուստը փոքր է: 

Եթե մոտ է 0-ին, կորուստը ձգտում է անվերջության:Եթե իրական պատասխանը $y=0$ է:Առաջին մասն է դառնում զրո ($0 \cdot \log...$), և մնում է $-\log(1-p_i)$: Եթե $p_i$ մոտ է 0-ին, կորուստը փոքր է:



Պատկերացնենք՝ մենք ունենք տվյալներ այն մասին, թե քանի ժամ է ուսանողը պատրաստվել քննությանը ($x$): Մեր նպատակն է կանխատեսել արդյունքը ($y$), որտեղ $1$-ը «հանձնելն» է, իսկ $0$-ն՝ «չհանձնելը»:

1. Մոդելի մուտքը (Input)Ենթադրենք՝ ուսանողը սովորել է 15 ժամ:Մոդելն օգտագործում է իր սովորած կշիռները ($w$) և շեղումը ($b$): Ասենք՝ $w = 0.5$ և $b = -5$:Հաշվարկում ենք գծային մասը.$$z = (0.5 \times 15) + (-5) = 7.5 - 5 = 2.5$$

2. Հավանականության հաշվարկ (Sigmoid)Այս $2.5$ արժեքը դեռ հասկանալի չէ: Մենք այն դնում ենք Սիգմոիդ ֆունկցիայի մեջ.$$\sigma(2.5) \approx 0.92$$Սա նշանակում է, որ մոդելը կանխատեսում է 92% հավանականություն, որ ուսանողը կհանձնի քննությունը:
3. Որոշման կայացում (Decision)Քանի որ $0.92 > 0.5$, մոդելը վերջնական կանխատեսում է անում՝ Դաս 1 (Կհանձնի):

Իսկ ի՞նչ դեր ունի Log Loss-ը այստեղ:Պատկերացրու երկու իրավիճակ.

Իրավիճակ Ա: Ուսանողը հանձնեց քննությունը ($y=1$): Մոդելը կանխատեսել էր $0.92$: Կորուստը (Loss) կլինի շատ փոքր, քանի որ մոդելը մոտ էր ճշմարտությանը:

Իրավիճակ Բ: Ուսանողը չհանձնեց ($y=0$): Բայց մոդելը շատ վստահ էր (92%), որ նա կհանձնի: Այս դեպքում Log Loss-ը կտա շատ մեծ «տուգանք», որովհետև մոդելը թույլ է տվել «վստահ սխալ»:

Ուսանողների հազարավոր օրինակների վրա ալգորիթմը անընդհատ հաշվում է այս կորուստները (Log Loss) և փոխում $w$ ու $b$ արժեքները այնքան ժամանակ, մինչև ընդհանուր սխալը դառնա նվազագույն: Արդյունքում ստացվում է այնպիսի «Որոշման սահման» (Decision Boundary), որը լավագույնս բաժանում է հանձնողներին չհանձնողներից:


Օպտիմիզացում (Optimization)

Նպատակը (Goal):$$\min_{w} L(w)$$Մեր նպատակն է գտնել այնպիսի կշիռներ ($w$), որոնք նվազագույնի կհասցնեն կորստի ֆունկցիան ($L$): Այսինքն՝ մոդելը պետք է թույլ տա հնարավորինս քիչ սխալներ:


Գծային ռեգրեսիայում մենք կարող ենք օգտագործել մատրիցային մի բանաձև (Normal Equation) և միանգամից ստանալ պատասխանը: Լոգիստիկ ռեգրեսիայում դա անհնար է:

Գրադիենտի բանաձևը:

$$\nabla_w L(w) = \sum_{i=1}^{N} (\sigma(w^T x_i) - y_i)x_i$$

Սա կորստի ֆունկցիայի ածանցյալն է (գրադիենտը): Այն ցույց է տալիս, թե ինչ ուղղությամբ և որքանով պետք է փոխենք կշիռները, որպեսզի սխալը փոքրանա:

Գրադիենտը զրոյի հավասարեցնելը ($\nabla_w L(w) = 0$) տալիս է ոչ գծային համակարգ, որը չունի անալիտիկ լուծում $w$-ի համար:


Ի՞նչ է նշանակում «չկա անալիտիկ լուծում»:

Դա նշանակում է, որ հնարավոր չէ գրել մի թղթի վրա $w = \dots$ տիպի բանաձև և ստանալ ճիշտ պատասխանը: Քանի որ բանաձևի մեջ կա սիգմոիդ ($\sigma$), հավասարումը դառնում է «խճճված» (ոչ գծային):

Ինչպե՞ս է լուծվում այս խնդիրը:

Քանի որ մենք չենք կարող միանգամից հաշվել $w$-ն, մենք օգտագործում ենք իտերատիվ մեթոդներ, օրինակ՝ Gradient Descent (Գրադիենտային վայրէջք):Մենք սկսում ենք պատահական $w$-ից:Հաշվում ենք գրադիենտը (սխալի ուղղությունը):Մի փոքր քայլ ենք անում հակառակ ուղղությամբ:Կրկնում ենք սա հազարավոր անգամներ, մինչև հասնենք ամենացածր կորստին:

Նմանությունը գծային ռեգրեսիայի հետ:Հետաքրքիր է, որ գրադիենտի վերջնական տեսքը $(\sigma(w^T x_i) - y_i)x_i$ շատ նման է գծային ռեգրեսիայի գրադիենտին: Տարբերությունն այն է, որ այստեղ կանխատեսումը կատարվում է սիգմոիդի միջոցով:


Գրադիենտային վայրէջքը (Gradient Descent):

Ի՞նչ է գրադիենտային վայրէջքը:
Օպտիմիզացման տեխնիկա/ալգորիթմ է: Այն օգտագործվում է մոդելի պարամետրերը (կշիռները) լավացնելու համար:

Հիմնականում օգտագործվում է վերահսկվող մեքենայական ուսուցման (supervised ML) և խորը ուսուցման (deep learning) մեջ:

Կոչվում է նաև առաջին կարգի օպտիմիզացման ալգորիթմ: Սա նշանակում է, որ այն հիմնված է ֆունկցիայի առաջին կարգի ածանցյալի վրա:

ML մոդելներում պարամետրերի օպտիմիզացման համար ամենաշատ օգտագործվող ալգորիթմներից մեկն է:

Տերմինի իմաստը:
Գրադիենտ (Gradient): Նշանակում է առաջին կարգի ածանցյալ կամ կորի թեքություն (slope): Այն ցույց է տալիս, թե որքանով է փոխվում ֆունկցիան տվյալ կետում:

Վայրէջք (Descent): Նշանակում է շարժում դեպի ավելի ցածր կետ:

Ալգորիթմի էությունը: Այն օգտագործում է գրադիենտը/թեքությունը, որպեսզի «իջնի» և հասնի կորստի ֆունկցիայի (loss function) նվազագույն կամ ամենացածր կետին:

Պատկերացրու, որ դու գտնվում ես մի մառախլապատ սարի գագաթին և պետք է իջնես հովիտ (ամենացածր կետը), բայց ոչինչ չես տեսնում: Ի՞նչ ես անում.Ոտքով շոշափում ես քեզնից մի քայլ այն կողմ գտնվող հողը, որպեսզի զգաս թեքությունը:Եթե զգում ես, որ ուղղությունը դեպի ներքև է, մի փոքր քայլ ես անում այդ ուղղությամբ:Կրկնում ես սա այնքան ժամանակ, մինչև շուրջբոլորդ հողը հարթվի (թեքությունը դառնա զրո)։ Դա նշանակում է, որ դու հասել ես հովտի հատակին:Մեքենայական ուսուցման մեջ «սարը» մեր կորստի ֆունկցիան է, իսկ «իջնելը»՝ սխալը նվազեցնելը: Մենք փոփոխում ենք $w$ (կշիռները), որպեսզի հասնենք այնպիսի արժեքների, որոնց դեպքում մոդելը կանի նվազագույն սխալ:

Գրադիենտային վայրէջքի ալգորիթմն իրականացնելիս մեքենան իտերատիվ կերպով (քայլ առ քայլ) հաշվարկում է հաջորդ կետը, որին պետք է հասնի: Դրա համար օգտագործվում է ընթացիկ դիրքի գրադիենտը, որը հանվում է պարամետրի արժեքից՝ նախապես բազմապատկվելով ուսուցման արագությամբ (learning rate):

Մաթեմատիկական տեսքը:

$$p_{n+1} = p_n - l \nabla f(p_n)$$

Որտեղ՝

$p_{n+1}$: Պարամետրի նոր (թարմացված) արժեքը:
$p_n$: Պարամետրի ընթացիկ արժեքը:
$l$: Ուսուցման արագությունը (Learning Rate), որը որոշում է քայլի չափը:
$\nabla f(p_n)$: Սպասվող կորստի ֆունկցիայի գրադիենտը ընթացիկ կետում:

Ինչու՞ հանման նշան ($-$):Գրադիենտը ցույց է տալիս ֆունկցիայի ամենակտրուկ աճի ուղղությունը: Քանի որ մեր նպատակը սխալը (կորուստը) նվազեցնելն է, մենք պետք է շարժվենք գրադիենտի հակառակ ուղղությամբ: Ահա թե ինչու ենք հանում գրադիենտը ընթացիկ արժեքից:

Ինչու՞ է պետք $l$ (Learning Rate):

Առանց սրա մենք կարող ենք անել չափազանց մեծ քայլեր: $l$-ը սովորաբար շատ փոքր թիվ է (օրինակ՝ $0.01$ կամ $0.001$): Այն թույլ է տալիս մեզ զգուշորեն և աստիճանաբար մոտենալ նվազագույն կետին:

Իտերացիայի գաղափարը ($n$ և $n+1$):Սա շարունակական գործընթաց է: Մենք չենք գտնում ճիշտ պատասխանը մեկ քայլով: Ալգորիթմը կատարում է հարյուրավոր կամ հազարավոր նման քայլեր, մինչև $p_{n+1}$ և $p_n$ արժեքների միջև տարբերությունը դառնա աննշան (ինչը նշանակում է, որ հասել ենք «հովտի հատակին»):


Ուսուցման արագություն (Learning Rate)


Ուսուցման արագությունը (LR) կարևոր պարամետր է:

Այն որոշում է այն տեմպը, որով ալգորիթմը սովորում կամ բարելավում է իր անհայտ պարամետրերը:

Միայն գրադիենտները բավարար չեն՝ համոզվելու համար, որ մենք կհասնենք լավագույն լուծմանը (best fit) ցածր հաշվողական ժամանակում:

Ուսուցման պարամետրը ($l$) որոշում է քայլի չափը (step size) յուրաքանչյուր իտերացիայի ժամանակ:

Կարող են լինել երեք դեպքեր, երբ մենք չենք օգտագործում ուսուցման պարամետրը (այսինքն՝ հենվում ենք միայն գրադիենտի վրա)՝
* Գրադիենտը չափազանց փոքր է, և մենք հասնում ենք լավագույն կետին չափազանց շատ իտերացիաների արդյունքում:
* Գրադիենտը չափազանց մեծ է, և մենք «թռչում ենք» լավագույն կետի վրայով՝ ավելի հեռանալով դրանից:
* Գրադիենտը հենց այնքան է, որքան պետք է, և մենք հասնում ենք լավագույն կետին առանց խնդիրների (սա հազվադեպ պատահող հաջողություն է):

Համոզվելու համար, որ քայլի չափը ճիշտ է, մենք օգտագործում ենք ուսուցման պարամետրը, որը մասշտաբավորում (կարգավորում) է պարամետրի նախորդ արժեքի փոփոխությունը՝ նոր արժեք ստանալու համար։

Պատկերացրեք Learning Rate-ը որպես մարդու քայլի երկարություն, որը փորձում է իջնել սարի ստորոտը (նվազագույն սխալի կետը)՝

Եթե LR-ը չափազանց փոքր է (Low Learning Rate):Դուք անում եք մրջյունի քայլեր: Դուք անպայման կհասնեք նպատակին, բայց դա կխլի շատ ժամանակ և հաշվողական ռեսուրս: Սա կոչվում է դանդաղ կոնվերգենցիա (convergence):

Եթե LR-ը չափազանց մեծ է (High Learning Rate):Դուք անում եք հսկայական թռիչքներ: Դուք կարող եք թռչել հովտի մի կողմից մյուսը՝ անցնելով նպատակակետի վրայով: Արդյունքում՝ ալգորիթմը կարող է երբեք չգտնել նվազագույն կետը կամ նույնիսկ սկսել հեռանալ դրանից (divergence):

Իդեալական LR:Այն թույլ է տալիս կատարել այնպիսի քայլեր, որոնք բավականաչափ մեծ են արագ առաջ գնալու համար, բայց բավականաչափ փոքր՝ թիրախը բաց չթողնելու համար:

Գործնականում: Մեքենայական ուսուցման մասնագետները հաճախ փորձարկում են տարբեր արժեքներ (օրինակ՝ $0.1, 0.01, 0.001$), որպեսզի գտնեն այն «ոսկե միջինը», որը թույլ կտա մոդելին սովորել արագ և ճշգրիտ:




Այս պատկերները  բացատրում են Ուսուցման արագության (Learning Rate) ազդեցությունը մոդելի օպտիմիզացման գործընթացի վրա։ Կապույտ կորը ներկայացնում է կորստի ֆունկցիան, իսկ վարդագույն կետերը՝ ալգորիթմի քայլերը դեպի նվազագույն կետը։


1. Լավ LR (Good LR, converges quickly)
Իրավիճակը: Ուսուցման արագությունը ընտրված է օպտիմալ։

Արդյունքը: Քայլերը բավականաչափ մեծ են արագ առաջ գնալու համար, բայց ճշգրիտ կերպով մոտենում են կենտրոնին։ Մոդելը «համամիտվում» է (converges) արագ և հաստատուն։

2. Չափազանց մեծ LR (Large LR, diverges)
Իրավիճակը: Ուսուցման արագությունը չափազանց մեծ է։

Արդյունքը: Ալգորիթմը կատարում է հսկայական թռիչքներ և «թռչում է» նվազագույն կետի վրայով։ Փոխանակ իջնելու, այն սկսում է ավելի ու ավելի բարձրանալ կորի վրայով։ Մոդելը «տարամիտվում» է (diverges), և մենք երբեք լուծում չենք գտնի։

3. Մեծ LR (Large LR, converges slowly)
Իրավիճակը: Արագությունը մեծ է, բայց ոչ կրիտիկական։

Արդյունքը: Ալգորիթմը սկսում է «ցատկոտել» մի կողմից մյուսը։ Չնայած այն ի վերջո հասնում է նվազագույն կետին, սակայն անում է բազմաթիվ ավելորդ քայլեր (զիգզագաձև), ինչը դանդաղեցնում է գործընթացը։

4. Փոքր LR (Small LR, converges slowly)
Իրավիճակը: Ուսուցման արագությունը չափազանց փոքր է։

Արդյունքը: Քայլերը շատ կարճ են (մրջյունի քայլեր)։ Ալգորիթմը շատ զգույշ և հաստատուն իջնում է ներքև, բայց դա խլում է հսկայական ժամանակ և հաշվողական ռեսուրս։ Սա նույնպես համարվում է դանդաղ համամիտում։

Ամփոփում:

Small LR = Շատ դանդաղ, բայց անվտանգ։

Large LR = Կամ շատ քաոսային (զիգզագ), կամ ընդհանրապես ձախողված (divergence)։

Good LR = Հավասարակշռված քայլեր, որոնք ամենաարագն են հասնում նպատակակետին։


Դիտարկենք մաթեմատիկական պայմանները, որոնց պետք է բավարարի ֆունկցիան, որպեսզի հնարավոր լինի կիրառել Գրադիենտային վայրէջքի ալգորիթմը:


Գրադիենտային վայրէջքի ալգորիթմը նախատեսված չէ բոլոր տեսակի ֆունկցիաների համար:

Ֆունկցիան պետք է բավարարի երկու պայմանի, որպեսզի Գրադիենտային վայրէջքը կիրառելի լինի դրա վրա:

Այն պետք է լինի ածանցելի (differentiable):
Սա նշանակում է, որ ֆունկցիայի կորի յուրաքանչյուր կետում մենք պետք է կարողանանք հաշվել ածանցյալը (թեքությունը): Եթե ֆունկցիան ունի կտրուկ անկյուններ կամ ընդհատումներ, ալգորիթմը «չի իմանա», թե որ ուղղությամբ շարժվի:

Այն պետք է լինի ուռուցիկ ֆունկցիա (convex function):
Ուռուցիկ ֆունկցիան (ինչպես ձախակողմյան գրաֆիկում) ունի ընդամենը մեկ նվազագույն կետ (Global Minimum): Սա երաշխավորում է, որ ալգորիթմը միշտ կգտնի լավագույն լուծումը:



Գրաֆիկների բացատրությունը
Ուռուցիկ (Convex): Պատկերացրեք մի թաս կամ ձագար: Որտեղից էլ գնդակը բաց թողնեք, այն միշտ կգլորվի և կհասնի հենց կենտրոնին (ամենացածր կետին): Սա իդեալական վիճակն է Գրադիենտային վայրէջքի համար:

Ոչ ուռուցիկ (Non-convex): Այսպիսի ֆունկցիան ունի մի քանի «հովիտներ» և «բլուրներ»:

Ալգորիթմը կարող է «խցանվել» լոկալ նվազագույնի (Local Minimum) մեջ, որն իրականում ամենացածր կետը չէ:

Արդյունքում մոդելը կարող է սխալ սովորել՝ կարծելով, թե գտել է լավագույն լուծումը, մինչդեռ ավելի լավ տարբերակ գոյություն ունի:


Գրադիենտային վայրէջք Լոգիստիկ ռեգրեսիայի համար

1. Կորստի ֆունկցիան (Cross-Entropy loss):

$$L(w) = -\sum_{i=1}^{N} [y_i \log \sigma(w^T x_i) + (1 - y_i) \log(1 - \sigma(w^T x_i))]$$


Սա մեր չափման միավորն է, որը ցույց է տալիս, թե որքանով է մոդելի կանխատեսած հավանականությունը տարբերվում իրական պատասխանից:

Խնդրը:
$$\min_{w} L(w)$$

Մենք պետք է գտնենք այնպիսի կշիռներ ($w$), որոնք այս կորուստը կդարձնեն հնարավորինս փոքր:

Գրադիենտը գոյություն ունի, ֆունկցիան ուռուցիկ է (ինչը լավ է), բայց $\nabla L(w) = 0$ հավասարումը անալիտիկորեն լուծելն անհնար է: Հետևաբար, մենք օգտագործում ենք օպտիմիզացման տեխնիկա:

Կորստի ֆունկցիայի գրադիենտը:

$$\nabla_w L(w) = \sum_{i=1}^{N} (\sigma(w^T x_i) - y_i)x_i$$

Իտերատիվ թարմացման կանոնը:

$$w^{(t+1)} = w^{(t)} - \eta \sum_{i=1}^{N} (\sigma(w^{(t)T} x_i) - y_i)x_i$$

Մենք սկսում ենք նախնական $w^{(0)}$ արժեքից և քայլ առ քայլ թարմացնում ենք այն՝ օգտագործելով $\eta$ (ուսուցման արագությունը):

Քանի որ մենք չունենք մեկ պարզ բանաձև լավագույն կշիռները միանգամից գտնելու համար, մենք համակարգչին ստիպում ենք «քայլել» կորստի ֆունկցիայի կորի վրայով: Յուրաքանչյուր քայլին ($t, t+1, \dots$) մոդելը նայում է իր սխալին $(\sigma(w^T x_i) - y_i)$, բազմապատկում է այն մուտքային տվյալներով և մի փոքր փոխում իր կշիռները հակառակ ուղղությամբ: Այս գործընթացը շարունակվում է այնքան ժամանակ, մինչև կորուստը հասնի իր նվազագույն արժեքին, և մենք ստանանք լավագույնս պատրաստված մոդելը:




Գրաֆիկի բաղադրիչներըԿարմիր խաչիկներ (Data Points): Սրանք մեր իրական տվյալներն են ($y$): Ինչպես տեսնում ես, դրանք գտնվում են միայն երկու մակարդակի վրա՝ կա՛մ $0$, կա՛մ $1$: Սա բնորոշ է դասակարգման խնդիրներին:

Կանաչ S-աձև կոր (Sigmoid Curve): Սա մոդելի կանխատեսած հավանականությունն է ($p$): Այն ցույց է տալիս, թե ինչպես է $x$-ի աճին զուգընթաց փոխվում $y=1$ լինելու հավանականությունը:

Մաթեմատիկական բանաձևը (Logit Function)Պատկերի կենտրոնում գրված է.$$\ln\left(\frac{p}{1-p}\right) = b_0 + b_1 \cdot x$$

Այս բանաձևը կոչվում է Log-Odds կամ Logit

Odds (Շանսեր): $\frac{p}{1-p}$ հարաբերակցությունն է: Օրինակ, եթե հաղթելու հավանականությունը $0.8$ է, ապա շանսերը կլինեն $0.8 / 0.2 = 4$: Այսինքն՝ «4-ը 1-ի դեմ» հարաբերակցությամբ:

Log-Odds: Երբ մենք վերցնում ենք շանսերի բնական լոգարիթմը ($\ln$), մենք ստանում ենք մի արժեք, որը կարող է լինել ցանկացած թիվ (բացասական անվերջությունից մինչև դրական անվերջություն):

Գծային կապ: Սա թույլ է տալիս մեզ հաստատել գծային կապ անկախ փոփոխականի ($x$) և հավանականության լոգարիթմական արտահայտության միջև ($b_0 + b_1x$):

Լոգիստիկ ռեգրեսիան իրականում «թաքնված» գծային մոդել է: Չնայած գրաֆիկի վրա մենք տեսնում ենք կոր գիծ, մաթեմատիկորեն մենք հաշվարկում ենք գծային կապ $x$-ի և հավանականության լոգարիթմի միջև: Սա օգնում է ալգորիթմին հեշտությամբ գտնել օպտիմալ պարամետրերը ($b_0, b_1$) գրադիենտային վայրէջքի միջոցով:

Ամփոփում: Այս պատկերը ցույց է տալիս, թե ինչպես է լոգիստիկ ռեգրեսիան կապում գծային հավասարումը հավանականային կորի հետ, որպեսզի կատարի ճշգրիտ դասակարգում:




Այս պատկերը ցույց է տալիս Լոգիստիկ ռեգրեսիայի գործնական կիրառումը, որտեղ մոդելը կոնկրետ մուտքային արժեքների ($x$) համար հաշվարկում է համապատասխան հավանականությունները ($\hat{p}$):


Գրաֆիկը պատկերում է, թե ինչպես է փոխվում իրադարձության տեղի ունենալու հավանականությունը՝ կախված $x$ փոփոխականից.
* Երբ $x = 20$: Հավանականությունը շատ փոքր է՝ $0.7\%$: Սա նշանակում է, որ այս կետում մոդելը գրեթե վստահ է, որ արդյունքը կլինի «0» դասը:
* Երբ $x = 30$: Հավանականությունը բարձրանում է մինչև $23\%$: Չնայած աճին, այն դեռ 50%-ից ցածր է, ուստի կանխատեսումը դեռ «0» է:
* Երբ $x = 40$: Հավանականությունը կտրուկ աճում է և հասնում $85\%$: Այստեղ մոդելն արդեն փոխում է իր որոշումը և կանխատեսում է «1» դասը:
* Երբ $x = 50$: Հավանականությունը հասնում է $99.4\%$: Մոդելը գրեթե բացարձակ վստահ է, որ արդյունքը «1» դասն է:

Ոչ գծային աճ: Ուշադրություն դարձրու, թե ինչպես է $x$-ի փոփոխությունը ազդում արդյունքի վրա: $20$-ից $30$ անցնելիս հավանականությունն աճում է ընդամենը $22.3\%$-ով, բայց $30$-ից $40$ անցնելիս աճը կազմում է $62\%$: Սա Սիգմոիդ ֆունկցիայի բնորոշ հատկությունն է՝ կենտրոնական հատվածում փոփոխություններն ավելի կտրուկ են:

Սահմանափակվածություն: Անկախ նրանից, թե որքան կմեծանա $x$-ը, գիծը երբեք չի անցնի $100\%$-ը, և որքան էլ փոքրանա՝ չի իջնի $0\%$-ից ներքև:

Որոշման շեմ (Threshold): Սովորաբար, եթե կորը հատում է $50\%$ հորիզոնական գիծը (այս գրաֆիկում դա տեղի է ունենում $x=30$ և $x=40$ արժեքների արանքում), մոդելը փոխում է իր վերջնական դասակարգումը:

Օրինակ: Եթե սա լիներ բանկային վարկի մոդել, որտեղ $x$-ը հաճախորդի եկամուտն է, ապա 20,000 կամ 30,000 դրամ եկամուտ ունեցողներին վարկը կմերժվեր ($\hat{p} < 50\%$), իսկ 40,000-ից սկսած՝ կհաստատվեր ($\hat{p} > 50\%$):


Այս պատկերը բացատրում է Լոգիստիկ ռեգրեսիայի վերջնական փուլը՝ ինչպես է մոդելը հաշվարկված հավանականությունը վերածում հստակ դասակարգման (0 կամ 1)՝ օգտագործելով որոշման շեմը (Decision Threshold):

Կանխատեսված կախյալ փոփոխական ($\hat{y}$)

Գրաֆիկի վրա տեսնում ենք հետևյալ հիմնական տարրերը`

Կանաչ կոր (Sigmoid): Այն ներկայացնում է հավանականությունը ($p$), որը միշտ գտնվում է 0-ի և 1-ի միջև:

Հորիզոնական գիծ ($0.5$): Սա ստանդարտ Որոշման շեմն է: Այն ծառայում է որպես սահմանագիծ երկու դասերի միջև:

Կապույտ և սպիտակ խաչիկներ: Կապույտ կետերը հաշվարկված հավանականություններն են կորի վրա, իսկ սպիտակ խաչիկները՝ մոդելի վերջնական որոշումը ($\hat{y}$):


Մոդելը հետևում է մի պարզ կանոնի`

Եթե $p < 0.5$ (Ձախ հատված):Կապույտ սլաքները ցույց են տալիս դեպի ներքև:Բոլոր հավանականությունները, որոնք 0.5-ից ցածր են, «կլորացվում» են դեպի $\hat{y} = 0$:Սա տեսնում ենք գրաֆիկի ձախ կողմում գտնվող երկու սպիտակ խաչիկների օրինակով:

Եթե $p \geq 0.5$ (Աջ հատված):Կապույտ սլաքները ցույց են տալիս դեպի վերև:Բոլոր հավանականությունները, որոնք 0.5 կամ ավելի բարձր են, «կլորացվում» են դեպի $\hat{y} = 1$:Սա պատկերված է աջ կողմի երկու սպիտակ խաչիկներով:

Այս պատկերը ցույց է տալիս անցումը հավանականությունից դեպի կոնկրետ դաս: Օրինակ, եթե մոդելը հաշվարկել է, որ հիվանդության հավանականությունը $30\%$ է ($0.3$), ապա քանի որ այն $0.5$-ից փոքր է, մոդելը կանխատեսում է «առողջ» ($0$): Իսկ եթե հավանականությունը $75\%$ է ($0.75$), ապա կանխատեսումը կլինի «հիվանդ» ($1$):



Բազմադաս լոգիստիկ ռեգրեսիա (Multiclass Logistic Regression)

Մինչ այժմ:Երկուական դասակարգում ($y \in \{0, 1\}$):

Մենք որոշում էինք կայացնում միայն երկու տարբերակի միջև (օրինակ՝ «Այո» կամ «Ոչ»):

Այժմ:Բազմաթիվ դասեր ($y \in \{1, 2, \dots, K\}$): 

Մենք աշխատում ենք խնդիրների հետ, որտեղ հնարավոր պատասխանները երկուսից ավելի են (օրինակ՝ թվանշանների ճանաչում $0$-ից $9$, կամ կենդանիների տեսակների որոշում):

Մոդելավորում:

Մենք փորձում ենք հաշվարկել $p(y=k \mid x)$:Սա նշանակում է՝ «Որքա՞ն է հավանականությունը, որ տվյալ $x$ մուտքային փոփոխականը պատկանում է $k$-րդ դասին»: Մենք հաշվում ենք այս հավանականությունը բոլոր $K$ դասերի համար:


Հիմնական գաղափարը (Idea)

Յուրաքանչյուր դասին վերագրել միավոր (Assign a score to each class):Մոդելը յուրաքանչյուր հնարավոր դասի համար հաշվում է մի թիվ (score), որը ցույց է տալիս, թե որքանով է տվյալ օբյեկտը «նման» այդ դասին: Սովորաբար դա արվում է գծային ֆունկցիայի միջոցով ($z_k = w_k^T x$):

Միավորները վերածել հավանականությունների (Convert scores into probabilities):

Քանի որ ստացված միավորները կարող են լինել ցանկացած թիվ (բացասական կամ շատ մեծ), մենք օգտագործում ենք հատուկ ֆունկցիա՝ Softmax, որպեսզի այդ թվերը դարձնենք հավանականություններ:

Պատկերացրեք՝ մոդելին ցույց ենք տալիս կատվի նկար: Մոդելը հաշվում է միավորներ 3 դասի համար`
* Շուն՝ $-1.2$
* Կատու՝ $4.5$
* Թռչուն՝ $0.8$

Softmax ֆունկցիան վերցնում է այս թվերը և դրանք վերածում տոկոսների (հավանականությունների), որոնց գումարը պետք է լինի ուղիղ $1$ (կամ $100\%$): Արդյունքում մենք կստանանք մի պատկեր, որտեղ կատվի հավանականությունը կլինի ամենաբարձրը (օրինակ՝ $92\%$):


Softmax ֆունկցիան հանդիսանում է բազմադաս լոգիստիկ ռեգրեսիայի հիմքը։ Այն թույլ է տալիս մոդելին կանխատեսել հավանականություններ մի քանի դասերի համար միաժամանակ։

Softmax ֆունկցիայի բանաձևը

Բանաձևը սահմանվում է հետևյալ կերպ`

$$p(y = k \mid x) = \frac{e^{w_k^T x}}{\sum_{j} e^{w_j^T x}}$$

Այստեղ $w_k^T x$-ը $k$-րդ դասի համար հաշվարկված միավորն է (score)։ Ֆունկցիան վերցնում է այդ միավորի էքսպոնենտը և բաժանում բոլոր հնարավոր դասերի էքսպոնենտների գումարի վրա։

$w_k^T x$ (Միավոր/Score):
Սա գծային մոդելի արդյունքն է $k$-րդ դասի համար: Այն ցույց է տալիս, թե որքան ուժեղ է մոդելը «կողմ» տվյալ դասին: Սա կարող է լինել ցանկացած իրական թիվ (դրական, բացասական կամ զրո):

$e^{w_k^T x}$ (Էքսպոնենցիալ տրանսֆորմացիա):Մենք վերցնում ենք միավորի էքսպոնենտը երկու պատճառով՝

Դրականություն: Էքսպոնենտը միշտ դրական է, ինչը պարտադիր է հավանականություն հաշվարկելու համար:

Ուժեղացում: Այն մեծացնում է տարբերությունը մեծ և փոքր միավորների միջև՝ ավելի հստակ դարձնելով հաղթող դասի ընտրությունը:

$\sum_{j} e^{w_j^T x}$ (Հայտարարը/Նորմալացում):
Սա բոլոր հնարավոր դասերի ($j=1, \dots, K$) էքսպոնենտների գումարն է: Բաժանելով այս ընդհանուր գումարի վրա՝ մենք ստիպում ենք, որ բոլոր հավանականությունների գումարը լինի ուղիղ $1$:


Պատկերացնենք՝ ունենք 3 դաս (օրինակ՝ նկարում պատկերված է կատու, շուն կամ թռչուն) և մոդելը տվել է հետևյալ միավորները (scores)՝
* Կատու ($w_1^T x$): $2.0$
* Շուն ($w_2^T x$): $1.0$
* Թռչուն ($w_3^T x$): $0.1$

Քայլ 1: Հաշվում ենք էքսպոնենտները

* $e^{2.0} \approx 7.39$
* $e^{1.0} \approx 2.72$
* $e^{0.1} \approx 1.11$

Քայլ 2: Հաշվում ենք ընդհանուր գումարը (Հայտարարը)

$Sum = 7.39 + 2.72 + 1.11 = 11.22$

Քայլ 3: Հաշվում ենք վերջնական հավանականությունները

* Կատու: $7.39 / 11.22 \approx \mathbf{0.66}$ ($66\%$)
* Շուն: $2.72 / 11.22 \approx \mathbf{0.24}$ ($24\%$)
* Թռչուն: $1.11 / 11.22 \approx \mathbf{0.10}$ ($10\%$)

Հիմնական հատկությունները (Properties)

Արդյունքները միշտ դրական են։ Քանի որ էքսպոնենտը ($e^z$) միշտ դրական թիվ է, ստացված բոլոր հավանականությունները լինելու են $> 0$։

Հավանականությունների գումարը հավասար է 1-ի. Սա ամենակարևոր հատկությունն է։ Բաժանելով ընդհանուր գումարի վրա՝ մենք ստանում ենք արժեքների բաշխում, որոնց գումարը տալիս է ուղիղ $1$ (կամ $100\%$)։

Sigmoid-ի ընդհանրացումը՝ Softmax-ը հանդիսանում է սիգմոիդ ֆունկցիայի մաթեմատիկական ընդլայնումը երկու դասից ավելի դեպքերի համար։

Բարձր միավոր $\rightarrow$ Բարձր հավանականություն. Որքան մեծ է որոշակի դասի համար հաշվարկված $w_k^T x$ արժեքը, այնքան ավելի մեծ կլինի Softmax-ի տված հավանականությունը հենց այդ դասի համար։




Բազմադաս (Multiclass) լոգիստիկ ռեգրեսիայի ուսուցման (Training) և կանխատեսման (Prediction) գործընթացը  ցույց է տալիս, թե ինչպես է մոդելը գնահատում իր սխալը և ինչպես է ընտրում վերջնական դասը։

1. Կորստի ֆունկցիա (Cross-Entropy Loss)Բազմադաս դեպքում մենք օգտագործում ենք խաչաձև էնտրոպիայի ընդհանրացված տարբերակը։ Բանաձևն է՝ $$\mathcal{L} = -\sum_{i} \sum_{k} y_{ik} \log p_{ik}$$

$\sum_{i}$: Գումարումն ըստ բոլոր ուսուցանվող օրինակների ($i$)։

$\sum_{k}$: Գումարումն ըստ բոլոր հնարավոր դասերի ($k$)։

$y_{ik}$: Սա կոչվում է One-Hot Encoding։ Այն հավասար է $1$-ի, եթե $i$ օրինակն իրականում պատկանում է $k$ դասին, և $0$՝ հակառակ դեպքում։

$p_{ik}$: Սա Softmax ֆունկցիայի միջոցով հաշվարկված հավանականությունն է, որ $i$ օրինակը պատկանում է $k$ դասին։

($\log$):Լոգարիթմն օգնում է «չափել» սխալը այնպես, որ այն համապատասխանի մեր կանխատեսումների վստահությանը.Սխալի ուժեղացում: Երբ մոդելը շատ վստահ է սխալ պատասխանի վրա (օրինակ՝ ճիշտ դասի հավանականությունը տալիս է $0.01$), լոգարիթմի արժեքը դառնում է շատ մեծ թիվ: Սա թույլ է տալիս մոդելին «խստորեն պատժել» նման սխալների համար:

Մաթեմատիկական հարմարավետություն: Հավանականությունները բազմապատկելու փոխարեն (ինչը կարող է հանգեցնել շատ փոքր թվերի), լոգարիթմը թույլ է տալիս դրանք գումարել, ինչն ավելի հեշտ է հաշվարկների և ածանցման համար:

($-1$):Մինուս նշանը պարզապես փոխում է արժեքի ուղղությունը՝

Մաքսիմալացումից դեպի մինիմալացում: Հավանականության լոգարիթմը ($0$-ից $1$ միջակայքում) միշտ բացասական թիվ է: Քանի որ մենք ուզում ենք մինիմալացնել կորուստը (Loss), մենք բազմապատկում ենք այն $-1$-ով, որպեսզի ստանանք դրական արժեք:

Տրամաբանություն: Որքան մեծ է ճիշտ դասի հավանականությունը ($p_{ik}$), այնքան փոքր կլինի վերջնական կորուստը ($\mathcal{L}$): Եթե մոդելը կանխատեսում է $100\%$ հավանականություն ճիշտ դասի համար, ապա $\log(1) = 0$, և կորուստը դառնում է $0$:
Քանի որ $y_{ik}$-ն $1$ է միայն ճիշտ դասի համար, բանաձևը գործնականում հաշվի է առնում միայն ճիշտ դասի համար կանխատեսված հավանականության լոգարիթմը։ Մոդելը փորձում է մինիմալացնել այս արժեքը, ինչը նշանակում է առավելագույնի հասցնել ճիշտ դասի հավանականությունը։

Կանխատեսում (Prediction)

Երբ մոդելն արդեն սովորել է, կանխատեսումը կատարվում է հետևյալ կերպ՝

$$\text{class} = \arg\max_k p(y=k|x)$$

$\arg\max_k$: Սա նշանակում է՝ «ընտրել այն $k$ դասի ինդեքսը, որի համար հավանականությունն ամենամեծն է»։

$p(y=k|x)$: Սա Softmax-ի տված հավանականությունն է յուրաքանչյուր դասի համար։

Եթե ունենք երեք դաս և հաշվարկված հավանականություններն են՝ $0.1, 0.7, 0.2$, ապա $\arg\max$-ը կընտրի երկրորդ դասը ($0.7$), քանի որ այն ամենամեծն է։


Փոքրագույն քառակուսիների մեթոդը (Least Squares)։

Դիտարկենք այն դեպքը երբ դասակարգման (classification) խնդիրը փորձում ենք լուծել ռեգրեսիայի մեթոդներով՝ օգտագործելով Փոքրագույն քառակուսիների մեթոդը (Least Squares)։

1-of-K կոդավորման սխեմա:Դիտարկում ենք դասակարգման խնդիր $K$ հատ դասերով։ Յուրաքանչյուր թիրախային վեկտոր $t$ կոդավորվում է այնպես, որ այն ունի $1$ արժեքը միայն իր դասին համապատասխանող դիրքում, իսկ մնացած տեղերում՝ $0$ (սա նույն One-Hot Encoding-ն է)։

Գծային մոդել յուրաքանչյուր դասի համար:

Յուրաքանչյուր $C_k$ դաս նկարագրվում է իր սեփական գծային մոդելով՝

$$y_k(\mathbf{x}) = \mathbf{w}_k^T \mathbf{x} + w_{k0}$$

Այստեղ $y_k(\mathbf{x})$-ը փորձում է կանխատեսել $k$-րդ դասին պատկանելու «միավորը»։

Եթե մենք միավորենք բոլոր $K$ դասերի կշիռները մեկ ընդհանուր $\widetilde{\mathbf{W}}$ մատրիցի մեջ, կստանանք հետևյալ հակիրճ բանաձևը՝

$$\mathbf{y}(\mathbf{x}) = \widetilde{\mathbf{W}}^T \widetilde{\mathbf{x}}$$

Այստեղ $\widetilde{\mathbf{x}}$-ը ընդլայնված մուտքային վեկտորն է (որը ներառում է նաև հաստատուն մասը՝ $w_{k0}$-ն)։

Ֆունկցիայի տեսակը: Ի տարբերություն Լոգիստիկ ռեգրեսիայի (որն օգտագործում է Սիգմոիդ կամ Softmax), այս մոդելը չունի ոչ գծային սեղմող ֆունկցիա։ Արդյունքը կարող է լինել ցանկացած թիվ, ոչ միայն $0$-ից $1$ միջակայքում։

Կորստի ֆունկցիա: Այստեղ օգտագործվում է քառակուսային սխալը (MSE), այլ ոչ թե Cross-Entropy-ն։

Խնդիրը: Փոքրագույն քառակուսիների մեթոդը դասակարգման համար շատ զգայուն է «արտաքին շեղումների» (outliers) նկատմամբ և կարող է վատ աշխատել, եթե տվյալները խիստ տարանջատված չեն։

Գործնականում: Չնայած այս մեթոդն ունի անալիտիկ (ճշգրիտ) լուծում և արագ է հաշվարկվում, այն հազվադեպ է օգտագործվում ժամանակակից դասակարգման խնդիրներում, քանի որ լոգիստիկ ռեգրեսիան և Softmax-ը տալիս են շատ ավելի ճշգրիտ հավանականային արդյունքները։



Փոքրագույն քառակուսիների մեթոդի (Least Squares) մաթեմատիկական լուծումը դասակարգման խնդիրների համար։ 

Այստեղ ցույց է տրվում, թե ինչպես կարելի է անալիտիկ ճանապարհով գտնել օպտիմալ կշիռները։

Մենք հաշվարկում ենք գործակիցների մատրիցը՝ մինիմալացնելով քառակուսային սխալների գումարը՝
$$\frac{1}{2} \text{Tr} \{ (\widetilde{\mathbf{X}} \widetilde{\mathbf{W}} - \mathbf{T})^T (\widetilde{\mathbf{X}} \widetilde{\mathbf{W}} - \mathbf{T}) \}$$

$\widetilde{\mathbf{X}}$: Մուտքային տվյալների մատրիցն է (design matrix)։

$\widetilde{\mathbf{W}}$: Կշիռների մատրիցն է, որը մենք փնտրում ենք։

$\mathbf{T}$: Թիրախային արժեքների մատրիցն է (տվյալ օրինակի ճիշտ դասերը՝ One-Hot կոդավորմամբ)։

$\text{Tr}$ (Trace): Մատրիցի հետքն է (անկյունագծային տարրերի գումարը)։ Մատրիցային տեսքով սա համարժեք է բոլոր տարրերի քառակուսային սխալների գումարին։

Օպտիմալ լուծումը (Normal Equations)

Ածանցելով ըստ կշիռների մատրիցի ($\widetilde{\mathbf{W}}$) և արդյունքը հավասարեցնելով զրոյի, ստանում ենք կշիռների ուղիղ հաշվարկման բանաձևը՝

$$\widetilde{\mathbf{W}} = (\widetilde{\mathbf{X}}^T \widetilde{\mathbf{X}})^{-1} \widetilde{\mathbf{X}}^T \mathbf{T} = \widetilde{\mathbf{X}}^\dagger \mathbf{T}$$

$\widetilde{\mathbf{X}}^\dagger$: Սա կոչվում է Մուր-Պենրոուզի կեղծ հակադարձ մատրից (Moore-Penrose pseudo-inverse)։Սա նշանակում է, որ ի տարբերություն Լոգիստիկ ռեգրեսիայի, որտեղ մենք օգտագործում ենք Գրադիենտային վայրէջք (Gradient Descent), այստեղ լուծումը կարելի է գտնել մեկ քայլով՝ մատրիցային գործողությունների միջոցով։

Վերջնական կանխատեսող ֆունկցիան ունի հետևյալ տեսքը՝

$$\mathbf{y}(\mathbf{x}) = \widetilde{\mathbf{W}}^T \widetilde{\mathbf{x}}$$

Այս ֆունկցիան վերադարձնում է վեկտոր, որի յուրաքանչյուր տարր համապատասխանում է մի դասի։ Մենք ընտրում ենք այն դասը, որի արժեքը ստացված վեկտորում ամենամեծն է։


Ձախ պատկերը (Նորմալ վիճակ)
Այստեղ տվյալները երկու խմբի են (կարմիր «x» և կապույտ «o»)։

Կանաչ գիծը լոգիստիկ ռեգրեսիայի (կամ SVM-ի) սահմանն է, իսկ մանուշակագույնը՝ Least Squares-ինը։

Ինչպես տեսնում ես, երկու գծերն էլ գրեթե նույն տեղում են և շատ լավ բաժանում են խմբերը։

Աջ պատկերը (Outliers - Շեղվող կետեր)
Ահա այստեղ է երևում խնդիրը։ Նկարի ներքևի աջ անկյունում հայտնվել են նոր կապույտ կետեր («o»), որոնք արդեն իսկ ճիշտ դասակարգված են (դրանք հեռու են բաժանարար գծից և գտնվում են կապույտների տարածքում)։

Կանաչ գիծը (Լոգիստիկ ռեգրեսիա): Այն գրեթե չի փոխվել։ Մոդելը «հասկանում է», որ այդ նոր կետերը կապույտ են և արդեն ճիշտ կողմում են, ուստի կարիք չկա գիծը տեղաշարժել։

Մանուշակագույն գիծը (Least Squares): Այն խիստ թեքվել է դեպի աջ։ Քանի որ Least Squares-ը փորձում է մինիմալացնել հեռավորության քառակուսին, նա տեսնում է, որ այդ նոր կապույտ կետերը շատ հեռու են իրենից և փորձում է «մոտենալ» դրանց։

Least Squares-ը չափազանց զգայուն է «արտաքին շեղումների» (outliers) նկատմամբ։ Այն փորձում է լինել հնարավորինս մոտ բոլոր կետերին, նույնիսկ եթե դա փչացնում է բաժանարար սահմանը։ Իսկ Լոգիստիկ ռեգրեսիան ավելի կայուն է, քանի որ նրա համար կարևոր է միայն, որ կետերը լինեն սահմանի ճիշտ կողմում։



Ձախ պատկերը (Least Squares Discriminant)
Խնդիրը: Դու կարող ես տեսնել երեք դաս՝ կարմիր (x), կանաչ (+) և կապույտ (o):

Արդյունքը: Փոքրագույն քառակուսիների մեթոդը փորձում է գծային սահմաններ գտնել, բայց կանաչ դասի համար նախատեսված տարածքը ստացվել է չափազանց նեղ:

Հետևանքը: Կանաչ կետերի մեծ մասը սխալմամբ հայտնվել են կարմիր կամ կապույտ գոտիներում: Սա նշանակում է, որ մոդելը «չի տեսնում» կանաչ դասը որպես առանձին տիրույթ, քանի որ այն գտնվում է մյուս երկուսի արանքում:

Least Squares-ը փորձում է կանխատեսել թիրախային արժեքները ($0$ կամ $1$) որպես իրական թվեր: Երբ դասերը դասավորված են այնպես, որ մեկը գտնվում է մյուսների միջև, քառակուսային սխալը մինիմալացնելիս «միջանկյալ» դասի սահմանները հաճախ սեղմվում կամ անհետանում են:


Աջ պատկերը (Logistic Regression)

Լուծումը: Այստեղ օգտագործվել է լոգիստիկ ռեգրեսիան (Softmax-ի սկզբունքով):

Արդյունքը: Մոդելը հստակորեն առանձնացրել է երեք տիրույթներն էլ: Կանաչ դասը ստացել է իր սեփական լայն գոտին, և բոլոր կետերը ճիշտ են դասակարգվել:

Logistic Regression-ը օգտագործում է հավանականություններ և լոգարիթմական կորուստ (Cross-Entropy), ինչը թույլ է տալիս մոդելին ավելի ճկուն լինել և յուրաքանչյուր դասի համար ստեղծել իրեն համապատասխան որոշման տիրույթը:

Ֆիշերի գծային դիսկրիմինանտին (Fisher’s Linear Discriminant)

Եթե նախորդ մեթոդները փորձում էին ուղղակի գիծ քաշել դասերի արանքում, ապա Ֆիշերի մոտեցումը մի փոքր այլ է՝ այն հիմնված է չափողականության նվազեցման (dimensionality reduction) գաղափարի վրա:


Ֆիշերի գծային դիսկրիմինանտի հիմնական գաղափարը

Պրոյեկցիա մեկ չափողականության վրա:
Մենք ունենք բազմաչափ տվյալներ (օրինակ՝ երկչափ հարթության վրա), և ուզում ենք դրանք «պրոյեկտել» (իջեցնել) մի ուղիղ գծի վրա՝ օգտագործելով $y = \mathbf{w}^T\mathbf{x}$ բանաձևը:

Դասակարգման շեմը:
Եթե պրոյեկցիայի արդյունքում ստացված $y$ արժեքը մեծ է ինչ-որ $w_0$ շեմից, ապա կետը վերագրում ենք $C_1$ դասին, հակառակ դեպքում՝ $C_2$ դասին: Սա մեզ տալիս է ստանդարտ գծային դասակարգիչ:

Ինչու՞ է սա խնդրահարույց և ինչպե՞ս է լուծվում
Ինֆորմացիայի կորուստ:
Երբ մենք բազմաչափ տվյալները սեղմում ենք մինչև մեկ չափողականություն, տեղի է ունենում ինֆորմացիայի զգալի կորուստ:

Դասերի վերածածկում (Overlap):
Նույնիսկ եթե դասերը բազմաչափ տարածության մեջ լավ առանձնացված են, սխալ ընտրված պրոյեկցիայի դեպքում մեկ չափողականության վրա դրանք կարող են խառնվել իրար (վերածածկվել):


Ֆիշերի լուծումը

Մենք կարող ենք կարգավորել $\mathbf{w}$ կշիռների վեկտորը այնպես, որ ընտրենք այնպիսի պրոյեկցիա, որը մաքսիմալացնում է դասերի միջև տարանջատումը (class separation):

Պարզ ասած. Ֆիշերի մեթոդը փնտրում է այնպիսի ուղղություն, որի վրա պրոյեկտելիս՝Դասերի կենտրոնները լինեն իրարից հնարավորինս հեռու:

Յուրաքանչյուր դասի ներսում կետերը լինեն հնարավորինս խիտ (փոքր ցրվածությամբ):Սա օգնում է խուսափել այն խնդիրներից, որոնք տեսանք Least Squares-ի դեպքում, քանի որ այստեղ մենք հաշվի ենք առնում տվյալների բաշխվածությունը (կովարիացիան) յուրաքանչյուր դասի ներսում:


Ֆիշերի գծային դիսկրիմինանտի առաջին քայլը՝ ինչպես չափել դասերի միջև հեռավորությունը պրոյեկցիայից հետո:

1. Դասերի միջին արժեքները (Class Means)

Նախ, դիտարկում ենք երկու դաս՝ $C_1$ և $C_2$, որտեղ $N_1$-ը և $N_2$-ը համապատասխանաբար կետերի քանակն են յուրաքանչյուր դասում: Մենք հաշվում ենք յուրաքանչյուր դասի կենտրոնը (միջին վեկտորը) սկզբնական բազմաչափ տարածությունում՝

$$\mathbf{m}_1 = \frac{1}{N_1} \sum_{n \in C_1} \mathbf{x}_n, \quad \mathbf{m}_2 = \frac{1}{N_2} \sum_{n \in C_2} \mathbf{x}_n$$


2. Պրոյեկտված միջիններ (Projected Means)

Երբ մենք տվյալները պրոյեկտում ենք $\mathbf{w}$ ուղղությամբ ուղղի վրա, յուրաքանչյուր դասի միջին արժեքը այդ ուղղի վրա դառնում է սկալյար թիվ ($m_k$)
$$m_k = \mathbf{w}^T \mathbf{m}_k$$

3. Տարանջատման չափանիշը (Separation Measure)

Դասերի միջև տարանջատումը չափելու ամենապարզ ձևը պրոյեկտված միջինների տարբերությունն է: Մենք ցանկանում ենք ընտրել այնպիսի $\mathbf{w}$ ուղղություն, որը կմաքսիմալացնի այդ տարբերությունը՝  $$m_2 - m_1 = \mathbf{w}^T (\mathbf{m}_2 - \mathbf{m}_1)$$

Չնայած միջինների տարբերությունը մաքսիմալացնելը տրամաբանական է թվում, սակայն դա բավարար չէ: Խնդիրն այն է, որ եթե դասերի ներսում կետերը շատ ցրված են, նույնիսկ եթե կենտրոնները հեռու լինեն, պրոյեկցիաները միևնույն է կարող են վերածածկվել (overlapping):

Հենց սա է պատճառը, որ Ֆիշերն առաջարկեց հաշվի առնել նաև դասի ներսում ցրվածությունը (variance):

Սակայն միջինների հեռավորությունը ($m_2 - m_1$) մաքսիմալացնելը բավարար չէ լավ դասակարգիչ ստանալու համար: Ահա հիմնական կետերը։ 

Եթե մենք փորձենք մաքսիմալացնել $m_2 - m_1 = \mathbf{w}^T(\mathbf{m}_2 - \mathbf{m}_1)$ արտահայտությունը առանց սահմանափակումների, մենք կարող ենք ստանալ անվերջ մեծ թիվ՝ պարզապես մեծացնելով $\mathbf{w}$ վեկտորի երկարությունը:

Այս խնդիրը լուծելու համար դրվում է սահմանափակում՝ $\mathbf{w}$ վեկտորը պետք է ունենա միավոր երկարություն ($||\mathbf{w}||_2 = 1$):

Օգտագործելով Լագրանժի բազմապատկիչների մեթոդը, ստացվում է, որ լավագույն ուղղությունը պարզապես միջինների տարբերությանն ուղղահայաց վեկտորն է՝ $\mathbf{w} \propto (\mathbf{m}_2 - \mathbf{m}_1)$:


Սակայն նույնիսկ եթե երկչափ տարածությունում դասերը լավ առանձնացված են, միայն միջինները միացնող գծի վրա պրոյեկտելը կարող է բերել դասերի զգալի վերածածկման (overlap):
Սա տեղի է ունենում այն պատճառով, որ այս մոտեցումը հաշվի չի առնում դասերի ներսում տվյալների տարածվածությունը կամ ձևը:

Խնդրի հիմնական պատճառը դասերի բաշխվածության կովարիացիաներն են:
Եթե տվյալները դասավորված են «էլիպսաձև» (ոչ թե կլոր) և այդ էլիպսները թեքված են, ապա միջինների ուղղությամբ պրոյեկտելիս կետերը կխառնվեն իրար:

Սա նշանակում է, որ լավ պրոյեկցիան պետք է հաշվի առնի ոչ միայն այն, թե որտեղ են դասերի կենտրոնները, այլ նաև այն, թե ինչպես են կետերը ցրված այդ կենտրոնների շուրջը:




Ձախ պատկերը. Պրոյեկցիա միջինները միացնող գծի վրա

Ի՞նչ է կատարվում. Այստեղ $\mathbf{w}$ ուղղությունն ընտրվել է այնպես, որ այն զուգահեռ լինի երկու դասերի կենտրոնները (միջինները) միացնող գծին:

Խնդիրը. Թեև կենտրոնների միջև հեռավորությունը մաքսիմալ է, դասերի ներսում տվյալների բաշխվածությունը (էլիպսաձև տեսքը) այնպիսին է, որ պրոյեկցիայի արդյունքում (ներքևի հիստոգրամը) կարմիր և կապույտ գոտիները ուժեղ վերածածկվում են (overlap):


Արդյունքը. Միայն միջինների վրա հիմնված այս դասակարգիչը շատ սխալներ կթույլ տա:


Աջ պատկերը. Ֆիշերի գծային դիսկրիմինանտ

Ի՞նչ է կատարվում. Այստեղ հաշվի է առնվել նաև դասերի ներքին կովարիացիան (ցրվածությունը): $\mathbf{w}$ ուղղությունը թեքվել է այնպես, որ պրոյեկտված կետերը լինեն հնարավորինս խիտ իրենց խմբերում

:Լուծումը. Ինչպես տեսնում ես հիստոգրամի վրա, կարմիր և կապույտ խմբերը հիմա լիովին առանձնացված են:

Արդյունքը. Չնայած այս ուղղության վրա միջինների միջև հեռավորությունը կարող է ավելի փոքր լինել, քան ձախ նկարում, սակայն դասակարգումը կատարվում է անթերի, քանի որ վերածածկում չկա:


Ֆիշերի չափանիշի (Fisher criterion) մաթեմատիկական ձևակերպմանը, որը լուծում է նախորդ օրինակներում տեսած «վերածածկման» խնդիրը:

Ֆիշերի գաղափարն է մաքսիմալացնել մի ֆունկցիա, որը միաժամանակ կատարում է երկու գործողություն`
* Մաքսիմալացնում է հեռավորությունը պրոյեկտված դասերի միջինների միջև (large separation between projected class means):
* Մինիմալացնում է ցրվածությունը յուրաքանչյուր դասի ներսում (small variance within each class):
Այս երկու գործողությունների համատեղումը թույլ է տալիս հասցնել դասերի վերածածկումը նվազագույնի:


 Ներդասային ցրվածություն (Within-class variance)
 
 Յուրաքանչյուր $C_k$ դասի համար հաշվարկվում է պրոյեկտված տվյալների շեղումը տվյալ դասի միջինից ($m_k$):

 $$s_k^2 = \sum_{n \in C_k} (y_n - m_k)^2$$

 Սա ցույց է տալիս, թե որքանով են «ցրված» կամ «խիտ» դասավորված կետերը պրոյեկտված ուղղի վրա:

 Ֆիշերի չափանիշը (The Fisher Criterion)

 Ֆիշերի չափանիշը ($J(\mathbf{w})$) սահմանվում է որպես միջդասային հեռավորության քառակուսու հարաբերությունը ներդասային ցրվածությունների գումարին:

 $$J(\mathbf{w}) = \frac{(m_2 - m_1)^2}{s_1^2 + s_2^2}$$

 Համարիչը (Numerator): Երբ այն մեծ է, դասերի կենտրոնները իրարից հեռու են:

 Հայտարարը (Denominator): Երբ այն փոքր է, յուրաքանչյուր դասի կետերը շատ խիտ են հավաքված իրենց կենտրոնի շուրջը:
 
 Արդյունքը: Մենք փնտրում ենք այնպիսի $\mathbf{w}$ վեկտոր, որը կտա $J(\mathbf{w})$-ի հնարավոր առավելագույն արժեքը:

Այս չափանիշը պետք է վերաձևակերպվի $\mathbf{w}$ վեկտորի նկատմամբ, որպեսզի հնարավոր լինի գտնել լավագույն ուղղությունը մատրիցային տեսքով:

1. Ցրվածության մատրիցներ (Scatter Matrices)

Որպեսզի կարողանանք հաշվել $J(\mathbf{w})$-ն կամայական $\mathbf{w}$ ուղղության համար, մենք ներմուծում ենք երկու հիմնական մատրից։
Միջդասային ցրվածության մատրից ($\mathbf{S}_B$ - Between-class scatter matrix):Այն բնութագրում է դասերի կենտրոնների միջև հեռավորությունը սկզբնական տարածությունում:

$$\mathbf{S}_B = (\mathbf{m}_2 - \mathbf{m}_1)(\mathbf{m}_2 - \mathbf{m}_1)^T$$

Սա արտաքին արտադրյալ է (outer product), որը ցույց է տալիս միջինների տարբերության ուղղությունը:


Ներդասային ցրվածության մատրից ($\mathbf{S}_W$ - Within-class scatter matrix):Այն բնութագրում է տվյալների ընդհանուր ցրվածությունը յուրաքանչյուր դասի ներսում:

$$\mathbf{S}_W = \sum_{n \in C_1} (\mathbf{x}_n - \mathbf{m}_1)(\mathbf{x}_n - \mathbf{m}_1)^T + \sum_{n \in C_2} (\mathbf{x}_n - \mathbf{m}_2)(\mathbf{x}_n - \mathbf{m}_2)^T$$

Սա երկու դասերի կովարիացիոն մատրիցների գումարն է (առանց նորմավորման):


2. Ֆիշերի չափանիշի վերաձևակերպումը

Օգտագործելով այս մատրիցները, մենք կարող ենք վերաշարադրել $J(\mathbf{w})$ ֆունկցիան հետևյալ տեսքով:

$$J(\mathbf{w}) = \frac{\mathbf{w}^T \mathbf{S}_B \mathbf{w}}{\mathbf{w}^T \mathbf{S}_W \mathbf{w}}$$

Այս տեսքը կոչվում է Ռելեի հարաբերություն (Rayleigh quotient):

Մեր նպատակն է մաքսիմալացնել $J(\mathbf{w})$-ն: Եթե մենք ածանցենք այս արտահայտությունը ըստ $\mathbf{w}$-ի և հավասարեցնենք զրոյի, կտեսնենք, որ օպտիմալ լուծումը բավարարում է հետևյալ պայմանին:

$$\mathbf{S}_B \mathbf{w} = \lambda \mathbf{S}_W \mathbf{w}$$

Քանի որ երկու դասի դեպքում $\mathbf{S}_B \mathbf{w}$ վեկտորը միշտ ուղղված է լինում $(\mathbf{m}_2 - \mathbf{m}_1)$ ուղղությամբ, մենք կարող ենք ստանալ վերջնական պարզ լուծումը:

$$\mathbf{w} \propto \mathbf{S}_W^{-1} (\mathbf{m}_2 - \mathbf{m}_1)$$

Եթե դասի ներսում ցրվածությունը բոլոր ուղղություններով նույնն է (իզոտրոպ է), ապա $\mathbf{S}_W$-ն նման կլինի միավոր մատրիցի, և $\mathbf{w}$-ն ուղղակի կլինի միջինների տարբերությունը: Բայց եթե տվյալները «ձգված» են , ապա $\mathbf{S}_W^{-1}$ գործակիցը «պտտում» է $\mathbf{w}$ վեկտորը այնպես, որ այն դառնա ուղղահայաց ցրվածության ամենամեծ առանցքին՝ ապահովելով դասերի մաքսիմալ տարանջատում:
